This repository contains an **independent**, professional-grade Python implementation of the research methodology from the **arXiv preprint** entitled **"How Simple Can It Get? From Interpretable Equations to Readable Rules for Financial Decision Making"** by:

*   **Adia Lumadjeng** (University of Amsterdam)
*   **Ilker Birbil** (University of Amsterdam)
*   **Erman Acar** (University of Amsterdam)

The paper provides no code; this project translates its mathematical specifications into robust, versioned, executable software artifacts. The project delivers a **decision-grade explanation and governance system** that converts an already interpretable but operationally unreadable learned equation — an ECSEL single-term monomial — into the simplest human-actionable representation appropriate to each stakeholder (customer, loan officer, fraud investigator, model validator, or regulator), with the predictive, fidelity, and calibration costs of that conversion explicitly quantified and, where possible, predicted before deployment.

## Introduction

This project provides a Python implementation of the analytical framework presented in Lumadjeng, Birbil, and Acar (2026). The core of this repository is the iPython Notebook `from_interpretable_equations_to_readable_rules_for_financial_decision_making_draft.ipynb`, which contains a comprehensive suite of functions (Tasks 1 to 16-plus) to replicate the paper's findings. The pipeline studies the **reverse direction** of interpretability: starting from an already-interpretable classifier expressed as a single equation over the input features, it progressively simplifies the equation into more readable forms — a pruned monomial, a directional if–then rule, and the integer scorecards and tallies that finance already deploys — and directly quantifies what is lost under each simplification, because the equation is itself the predictive model rather than a post-hoc explanation.

This codebase operationalizes the paper's framework, allowing users to:

-   Rigorously validate the four raw financial datasets and the entire experimental configuration against the paper's constants via a single `config.yaml` file (single source of truth; every `UNSPECIFIED IN MANUSCRIPT` parameter carries an explicit replicator placeholder).
-   Cleanse raw data with leakage-free, training-only decisions and freeze the cleansed state in checksummed artifacts.
-   Construct deterministic feature matrices of exactly **42** (`Loan`), **6** (`FraudEcom`), **26** (`Default`), and **30** (`Creditcard`) columns.
-   Create five stratified 60/20/20 train/validation/test splits (seeds `[0, 1, 2, 3, 4]`) and apply the train-only min–max scaling $\tilde{x}_j = 0.01 + 10.0\,\frac{x_j - \min_j}{\max_j - \min_j}$ into $[0.01, 10.01]$.
-   Select learning rate, $\ell_1$ strength, and training length with Optuna (TPE sampler, 100 trials, five-fold cross-validated PR-AUC on the training split).
-   Train the ECSEL single-term monomial $z(x) = \alpha \prod_j x_j^{\beta_j}$ with automatic differentiation on a numerically stable log-space objective, verifying the log-space identity $\log|z(x)| = \log|\alpha| + \sum_j \beta_j \log x_j$ to within $10^{-6}$.
-   Prune to the smallest support whose validation PR-AUC stays within $0.01$ of the full monomial and verify the worst-case bound $|s_{\mathrm{full}} - s_{\mathrm{pruned}}| \le L\sum_{j \in S}|\beta_j|$ with $L = \max\{|\log 0.01|, |\log 10.01|\}$.
-   Build the directional rule $s_{\mathrm{dir}}(x) = \sum_{j \in S}\mathrm{sign}(\beta_j)\log x_j$ and predict its rank fidelity a-priori via $\rho = \frac{\beta^\top \Sigma d}{\sqrt{\beta^\top \Sigma \beta}\sqrt{d^\top \Sigma d}}$ and Greiner's relation $\tau_{\mathrm{pred}} = \frac{2}{\pi}\arcsin(\rho)$.
-   Construct integer scorecards $V(x) = \sum_j p_j\mathbf{1}_j(x)$ with $p_j = \max(1, \mathrm{round}(5|\beta_j|/\max_k|\beta_k|))$ and tallies $T(x) = \sum_j \mathbf{1}_j(x)$ with validation-selected F1 thresholds.
-   Calibrate all five representations with validation-fit isotonic regression and report the expected calibration error $\mathrm{ECE} = \sum_b \frac{|b|}{n}|\mathrm{acc}(b) - \mathrm{conf}(b)|$ on the test set.
-   Measure tie-corrected rank fidelity $\tau_b = \frac{C - D}{\sqrt{(C + D + T_s)(C + D + T_f)}}$ between every simplified form and the full monomial.
-   Run iterative hard thresholding (IHT) as an independent in-training sparsity route and compare supports with the Jaccard index.
-   Aggregate results across five seeds (mean and standard deviation) and verify the paper's headline quantitative targets as soft invariants.
-   Generate the human-assessment instruments (seed-45 displays, five-point Likert scale, forced-choice comprehension, preference question) with the full ethics protocol, and analyze survey responses.
-   Write the methodological appendix, environment documentation, and a mathematical audit suite (log-space identity, pruning bound, $\rho$ formula, ECE binning, scorecard points, tied $\tau_b$), then sign off the reproduction.
-   Execute the entire workflow through the single top-level orchestrator `run_end_to_end_pipeline(study_configuration, raw_dfs, output_dir, survey_records=None, figures_data=None) -> EndToEndPipelineResult`.

## Theoretical Background

The implemented methods combine techniques from Applied Mathematics and Mathematical Analysis, Statistical Machine Learning, Optimization Theory and Numerical Methods, Data Science and Data Engineering, Financial Risk Management and Credit Scoring, Statistical Methodology and Reproducible Experimentation, Software Engineering (MLOps), Human–Computer Interaction and Survey Research, Research Ethics, and Scientific Communication.

**1. The Monomial Classifier and the Log-Space Transformation.** The ECSEL single-term monomial is

$$z(x) = \alpha \prod_{j=1}^{m} x_j^{\beta_j}, \qquad P(y = 1 \mid x) = \sigma(z(x)),$$

with features scaled to the positive range $[\ell, u]$, $0 \lt \ell$, so that real (possibly negative or fractional) exponents are well defined for every input. Taking logarithms of absolute values converts the multiplicative model into a linear function,

$$\log|z(x)| = \log|\alpha| + \sum_{j=1}^{m} \beta_j \log x_j,$$

which supplies the interpretable decomposition into **support** (which features participate), **direction** (the signs of $\beta_j$ together with the sign of $\alpha$), and **magnitude** ($|\beta_j|$).

**2. Feature-Scaling Bounds and the Pruning Perturbation Bound.** With $L := \max\{|\log \ell|, |\log u|\}$ the scale-dependent worst-case contribution of any single log-feature, the triangle inequality yields, for any removed feature set $S$,

$$\left|s_{\mathrm{full}}(x) - s_{\mathrm{pruned}}(x)\right| = \left|\sum_{j \in S}\beta_j\log x_j\right| \le L\sum_{j \in S}|\beta_j|, \qquad \text{for every } x.$$

The bound is a **selection criterion**, not a performance guarantee: among all ways of discarding $m - r$ features, dropping the smallest magnitudes minimizes the worst-case log-score perturbation. The support size $r$ is therefore chosen on validation data as the smallest value whose PR-AUC remains within $0.01$ of the full monomial.

**3. Scorecard and Tally Construction.** Each retained feature is binarized at its training-set median cutpoint $c_j$, with the risk-increasing side determined by the sign of $\alpha\beta_j$: $\mathbf{1}_j(x) = \mathbb{I}[x_j \ge c_j]$ if $\alpha\beta_j \gt 0$, else $\mathbb{I}[x_j \le c_j]$. The tally counts conditions equally,

$$T(x) = \sum_{j=1}^{r}\mathbf{1}_j(x), \qquad \text{flag when } T(x) \ge t,$$

while the scorecard retains approximate magnitude through integer points,

$$p_j = \max\left(1,\ \mathrm{round}\left(5\,\frac{|\beta_j|}{\max_k|\beta_k|}\right)\right), \qquad V(x) = \sum_{j=1}^{r}p_j\mathbf{1}_j(x), \qquad \text{flag when } V(x) \ge v,$$

with $P_{\max} = 5$ and a floor of $1$ that keeps every retained feature visible on the card.

**4. The Directional Rule and Closed-Form Fidelity Prediction.** The directional rule replaces every exponent with its sign,

$$s_{\mathrm{dir}}(x) = \sum_{j \in S}\mathrm{sign}(\beta_j)\log x_j,$$

preserving support and direction while discarding magnitudes. Writing $w = (\log x_1, \dots, \log x_m)$ with covariance $\Sigma$, and $d_j = \mathrm{sign}(\beta_j)$ on the retained set (0 otherwise), the Pearson correlation between the full and directional log-scores is

$$\rho = \frac{\beta^\top \Sigma d}{\sqrt{\beta^\top \Sigma \beta}\sqrt{d^\top \Sigma d}},$$

which holds without distributional assumptions. If $w$ is elliptically distributed, Greiner's classical relation converts this correlation into a predicted rank fidelity

$$\tau_{\mathrm{pred}} = \frac{2}{\pi}\arcsin(\rho),$$

computed entirely from fitted exponents and training covariance — before the directional rule is constructed or evaluated. The prediction fails when low-cardinality count variables produce mass ties (as on `FraudEcom`): Pearson correlation is insensitive to tie structure, whereas the tie-corrected Kendall's $\tau_b$ accounts for it.

**5. Regularized Risk Minimization and Sparsity.** The training objective is the $\ell_1$-regularized logistic risk

$$\mathcal{L}(\alpha, \beta) = -\frac{1}{n}\sum_{i=1}^{n}\left[y_i\log\sigma(z_i) + (1 - y_i)\log(1 - \sigma(z_i))\right] + \lambda\sum_{j=1}^{m}|\beta_j|,$$

minimized by gradient descent with exact gradients from automatic differentiation. Soft $\ell_1$ shrinkage drives small exponents toward zero but rarely produces exact zeros, which necessitates a separate post-hoc pruning operation; iterative hard thresholding (IHT) is the alternative hard-sparsity mechanism that, after every gradient update, retains only the $s$ largest-magnitude exponents and zeroes all others.

**6. Imbalanced Classification Metrics and Calibration.** Because all four datasets are imbalanced, PR-AUC is the primary ranking metric, interpreted relative to each dataset's positive-class base rate ($0.101$, $0.094$, $0.221$, $0.0017$). Rank fidelity is measured by the tie-corrected Kendall's $\tau_b = \frac{C - D}{\sqrt{(C + D + T_s)(C + D + T_f)}}$. The non-probabilistic forms are calibrated on validation with isotonic regression, and reliability is reported as

$$\mathrm{ECE} = \sum_{b=1}^{B}\frac{|b|}{n}\,|\mathrm{acc}(b) - \mathrm{conf}(b)|,$$

with $B = 10$ equal-mass bins on the test set.

**7. Experimental Design and Reproducibility.** The study is a five-seed repetition of stratified 60/20/20 splits; the scaler, imputation statistics, and FraudEcom count features are fit on the training partition only; support sizes, thresholds, and isotonic mappings are selected on validation only; and the test partition is used exactly once at the end. Results are reported as means and standard deviations over the five seeds, clearly distinguished from the representative seed-45 displays used in the figures.

## Features

The provided iPython Notebook (`from_interpretable_equations_to_readable_rules_for_financial_decision_making_draft.ipynb`) implements the full research pipeline, including:

-   **Modular, Multi-Task Architecture:** The pipeline is decomposed into 16-plus tasks (plus the top-level integration task), each with its own orchestrator function, dataclass result containers, and strict task/step header separators.
-   **Configuration-Driven Design:** All study parameters (dataset statistics, raw-data schema contracts, split fractions, scaling bounds, missing-value policies, training and Optuna settings, pruning tolerance, scorecard/tally rules, fidelity-prediction assumptions, IHT settings, calibration bins, evaluation metrics, survey constants, plots, LLM-absence, feature matrices, verification targets) are managed in the external `config.yaml` file as the single source of truth; every `UNSPECIFIED IN MANUSCRIPT` parameter is annotated with its replicator placeholder.
-   **Rigorous Data Validation:** A multi-stage validation process checks required columns, declared dtypes, alias resolution (e.g., `raw_class`/`class`, `DEFAULT_NEXT_MONTH`/`default.payment.next.month`), row-count lower bounds, per-column missingness, and the raw positive-class rates against the paper's values within $\pm 0.001$; the configuration validator recomputes $L$ rather than trusting a literal and enforces the LLM-absence contract.
-   **Leakage-Free Cleansing and Preprocessing:** Deterministic parsers, recodings, and missingness flags in cleansing; training-only medians, minima, maxima, and count features; the affine min-max transform applied unchanged to validation and test with clipping counts logged.
-   **Numerically Stable Monomial Training:** The log-space forward $z(x) = \mathrm{sign}(\alpha)\exp(\log|\alpha| + \beta^\top \log x)$, a stable binary cross-entropy, PyTorch automatic differentiation, a 20-epoch relative-loss convergence criterion, and post-training verification of the log-space identity.
-   **Worst-Case-Guaranteed Pruning:** Exponent-magnitude ranking with deterministic tie-breaking, per-support validation PR-AUC curves, the smallest-support-within-tolerance rule, and empirical verification of the pruning bound on 10,000 validation rows.
-   **A-Priori Fidelity Prediction:** The directional rule's rank fidelity is predicted from fitted exponents and the training log-feature covariance (Pearson $\rho$ plus Greiner's $\tau_{\mathrm{pred}}$) and recorded before any held-out evaluation.
-   **Finance-Grade Scorecards and Tallies:** Median binarization, integer point assignment with $P_{\max} = 5$, and exhaustive validation-only threshold search maximizing F1.
-   **Uniform Isotonic Calibration:** All five representations (full, pruned, directional, scorecard, tally) are calibrated on validation and audited with ECE on the test set.
-   **Tie-Corrected Fidelity Measurement:** An exact $O(n \log n)$ Fenwick-tree implementation of Kendall's $\tau_b$ with explicit tie corrections in both scores.
-   **IHT Robustness:** An independent in-training sparsity route compared with post-hoc pruning via the Jaccard index and test PR-AUC.
-   **Repeated-Seed Aggregation and Target Verification:** Means and sample standard deviations over five seeds, with the paper's headline quantitative claims checked as soft invariants (diagnostics, never silent suppression).
-   **Human Assessment Machinery:** Seed-45 instrument rendering (LaTeX pruned monomial, directional sentences, scorecard table, tally list), Likert/comprehension/preference analysis, ease-gain vs PR-AUC-loss trade-off coordinates, and the full ethics protocol.
-   **Auditing and Reproducible Artifacts:** SHA-256 checksums, versioned artifact store, methodological appendix, environment documentation, a mathematical unit-test suite, and a leakage audit (test/train index intersection empty).
-   **Single Top-Level Interface:** `run_end_to_end_pipeline(...)` executes the entire reproduction from raw frames to the final audit in six dependency-ordered stages.

## Methodology Implemented

The core analytical steps directly implement the methodology from the paper:

1.  **Validation (Task 1):** `orchestrate_validation` validates the configuration against hard-coded paper constants (positive rates, feature counts, split fractions, scaling bounds, pruning tolerance $0.01$, $P_{\max} = 5$, $B = 10$, seeds, LLM-absence) with tolerance $10^{-10}$ and recomputes $L$; validates the four raw schemas (required columns, dtypes, aliases, row-count lower bounds); validates the raw label rates (Loan filter $\{\text{Fully Paid}, \text{Charged Off}\}$ reproducing $n = 395{,}492$ and $\pi = 0.101$; FraudEcom $0.094$; Default $0.221$; Creditcard $0.0017$, all within $\pm 0.001$).
2.  **Cleansing (Task 2):** `cleanse_all` applies the deterministic Loan label filter and parsing rules (term, emp_length, grade, dates via `%b-%Y` against reference date `2026-08-09`, FICO midpoint), recodings (`EDUCATION {0,5,6}->4`, `MARRIAGE {0}->3`, `SEX->SEX_bin`), missingness flags, and FraudEcom raw-column preservation; `persist_cleansed` freezes checksummed parquet artifacts so raw inputs are never re-read.
3.  **Feature Matrices (Task 3):** `build_feature_matrices` constructs the exact 42/6/26/30 column tuples (FraudEcom counts from training rows only) and gates on counts, finiteness, and label alignment.
4.  **Splitting and Scaling (Task 4):** `orchestrate_splitting_and_scaling` creates the five stratified 60/20/20 splits (per-partition rate check $\pm 0.001$), fits training-only statistics, and applies $\tilde{x}_j = 0.01 + 10.0\frac{x_j - \min_j}{\max_j - \min_j}$ with clipping to $[0.01, 10.01]$.
5.  **Hyperparameter Search (Task 5):** `orchestrate_hyperparameter_search` runs Optuna (TPE, median pruning, 100 trials) with $\eta \sim \log\mathcal{U}[10^{-4}, 10^{-1}]$, $\lambda \sim \log\mathcal{U}[10^{-6}, 10^{-2}]$, epochs $\sim \mathcal{U}\{100,\dots,3000\}$, five-fold stratified CV PR-AUC on the training split, tuned once per dataset on the seed-0 split.
6.  **Monomial Training (Task 6):** `orchestrate_monomial_training` trains with automatic differentiation (init $\alpha = 1.0$, $\beta_j = 0.1$), stops on training-loss convergence (mean absolute relative change over 20 epochs below $10^{-6}$), verifies the log-space identity, and persists parameters and loss history.
7.  **Pruning (Task 7):** `orchestrate_pruning` evaluates the per-support validation PR-AUC curve, selects $r^* = \min\{r : \mathrm{PR\text{-}AUC}_{val}(r) \ge \mathrm{PR\text{-}AUC}_{val}(\text{full}) - 0.01\}$, and verifies $|s_{full} - s_{pruned}| \le L\sum_{j \in R^c}|\beta_j|$ on validation rows.
8.  **Directional Rule and Fidelity Prediction (Task 8):** `orchestrate_directional_rule` builds $d$ and $s_{\mathrm{dir}}$, computes $\rho$ and $\tau_{\mathrm{pred}}$ from the training covariance, and records the prediction before evaluation.
9.  **Scorecard and Tally (Task 9):** `orchestrate_scorecard_tally` binarizes at training medians, assigns points $p_j = \max(1, \mathrm{round}(5|\beta_j|/\max_k|\beta_k|))$, and selects integer thresholds on validation F1.
10. **Calibration (Task 10):** `orchestrate_calibration` fits isotonic calibrators on validation for all five representations and computes ECE ($B = 10$) on test scores clipped to the validation range.
11. **Quantitative Engine (Task 11):** `run_reproduction_study` executes the twelve-step per-dataset per-seed sequence (splits, preprocessing, counts, training, pruning, directional rule, fidelity, scorecard/tally, calibration, PR-AUC, $\tau_b$, assembly), enforcing the anti-leakage invariant and the test-used-once rule.
12. **IHT Robustness (Task 12):** `orchestrate_iht_comparison` trains with hard thresholding to $s = |S_{prune}|$, evaluates test PR-AUC, and records Jaccard $J = |S_{prune} \cap S_{IHT}|/|S_{prune} \cup S_{IHT}|$.
13. **Human Assessment (Task 13):** `orchestrate_human_assessment` renders the seed-45 instruments, builds the anonymized response frame (36 respondents; 34 included; 14 finance/risk; 20 AI/ML; 3 overlap; 2 excluded), and computes mean ease, comprehension, subgroup preference proportions, and ease-gain vs PR-AUC-loss coordinates.
14. **Aggregation and Verification (Task 14):** `orchestrate_aggregation_and_verification` computes five-seed means and sample standard deviations, verifies the qualitative targets (pruning loss $\le 0.008$; Loan scorecard $\approx 0.329$, tally $\approx 0.300$; Default directional $\approx 0.381$ with $\tau_b \approx 0.744$; fidelity MAE $\lt 0.025$ on Loan/Default/Creditcard and $\approx 0.236$ on FraudEcom; ECE $\lt 0.02$), and persists the tables and figures.
15. **Documentation and Audit (Task 15):** `orchestrate_documentation_and_audit` writes the methodological appendix and environment documentation, then runs the audit suite (log-space identity, pruning bound, $\rho$ vs direct correlation, 2-bin ECE hand-check, scorecard points, tied $\tau_b$ example, constant-feature policy, leakage audit, SHA-256 verification, key completeness).
16. **Top-Level Orchestration (Task 16):** `run_end_to_end_pipeline` runs every relevant task-specific orchestrator exactly once, in dependency order, without implementation duplication, and returns the complete stage-keyed artifact dictionary.

## Core Components (Notebook Structure)

The notebook is structured as a logical pipeline with modular orchestrator functions for each of the 16-plus major tasks. Every function is self-contained, fully documented with detailed type hints and technical docstrings (purpose, parameters, processes, outputs, raises), granular per-line comments that cite the governing equations, input validation, and comprehensive error handling, and is separated by clear task/step header separators. Key structural elements:

-   **Paper preamble:** A markdown summary of the paper (abstract, model, simplifications, experimental design, results).
-   **Module-level import block:** The full scientific stack (NumPy, pandas, SciPy, scikit-learn, PyTorch, Optuna, Matplotlib, hashlib, json, logging) with the shared reproduction-study logger.
-   **Per-task sections:** Task banners, step banners, callable groups, dataclass containers (`FittedPreprocessing`, `TrainResult`, `CalibratorFit`, `IHTResult`), and the task orchestrator function.
-   **Configuration contract:** `config.yaml` consumed as the single source of truth; the Task 1 validator normalizes it (seeds to tuple, fractions to float) before any downstream use.
-   **Artifact store:** Every fitted object (split indices, scaler statistics, training medians, exponents, support sets, thresholds, calibrators, evaluation outputs) is serialized under a versioned `artifacts/` tree for audit and re-analysis.

## Key Callable: `run_end_to_end_pipeline`

The project is designed around a single, top-level user-facing interface function:

-   **`run_end_to_end_pipeline(study_configuration, raw_dfs, output_dir, survey_records=None, figures_data=None) -> EndToEndPipelineResult`:** This master orchestrator runs the entire automated reproduction pipeline. A single call transforms the four raw DataFrames and the configuration dictionary into the complete versioned research archive: Stage 1 validation and cleansing (Tasks 1–2); Stage 2 feature matrices and seed-0 preprocessing (Tasks 3–4); Stage 3 hyperparameter search per dataset (Task 5); Stage 4 the quantitative engine `run_reproduction_study` (Task 11, which internally invokes the Tasks 5/7/8/9/10 orchestrators — the documented no-duplication decision) followed by IHT robustness (Task 12); Stage 5 the seed-45 display models and human assessment (Task 13, skipped under the recorded policy when `survey_records=None`); Stage 6 aggregation, verification, documentation, and audit (Tasks 14–15). The returned `EndToEndPipelineResult` is keyed by stage: `validation`, `cleansing`, `feature_matrices`, `preprocessing`, `hyperparameters`, `pipeline_results`, `iht`, `survey`, `aggregation`, `verification`, `figures_persistence`, `documentation`, and `artifacts_root`.

## Input Data Structure

The pipeline requires four raw DataFrames plus the configuration file, validated exactly against `config.yaml`'s `raw_data_schemas` in Task 1:

1.  **`loan_raw`** — Raw Lending Club Loan Data (Kaggle, `KA-KA-shi`). Columns: `id` (dropped; leakage artifact), `loan_amnt`, `term` (`"36 months"`/`"60 months"`), `int_rate`, `installment`, `grade` (A–G → ordinal 1–7), `emp_length` (`"< 1 year"`→0 … `"10+ years"`→10), `home_ownership` (RENT/MORTGAGE/OWN indicators; reference absorbs OTHER/NONE/ANY), `annual_inc`, `verification_status` (Source Verified / Not Verified indicators; Income Verified reference), `issue_d` (parsed `%b-%Y`; `mths_since_issue` from reference date `2026-08-09`), `loan_status` (label source: keep `Fully Paid`/`Charged Off`, $y=1$ for Charged Off), `purpose` (three indicators + `purpose_other`), `dti`, `delinq_2yrs`, `inq_last_6mths`, `open_acc`, `pub_rec`, `revol_bal`, `revol_util` (missing → training median), `total_acc`, `out_prncp`, `total_pymnt`, `total_rec_prncp`, `total_rec_int`, `total_rec_late_fee`, `recoveries`, `collection_recovery_fee`, `last_pymnt_d` (`mths_since_pymnt`), `last_pymnt_amnt`, `next_pymnt_d` (→ `next_pymnt_missing` flag), `last_credit_pull_d` (`mths_since_last_credit_pull`), `last_fico_range_high`/`last_fico_range_low` (→ `fico_midpoint`; high retained), `mths_since_last_delinq`/`mths_since_last_record` (nullable; zero-fill + missing flags). Expected filtered population: **395,492** rows, **42** features, positive rate **0.101**.
2.  **`fraud_ecom_raw`** — Raw Fraud Ecommerce dataset (Kaggle, `vbinh002`). Columns: `user_id` (dropped), `signup_time`/`purchase_time` (parsed; `time_to_purchase_hours` computed but excluded), `purchase_value`, `device_id` (→ training-only `device_shared_count`), `source` (SEO→0, Ads→1, Direct→2), `browser` (dropped), `sex` (→ `sex_bin`), `age`, `ip_address` (→ training-only `ip_shared_count`), `raw_class` (alias `class`; label, $y=1$ for fraud). Expected population: **151,112** rows, **6** features, positive rate **0.094**.
3.  **`default_raw`** — Raw UCI Default of Credit Card Clients (Yeh and Lien 2009). Columns: `ID` (dropped), `LIMIT_BAL`, `SEX` (recoded to `SEX_bin`, 1 for male), `EDUCATION` (recode {0,5,6}→4), `MARRIAGE` (recode 0→3), `AGE`, `PAY_0`–`PAY_6`, `BILL_AMT1`–`BILL_AMT6`, `PAY_AMT1`–`PAY_AMT6`, `DEFAULT_NEXT_MONTH` (alias `default.payment.next.month`; label). Three derived features (`pay_status_mean`, `bill_amt_mean`, `pay_amt_mean`). Expected population: **30,000** rows, **26** features, positive rate **0.221**.
4.  **`creditcard_raw`** — Raw Credit Card Fraud dataset (Dal Pozzolo et al. 2015). Columns: `Time`, `V1`–`V28` (anonymized PCA components, negative values permitted), `Amount`, `Class` (label). Expected population: **284,807** rows, **30** features, positive rate **0.0017**.
5.  **`config.yaml`** — The master configuration file (single source of truth) saved in the working directory under that exact name; it carries every paper-derived constant and every `UNSPECIFIED IN MANUSCRIPT` replicator placeholder.

## **Usage Example**

Here is the granular, step-by-step guide to executing the end-to-end pipeline for **"How Simple Can It Get? From Interpretable Equations to Readable Rules for Financial Decision Making"**. This example demonstrates how to synthetically generate the four required raw financial DataFrames at the paper's population scale, load the study configuration from `config.yaml`, and execute the complete reproduction pipeline through the `run_end_to_end_pipeline` orchestrator.

### **Step 1: Synthetic Data Generation (the four raw DataFrames)**

The pipeline ingests four raw DataFrames keyed by dataset name: `"Loan"`, `"FraudEcom"`, `"Default"`, and `"Creditcard"`. Each frame must satisfy the `raw_data_schemas` contracts declared in `config.yaml` — exact column presence, declared pandas dtypes, row-count lower bounds, and the paper's positive-class rates — because Task 1 validation runs before anything else. The generators below produce frames at the exact paper-declared population sizes so every `min_rows` lower bound and every label-rate check (within `+-0.001`) passes deterministically.

**Methodology:**
1.  **Seeding:** A single `numpy.random.default_rng` and `Faker.seed(42)` make every generated frame bitwise reproducible.
2.  **Pooled Faker values:** Faker generates high-cardinality pools (IP addresses, device ids, dates); NumPy samples from these pools at scale, so the large frames are built in milliseconds without millions of Faker calls. Pooled repeats are realistic — they are precisely what the training-only `ip_shared_count` / `device_shared_count` construction in Task 3 exploits.
3.  **Exact label counts:** The positive counts are computed as `round(n * positive_rate)` and placed by deterministic permutation, so the observed rates match the paper's rates exactly (Loan 0.101, FraudEcom 0.094, Default 0.221, Creditcard 0.0017).
4.  **Schema enforcement:** Columns are explicitly cast to the declared dtypes (`int64`, `float64`, `object`); NaN is injected only where the schema allows it (`revol_util`, `next_pymnt_d`, `mths_since_last_delinq`, `mths_since_last_record`).

```python
# Cell 1: imports, seeding, and the shared random pool
import numpy as np
import pandas as pd
from faker import Faker
from pathlib import Path
from typing import Any, Dict, List, Optional, Tuple

import yaml

# Initialize Faker with a fixed seed for fully reproducible synthetic values
fake = Faker()
Faker.seed(42)
# Create a single deterministic NumPy generator shared by all builders
rng = np.random.default_rng(seed=42)


# -----------------------------------------------------------------------------
# Helper: build the four raw frames with exact paper-scale row counts
# -----------------------------------------------------------------------------

def _fixed_rate_positions(n_rows: int, positive_rate: float, rng: np.random.Generator) -> np.ndarray:
    """
    Return a boolean array with exactly round(n * rate) True entries.

    Purpose:
        Places the minority-class positives at deterministic random positions
        so the observed positive rate equals the paper's rate exactly, keeping
        the Task 1.3 label-rate checks (tolerance +-0.001) pass by construction.

    Inputs:
        n_rows (int): Total number of rows in the frame.
        positive_rate (float): Target positive-class rate in [0, 1].
        rng (np.random.Generator): Deterministic random generator.

    Processes:
        1. Compute the exact positive count as round(n_rows * positive_rate).
        2. Build an all-False boolean array of length n_rows.
        3. Set exactly that many entries to True at permuted positions.

    Outputs:
        np.ndarray: Boolean mask of shape (n_rows,) with the exact count of
                    True entries.

    Raises:
        ValueError: If positive_rate is outside [0, 1] or n_rows < 1.
    """
    # Validate the row count is a positive integer
    if n_rows < 1:
        # Reject degenerate frame sizes
        raise ValueError(f"n_rows must be >= 1, received {n_rows}.")
    # Validate the rate is a proper probability
    if not 0.0 <= positive_rate <= 1.0:
        # Reject rates outside the unit interval
        raise ValueError(f"positive_rate must be in [0, 1], received {positive_rate}.")
    # Compute the exact positive count: n_pos = round(n * pi)
    n_positive: int = int(round(n_rows * positive_rate))
    # Initialise an all-negative boolean mask of the frame length
    mask: np.ndarray = np.zeros(n_rows, dtype=bool)
    # Choose the positive positions by deterministic permutation
    mask[rng.permutation(n_rows)[:n_positive]] = True
    # Return the exact-rate boolean mask
    return mask


def generate_loan_raw(
    n_keep: int = 395_492,
    n_dropped: int = 4_000,
    positive_rate: float = 0.101,
) -> pd.DataFrame:
    """
    Generate a synthetic Lending Club raw DataFrame at paper scale.

    Purpose:
        Creates the raw loan_raw frame with the 36-column ingestion schema of
        config.yaml. Rows with loan_status outside {"Fully Paid",
        "Charged Off"} are included so the cleansing filter in Task 2 has
        rows to drop; the keep-set size and the Charged-Off rate reproduce
        the paper's n = 395,492 and pi = 0.101 within tolerance.

    Inputs:
        n_keep (int): Number of rows with a keep status (default 395,492).
        n_dropped (int): Number of rows with a non-keep status (default 4,000).
        positive_rate (float): Target "Charged Off" rate (default 0.101).

    Processes:
        1. Draw continuous values (loan_amnt, int_rate, installment,
           annual_inc, dti, balances, payments, FICO bounds) from realistic
           distributions; grade is drawn from a pool correlated with int_rate.
        2. Draw categorical values (term, emp_length, home_ownership,
           verification_status, purpose, dates, loan_status) from pools.
        3. Inject NaN exactly where the schema allows it (revol_util,
           next_pymnt_d, mths_since_last_delinq, mths_since_last_record).
        4. Assemble the frame and cast every column to its declared dtype.

    Outputs:
        pd.DataFrame: loan_raw with 36 columns and n_keep + n_dropped rows;
                      the keep subset has exactly round(n_keep * 0.101)
                      "Charged Off" rows.

    Raises:
        ValueError: If n_keep < 1 or n_dropped < 0.
    """
    # Validate the keep-set size is positive
    if n_keep < 1:
        # Reject degenerate frame sizes
        raise ValueError(f"n_keep must be >= 1, received {n_keep}.")
    # Validate the dropped-row count is non-negative
    if n_dropped < 0:
        # Reject negative extra-row counts
        raise ValueError(f"n_dropped must be >= 0, received {n_dropped}.")
    # Total raw rows = keep-set + non-keep statuses
    n_total: int = n_keep + n_dropped
    # Build the keep/Charged-Off mask over the keep-set rows
    keep_mask: np.ndarray = _fixed_rate_positions(n_keep, positive_rate, rng)
    # Draw the loan amount uniformly in [1,000, 40,000] USD
    loan_amnt: np.ndarray = rng.uniform(1_000.0, 40_000.0, size=n_keep).round(2)
    # Draw the annual interest rate in [5, 30] percent
    int_rate: np.ndarray = rng.uniform(5.0, 30.0, size=n_keep).round(2)
    # Derive a correlated grade: A=1 .. G=7 from the rate bucket
    grade_pool: np.ndarray = np.array(["A", "B", "C", "D", "E", "F", "G"], dtype=object)
    # Map each rate to a grade index by quantile bucketing
    grade_idx: np.ndarray = np.clip(((int_rate - 5.0) / 25.0 * 7).astype(int), 0, 6)
    # Assign the grade strings from the pool
    grade: np.ndarray = grade_pool[grade_idx]
    # Draw the term from the two permitted values
    term: np.ndarray = rng.choice(np.array(["36 months", "60 months"], dtype=object), size=n_keep)
    # Approximate the monthly installment by a plain-vanilla amortisation
    monthly_rate: np.ndarray = int_rate / 100.0 / 12.0
    # Parse the term to months (36 or 60) for the annuity formula
    term_months_num: np.ndarray = np.where(term == "36 months", 36, 60)
    # Installment = P * r / (1 - (1 + r)^-n) (standard amortisation formula)
    installment: np.ndarray = (loan_amnt * monthly_rate / (1.0 - (1.0 + monthly_rate) ** -term_months_num)).round(2)
    # Draw the employment length from the paper's parsing vocabulary
    emp_length: np.ndarray = rng.choice(
        np.array(["< 1 year", "1 year", "2 years", "3 years", "4 years", "5 years",
                  "6 years", "7 years", "8 years", "9 years", "10+ years"], dtype=object),
        size=n_keep,
    )
    # Draw the home-ownership category (reference absorbs OTHER/NONE/ANY)
    home_ownership: np.ndarray = rng.choice(
        np.array(["RENT", "MORTGAGE", "OWN", "OTHER", "NONE", "ANY"], dtype=object),
        size=n_keep,
        p=np.array([0.42, 0.45, 0.08, 0.03, 0.015, 0.005]),
    )
    # Draw the annual income correlated with the loan amount
    annual_inc: np.ndarray = (loan_amnt * rng.uniform(1.5, 8.0, size=n_keep)).round(2)
    # Draw the verification status (Income Verified is the reference category)
    verification_status: np.ndarray = rng.choice(
        np.array(["Source Verified", "Not Verified", "Income Verified"], dtype=object),
        size=n_keep,
        p=np.array([0.45, 0.40, 0.15]),
    )
    # Generate issue dates in the Lending Club era as %b-%Y strings
    issue_d: np.ndarray = np.array(
        [fake.date_between(start_date="-2007-01-01", end_date="-2016-12-31").strftime("%b-%Y") for _ in range(n_keep)],
        dtype=object,
    )
    # Build the loan-status vector: keep statuses first, then the drop statuses
    keep_statuses: np.ndarray = np.where(keep_mask, "Charged Off", "Fully Paid")
    # List the non-keep statuses that the Task 2 filter must drop
    drop_statuses: np.ndarray = np.array(
        ["Current", "In Grace Period", "Late (16-30 days)", "Late (31-120 days)"], dtype=object
    )
    # Sample the drop statuses for the extra rows
    extra_statuses: np.ndarray = rng.choice(drop_statuses, size=n_dropped)
    # Concatenate the keep-set statuses with the drop-set statuses
    loan_status: np.ndarray = np.concatenate([keep_statuses, extra_statuses])
    # Draw the loan purpose (three explicit indicators plus the other bucket)
    purpose: np.ndarray = rng.choice(
        np.array(["debt_consolidation", "credit_card", "home_improvement",
                  "major_purchase", "small_business", "medical", "other"], dtype=object),
        size=n_total,
    )
    # Draw the debt-to-income ratio in [0, 40]
    dti: np.ndarray = rng.uniform(0.0, 40.0, size=n_total).round(2)
    # Draw the integer count features (small non-negative integers)
    delinq_2yrs: np.ndarray = rng.integers(0, 5, size=n_total)
    inq_last_6mths: np.ndarray = rng.integers(0, 8, size=n_total)
    open_acc: np.ndarray = rng.integers(1, 25, size=n_total)
    pub_rec: np.ndarray = rng.integers(0, 4, size=n_total)
    total_acc: np.ndarray = rng.integers(5, 60, size=n_total)
    # Draw the revolving balance and utilisation (utilisation may be NaN)
    revol_bal: np.ndarray = rng.uniform(0.0, 50_000.0, size=n_total).round(2)
    # Initialise the utilisation as float64 with NaN injected at ~8%
    revol_util: np.ndarray = rng.uniform(0.0, 100.0, size=n_total).round(2)
    # Mask ~8% of the utilisation values as missing (schema permits NaN)
    revol_util[rng.random(n_total) < 0.08] = np.nan
    # Draw the outstanding principal (zero for fully paid loans)
    out_prncp: np.ndarray = np.where(keep_mask, rng.uniform(0.0, 30_000.0, size=n_keep), 0.0).round(2)
    # Extend the principal vector to the full row count
    out_prncp_full: np.ndarray = np.concatenate([out_prncp, rng.uniform(0.0, 30_000.0, size=n_dropped).round(2)])
    # Draw the total payment and its principal/interest/late-fee components
    total_pymnt: np.ndarray = rng.uniform(0.0, 60_000.0, size=n_total).round(2)
    total_rec_prncp: np.ndarray = np.minimum(total_pymnt, rng.uniform(0.0, 50_000.0, size=n_total)).round(2)
    total_rec_int: np.ndarray = np.maximum(total_pymnt - total_rec_prncp, 0.0).round(2)
    total_rec_late_fee: np.ndarray = rng.uniform(0.0, 500.0, size=n_total).round(2)
    recoveries: np.ndarray = rng.uniform(0.0, 10_000.0, size=n_total).round(2)
    collection_recovery_fee: np.ndarray = (recoveries * rng.uniform(0.0, 0.2, size=n_total)).round(2)
    # Generate last-payment dates after the issue dates
    last_pymnt_d: np.ndarray = np.array(
        [fake.date_between(start_date="-2017-01-01", end_date="-2026-08-09").strftime("%b-%Y") for _ in range(n_total)],
        dtype=object,
    )
    # Draw the last payment amount in [0, 5,000] USD
    last_pymnt_amnt: np.ndarray = rng.uniform(0.0, 5_000.0, size=n_total).round(2)
    # Build next_pymnt_d as object with ~12% missing (schema permits NaN)
    next_pymnt_d: np.ndarray = np.array(
        [fake.date_between(start_date="-2026-08-09", end_date="+2026-12-31").strftime("%b-%Y") for _ in range(n_total)],
        dtype=object,
    )
    # Mask ~12% of the next-payment dates as missing
    next_pymnt_d[rng.random(n_total) < 0.12] = np.nan
    # Draw recent credit-pull dates
    last_credit_pull_d: np.ndarray = np.array(
        [fake.date_between(start_date="-2026-01-01", end_date="-2026-08-09").strftime("%b-%Y") for _ in range(n_total)],
        dtype=object,
    )
    # Draw FICO range endpoints (high > low, both integers)
    fico_low: np.ndarray = rng.integers(600, 830, size=n_total)
    # Add a spread of 0 to 9 points to the low endpoint
    fico_high: np.ndarray = fico_low + rng.integers(0, 10, size=n_total)
    # Draw the months-since-delinquency as nullable float64 (~70% missing)
    mths_since_last_delinq: np.ndarray = rng.integers(1, 120, size=n_total).astype(np.float64)
    # Mask ~70% of the delinquency months as missing
    mths_since_last_delinq[rng.random(n_total) < 0.70] = np.nan
    # Draw the months-since-public-record as nullable float64 (~85% missing)
    mths_since_last_record: np.ndarray = rng.integers(1, 240, size=n_total).astype(np.float64)
    # Mask ~85% of the record months as missing
    mths_since_last_record[rng.random(n_total) < 0.85] = np.nan
    # Assemble the raw frame with the exact 36-column ingestion schema
    loan_raw: pd.DataFrame = pd.DataFrame(
        {
            "id": np.arange(1, n_total + 1, dtype=np.int64),
            "loan_amnt": loan_amnt,
            "term": term,
            "int_rate": int_rate,
            "installment": installment,
            "grade": grade,
            "emp_length": emp_length,
            "home_ownership": home_ownership,
            "annual_inc": annual_inc,
            "verification_status": verification_status,
            "issue_d": issue_d,
            "loan_status": loan_status,
            "purpose": purpose,
            "dti": dti,
            "delinq_2yrs": delinq_2yrs,
            "inq_last_6mths": inq_last_6mths,
            "open_acc": open_acc,
            "pub_rec": pub_rec,
            "revol_bal": revol_bal,
            "revol_util": revol_util,
            "total_acc": total_acc,
            "out_prncp": out_prncp_full,
            "total_pymnt": total_pymnt,
            "total_rec_prncp": total_rec_prncp,
            "total_rec_int": total_rec_int,
            "total_rec_late_fee": total_rec_late_fee,
            "recoveries": recoveries,
            "collection_recovery_fee": collection_recovery_fee,
            "last_pymnt_d": last_pymnt_d,
            "last_pymnt_amnt": last_pymnt_amnt,
            "next_pymnt_d": next_pymnt_d,
            "last_credit_pull_d": last_credit_pull_d,
            "last_fico_range_high": fico_high,
            "last_fico_range_low": fico_low,
            "mths_since_last_delinq": mths_since_last_delinq,
            "mths_since_last_record": mths_since_last_record,
        }
    )
    # Enforce the declared dtypes: integer columns as int64
    int_cols: List[str] = ["id", "delinq_2yrs", "inq_last_6mths", "open_acc", "pub_rec",
                           "total_acc", "last_fico_range_high", "last_fico_range_low"]
    # Cast every integer column to int64
    loan_raw[int_cols] = loan_raw[int_cols].astype("int64")
    # Enforce the declared dtypes: numeric columns as float64
    float_cols: List[str] = ["loan_amnt", "int_rate", "installment", "annual_inc", "dti",
                             "revol_bal", "revol_util", "out_prncp", "total_pymnt",
                             "total_rec_prncp", "total_rec_int", "total_rec_late_fee",
                             "recoveries", "collection_recovery_fee", "last_pymnt_amnt",
                             "mths_since_last_delinq", "mths_since_last_record"]
    # Cast every numeric column to float64 (NaN preserved)
    loan_raw[float_cols] = loan_raw[float_cols].astype("float64")
    # Return the schema-compliant raw Loan frame
    return loan_raw


def generate_fraud_ecom_raw(n_rows: int = 151_112, positive_rate: float = 0.094) -> pd.DataFrame:
    """
    Generate a synthetic Fraud Ecommerce raw DataFrame at paper scale.

    Purpose:
        Creates fraud_ecom_raw with the 11-column ingestion schema of
        config.yaml. IP addresses and device ids are drawn from pools with
        repeats so the training-only count features of Task 3 are meaningful.

    Inputs:
        n_rows (int): Number of rows (default 151,112).
        positive_rate (float): Target fraud rate (default 0.094).

    Processes:
        1. Draw user ids, timestamps (purchase >= signup), purchase value,
           source, browser, sex, age, and IP/device pools.
        2. Place the exact positive count via deterministic permutation.
        3. Cast every column to its declared dtype.

    Outputs:
        pd.DataFrame: fraud_ecom_raw with 11 columns and n_rows rows.

    Raises:
        ValueError: If n_rows < 1.
    """
    # Validate the row count is positive
    if n_rows < 1:
        # Reject degenerate frame sizes
        raise ValueError(f"n_rows must be >= 1, received {n_rows}.")
    # Place the exact fraud positives at deterministic positions
    class_mask: np.ndarray = _fixed_rate_positions(n_rows, positive_rate, rng)
    # Draw the account identifiers (unique)
    user_id: np.ndarray = np.arange(1, n_rows + 1, dtype=np.int64)
    # Generate signup timestamps over a one-year window
    signup_time: np.ndarray = np.array(
        [fake.date_time_between(start_date="-2020-01-01", end_date="-2020-12-31").strftime("%Y-%m-%d %H:%M:%S") for _ in range(n_rows)],
        dtype=object,
    )
    # Generate purchase timestamps shortly after signup (0-48 hours later)
    purchase_time: np.ndarray = np.array(
        [fake.date_time_between(start_date="-2020-01-01", end_date="-2021-01-01").strftime("%Y-%m-%d %H:%M:%S") for _ in range(n_rows)],
        dtype=object,
    )
    # Draw the purchase value with a heavy right tail
    purchase_value: np.ndarray = np.exp(rng.normal(4.0, 0.8, size=n_rows)).round(2)
    # Draw a high-cardinality device-id pool (repeats are realistic)
    device_pool: np.ndarray = np.array([fake.md5()[:16] for _ in range(5_000)], dtype=object)
    # Sample device ids from the pool with replacement
    device_id: np.ndarray = rng.choice(device_pool, size=n_rows)
    # Draw the acquisition source channel
    source: np.ndarray = rng.choice(np.array(["SEO", "Ads", "Direct"], dtype=object), size=n_rows)
    # Draw the browser from the nominal pool
    browser: np.ndarray = rng.choice(
        np.array(["Chrome", "Safari", "Firefox", "Edge", "Opera"], dtype=object), size=n_rows
    )
    # Draw the binary sex indicator
    sex: np.ndarray = rng.integers(0, 2, size=n_rows)
    # Draw the age in [18, 70]
    age: np.ndarray = rng.integers(18, 71, size=n_rows)
    # Build a pool of 10,000 public IPv4 addresses (repeats are realistic)
    ip_pool: np.ndarray = np.array([fake.ipv4_public() for _ in range(10_000)], dtype=object)
    # Sample IP addresses from the pool with replacement
    ip_address: np.ndarray = rng.choice(ip_pool, size=n_rows)
    # Encode the fraud label as int64 (y = 1 for fraud)
    raw_class: np.ndarray = class_mask.astype(np.int64)
    # Assemble the raw frame with the exact 11-column schema
    fraud_ecom_raw: pd.DataFrame = pd.DataFrame(
        {
            "user_id": user_id,
            "signup_time": signup_time,
            "purchase_time": purchase_time,
            "purchase_value": purchase_value,
            "device_id": device_id,
            "source": source,
            "browser": browser,
            "sex": sex,
            "age": age,
            "ip_address": ip_address,
            "raw_class": raw_class,
        }
    )
    # Enforce the declared integer dtypes
    int_cols: List[str] = ["user_id", "sex", "age", "raw_class"]
    # Cast every integer column to int64
    fraud_ecom_raw[int_cols] = fraud_ecom_raw[int_cols].astype("int64")
    # Cast the purchase value to float64
    fraud_ecom_raw["purchase_value"] = fraud_ecom_raw["purchase_value"].astype("float64")
    # Return the schema-compliant raw FraudEcom frame
    return fraud_ecom_raw


def generate_default_raw(n_rows: int = 30_000, positive_rate: float = 0.221) -> pd.DataFrame:
    """
    Generate a synthetic UCI Default of Credit Card Clients raw DataFrame.

    Purpose:
        Creates default_raw with the 25-column ingestion schema of
        config.yaml, including the standard PAY_* repayment codes (-2..8),
        the six-month BILL_AMT/PAY_AMT panel, and the default label.

    Inputs:
        n_rows (int): Number of rows (default 30,000).
        positive_rate (float): Target default rate (default 0.221).

    Processes:
        1. Draw demographics (LIMIT_BAL, SEX, EDUCATION, MARRIAGE, AGE).
        2. Draw the six-month repayment panel (PAY_0..PAY_6, BILL_AMT1..6,
           PAY_AMT1..6) with realistic codes and amounts.
        3. Place the exact default count via deterministic permutation.
        4. Cast every column to int64.

    Outputs:
        pd.DataFrame: default_raw with 25 columns and n_rows rows.

    Raises:
        ValueError: If n_rows < 1.
    """
    # Validate the row count is positive
    if n_rows < 1:
        # Reject degenerate frame sizes
        raise ValueError(f"n_rows must be >= 1, received {n_rows}.")
    # Place the exact default positives at deterministic positions
    default_mask: np.ndarray = _fixed_rate_positions(n_rows, positive_rate, rng)
    # Draw unique client identifiers
    ids: np.ndarray = np.arange(1, n_rows + 1, dtype=np.int64)
    # Draw the available credit limit in NTD
    limit_bal: np.ndarray = rng.integers(10_000, 1_000_001, size=n_rows)
    # Draw the raw sex code (1 = male, 2 = female)
    sex: np.ndarray = rng.integers(1, 3, size=n_rows)
    # Draw the education code including the recodable values 0, 5, 6
    education: np.ndarray = rng.choice(np.array([0, 1, 2, 3, 4, 5, 6]), size=n_rows, p=np.array([0.02, 0.35, 0.35, 0.20, 0.05, 0.02, 0.01]))
    # Draw the marriage code including the recodable value 0
    marriage: np.ndarray = rng.choice(np.array([0, 1, 2, 3]), size=n_rows, p=np.array([0.02, 0.45, 0.35, 0.18]))
    # Draw the age in [20, 70]
    age: np.ndarray = rng.integers(20, 71, size=n_rows)
    # Draw the repayment status codes for the six months (most recent first)
    pay_columns: Dict[str, np.ndarray] = {}
    # The standard PAY code vocabulary ranges from -2 (no consumption) to 8
    pay_codes: np.ndarray = np.array([-2, -1, 0, 1, 2, 3, 4, 5, 6, 7, 8])
    # Build each of the six repayment columns from the code vocabulary
    for col in ["PAY_0", "PAY_2", "PAY_3", "PAY_4", "PAY_5", "PAY_6"]:
        # Sample the repayment code for the month
        pay_columns[col] = rng.choice(pay_codes, size=n_rows, p=np.array([0.05, 0.05, 0.50, 0.20, 0.10, 0.04, 0.03, 0.02, 0.005, 0.003, 0.002]))
    # Build the six bill-amount columns (declining panel)
    bill_columns: Dict[str, np.ndarray] = {}
    # Generate each month's bill amount with a decreasing drift
    for i in range(1, 7):
        # Bill amounts decline gently across the six-month window
        bill_columns[f"BILL_AMT{i}"] = rng.integers(0, 600_000, size=n_rows)
    # Build the six payment-amount columns (often zero for defaults)
    pay_amt_columns: Dict[str, np.ndarray] = {}
    # Generate each month's payment amount with mass at zero
    for i in range(1, 7):
        # Zero payments are common among defaulters
        pay_amt_columns[f"PAY_AMT{i}"] = np.where(rng.random(n_rows) < 0.3, 0, rng.integers(1, 100_000, size=n_rows))
    # Encode the default label as int64 (y = 1 for DEFAULT_NEXT_MONTH == 1)
    default_next_month: np.ndarray = default_mask.astype(np.int64)
    # Assemble the raw frame with the exact 25-column schema
    default_raw: pd.DataFrame = pd.DataFrame(
        {"ID": ids, "LIMIT_BAL": limit_bal, "SEX": sex, "EDUCATION": education,
         "MARRIAGE": marriage, "AGE": age, **pay_columns, **bill_columns,
         **pay_amt_columns, "DEFAULT_NEXT_MONTH": default_next_month}
    )
    # Cast every column to the declared int64 dtype
    default_raw = default_raw.astype("int64")
    # Return the schema-compliant raw Default frame
    return default_raw


def generate_creditcard_raw(n_rows: int = 284_807, positive_rate: float = 0.0017) -> pd.DataFrame:
    """
    Generate a synthetic Credit Card Fraud raw DataFrame at paper scale.

    Purpose:
        Creates creditcard_raw with the 31-column ingestion schema of
        config.yaml: Time, V1..V28 (anonymized PCA components, negative
        values allowed), Amount, and the Class label.

    Inputs:
        n_rows (int): Number of rows (default 284,807).
        positive_rate (float): Target fraud rate (default 0.0017).

    Processes:
        1. Draw the elapsed-seconds Time column.
        2. Draw the 28 PCA components from a standard normal (signed).
        3. Draw the Amount with a heavy right tail.
        4. Place the exact fraud count via deterministic permutation.

    Outputs:
        pd.DataFrame: creditcard_raw with 31 columns and n_rows rows.

    Raises:
        ValueError: If n_rows < 1.
    """
    # Validate the row count is positive
    if n_rows < 1:
        # Reject degenerate frame sizes
        raise ValueError(f"n_rows must be >= 1, received {n_rows}.")
    # Place the exact fraud positives at deterministic positions
    fraud_mask: np.ndarray = _fixed_rate_positions(n_rows, positive_rate, rng)
    # Draw the elapsed seconds since the first transaction (0..172,800)
    time_sec: np.ndarray = rng.uniform(0.0, 172_800.0, size=n_rows)
    # Build the 28 anonymized PCA components as signed standard normals
    pca_components: Dict[str, np.ndarray] = {f"V{i}": rng.normal(0.0, 1.0, size=n_rows) for i in range(1, 29)}
    # Draw the transaction amount with a heavy right tail (EUR)
    amount: np.ndarray = np.exp(rng.normal(3.0, 1.2, size=n_rows)).round(2)
    # Encode the fraud label as int64 (y = 1 for fraud)
    class_label: np.ndarray = fraud_mask.astype(np.int64)
    # Assemble the raw frame with the exact 31-column schema
    creditcard_raw: pd.DataFrame = pd.DataFrame(
        {"Time": time_sec, **pca_components, "Amount": amount, "Class": class_label}
    )
    # Enforce the declared dtypes: all numeric columns as float64
    creditcard_raw[["Time", "Amount"] + [f"V{i}" for i in range(1, 29)]] = creditcard_raw[
        ["Time", "Amount"] + [f"V{i}" for i in range(1, 29)]
    ].astype("float64")
    # Cast the label column to int64
    creditcard_raw["Class"] = creditcard_raw["Class"].astype("int64")
    # Return the schema-compliant raw Creditcard frame
    return creditcard_raw
```

### **Step 2: Loading the Configuration (`config.yaml` → `config`)**

The pipeline is driven by the single source of truth stored in `config.yaml`, which is assumed to be saved in the working directory. The file is read into memory and converted into the pipeline-usable Python dictionary `config` via `yaml.safe_load`. No manual conversion is required: the Task 1 configuration validator (`validate_configuration`) normalizes the dictionary internally (seeds coerced to a tuple, split fractions cast to `float`) and aborts with a `ConfigError` if any paper constant is violated.

**Methodology:**
1.  **File I/O:** Open `config.yaml` in read mode.
2.  **Parsing:** Use `yaml.safe_load` to convert the YAML structure into a nested Python dictionary.
3.  **Validation:** Raise informative errors for missing files and malformed YAML; verify the top-level sections are present.

```python
# Cell 3: load the study configuration from the working-directory YAML file

def load_study_configuration(filepath: str = "config.yaml") -> Dict[str, Any]:
    """
    Load the study configuration parameters from a YAML file into a Python dict.

    Purpose:
        To ingest the deterministic hyperparameters, data-split definitions,
        schema contracts, and verification targets defined in the external
        configuration file. This ensures reproducibility by separating code
        from configuration: every UNSPECIFIED IN MANUSCRIPT placeholder is
        resolved exactly once, in this file.

    Inputs:
        filepath (str): The relative or absolute path to the YAML
                        configuration file. Default is "config.yaml".

    Processes:
        1. Validate that filepath is a string.
        2. Open the file in read mode (UTF-8).
        3. Parse the YAML structure with yaml.safe_load.
        4. Verify that the parsed result is a non-empty dictionary.

    Outputs:
        Dict[str, Any]: The nested configuration dictionary, directly usable
                        as the study_configuration argument of
                        run_end_to_end_pipeline.

    Raises:
        TypeError: If filepath is not a string.
        FileNotFoundError: If the file does not exist.
        yaml.YAMLError: If the file contains invalid YAML syntax.
        ValueError: If the parsed YAML is not a non-empty dictionary.
    """
    # Validate the input type before any file access
    if not isinstance(filepath, str):
        # Reject non-string paths with an informative message
        raise TypeError(f"filepath must be a str, received {type(filepath).__name__}.")
    # Open the YAML file in read mode with explicit UTF-8 encoding
    with open(filepath, "r", encoding="utf-8") as yaml_stream:
        # Parse the YAML content safely (no arbitrary object construction)
        config: Any = yaml.safe_load(yaml_stream)
    # Verify the parsed content is a dictionary
    if not isinstance(config, dict):
        # Reject YAML files that do not declare a mapping
        raise ValueError(f"config.yaml must contain a YAML mapping, received {type(config).__name__}.")
    # Verify the mapping is not empty
    if len(config) == 0:
        # Reject empty configuration files
        raise ValueError("config.yaml is empty; no study parameters were loaded.")
    # Confirm the load to the console
    print(f"Successfully loaded configuration from {filepath}")
    # Return the pipeline-usable configuration dictionary
    return config


# Load the configuration (config.yaml is assumed to be in the working directory)
config: Dict[str, Any] = load_study_configuration()

# Print the top-level sections of the loaded configuration
print("\nTop-level configuration sections:")
# Iterate over the loaded top-level keys and print them in order
for section_key in config.keys():
    print(f"  - {section_key}")
```

### **Step 3: Pipeline Input Parameters and Defaults**

The `run_end_to_end_pipeline` function exposes five input parameters. Each is described below with its data structure and its default value in this example:

1.  **`study_configuration: Dict[str, Any]`** — The single source of truth. In this example it is the `config` dictionary loaded from `config.yaml` in Step 2. It contains every paper-derived constant (dataset sizes, base rates, split fractions, scaling bounds, $L = \max\{|\log 0.01|, |\log 10.01|\}$, tolerance 0.01, $P_{\max}=5$, $B=10$, seeds) and every replicator-chosen UNSPECIFIED IN MANUSCRIPT placeholder.
2.  **`raw_dfs: Dict[str, pd.DataFrame]`** — The four raw DataFrames keyed by dataset name (`"Loan"`, `"FraudEcom"`, `"Default"`, `"Creditcard"`), synthetically generated in Step 1 at the paper's population sizes.
3.  **`output_dir: Path`** — The project output root. Every stage persists under `output_dir/artifacts` (versioned sub-directories: `validation/`, `cleansed/`, `splits/`, `preprocessing/`, `scaled/`, `hyperparameters/`, `monomial/`, `pruning/`, `scorecard/`, `iht/`, `survey/`, plus the final tables and figures). Default in this example: `Path("study_outputs")`.
4.  **`survey_records: Optional[List[Dict[str, Any]]]`** — The raw survey response records for the Task 13 human-assessment stage. **Default: `None`**, which activates the pipeline's explicit, recorded skip policy: the survey stage is skipped (the paper's 36-respondent study is not fabricatable). When real records exist, each entry carries respondent id, professional background, per-representation Likert ratings, the forced-choice comprehension answer, and the preference choice.
5.  **`figures_data: Optional[Dict[str, Any]]`** — Figure-data dictionaries keyed `"fig1"`..`"fig4"` for the Task 14 persistence stage. **Default: `None`**, in which case the figures are derived internally from the survey analysis and the seed-45 display models.

### **Step 4: Executing the End-to-End Pipeline**

With the four synthetic raw frames (`raw_dfs`) and the loaded configuration (`config`) in memory, the top-level orchestrator is invoked. The function manages the complete lifecycle: Task 1 validation of schemas, configuration constants, and label rates; Task 2 leakage-free cleansing with checksummed persistence; Task 3 feature-matrix construction (42/6/26/30); Task 4 stratified splits and train-only min-max scaling; Task 5 Optuna hyperparameter search; Task 11 as the quantitative engine (training, pruning, directional rule with closed-form fidelity prediction, scorecard/tally, isotonic calibration); Task 12 IHT robustness; Task 13 survey (skipped); Task 14 aggregation and target verification; and Task 15 documentation and audit.

**Methodology:**
1.  **Function Call:** Pass `config`, `raw_dfs`, and `output_dir` to `run_end_to_end_pipeline`.
2.  **Compute gate:** Because the full run executes 100 Optuna trials and five-seed training on 395,492-row data (multi-hour CPU budget), the call is guarded by an explicit `EXECUTE_PIPELINE` flag. Set it to `True` to run; the inspection code that follows is valid either way.
3.  **Output Handling:** The returned dictionary contains the per-stage artifacts — `validation`, `cleansing`, `feature_matrices`, `preprocessing`, `hyperparameters`, `pipeline_results`, `iht`, `survey`, `aggregation`, `verification`, `figures_persistence`, `documentation`, and `artifacts_root`.

### **Summary of the Execution Flow**

1.  **Data Ingestion:** The four synthetic raw DataFrames (`loan_raw` at 399,492 rows with 395,492 keep-rows, `fraud_ecom_raw` at 151,112, `default_raw` at 30,000, `creditcard_raw` at 284,807) are generated with Faker + NumPy at the paper's exact population sizes and positive rates (0.101 / 0.094 / 0.221 / 0.0017), then keyed as `raw_dfs`.
2.  **Configuration Ingestion:** `config.yaml` (working directory) is read with `yaml.safe_load` into the dictionary `config` — the single source of truth carrying every paper constant and every UNSPECIFIED IN MANUSCRIPT placeholder; the Task 1 validator normalizes it (seeds → tuple, fractions → float) and aborts on any contradiction.
3.  **Validation (Task 1):** `orchestrate_validation` checks the configuration against paper constants (including the recomputed $L = \max\{|\log 0.01|, |\log 10.01|\}$), validates all four raw schemas (required columns, dtypes, row-count lower bounds, aliases), and verifies the raw label rates within $\pm 0.001$.
4.  **Cleansing (Task 2):** `cleanse_all` applies the deterministic parsers, recodings, and the Loan status filter ($y=1$ for "Charged Off"), builds missingness flags only, and persists checksummed parquet artifacts; the raw inputs are never re-read afterwards.
5.  **Feature Engineering and Preprocessing (Tasks 3–4):** Deterministic matrices (42/6/26/30) are built — FraudEcom count features from training rows only — then stratified 60/20/20 splits are made per seed and the affine transform $\tilde{x}_j = 0.01 + 10.0\,(x_j - \min_j)/(\max_j - \min_j)$ is fit on training only and applied to all partitions with clipping to $[0.01, 10.01]$.
6.  **Hyperparameter Search (Task 5):** Optuna (TPE, median pruning, 100 trials) tunes $\eta$, $\lambda$, and epochs once per dataset on the seed-0 split via five-fold cross-validated PR-AUC; the record is reused by all five seeds.
7.  **Quantitative Engine (Task 11):** For every dataset and seed, the monomial $z(x) = \alpha \prod_j x_j^{\beta_j}$ is trained with automatic differentiation; pruning selects the smallest support within the 0.01 validation tolerance and verifies the worst-case bound $|s_{full} - s_{pruned}| \le L \sum_{j \in S}|\beta_j|$; the directional rule is built and its fidelity predicted a-priori via $\rho = \beta^\top\Sigma d/(\sqrt{\beta^\top\Sigma\beta}\sqrt{d^\top\Sigma d})$ and $\tau_{pred} = \frac{2}{\pi}\arcsin(\rho)$; scorecard $V(x)$ and tally $T(x)$ thresholds are selected on validation F1; all five representations are isotonically calibrated and evaluated (PR-AUC, $\tau_b$, ECE).
8.  **IHT Robustness (Task 12):** Iterative hard thresholding trains an independent sparse monomial and the Jaccard overlap with the pruned support is recorded.
9.  **Survey (Task 13):** Skipped under the explicit skip policy because `survey_records=None` — no synthetic human-subjects data is fabricated.
10. **Aggregation and Verification (Task 14):** Means and standard deviations over the five seeds are computed and the paper's headline targets (pruning loss ≤ 0.008, Loan scorecard ≈ 0.329 / tally ≈ 0.300, Default directional ≈ 0.381 with $\tau_b$ ≈ 0.744, fidelity MAE < 0.025 vs ≈ 0.236, ECE < 0.02) are checked as soft invariants with a diagnostic report.
11. **Documentation and Audit (Task 15):** The methodological appendix, environment documentation, and the mathematical audit suite (log-space identity, pruning bound, $\rho$ formula, ECE binning, scorecard points, tied $\tau_b$) are produced, and the reproduction is signed off.
12. **Output:** The `EndToEndPipelineResult` dictionary (stages: `validation`, `cleansing`, `feature_matrices`, `preprocessing`, `hyperparameters`, `pipeline_results`, `iht`, `survey`, `aggregation`, `verification`, `figures_persistence`, `documentation`, `artifacts_root`) is returned, with every artifact persisted under `output_dir/artifacts` for audit and re-analysis.

## Output Structure

The pipeline returns an `EndToEndPipelineResult` dictionary containing the stage keys:

-   **`validation`** — The merged Task 1 report (`config`, `schemas`, `label_rates`) with the recomputed log-bound constant $L$ and the per-dataset label-rate verdicts.
-   **`cleansing`** — The Task 2 persistence report (checksummed parquet paths and the cleansing report).
-   **`feature_matrices` / `preprocessing`** — The Task 3/4 persistence reports (feature matrices, split indices, fitted statistics, scaled matrices, clip counts).
-   **`hyperparameters`** — The per-dataset Optuna records ($\eta^*$, $\lambda^*$, epochs, best CV PR-AUC) and study databases.
-   **`pipeline_results`** — The nested per-dataset per-seed quantitative records: fitted `(alpha, beta)`, support sets, retained exponents, per-representation test scores, PR-AUC, $\tau_b$, ECE, $\rho$, $\tau_{\mathrm{pred}}$, thresholds, calibrated scores, and prediction artifacts.
-   **`iht`** — The IHT comparison records (Jaccard index, $r_{prune}$, $r_{IHT}$, PR-AUCs, feature names).
-   **`survey`** — The Task 13 stage output (instruments, anonymized responses, analysis, trade-off table) or the recorded skip verdict.
-   **`aggregation` / `verification`** — The five-seed mean/SD tables and the soft-invariant target verification report.
-   **`figures_persistence`** — The persisted table and figure equivalents (seed-45 displays labeled as representative, not averaged).
-   **`documentation`** — The methodological appendix path, environment documentation, and the machine-readable `audit_report.json`.
-   **`artifacts_root`** — The versioned artifact archive root (`output_dir/artifacts`).

# Paper  

Title: "*How Simple Can It Get? From Interpretable Equations to Readable Rules for Financial Decision Making*"

Authors: Adia Lumadjeng, Ilker Birbil, Erman Acar

E-Journal Submission Date: 10 August 2026

Link: https://arxiv.org/abs/2608.09433v1

Abstract:

In regulated domains such as finance, a model that cannot be explained cannot be deployed, yet many interpretable classifiers defeat their own purpose by producing formulas with dozens of features that no regulator could read. We take the reverse direction. Starting from an interpretable classifier expressed as a single equation over the input features, we progressively simplify it into more readable forms, including a pruned monomial, a directional if--then rule, and the integer scorecards and tallies that finance already deploys. Because the equation is itself the predictive model rather than a post-hoc explanation we can directly quantify what is lost under each simplification. Across four financial datasets, we find that pruning is nearly free and that fidelity can erode faster than predictive performance, allowing simpler rules to remain effective classifiers without faithfully reproducing the original model. A human assessment shows that simplification improves perceived readability, while preferences for different representations vary by professional background. Beyond measuring these losses empirically, we show that some can be anticipated from the original model: we derive a bound on the change caused by pruning and predict how faithfully a rule retaining only the direction of each feature's effect preserves the original ranking.

# Paper Summary

# How Simple Can It Get? From Interpretable Equations to Readable Rules for Financial Decision Making

**A Formal Summary**

*Source paper: Adia Lumadjeng, Ilker Birbil, and Erman Acar (University of Amsterdam)*



## Executive Summary

In regulated financial domains, a model that cannot be explained cannot be deployed. The prevailing response has been interpretable-by-design classifiers, whose reasoning is legible from the model itself. This paper identifies a gap that such designs do not close: structural transparency does not guarantee practical readability. An interpretable-by-design classifier trained on loan default produces a single equation spanning forty-two features — nothing is hidden, yet no regulator could reasonably read it.

The authors therefore invert the usual research direction. Rather than constructing transparent models that retain accuracy, they begin from an already-interpretable model and ask how far its representation can be simplified before something important is lost. Starting from a fitted single-term monomial classifier, they derive four progressively simpler representations — a pruned monomial, a directional if–then rule, an integer scorecard, and a tally — each removing a different, identifiable component of the original model's information.

The decisive methodological move is that the equation *is* the predictive model, not a post-hoc explanation of one. Each simplified rule can therefore be evaluated as the classifier it displays, making the cost of every simplification directly measurable rather than assumed.

Three findings stand out. First, the reductions do not lie on a single accuracy–readability trade-off: pruning weakly contributing features is nearly free, removing effect magnitudes is often cheap, but binarizing continuous feature values is expensive. Second, predictive performance and fidelity to the original model can diverge — a simpler rule may remain an effective classifier while no longer faithfully reproducing the model it claims to represent. Third, some of these losses are predictable in advance: the authors derive a worst-case bound on the perturbation caused by pruning and a closed-form prediction of the directional rule's rank fidelity, computable immediately after training. A human assessment confirms that simplification improves perceived ease, but shows that which representation practitioners actually prefer depends on professional background.



## Main Discussion

### 1. Problem Setting and Motivation

Credit default and fraud detection are imbalanced binary classification problems in which decisions must be explainable to a customer, auditor, or supervisor. The interpretability literature has largely worked forward — building models that stay transparent while retaining accuracy. The authors argue this leaves a question unaddressed: given an already-interpretable model, how far can its *representation* be simplified while preserving predictive performance, fidelity to the original model, and gaining genuine human readability?

This distinction matters specifically for interpretable-by-design methods. The authors note that the original ECSEL work evaluates the fitted model at full size while using pruning only to produce more legible displayed equations — it never evaluates how those simplified displays perform as classifiers, nor how faithfully they preserve the fitted model's behaviour. This paper makes that transition explicit: the rule shown to the reader is also the rule whose predictions are evaluated.

### 2. The Model

The starting point is ECSEL, which learns a signomial equation serving simultaneously as classifier and explanation. The authors restrict it to its simplest single-term form (`K = 1`), a monomial:

$$z(x) = \alpha \prod_{j=1}^{m} x_j^{\beta_j}, \qquad P(y = 1 \mid x) = \sigma(z(x)).$$

Because exponents are real-valued, every feature is scaled to a fixed positive range $[\ell, u]$ with $0 < \ell$, fit on the training split alone. Taking logarithms linearises the model:

$$\log|z(x)| = \log|\alpha| + \sum_{j=1}^{m} \beta_j \log x_j.$$

This decomposition exposes exactly three interpretable components — **support** (which features participate), **direction** (the signs of $\beta_j$ and $\alpha$), and **magnitude** ($|\beta_j|$) — to which the authors add a fourth axis, whether **feature values** are continuous or binarized. These four components can be removed independently, which is what makes the cost of each simplification separable rather than conflated into a single vague notion of "readability."

### 3. The Controlled Simplifications

| Representation | Support | Values | Direction | Magnitude |
||||||
| Full monomial | All | Continuous | ✓ | Exact |
| Pruned monomial | Reduced | Continuous | ✓ | Exact |
| Directional rule | Reduced | Continuous | ✓ | — |
| Scorecard | Reduced | Binary | ✓ | Quantized |
| Tally | Reduced | Binary | ✓ | — |

**Pruned monomial.** Retains the $r$ features with the largest $|\beta_j|$, zeroing the rest. The authors justify this criterion with a worst-case bound: with $L := \max\{|\log \ell|, |\log u|\}$ and $S$ the removed set,

$$|s_{\mathrm{full}}(x) - s_{\mathrm{pruned}}(x)| \le L \sum_{j \in S} |\beta_j| \quad \text{for every } x.$$

Crucially, the authors are careful about what this establishes. It is a *selection criterion*, not a performance guarantee: among all ways of discarding $m - r$ features, dropping the smallest magnitudes minimises the worst-case log-score perturbation, but nothing is promised about predictive performance or ranking. The support size $r$ is therefore chosen empirically on validation data.

**Scorecard and tally.** Each retained feature becomes a binary condition at its training-set median, with the direction set by the sign of $\alpha\beta_j$. The tally counts satisfied conditions equally; the scorecard assigns integer points $p_j = \max(1, \mathrm{round}(P_{\max}|\beta_j|/\max_k|\beta_k|))$, the floor of one ensuring every retained feature stays visible on the card. The tally is exactly the scorecard with all points set to one.

**Directional rule.** Replaces every exponent with its sign, giving $s_{\mathrm{dir}}(x) = \sum_{j} \mathrm{sign}(\beta_j) \log x_j$. It preserves support, direction, and continuous values while discarding relative strength entirely.

### 4. Predicting Fidelity Before Simplification

The paper's sharpest theoretical contribution concerns the directional rule, whose cost is least predictable from its form alone. Writing $w = (\log x_1, \dots, \log x_m)$ with covariance $\Sigma$, and $d_j = \mathrm{sign}(\beta_j)$ on the retained set, the correlation between full and directional log-scores is

$$\rho = \frac{\beta^\top \Sigma d}{\sqrt{\beta^\top \Sigma \beta}\sqrt{d^\top \Sigma d}},$$

which holds without distributional assumptions. If $w$ is elliptically distributed, Greiner's classical relation converts this into a predicted rank fidelity, $\tau = \tfrac{2}{\pi}\arcsin(\rho)$.

The practical significance is that both $\beta$ and $\Sigma$ are available immediately after training. The expected agreement between the directional rule and the original classifier can therefore be estimated **before the simplified rule is constructed or evaluated**, requiring no held-out data.

### 5. Experimental Design

Four public financial datasets span an order of magnitude in size and positive-class rates from 0.17% to 22.1%: **Creditcard** (284,807 rows, 30 features, 0.17%), **FraudEcom** (151,112, 6, 9.4%), **Loan** (395,492, 42, 10.1%), and **Default** (30,000, 26, 22.1%). They also differ in feature type — named continuous attributes, small integer counts, and anonymized principal components respectively — a difference that proves consequential.

Protocol discipline is strict: 60/20/20 stratified splits over five random seeds; scaling to $[0.01, 10.01]$ fit on training data only; hyperparameters via Optuna; pruning level and all decision thresholds selected on validation; the test split used only for final evaluation. PR-AUC is the primary metric given class imbalance, tie-corrected Kendall's $\tau_b$ measures fidelity, and each non-probabilistic representation is recalibrated with isotonic regression and audited by expected calibration error.

### 6. Findings

**Pruning is nearly free.** The largest test PR-AUC reduction is 0.008 (FraudEcom); Loan and Creditcard lose 0.001 and 0.002. Fidelity stays high, $\tau_b$ between 0.70 and 0.85. On Loan, forty-two features reduce to seven in the representative display with virtually no predictive loss.

**Binarizing feature values is costly.** On Loan, PR-AUC falls from 0.885 (pruned) to 0.329 (scorecard) and 0.300 (tally). On Creditcard both point-based forms collapse to roughly 0.018 and 0.017, near the 0.002 base rate — attributed to thresholding anonymized principal components, which discards most of the information carried in their continuous values. Continuous feature values evidently carry more predictive information than precise exponent magnitudes.

**Performance and fidelity diverge.** On Default, the directional rule essentially matches the full monomial (PR-AUC 0.381 versus 0.380) despite rank fidelity of only 0.744. On Loan, fidelity falls to 0.568 while predictive performance remains far above the point-based forms. A simplified rule can therefore remain an effective classifier without faithfully reproducing the model it purports to represent — the paper's most consequential negative result for practitioners who treat readable explanations as trustworthy proxies.

**Simplified scores calibrate reliably.** After isotonic recalibration, ECE remains low across all representations and datasets, showing that ranking loss does not preclude reliable probability estimates.

**Fidelity prediction largely holds.** On Loan, Default, and Creditcard the prediction is accurate across fifteen fitted models: mean absolute error 0.017, maximum error 0.039, over observed fidelities from 0.25 to 0.83. FraudEcom is the instructive exception (MAE 0.236): its dominant features are low-cardinality count variables producing many tied scores. Pearson correlation is insensitive to tie structure whereas $\tau_b$ is not, so the failure is one of the elliptical distributional assumption, not of the correlation formula.

**Human assessment.** Of 36 respondents, 34 were analysed — 14 with finance/risk backgrounds and 20 AI/ML researchers. Objective comprehension was high across all four forms (92–100%), so differences concern perceived ease and preference rather than basic understanding. All three simplified rules were perceived as easier than the pruned monomial. Preferences, however, split sharply by background: finance/risk respondents chose the directional rule in 57% of assessments versus 29% for point-based forms, while AI/ML researchers chose point-based forms in 72% versus 18% for the directional rule. The pruned monomial was rarely preferred by either group.

**Robustness.** Comparing post-hoc pruning against in-training iterative hard thresholding, both routes retain nearly identical feature sets — Jaccard overlap 1.00 on Loan, FraudEcom, and Creditcard, and 0.83 on Default — with comparable predictive performance. The extracted structure therefore reflects learned data structure rather than a pruning artifact.

### 7. Limitations

The authors are explicit about scope. The study covers only single monomials ($K = 1$) and four financial datasets. The readable forms use deliberately fixed simplifications, such as median binarization, to isolate the cost of removing specific information rather than to optimise each representation — so the results characterise the cost of these simplifications, not the best achievable performance of each rule form. The human assessment is exploratory, with 36 respondents, and measures stated preferences rather than use in real decisions.



## Conclusion

The paper's central contribution is conceptual as much as empirical: it reframes readable explanations as **measurable reductions of an interpretable model** rather than as representations that may be assumed faithful because they are easy to read. By making the displayed rule identical to the evaluated classifier, the authors convert a question usually settled by intuition into one settled by measurement.

The resulting picture resists the familiar "accuracy–readability trade-off" framing. Different simplifications remove different information, and their cost depends on where a given dataset's predictive signal resides. Removing weakly contributing features is nearly free everywhere. Removing effect magnitudes is often cheap, and — uniquely among the simplifications studied — its effect on fidelity can be anticipated from the fitted model before the rule is built. Binarizing continuous values consistently incurs the largest predictive losses, severely so when features are latent components rather than meaningful financial variables.

The divergence between predictive performance and fidelity is the finding with the most direct governance implications. A scorecard or directional rule that classifies well may nonetheless rank cases differently from the model it was derived from. For institutions presenting simplified rules to customers, auditors, or supervisors as accounts of a deployed model, accuracy alone is not evidence of faithfulness; the two must be measured separately.

Finally, the human assessment cautions against a single notion of readability. Simplification reliably improved perceived ease, but finance/risk practitioners and AI/ML researchers preferred opposite representations — suggesting that the appropriate readable form is a function of audience, not a property of the rule. Consistent with the authors' own ethics statement, simplification should complement rather than replace domain validation, fairness assessment, and regulatory review.

 ## **Discussion of the Inputs, Processes and Outputs (IPOs) of Key Callables**

 Below is a discussion of the Inputs, Processes and Outputs (IPOs) of the sixteen task-specific orchestrator callables in the "How Simple Can It Get? From Interpretable
 Equations to Readable Rules for Financial Decision Making" end-to-end reproduction pipeline (Lumadjeng, Birbil, Acar 2026), as implemented in the reproduction-study
 workbook:

 ### Task 1 — orchestrate_validation(loan_raw, fraud_ecom_raw, default_raw, creditcard_raw, study_configuration, output_dir) → ValidationResult

 Inputs: loan_raw, fraud_ecom_raw, default_raw, creditcard_raw: pd.DataFrame — raw Kaggle/UCI frames; study_configuration: Dict single source of truth; output_dir: Path.

 0.01|,|log 10.01|}=4.605170185988091; Step 1.1 validate_all_raw_schemas required-column + dtype checks with alias resolution, row-count lower bounds, missingness report;
 Step 1.3 validate_all_label_rates — Loan filter {Fully Paid,Charged Off} reproducing n=395,492 and rate 0.101 within 0.5%/0.001, others within ±0.001; persists
 validation_report.json + label_check_report.json.

 ### Task 5 — orchestrate_hyperparameter_search(X_train, y_train, cfg, dataset_name, output_dir, seed=0) → HyperparameterResult

 Inputs: Seed-0 scaled X_train, y_train; cfg (search spaces, 100-trial budget); output_dir.

 convergence (never fold-validation PR-AUC — anti-leakage); persists SQLite study + checksummed JSON (η^*,λ^*,epochs^*, best CV PR-AUC).

 Transformation: Training-only matrix/labels → frozen per-dataset hyperparameter record reused by all five seeds.

 Role: §4 Setup ("Hyperparameters were optimized using Optuna on the training split, with the best configuration selected by cross-validation") for the regularized objective
 ℒ=-1/n∑ᵢ[yᵢlog σ(zᵢ)+(1-yᵢ)log(1-σ(zᵢ))]+λ∑ⱼ|βⱼ|.

 ### Task 6 — orchestrate_monomial_training(X_train, y_train, hparams, cfg, dataset_name, seed, output_dir, X_val=None, y_val=None) → MonomialTrainingResult

 Process: Step 6.2 train_monomial — init α=1.0,βⱼ=0.1; precompute log X; automatic-differentiation gradient descent on the stable log-space forward
 z(x)=sign(α)exp(log|α|+β^⊤log x) and the ℓ₁-regularized loss; stop when mean absolute relative loss change over 20 epochs <10⁻⁶ or budget exhausted; validation PR-AUC after
 training; Step 6.3 verify_trained_monomial — finiteness on 100 rows, log-space identity log|z(x)|=log|α|+∑ⱼβⱼlog xⱼ within 10⁻⁶, smooth-loss gradient norm logged not
 aborted; persists params + loss history JSON.

 Transformation: Scaled features/labels → fitted verified monomial (α,β) — the immutable input to Tasks 7–9.

 Role: §3 "The Single Monomial": implements z(x)=α∏ⱼ₌₁ᵐ xⱼ^(βⱼ), P(y=1|x)=σ(z(x)) "learned by gradient descent, with sparsity regularization applied to the exponents … until
 convergence of the training objective"; the identity check is one of the paper's mathematical unit tests.

<br>

 ### Task 7 — orchestrate_pruning(alpha, beta, X_val, y_val, cfg, dataset_name, seed, output_dir, feature_names) → PruningResult

 Inputs: Fitted α,β; scaled validation Xᵥₐₗ,yᵥₐₗ; cfg (tolerance 0.01, L).

 Process: Step 7.1 evaluate_pruning_candidates — sort by |βⱼ| descending (tie-break index), per-r pruned logits via masked dot product, validation PR-AUC curve ∀ r∈{1..m};
 Step 7.2 select_pruning_level$ — $r^*=\min\{r:\mathrm{PR\text{-}AUC}_{val}(r)\ge\mathrm{PR\text{-}AUC}_{val}(\text{full})-0.01\}$, $r^*=m$ + warning if none; Step 7.3
 verify_pruning_bound` on validation rows — assert maxₓ|s_(full)(x)-s_(pruned)(x)|≤ L∑_(j∈ Rᶜ)|βⱼ|, abort on violation >10⁻⁶; persists support (names + exponents) and curve
 JSON.

 Transformation: Full monomial + validation data → pruned monomial (retained set R, removed set Rᶜ) with the worst-case bound empirically verified.

 Role: §3.1 "Controlling Feature Support": the Remark's bound |s_(full)-s_(pruned)|≤ L∑_(j∈ S)|βⱼ| with L=max {|log ℓ|,|log u|} "motivates exponent magnitude as the pruning
 criterion," and "r on validation data [is] the smallest value whose PR-AUC remains within a fixed tolerance" — the machinery behind "pruning is nearly free" (max loss
 0.008).

<br>

 ### Task 8 — orchestrate_directional_rule(alpha, beta, retained, X_train, X_val, X_test, cfg, dataset_name, seed, feature_names, output_dir) → DirectionalRuleResult

 Inputs: Fitted α,β; retained support; scaled train/val/test matrices (train used only for covariance).

 Process: Step 8.1 build_directional_rule — dⱼ=sign(βⱼ) on R, 0 otherwise; scores s_(dir)(x)=∑_(j∈ R)sign(βⱼ)log xⱼ on val/test; natural-language mapping from sign of αβⱼ;
 Step 8.2 predict_fidelity — training log-feature covariance Σ̂=1/(n-1)∑ᵢ(wᵢ-w̄)(wᵢ-w̄)^⊤, ρ=(β^⊤Σ̂ d)/(√(β^⊤Σ̂β)√(d^⊤Σ̂ d)) (clip |ρ|>1 + log), Greiner τ_(pred)=2/πarcsin(ρ);
 Step 8.3 record_fidelity_predictions before any held-out evaluation.

 Transformation: Exponents + support → sign-flattened scoring rule and closed-form a-priori fidelity prediction (ρ,τ_(pred)) from training data alone.

 Role: §3.1 "Removing Effect Magnitudes" + Proposition ("fidelity of the directional rule … can be estimated before the simplified rule is produced"): the only
 simplification with pre-deployment predictability; its records feed the paper's ρ/τ_(pred)/τ_b/MAE table (MAE 0.017 vs FraudEcom 0.236 tie-failure).

 ### Task 9 — orchestrate_scorecard_tally(alpha, beta, retained, X_train, X_val, X_test, y_val, y_test, cfg, dataset_name, seed, feature_names, output_dir) →

 pⱼ=max(1,round(5|βⱼ|/maxₖ|βₖ|)); scorecard_score V(x)=∑ⱼ pⱼ𝟏ⱼ(x); tally_score T(x)=∑ⱼ𝟏ⱼ(x); Step 9.3 select_thresholds — exhaustive integer search v∈{1..∑ pⱼ}, t∈{1..|R|}

 v^*,t^*.

 whose cost the paper measures as largest (Loan scorecard 0.329, tally 0.300).

## **Usage Example**

Below is a granular, step-by-step guide to executing the end-to-end pipeline for **"How Simple Can It Get? From Interpretable Equations to Readable Rules for Financial Decision Making"**. This example demonstrates how to synthetically generate the four required raw financial DataFrames at the paper's population scale, load the study configuration from `config.yaml`, and execute the complete reproduction pipeline through the `run_end_to_end_pipeline` orchestrator.

<br>

### **Step 1: Synthetic Data Generation (the four raw DataFrames)**

The pipeline ingests four raw DataFrames keyed by dataset name: `"Loan"`, `"FraudEcom"`, `"Default"`, and `"Creditcard"`. Each frame must satisfy the `raw_data_schemas` contracts declared in `config.yaml` — exact column presence, declared pandas dtypes, row-count lower bounds, and the paper's positive-class rates — because Task 1 validation runs before anything else. The generators below produce frames at the exact paper-declared population sizes so every `min_rows` lower bound and every label-rate check (within `+-0.001`) passes deterministically.

**Methodology:**
1.  **Seeding:** A single `numpy.random.default_rng` and `Faker.seed(42)` make every generated frame bitwise reproducible.
2.  **Pooled Faker values:** Faker generates high-cardinality pools (IP addresses, device ids, dates); NumPy samples from these pools at scale, so the large frames are built in milliseconds without millions of Faker calls. Pooled repeats are realistic — they are precisely what the training-only `ip_shared_count` / `device_shared_count` construction in Task 3 exploits.
3.  **Exact label counts:** The positive counts are computed as `round(n * positive_rate)` and placed by deterministic permutation, so the observed rates match the paper's rates exactly (Loan 0.101, FraudEcom 0.094, Default 0.221, Creditcard 0.0017).
4.  **Schema enforcement:** Columns are explicitly cast to the declared dtypes (`int64`, `float64`, `object`); NaN is injected only where the schema allows it (`revol_util`, `next_pymnt_d`, `mths_since_last_delinq`, `mths_since_last_record`).

```python
# Cell 1: imports, seeding, and the shared random pool
import numpy as np
import pandas as pd
from faker import Faker
from pathlib import Path
from typing import Any, Dict, List, Optional, Tuple

import yaml

# Initialize Faker with a fixed seed for fully reproducible synthetic values
fake = Faker()
Faker.seed(42)
# Create a single deterministic NumPy generator shared by all builders
rng = np.random.default_rng(seed=42)


# -----------------------------------------------------------------------------
# Helper: build the four raw frames with exact paper-scale row counts
# -----------------------------------------------------------------------------

def _fixed_rate_positions(n_rows: int, positive_rate: float, rng: np.random.Generator) -> np.ndarray:
    """
    Return a boolean array with exactly round(n * rate) True entries.

    Purpose:
        Places the minority-class positives at deterministic random positions
        so the observed positive rate equals the paper's rate exactly, keeping
        the Task 1.3 label-rate checks (tolerance +-0.001) pass by construction.

    Inputs:
        n_rows (int): Total number of rows in the frame.
        positive_rate (float): Target positive-class rate in [0, 1].
        rng (np.random.Generator): Deterministic random generator.

    Processes:
        1. Compute the exact positive count as round(n_rows * positive_rate).
        2. Build an all-False boolean array of length n_rows.
        3. Set exactly that many entries to True at permuted positions.

    Outputs:
        np.ndarray: Boolean mask of shape (n_rows,) with the exact count of
                    True entries.

    Raises:
        ValueError: If positive_rate is outside [0, 1] or n_rows < 1.
    """
    # Validate the row count is a positive integer
    if n_rows < 1:
        # Reject degenerate frame sizes
        raise ValueError(f"n_rows must be >= 1, received {n_rows}.")
    # Validate the rate is a proper probability
    if not 0.0 <= positive_rate <= 1.0:
        # Reject rates outside the unit interval
        raise ValueError(f"positive_rate must be in [0, 1], received {positive_rate}.")
    # Compute the exact positive count: n_pos = round(n * pi)
    n_positive: int = int(round(n_rows * positive_rate))
    # Initialise an all-negative boolean mask of the frame length
    mask: np.ndarray = np.zeros(n_rows, dtype=bool)
    # Choose the positive positions by deterministic permutation
    mask[rng.permutation(n_rows)[:n_positive]] = True
    # Return the exact-rate boolean mask
    return mask


def generate_loan_raw(
    n_keep: int = 395_492,
    n_dropped: int = 4_000,
    positive_rate: float = 0.101,
) -> pd.DataFrame:
    """
    Generate a synthetic Lending Club raw DataFrame at paper scale.

    Purpose:
        Creates the raw loan_raw frame with the 36-column ingestion schema of
        config.yaml. Rows with loan_status outside {"Fully Paid",
        "Charged Off"} are included so the cleansing filter in Task 2 has
        rows to drop; the keep-set size and the Charged-Off rate reproduce
        the paper's n = 395,492 and pi = 0.101 within tolerance.

    Inputs:
        n_keep (int): Number of rows with a keep status (default 395,492).
        n_dropped (int): Number of rows with a non-keep status (default 4,000).
        positive_rate (float): Target "Charged Off" rate (default 0.101).

    Processes:
        1. Draw continuous values (loan_amnt, int_rate, installment,
           annual_inc, dti, balances, payments, FICO bounds) from realistic
           distributions; grade is drawn from a pool correlated with int_rate.
        2. Draw categorical values (term, emp_length, home_ownership,
           verification_status, purpose, dates, loan_status) from pools.
        3. Inject NaN exactly where the schema allows it (revol_util,
           next_pymnt_d, mths_since_last_delinq, mths_since_last_record).
        4. Assemble the frame and cast every column to its declared dtype.

    Outputs:
        pd.DataFrame: loan_raw with 36 columns and n_keep + n_dropped rows;
                      the keep subset has exactly round(n_keep * 0.101)
                      "Charged Off" rows.

    Raises:
        ValueError: If n_keep < 1 or n_dropped < 0.
    """
    # Validate the keep-set size is positive
    if n_keep < 1:
        # Reject degenerate frame sizes
        raise ValueError(f"n_keep must be >= 1, received {n_keep}.")
    # Validate the dropped-row count is non-negative
    if n_dropped < 0:
        # Reject negative extra-row counts
        raise ValueError(f"n_dropped must be >= 0, received {n_dropped}.")
    # Total raw rows = keep-set + non-keep statuses
    n_total: int = n_keep + n_dropped
    # Build the keep/Charged-Off mask over the keep-set rows
    keep_mask: np.ndarray = _fixed_rate_positions(n_keep, positive_rate, rng)
    # Draw the loan amount uniformly in [1,000, 40,000] USD
    loan_amnt: np.ndarray = rng.uniform(1_000.0, 40_000.0, size=n_keep).round(2)
    # Draw the annual interest rate in [5, 30] percent
    int_rate: np.ndarray = rng.uniform(5.0, 30.0, size=n_keep).round(2)
    # Derive a correlated grade: A=1 .. G=7 from the rate bucket
    grade_pool: np.ndarray = np.array(["A", "B", "C", "D", "E", "F", "G"], dtype=object)
    # Map each rate to a grade index by quantile bucketing
    grade_idx: np.ndarray = np.clip(((int_rate - 5.0) / 25.0 * 7).astype(int), 0, 6)
    # Assign the grade strings from the pool
    grade: np.ndarray = grade_pool[grade_idx]
    # Draw the term from the two permitted values
    term: np.ndarray = rng.choice(np.array(["36 months", "60 months"], dtype=object), size=n_keep)
    # Approximate the monthly installment by a plain-vanilla amortisation
    monthly_rate: np.ndarray = int_rate / 100.0 / 12.0
    # Parse the term to months (36 or 60) for the annuity formula
    term_months_num: np.ndarray = np.where(term == "36 months", 36, 60)
    # Installment = P * r / (1 - (1 + r)^-n) (standard amortisation formula)
    installment: np.ndarray = (loan_amnt * monthly_rate / (1.0 - (1.0 + monthly_rate) ** -term_months_num)).round(2)
    # Draw the employment length from the paper's parsing vocabulary
    emp_length: np.ndarray = rng.choice(
        np.array(["< 1 year", "1 year", "2 years", "3 years", "4 years", "5 years",
                  "6 years", "7 years", "8 years", "9 years", "10+ years"], dtype=object),
        size=n_keep,
    )
    # Draw the home-ownership category (reference absorbs OTHER/NONE/ANY)
    home_ownership: np.ndarray = rng.choice(
        np.array(["RENT", "MORTGAGE", "OWN", "OTHER", "NONE", "ANY"], dtype=object),
        size=n_keep,
        p=np.array([0.42, 0.45, 0.08, 0.03, 0.015, 0.005]),
    )
    # Draw the annual income correlated with the loan amount
    annual_inc: np.ndarray = (loan_amnt * rng.uniform(1.5, 8.0, size=n_keep)).round(2)
    # Draw the verification status (Income Verified is the reference category)
    verification_status: np.ndarray = rng.choice(
        np.array(["Source Verified", "Not Verified", "Income Verified"], dtype=object),
        size=n_keep,
        p=np.array([0.45, 0.40, 0.15]),
    )
    # Generate issue dates in the Lending Club era as %b-%Y strings
    issue_d: np.ndarray = np.array(
        [fake.date_between(start_date="-2007-01-01", end_date="-2016-12-31").strftime("%b-%Y") for _ in range(n_keep)],
        dtype=object,
    )
    # Build the loan-status vector: keep statuses first, then the drop statuses
    keep_statuses: np.ndarray = np.where(keep_mask, "Charged Off", "Fully Paid")
    # List the non-keep statuses that the Task 2 filter must drop
    drop_statuses: np.ndarray = np.array(
        ["Current", "In Grace Period", "Late (16-30 days)", "Late (31-120 days)"], dtype=object
    )
    # Sample the drop statuses for the extra rows
    extra_statuses: np.ndarray = rng.choice(drop_statuses, size=n_dropped)
    # Concatenate the keep-set statuses with the drop-set statuses
    loan_status: np.ndarray = np.concatenate([keep_statuses, extra_statuses])
    # Draw the loan purpose (three explicit indicators plus the other bucket)
    purpose: np.ndarray = rng.choice(
        np.array(["debt_consolidation", "credit_card", "home_improvement",
                  "major_purchase", "small_business", "medical", "other"], dtype=object),
        size=n_total,
    )
    # Draw the debt-to-income ratio in [0, 40]
    dti: np.ndarray = rng.uniform(0.0, 40.0, size=n_total).round(2)
    # Draw the integer count features (small non-negative integers)
    delinq_2yrs: np.ndarray = rng.integers(0, 5, size=n_total)
    inq_last_6mths: np.ndarray = rng.integers(0, 8, size=n_total)
    open_acc: np.ndarray = rng.integers(1, 25, size=n_total)
    pub_rec: np.ndarray = rng.integers(0, 4, size=n_total)
    total_acc: np.ndarray = rng.integers(5, 60, size=n_total)
    # Draw the revolving balance and utilisation (utilisation may be NaN)
    revol_bal: np.ndarray = rng.uniform(0.0, 50_000.0, size=n_total).round(2)
    # Initialise the utilisation as float64 with NaN injected at ~8%
    revol_util: np.ndarray = rng.uniform(0.0, 100.0, size=n_total).round(2)
    # Mask ~8% of the utilisation values as missing (schema permits NaN)
    revol_util[rng.random(n_total) < 0.08] = np.nan
    # Draw the outstanding principal (zero for fully paid loans)
    out_prncp: np.ndarray = np.where(keep_mask, rng.uniform(0.0, 30_000.0, size=n_keep), 0.0).round(2)
    # Extend the principal vector to the full row count
    out_prncp_full: np.ndarray = np.concatenate([out_prncp, rng.uniform(0.0, 30_000.0, size=n_dropped).round(2)])
    # Draw the total payment and its principal/interest/late-fee components
    total_pymnt: np.ndarray = rng.uniform(0.0, 60_000.0, size=n_total).round(2)
    total_rec_prncp: np.ndarray = np.minimum(total_pymnt, rng.uniform(0.0, 50_000.0, size=n_total)).round(2)
    total_rec_int: np.ndarray = np.maximum(total_pymnt - total_rec_prncp, 0.0).round(2)
    total_rec_late_fee: np.ndarray = rng.uniform(0.0, 500.0, size=n_total).round(2)
    recoveries: np.ndarray = rng.uniform(0.0, 10_000.0, size=n_total).round(2)
    collection_recovery_fee: np.ndarray = (recoveries * rng.uniform(0.0, 0.2, size=n_total)).round(2)
    # Generate last-payment dates after the issue dates
    last_pymnt_d: np.ndarray = np.array(
        [fake.date_between(start_date="-2017-01-01", end_date="-2026-08-09").strftime("%b-%Y") for _ in range(n_total)],
        dtype=object,
    )
    # Draw the last payment amount in [0, 5,000] USD
    last_pymnt_amnt: np.ndarray = rng.uniform(0.0, 5_000.0, size=n_total).round(2)
    # Build next_pymnt_d as object with ~12% missing (schema permits NaN)
    next_pymnt_d: np.ndarray = np.array(
        [fake.date_between(start_date="-2026-08-09", end_date="+2026-12-31").strftime("%b-%Y") for _ in range(n_total)],
        dtype=object,
    )
    # Mask ~12% of the next-payment dates as missing
    next_pymnt_d[rng.random(n_total) < 0.12] = np.nan
    # Draw recent credit-pull dates
    last_credit_pull_d: np.ndarray = np.array(
        [fake.date_between(start_date="-2026-01-01", end_date="-2026-08-09").strftime("%b-%Y") for _ in range(n_total)],
        dtype=object,
    )
    # Draw FICO range endpoints (high > low, both integers)
    fico_low: np.ndarray = rng.integers(600, 830, size=n_total)
    # Add a spread of 0 to 9 points to the low endpoint
    fico_high: np.ndarray = fico_low + rng.integers(0, 10, size=n_total)
    # Draw the months-since-delinquency as nullable float64 (~70% missing)
    mths_since_last_delinq: np.ndarray = rng.integers(1, 120, size=n_total).astype(np.float64)
    # Mask ~70% of the delinquency months as missing
    mths_since_last_delinq[rng.random(n_total) < 0.70] = np.nan
    # Draw the months-since-public-record as nullable float64 (~85% missing)
    mths_since_last_record: np.ndarray = rng.integers(1, 240, size=n_total).astype(np.float64)
    # Mask ~85% of the record months as missing
    mths_since_last_record[rng.random(n_total) < 0.85] = np.nan
    # Assemble the raw frame with the exact 36-column ingestion schema
    loan_raw: pd.DataFrame = pd.DataFrame(
        {
            "id": np.arange(1, n_total + 1, dtype=np.int64),
            "loan_amnt": loan_amnt,
            "term": term,
            "int_rate": int_rate,
            "installment": installment,
            "grade": grade,
            "emp_length": emp_length,
            "home_ownership": home_ownership,
            "annual_inc": annual_inc,
            "verification_status": verification_status,
            "issue_d": issue_d,
            "loan_status": loan_status,
            "purpose": purpose,
            "dti": dti,
            "delinq_2yrs": delinq_2yrs,
            "inq_last_6mths": inq_last_6mths,
            "open_acc": open_acc,
            "pub_rec": pub_rec,
            "revol_bal": revol_bal,
            "revol_util": revol_util,
            "total_acc": total_acc,
            "out_prncp": out_prncp_full,
            "total_pymnt": total_pymnt,
            "total_rec_prncp": total_rec_prncp,
            "total_rec_int": total_rec_int,
            "total_rec_late_fee": total_rec_late_fee,
            "recoveries": recoveries,
            "collection_recovery_fee": collection_recovery_fee,
            "last_pymnt_d": last_pymnt_d,
            "last_pymnt_amnt": last_pymnt_amnt,
            "next_pymnt_d": next_pymnt_d,
            "last_credit_pull_d": last_credit_pull_d,
            "last_fico_range_high": fico_high,
            "last_fico_range_low": fico_low,
            "mths_since_last_delinq": mths_since_last_delinq,
            "mths_since_last_record": mths_since_last_record,
        }
    )
    # Enforce the declared dtypes: integer columns as int64
    int_cols: List[str] = ["id", "delinq_2yrs", "inq_last_6mths", "open_acc", "pub_rec",
                           "total_acc", "last_fico_range_high", "last_fico_range_low"]
    # Cast every integer column to int64
    loan_raw[int_cols] = loan_raw[int_cols].astype("int64")
    # Enforce the declared dtypes: numeric columns as float64
    float_cols: List[str] = ["loan_amnt", "int_rate", "installment", "annual_inc", "dti",
                             "revol_bal", "revol_util", "out_prncp", "total_pymnt",
                             "total_rec_prncp", "total_rec_int", "total_rec_late_fee",
                             "recoveries", "collection_recovery_fee", "last_pymnt_amnt",
                             "mths_since_last_delinq", "mths_since_last_record"]
    # Cast every numeric column to float64 (NaN preserved)
    loan_raw[float_cols] = loan_raw[float_cols].astype("float64")
    # Return the schema-compliant raw Loan frame
    return loan_raw


def generate_fraud_ecom_raw(n_rows: int = 151_112, positive_rate: float = 0.094) -> pd.DataFrame:
    """
    Generate a synthetic Fraud Ecommerce raw DataFrame at paper scale.

    Purpose:
        Creates fraud_ecom_raw with the 11-column ingestion schema of
        config.yaml. IP addresses and device ids are drawn from pools with
        repeats so the training-only count features of Task 3 are meaningful.

    Inputs:
        n_rows (int): Number of rows (default 151,112).
        positive_rate (float): Target fraud rate (default 0.094).

    Processes:
        1. Draw user ids, timestamps (purchase >= signup), purchase value,
           source, browser, sex, age, and IP/device pools.
        2. Place the exact positive count via deterministic permutation.
        3. Cast every column to its declared dtype.

    Outputs:
        pd.DataFrame: fraud_ecom_raw with 11 columns and n_rows rows.

    Raises:
        ValueError: If n_rows < 1.
    """
    # Validate the row count is positive
    if n_rows < 1:
        # Reject degenerate frame sizes
        raise ValueError(f"n_rows must be >= 1, received {n_rows}.")
    # Place the exact fraud positives at deterministic positions
    class_mask: np.ndarray = _fixed_rate_positions(n_rows, positive_rate, rng)
    # Draw the account identifiers (unique)
    user_id: np.ndarray = np.arange(1, n_rows + 1, dtype=np.int64)
    # Generate signup timestamps over a one-year window
    signup_time: np.ndarray = np.array(
        [fake.date_time_between(start_date="-2020-01-01", end_date="-2020-12-31").strftime("%Y-%m-%d %H:%M:%S") for _ in range(n_rows)],
        dtype=object,
    )
    # Generate purchase timestamps shortly after signup (0-48 hours later)
    purchase_time: np.ndarray = np.array(
        [fake.date_time_between(start_date="-2020-01-01", end_date="-2021-01-01").strftime("%Y-%m-%d %H:%M:%S") for _ in range(n_rows)],
        dtype=object,
    )
    # Draw the purchase value with a heavy right tail
    purchase_value: np.ndarray = np.exp(rng.normal(4.0, 0.8, size=n_rows)).round(2)
    # Draw a high-cardinality device-id pool (repeats are realistic)
    device_pool: np.ndarray = np.array([fake.md5()[:16] for _ in range(5_000)], dtype=object)
    # Sample device ids from the pool with replacement
    device_id: np.ndarray = rng.choice(device_pool, size=n_rows)
    # Draw the acquisition source channel
    source: np.ndarray = rng.choice(np.array(["SEO", "Ads", "Direct"], dtype=object), size=n_rows)
    # Draw the browser from the nominal pool
    browser: np.ndarray = rng.choice(
        np.array(["Chrome", "Safari", "Firefox", "Edge", "Opera"], dtype=object), size=n_rows
    )
    # Draw the binary sex indicator
    sex: np.ndarray = rng.integers(0, 2, size=n_rows)
    # Draw the age in [18, 70]
    age: np.ndarray = rng.integers(18, 71, size=n_rows)
    # Build a pool of 10,000 public IPv4 addresses (repeats are realistic)
    ip_pool: np.ndarray = np.array([fake.ipv4_public() for _ in range(10_000)], dtype=object)
    # Sample IP addresses from the pool with replacement
    ip_address: np.ndarray = rng.choice(ip_pool, size=n_rows)
    # Encode the fraud label as int64 (y = 1 for fraud)
    raw_class: np.ndarray = class_mask.astype(np.int64)
    # Assemble the raw frame with the exact 11-column schema
    fraud_ecom_raw: pd.DataFrame = pd.DataFrame(
        {
            "user_id": user_id,
            "signup_time": signup_time,
            "purchase_time": purchase_time,
            "purchase_value": purchase_value,
            "device_id": device_id,
            "source": source,
            "browser": browser,
            "sex": sex,
            "age": age,
            "ip_address": ip_address,
            "raw_class": raw_class,
        }
    )
    # Enforce the declared integer dtypes
    int_cols: List[str] = ["user_id", "sex", "age", "raw_class"]
    # Cast every integer column to int64
    fraud_ecom_raw[int_cols] = fraud_ecom_raw[int_cols].astype("int64")
    # Cast the purchase value to float64
    fraud_ecom_raw["purchase_value"] = fraud_ecom_raw["purchase_value"].astype("float64")
    # Return the schema-compliant raw FraudEcom frame
    return fraud_ecom_raw


def generate_default_raw(n_rows: int = 30_000, positive_rate: float = 0.221) -> pd.DataFrame:
    """
    Generate a synthetic UCI Default of Credit Card Clients raw DataFrame.

    Purpose:
        Creates default_raw with the 25-column ingestion schema of
        config.yaml, including the standard PAY_* repayment codes (-2..8),
        the six-month BILL_AMT/PAY_AMT panel, and the default label.

    Inputs:
        n_rows (int): Number of rows (default 30,000).
        positive_rate (float): Target default rate (default 0.221).

    Processes:
        1. Draw demographics (LIMIT_BAL, SEX, EDUCATION, MARRIAGE, AGE).
        2. Draw the six-month repayment panel (PAY_0..PAY_6, BILL_AMT1..6,
           PAY_AMT1..6) with realistic codes and amounts.
        3. Place the exact default count via deterministic permutation.
        4. Cast every column to int64.

    Outputs:
        pd.DataFrame: default_raw with 25 columns and n_rows rows.

    Raises:
        ValueError: If n_rows < 1.
    """
    # Validate the row count is positive
    if n_rows < 1:
        # Reject degenerate frame sizes
        raise ValueError(f"n_rows must be >= 1, received {n_rows}.")
    # Place the exact default positives at deterministic positions
    default_mask: np.ndarray = _fixed_rate_positions(n_rows, positive_rate, rng)
    # Draw unique client identifiers
    ids: np.ndarray = np.arange(1, n_rows + 1, dtype=np.int64)
    # Draw the available credit limit in NTD
    limit_bal: np.ndarray = rng.integers(10_000, 1_000_001, size=n_rows)
    # Draw the raw sex code (1 = male, 2 = female)
    sex: np.ndarray = rng.integers(1, 3, size=n_rows)
    # Draw the education code including the recodable values 0, 5, 6
    education: np.ndarray = rng.choice(np.array([0, 1, 2, 3, 4, 5, 6]), size=n_rows, p=np.array([0.02, 0.35, 0.35, 0.20, 0.05, 0.02, 0.01]))
    # Draw the marriage code including the recodable value 0
    marriage: np.ndarray = rng.choice(np.array([0, 1, 2, 3]), size=n_rows, p=np.array([0.02, 0.45, 0.35, 0.18]))
    # Draw the age in [20, 70]
    age: np.ndarray = rng.integers(20, 71, size=n_rows)
    # Draw the repayment status codes for the six months (most recent first)
    pay_columns: Dict[str, np.ndarray] = {}
    # The standard PAY code vocabulary ranges from -2 (no consumption) to 8
    pay_codes: np.ndarray = np.array([-2, -1, 0, 1, 2, 3, 4, 5, 6, 7, 8])
    # Build each of the six repayment columns from the code vocabulary
    for col in ["PAY_0", "PAY_2", "PAY_3", "PAY_4", "PAY_5", "PAY_6"]:
        # Sample the repayment code for the month
        pay_columns[col] = rng.choice(pay_codes, size=n_rows, p=np.array([0.05, 0.05, 0.50, 0.20, 0.10, 0.04, 0.03, 0.02, 0.005, 0.003, 0.002]))
    # Build the six bill-amount columns (declining panel)
    bill_columns: Dict[str, np.ndarray] = {}
    # Generate each month's bill amount with a decreasing drift
    for i in range(1, 7):
        # Bill amounts decline gently across the six-month window
        bill_columns[f"BILL_AMT{i}"] = rng.integers(0, 600_000, size=n_rows)
    # Build the six payment-amount columns (often zero for defaults)
    pay_amt_columns: Dict[str, np.ndarray] = {}
    # Generate each month's payment amount with mass at zero
    for i in range(1, 7):
        # Zero payments are common among defaulters
        pay_amt_columns[f"PAY_AMT{i}"] = np.where(rng.random(n_rows) < 0.3, 0, rng.integers(1, 100_000, size=n_rows))
    # Encode the default label as int64 (y = 1 for DEFAULT_NEXT_MONTH == 1)
    default_next_month: np.ndarray = default_mask.astype(np.int64)
    # Assemble the raw frame with the exact 25-column schema
    default_raw: pd.DataFrame = pd.DataFrame(
        {"ID": ids, "LIMIT_BAL": limit_bal, "SEX": sex, "EDUCATION": education,
         "MARRIAGE": marriage, "AGE": age, **pay_columns, **bill_columns,
         **pay_amt_columns, "DEFAULT_NEXT_MONTH": default_next_month}
    )
    # Cast every column to the declared int64 dtype
    default_raw = default_raw.astype("int64")
    # Return the schema-compliant raw Default frame
    return default_raw


def generate_creditcard_raw(n_rows: int = 284_807, positive_rate: float = 0.0017) -> pd.DataFrame:
    """
    Generate a synthetic Credit Card Fraud raw DataFrame at paper scale.

    Purpose:
        Creates creditcard_raw with the 31-column ingestion schema of
        config.yaml: Time, V1..V28 (anonymized PCA components, negative
        values allowed), Amount, and the Class label.

    Inputs:
        n_rows (int): Number of rows (default 284,807).
        positive_rate (float): Target fraud rate (default 0.0017).

    Processes:
        1. Draw the elapsed-seconds Time column.
        2. Draw the 28 PCA components from a standard normal (signed).
        3. Draw the Amount with a heavy right tail.
        4. Place the exact fraud count via deterministic permutation.

    Outputs:
        pd.DataFrame: creditcard_raw with 31 columns and n_rows rows.

    Raises:
        ValueError: If n_rows < 1.
    """
    # Validate the row count is positive
    if n_rows < 1:
        # Reject degenerate frame sizes
        raise ValueError(f"n_rows must be >= 1, received {n_rows}.")
    # Place the exact fraud positives at deterministic positions
    fraud_mask: np.ndarray = _fixed_rate_positions(n_rows, positive_rate, rng)
    # Draw the elapsed seconds since the first transaction (0..172,800)
    time_sec: np.ndarray = rng.uniform(0.0, 172_800.0, size=n_rows)
    # Build the 28 anonymized PCA components as signed standard normals
    pca_components: Dict[str, np.ndarray] = {f"V{i}": rng.normal(0.0, 1.0, size=n_rows) for i in range(1, 29)}
    # Draw the transaction amount with a heavy right tail (EUR)
    amount: np.ndarray = np.exp(rng.normal(3.0, 1.2, size=n_rows)).round(2)
    # Encode the fraud label as int64 (y = 1 for fraud)
    class_label: np.ndarray = fraud_mask.astype(np.int64)
    # Assemble the raw frame with the exact 31-column schema
    creditcard_raw: pd.DataFrame = pd.DataFrame(
        {"Time": time_sec, **pca_components, "Amount": amount, "Class": class_label}
    )
    # Enforce the declared dtypes: all numeric columns as float64
    creditcard_raw[["Time", "Amount"] + [f"V{i}" for i in range(1, 29)]] = creditcard_raw[
        ["Time", "Amount"] + [f"V{i}" for i in range(1, 29)]
    ].astype("float64")
    # Cast the label column to int64
    creditcard_raw["Class"] = creditcard_raw["Class"].astype("int64")
    # Return the schema-compliant raw Creditcard frame
    return creditcard_raw
```

```python
# Cell 2: verify the synthetic positive rates against the paper base rates
# Paper base rates: Loan 0.101, FraudEcom 0.094, Default 0.221, Creditcard 0.0017
# y = 1 for "Charged Off" (Loan), raw_class == 1 (FraudEcom),
# DEFAULT_NEXT_MONTH == 1 (Default), Class == 1 (Creditcard)
loan_filtered: pd.DataFrame = loan_raw[loan_raw["loan_status"].isin(["Fully Paid", "Charged Off"])]
# Compute the filtered Loan positive rate: fraction of "Charged Off"
loan_rate: float = float((loan_filtered["loan_status"] == "Charged Off").mean())
# Compute the FraudEcom positive rate
fraud_rate: float = float(fraud_ecom_raw["raw_class"].mean())
# Compute the Default positive rate
default_rate: float = float(default_raw["DEFAULT_NEXT_MONTH"].mean())
# Compute the Creditcard positive rate
creditcard_rate: float = float(creditcard_raw["Class"].mean())
# Report the observed rates next to the paper's rates
print(f"Loan filtered: n={len(loan_filtered)}, rate={loan_rate:.4f} (paper 0.101)")
print(f"FraudEcom: rate={fraud_rate:.4f} (paper 0.094)")
print(f"Default: rate={default_rate:.4f} (paper 0.221)")
print(f"Creditcard: rate={creditcard_rate:.4f} (paper 0.0017)")
```

### **Step 2: Loading the Configuration (`config.yaml` → `config`)**

The pipeline is driven by the single source of truth stored in `config.yaml`, which is assumed to be saved in the working directory. The file is read into memory and converted into the pipeline-usable Python dictionary `config` via `yaml.safe_load`. No manual conversion is required: the Task 1 configuration validator (`validate_configuration`) normalizes the dictionary internally (seeds coerced to a tuple, split fractions cast to `float`) and aborts with a `ConfigError` if any paper constant is violated.

**Methodology:**
1.  **File I/O:** Open `config.yaml` in read mode.
2.  **Parsing:** Use `yaml.safe_load` to convert the YAML structure into a nested Python dictionary.
3.  **Validation:** Raise informative errors for missing files and malformed YAML; verify the top-level sections are present.

```python
# Cell 3: load the study configuration from the working-directory YAML file

def load_study_configuration(filepath: str = "config.yaml") -> Dict[str, Any]:
    """
    Load the study configuration parameters from a YAML file into a Python dict.

    Purpose:
        To ingest the deterministic hyperparameters, data-split definitions,
        schema contracts, and verification targets defined in the external
        configuration file. This ensures reproducibility by separating code
        from configuration: every UNSPECIFIED IN MANUSCRIPT placeholder is
        resolved exactly once, in this file.

    Inputs:
        filepath (str): The relative or absolute path to the YAML
                        configuration file. Default is "config.yaml".

    Processes:
        1. Validate that filepath is a string.
        2. Open the file in read mode (UTF-8).
        3. Parse the YAML structure with yaml.safe_load.
        4. Verify that the parsed result is a non-empty dictionary.

    Outputs:
        Dict[str, Any]: The nested configuration dictionary, directly usable
                        as the study_configuration argument of
                        run_end_to_end_pipeline.

    Raises:
        TypeError: If filepath is not a string.
        FileNotFoundError: If the file does not exist.
        yaml.YAMLError: If the file contains invalid YAML syntax.
        ValueError: If the parsed YAML is not a non-empty dictionary.
    """
    # Validate the input type before any file access
    if not isinstance(filepath, str):
        # Reject non-string paths with an informative message
        raise TypeError(f"filepath must be a str, received {type(filepath).__name__}.")
    # Open the YAML file in read mode with explicit UTF-8 encoding
    with open(filepath, "r", encoding="utf-8") as yaml_stream:
        # Parse the YAML content safely (no arbitrary object construction)
        config: Any = yaml.safe_load(yaml_stream)
    # Verify the parsed content is a dictionary
    if not isinstance(config, dict):
        # Reject YAML files that do not declare a mapping
        raise ValueError(f"config.yaml must contain a YAML mapping, received {type(config).__name__}.")
    # Verify the mapping is not empty
    if len(config) == 0:
        # Reject empty configuration files
        raise ValueError("config.yaml is empty; no study parameters were loaded.")
    # Confirm the load to the console
    print(f"Successfully loaded configuration from {filepath}")
    # Return the pipeline-usable configuration dictionary
    return config


# Load the configuration (config.yaml is assumed to be in the working directory)
config: Dict[str, Any] = load_study_configuration()

# Print the top-level sections of the loaded configuration
print("\nTop-level configuration sections:")
# Iterate over the loaded top-level keys and print them in order
for section_key in config.keys():
    print(f"  - {section_key}")
```

### **Step 4: Executing the End-to-End Pipeline**

### **Summary of the Execution Flow**

1.  **Data Ingestion:** The four synthetic raw DataFrames (`loan_raw` at 399,492 rows with 395,492 keep-rows, `fraud_ecom_raw` at 151,112, `default_raw` at 30,000, `creditcard_raw` at 284,807) are generated with Faker + NumPy at the paper's exact population sizes and positive rates (0.101 / 0.094 / 0.221 / 0.0017), then keyed as `raw_dfs`.
2.  **Configuration Ingestion:** `config.yaml` (working directory) is read with `yaml.safe_load` into the dictionary `config` — the single source of truth carrying every paper constant and every UNSPECIFIED IN MANUSCRIPT placeholder; the Task 1 validator normalizes it (seeds → tuple, fractions → float) and aborts on any contradiction.
3.  **Validation (Task 1):** `orchestrate_validation` checks the configuration against paper constants (including the recomputed $L = \max\{|\log 0.01|, |\log 10.01|\}$), validates all four raw schemas (required columns, dtypes, row-count lower bounds, aliases), and verifies the raw label rates within $\pm 0.001$.
4.  **Cleansing (Task 2):** `cleanse_all` applies the deterministic parsers, recodings, and the Loan status filter ($y=1$ for "Charged Off"), builds missingness flags only, and persists checksummed parquet artifacts; the raw inputs are never re-read afterwards.
5.  **Feature Engineering and Preprocessing (Tasks 3–4):** Deterministic matrices (42/6/26/30) are built — FraudEcom count features from training rows only — then stratified 60/20/20 splits are made per seed and the affine transform $\tilde{x}_j = 0.01 + 10.0\,(x_j - \min_j)/(\max_j - \min_j)$ is fit on training only and applied to all partitions with clipping to $[0.01, 10.01]$.
6.  **Hyperparameter Search (Task 5):** Optuna (TPE, median pruning, 100 trials) tunes $\eta$, $\lambda$, and epochs once per dataset on the seed-0 split via five-fold cross-validated PR-AUC; the record is reused by all five seeds.
7.  **Quantitative Engine (Task 11):** For every dataset and seed, the monomial $z(x) = \alpha \prod_j x_j^{\beta_j}$ is trained with automatic differentiation; pruning selects the smallest support within the 0.01 validation tolerance and verifies the worst-case bound $|s_{full} - s_{pruned}| \le L \sum_{j \in S}|\beta_j|$; the directional rule is built and its fidelity predicted a-priori via $\rho = \beta^\top\Sigma d/(\sqrt{\beta^\top\Sigma\beta}\sqrt{d^\top\Sigma d})$ and $\tau_{pred} = \frac{2}{\pi}\arcsin(\rho)$; scorecard $V(x)$ and tally $T(x)$ thresholds are selected on validation F1; all five representations are isotonically calibrated and evaluated (PR-AUC, $\tau_b$, ECE).
8.  **IHT Robustness (Task 12):** Iterative hard thresholding trains an independent sparse monomial and the Jaccard overlap with the pruned support is recorded.
9.  **Survey (Task 13):** Skipped under the explicit skip policy because `survey_records=None` — no synthetic human-subjects data is fabricated.
10. **Aggregation and Verification (Task 14):** Means and standard deviations over the five seeds are computed and the paper's headline targets (pruning loss ≤ 0.008, Loan scorecard ≈ 0.329 / tally ≈ 0.300, Default directional ≈ 0.381 with $\tau_b$ ≈ 0.744, fidelity MAE < 0.025 vs ≈ 0.236, ECE < 0.02) are checked as soft invariants with a diagnostic report.
11. **Documentation and Audit (Task 15):** The methodological appendix, environment documentation, and the mathematical audit suite (log-space identity, pruning bound, $\rho$ formula, ECE binning, scorecard points, tied $\tau_b$) are produced, and the reproduction is signed off.
12. **Output:** The `EndToEndPipelineResult` dictionary (stages: `validation`, `cleansing`, `feature_matrices`, `preprocessing`, `hyperparameters`, `pipeline_results`, `iht`, `survey`, `aggregation`, `verification`, `figures_persistence`, `documentation`, `artifacts_root`) is returned, with every artifact persisted under `output_dir/artifacts` for audit and re-analysis.

 ## **Implemented Callables**

````python
# Task 1: Validate ingested raw data structures and configuration parameters

# ==============================================================================
# Task 1: Validate ingested raw data structures and configuration parameters
# ==============================================================================

# -----------------------------------------------------------------------------
# Paper-derived constants hard-coded in the validator (Section 1.2 contract).
# These values come from the manuscript and the study configuration preamble;
# the validator compares the configuration against them, never trusts the
# configuration to supply them.
# -----------------------------------------------------------------------------
PAPER_DATASET_STATS: Dict[str, Dict[str, float]] = {
    "Loan": {"n": 395_492.0, "features": 42.0, "positive_rate": 0.101},
    "FraudEcom": {"n": 151_112.0, "features": 6.0, "positive_rate": 0.094},
    "Default": {"n": 30_000.0, "features": 26.0, "positive_rate": 0.221},
    "Creditcard": {"n": 284_807.0, "features": 30.0, "positive_rate": 0.0017},
}

# Paper split fractions (train/validation/test), scaling bounds, and constants
PAPER_SPLIT_FRACTIONS: Tuple[float, float, float] = (0.60, 0.20, 0.20)
PAPER_SCALE_LOWER: float = 0.01
PAPER_SCALE_UPPER: float = 10.01
PAPER_PRUNING_TOLERANCE: float = 0.01
PAPER_SCORECARD_MAX_POINTS: float = 5.0
PAPER_ECE_BINS: float = 10.0
PAPER_SEEDS: Tuple[int, ...] = (0, 1, 2, 3, 4)

# Loan label keep-set: Charged Off is the positive class, Fully Paid negative
LOAN_KEEP_STATUSES: Tuple[str, str] = ("Fully Paid", "Charged Off")
LOAN_POSITIVE_STATUS: str = "Charged Off"

# Absolute tolerance applied to floating-point configuration checks
_FLOAT_TOLERANCE: float = 1e-10


# -----------------------------------------------------------------------------
# Task 1, Step 1: Validate the four raw DataFrames against their schemas
# -----------------------------------------------------------------------------

# Define a schema-validation failure type derived from the built-in ValueError
class SchemaValidationError(ValueError):
    """
    Raised when a raw DataFrame fails its declared schema validation.

    This exception aggregates every independent failure so that a single
    failing check never hides additional failures; the raised message
    lists missing columns, unexpected dtypes, and alias failures.

    Raises
    ------
    SchemaValidationError
        Raised by ``validate_raw_schema`` when any schema assertion fails.
    """


# Define a configuration-contradiction failure type derived from ValueError
class ConfigError(ValueError):
    """
    Raised when the study configuration contradicts a paper constant.

    The message includes the full dotted key path of the failed check so
    that the offending parameter can be located without a debugger.

    Raises
    ------
    ConfigError
        Raised by ``validate_configuration`` when any paper-derived
        constant check fails.
    """


# Define a label-rate reproduction failure type derived from ValueError
class LabelRateError(ValueError):
    """
    Raised when raw positive-class rates cannot reproduce the paper's.

    Carries the expected and observed counts/rates plus the full label
    frequency table for the diagnostic report.

    Attributes
    ----------
    diagnostics : Dict[str, Any]
        Structured diagnostics for the report writer: expected and
        observed counts/rates and the raw label frequency table.
    """

    # Define the constructor that binds the message and diagnostics payload
    def __init__(
        self,
        message: str,
        diagnostics: Dict[str, Any],
    ) -> None:
        """
        Construct the label-rate error with a message and diagnostics.

        Parameters
        ----------
        message : str
            Human-readable failure description naming the exact
            discrepancy, expected value, and observed value.
        diagnostics : Dict[str, Any]
            Structured payload with expected counts/rates, observed
            counts/rates, and the raw label frequency table.

        Returns
        -------
        None
            Constructor; the instance is initialized in place.
        """
        # Initialise the parent exception with the message text
        super().__init__(message)
        # Store the structured diagnostics for the report writer
        self.diagnostics: Dict[str, Any] = diagnostics


def _resolve_column(raw_df: pd.DataFrame, column: str, aliases: Dict[str, str]) -> Optional[str]:
    """
    Resolve a required column name through the alias map.

    Parameters
    ----------
    raw_df : pd.DataFrame
        Raw DataFrame whose columns are searched.
    column : str
        Canonical required column name.
    aliases : Dict[str, str]
        Mapping from canonical name to the raw alias name.

    Returns
    -------
    Optional[str]
        The column name present in the frame, or None when absent.
    """
    # Return the canonical name directly when it is present in the frame
    if column in raw_df.columns:
        return column
    # Look up the alias declared for this canonical column
    alias: Optional[str] = aliases.get(column)
    # Return the alias when it is present in the frame
    if alias is not None and alias in raw_df.columns:
        return alias
    # Return None when neither the canonical name nor the alias exists
    return None


def validate_raw_schema(
    raw_df: pd.DataFrame,
    schema: Dict[str, Any],
    dataset_name: str,
) -> Dict[str, Any]:
    """
    Validate one raw DataFrame against its declared schema contract.

    The contract distinguishes required columns (must be present) from
    allowed additional columns (logged, not fatal). Required columns are
    resolved through the schema's allowed-alias map, their dtypes are
    checked against the declared dtypes, the row count is checked as a
    lower bound, and missingness proportions are recorded.

    Parameters
    ----------
    raw_df : pd.DataFrame
        Raw DataFrame to validate. Must be read-only; the original index
        is preserved and never reordered.
    schema : Dict[str, Any]
        Schema contract with keys: "required_columns" (List[str]),
        "allowed_aliases" (Dict[str, str]), "dtypes" (Dict[str, str]),
        "min_rows" (int), and "label_column" (str).
    dataset_name : str
        Name of the dataset, used in error messages and the report.

    Returns
    -------
    Dict[str, Any]
        Per-dataset validation report with missing columns, extra columns,
        missingness proportions, row count, and index-uniqueness flag.

    Raises
    ------
    TypeError
        If `raw_df` is not a pd.DataFrame or `schema` is not a dict.
    SchemaValidationError
        If any required column is missing, any dtype mismatches, or the
        row count is below the declared lower bound.
    """
    # Validate the raw DataFrame type
    if not isinstance(raw_df, pd.DataFrame):
        raise TypeError(
            f"raw_df must be a pd.DataFrame; received "
            f"{type(raw_df).__name__} for dataset '{dataset_name}'."
        )
    # Validate the schema dictionary type
    if not isinstance(schema, dict):
        raise TypeError(
            f"schema must be a dict; received {type(schema).__name__} "
            f"for dataset '{dataset_name}'."
        )
    # Extract the required column list from the schema contract
    required_columns: List[str] = list(schema["required_columns"])
    # Extract the alias map from the schema contract
    allowed_aliases: Dict[str, str] = dict(schema.get("allowed_aliases", {}))
    # Extract the declared dtype map from the schema contract
    declared_dtypes: Dict[str, str] = dict(schema.get("dtypes", {}))
    # Extract the lower-bound row count from the schema contract
    min_rows: int = int(schema["min_rows"])
    # Initialise the aggregate failure message list
    failures: List[str] = []
    # Initialise the per-dataset report dictionary
    report: Dict[str, Any] = {
        "dataset": dataset_name,
        "row_count": int(len(raw_df)),
        "missing_columns": [],
        "extra_columns": [],
        "missingness": {},
        "index_unique": bool(raw_df.index.is_unique),
    }
    # Flag non-unique indices as a structural warning in the report
    if not raw_df.index.is_unique:
        failures.append(
            f"{dataset_name}: the DataFrame index is not unique; "
            "split alignment later requires a stable unique index."
        )
    # Check the row count lower bound against the declared minimum
    if len(raw_df) < min_rows:
        failures.append(
            f"{dataset_name}: row count {len(raw_df)} is below the "
            f"declared lower bound {min_rows}."
        )
    # Iterate over every required column to check presence and dtype
    for column in required_columns:
        # Resolve the column name through the alias map
        resolved: Optional[str] = _resolve_column(raw_df, column, allowed_aliases)
        # Record a failure when the column is absent under both names
        if resolved is None:
            failures.append(
                f"{dataset_name}: required column '{column}' (alias "
                f"'{allowed_aliases.get(column)}') is missing."
            )
            # Record the missing column in the report
            report["missing_columns"].append(column)
            # Skip dtype and missingness checks for an absent column
            continue
        # Compare the observed dtype against the declared dtype
        declared_dtype: Optional[str] = declared_dtypes.get(column)
        # Record a failure when the declared dtype is present and mismatched
        if declared_dtype is not None:
            # Normalise the observed dtype to a comparable string
            observed_dtype: str = str(raw_df[resolved].dtype)
            # Record a failure when the normalised dtypes disagree
            if observed_dtype != declared_dtype:
                failures.append(
                    f"{dataset_name}: column '{column}' has dtype "
                    f"'{observed_dtype}' but the schema declares "
                    f"'{declared_dtype}'."
                )
        # Compute and record the missingness proportion for the column
        report["missingness"][column] = float(raw_df[resolved].isna().mean())
    # Identify the additional columns that are not required
    required_set: set = set(required_columns)
    # Collect the alias values as additional acceptable names
    alias_values: set = set(allowed_aliases.values())
    # Record every column outside the required-plus-alias union as extra
    report["extra_columns"] = sorted(
        col
        for col in raw_df.columns
        if col not in required_set and col not in alias_values
    )
    # Raise the aggregated error when any failure was recorded
    if failures:
        raise SchemaValidationError(
            f"Schema validation failed for '{dataset_name}': "
            + " | ".join(failures)
        )
    # Return the per-dataset validation report
    return report


def validate_all_raw_schemas(
    raw_dfs: Dict[str, pd.DataFrame],
    raw_data_schemas: Dict[str, Dict[str, Any]],
    output_dir: Path,
) -> Dict[str, Any]:
    """
    Validate every raw DataFrame against its declared schema contract.

    Parameters
    ----------
    raw_dfs : Dict[str, pd.DataFrame]
        Mapping from dataset name to its raw DataFrame.
    raw_data_schemas : Dict[str, Dict[str, Any]]
        Mapping from dataset name to its schema contract.
    output_dir : Path
        Directory into which validation_report.json is written.

    Returns
    -------
    Dict[str, Any]
        Combined validation report keyed by dataset name.

    Raises
    ------
    SchemaValidationError
        If any dataset fails its schema validation.
    """
    # Validate that the raw data frames mapping is a dict
    if not isinstance(raw_dfs, dict):
        raise TypeError(
            f"raw_dfs must be a dict; received {type(raw_dfs).__name__}."
        )
    # Validate that the schema mapping is a dict
    if not isinstance(raw_data_schemas, dict):
        raise TypeError(
            f"raw_data_schemas must be a dict; "
            f"received {type(raw_data_schemas).__name__}."
        )
    # Validate that the output directory is a Path
    if not isinstance(output_dir, Path):
        raise TypeError(
            f"output_dir must be a pathlib.Path; "
            f"received {type(output_dir).__name__}."
        )
    # Create the output directory when it does not exist yet
    output_dir.mkdir(parents=True, exist_ok=True)
    # Initialise the combined report keyed by dataset name
    combined_report: Dict[str, Any] = {}
    # Track whether every dataset passed its schema validation
    all_passed: bool = True
    # Iterate over every dataset in the declared schema mapping
    for dataset_name, schema in raw_data_schemas.items():
        # Raise when a dataset in the schemas has no raw frame
        if dataset_name not in raw_dfs:
            raise SchemaValidationError(
                f"Schema declared for '{dataset_name}' but no raw DataFrame "
                "was supplied."
            )
        # Validate the raw frame against its schema contract
        try:
            # Collect the per-dataset validation report
            combined_report[dataset_name] = validate_raw_schema(
                raw_df=raw_dfs[dataset_name],
                schema=schema,
                dataset_name=dataset_name,
            )
        except SchemaValidationError as exc:
            # Record the failure and continue to surface all datasets
            combined_report[dataset_name] = {
                "dataset": dataset_name,
                "error": str(exc),
            }
            # Mark the aggregate outcome as failed
            all_passed = False
    # Compose the persisted validation report with a timestamp
    persisted: Dict[str, Any] = {
        "written_at": datetime.now(timezone.utc).isoformat(),
        "datasets": combined_report,
        "all_passed": all_passed,
    }
    # Persist the validation report to the output directory
    report_path: Path = output_dir / "validation_report.json"
    # Serialise the report with stable formatting
    report_path.write_text(
        json.dumps(persisted, indent=2, ensure_ascii=False) + "\n",
        encoding="utf-8",
    )
    # Raise the aggregated failure when any dataset failed
    if not all_passed:
        raise SchemaValidationError(
            "Schema validation failed for one or more datasets; see "
            f"validation_report.json at '{report_path}'."
        )
    # Return the combined validation report
    return combined_report


# -----------------------------------------------------------------------------
# Task 1, Step 2: Validate the study configuration dictionary against paper constants
# -----------------------------------------------------------------------------

def _check(
    condition: bool,
    key_path: str,
    message: str,
    failures: List[str],
) -> None:
    """
    Append a configuration failure when a check does not hold.

    Parameters
    ----------
    condition : bool
        Condition that must hold for the check to pass.
    key_path : str
        Dotted key path of the parameter being checked.
    message : str
        Human-readable description of the failure.
    failures : List[str]
        Mutable list to which the failure message is appended.
    """
    # Append the formatted failure when the condition is false
    if not condition:
        failures.append(f"{key_path}: {message}")


def _get_checked(
    config: Dict[str, Any],
    key_path: str,
    failures: List[str],
) -> Any:
    """
    Read a nested configuration value, recording a failure when missing.

    Parameters
    ----------
    config : Dict[str, Any]
        Configuration dictionary being validated.
    key_path : str
        Dotted key path, e.g. "datasets.Loan.positive_rate".
    failures : List[str]
        Mutable list to which a missing-key failure is appended.

    Returns
    -------
    Any
        The value at the key path, or None when the path is absent.
    """
    # Split the dotted path into its segment names
    segments: List[str] = key_path.split(".")
    # Start the traversal at the configuration root
    current: Any = config
    # Walk every segment of the dotted key path
    for segment in segments:
        # Record a failure when the segment is absent from the level
        if not isinstance(current, dict) or segment not in current:
            failures.append(f"{key_path}: key is missing.")
            # Return None for the absent path
            return None
        # Descend into the next level of the configuration
        current = current[segment]
    # Return the value found at the fully resolved path
    return current


def validate_configuration(cfg: Dict[str, Any]) -> Tuple[Dict[str, Any], Dict[str, Any]]:
    """
    Validate the study configuration against paper-derived constants.

    Checks dataset positive rates, feature counts, split fractions,
    scaling bounds, pruning tolerance, scorecard maximum points, ECE bin
    count, the five-seed list, the derived log-bound constant L, and the
    LLM-related fields (all must be None with llm_used False).

    Parameters
    ----------
    cfg : Dict[str, Any]
        The study configuration dictionary, the single source of truth.

    Returns
    -------
    Tuple[Dict[str, Any], Dict[str, Any]]
        Normalised configuration copy (seeds coerced to tuple, floats
        cast to float) and the validation report including computed L.

    Raises
    ------
    TypeError
        If `cfg` is not a dict.
    ConfigError
        If any check fails; the message lists every failed key path.
    """
    # Validate the configuration dictionary type
    if not isinstance(cfg, dict):
        raise TypeError(
            f"cfg must be a dict; received {type(cfg).__name__}."
        )
    # Initialise the aggregate failure list
    failures: List[str] = []
    # Iterate over every paper dataset statistic
    for dataset_name, paper_stats in PAPER_DATASET_STATS.items():
        # Read the configured row count for the dataset
        n_value: Any = _get_checked(
            cfg, f"datasets.{dataset_name}.n", failures
        )
        # Compare the configured row count against the paper value
        if n_value is not None:
            _check(
                float(n_value) == paper_stats["n"],
                f"datasets.{dataset_name}.n",
                f"expected {paper_stats['n']:.0f}, received {n_value}.",
                failures,
            )
        # Read the configured feature count for the dataset
        features_value: Any = _get_checked(
            cfg, f"datasets.{dataset_name}.feature_count", failures
        )
        # Compare the configured feature count against the paper value
        if features_value is not None:
            _check(
                float(features_value) == paper_stats["features"],
                f"datasets.{dataset_name}.feature_count",
                f"expected {paper_stats['features']:.0f}, received "
                f"{features_value}.",
                failures,
            )
        # Read the configured positive rate for the dataset
        rate_value: Any = _get_checked(
            cfg, f"datasets.{dataset_name}.positive_rate", failures
        )
        # Compare the configured positive rate within float tolerance
        if rate_value is not None:
            _check(
                math.isclose(
                    float(rate_value),
                    paper_stats["positive_rate"],
                    abs_tol=_FLOAT_TOLERANCE,
                ),
                f"datasets.{dataset_name}.positive_rate",
                f"expected {paper_stats['positive_rate']}, "
                f"received {rate_value}.",
                failures,
            )
    # Read the configured split fractions
    train_fraction: Any = _get_checked(
        cfg, "data_splitting.train_fraction", failures
    )
    val_fraction: Any = _get_checked(
        cfg, "data_splitting.val_fraction", failures
    )
    test_fraction: Any = _get_checked(
        cfg, "data_splitting.test_fraction", failures
    )
    # Compare each configured split fraction against the paper values
    if train_fraction is not None:
        _check(
            math.isclose(
                float(train_fraction), PAPER_SPLIT_FRACTIONS[0],
                abs_tol=_FLOAT_TOLERANCE,
            ),
            "data_splitting.train_fraction",
            f"expected {PAPER_SPLIT_FRACTIONS[0]}, received "
            f"{train_fraction}.",
            failures,
        )
    # Compare the validation fraction against the paper value
    if val_fraction is not None:
        _check(
            math.isclose(
                float(val_fraction), PAPER_SPLIT_FRACTIONS[1],
                abs_tol=_FLOAT_TOLERANCE,
            ),
            "data_splitting.val_fraction",
            f"expected {PAPER_SPLIT_FRACTIONS[1]}, received {val_fraction}.",
            failures,
        )
    # Compare the test fraction against the paper value
    if test_fraction is not None:
        _check(
            math.isclose(
                float(test_fraction), PAPER_SPLIT_FRACTIONS[2],
                abs_tol=_FLOAT_TOLERANCE,
            ),
            "data_splitting.test_fraction",
            f"expected {PAPER_SPLIT_FRACTIONS[2]}, received {test_fraction}.",
            failures,
        )
    # Read the configured scaling bounds
    lower_bound: Any = _get_checked(
        cfg, "feature_scaling.lower_bound", failures
    )
    upper_bound: Any = _get_checked(
        cfg, "feature_scaling.upper_bound", failures
    )
    # Compare the configured lower bound against the paper value
    if lower_bound is not None:
        _check(
            math.isclose(
                float(lower_bound), PAPER_SCALE_LOWER, abs_tol=_FLOAT_TOLERANCE
            ),
            "feature_scaling.lower_bound",
            f"expected {PAPER_SCALE_LOWER}, received {lower_bound}.",
            failures,
        )
    # Compare the configured upper bound against the paper value
    if upper_bound is not None:
        _check(
            math.isclose(
                float(upper_bound), PAPER_SCALE_UPPER, abs_tol=_FLOAT_TOLERANCE
            ),
            "feature_scaling.upper_bound",
            f"expected {PAPER_SCALE_UPPER}, received {upper_bound}.",
            failures,
        )
    # Compute the derived log-bound constant L from the scaling bounds
    computed_L: float = max(
        abs(math.log(PAPER_SCALE_LOWER)),
        abs(math.log(PAPER_SCALE_UPPER)),
    )
    # Read the configured log-bound constant when present
    configured_L: Any = _get_checked(cfg, "feature_scaling.log_bound_L", failures)
    # Compare the configured L against the computed value
    if configured_L is not None:
        _check(
            math.isclose(
                float(configured_L), computed_L, abs_tol=_FLOAT_TOLERANCE
            ),
            "feature_scaling.log_bound_L",
            f"expected {computed_L!r}, received {configured_L!r}.",
            failures,
        )
    # Read the pruning tolerance from the configuration
    tolerance: Any = _get_checked(cfg, "post_hoc_pruning.tolerance", failures)
    # Compare the pruning tolerance against the paper value
    if tolerance is not None:
        _check(
            math.isclose(
                float(tolerance), PAPER_PRUNING_TOLERANCE,
                abs_tol=_FLOAT_TOLERANCE,
            ),
            "post_hoc_pruning.tolerance",
            f"expected {PAPER_PRUNING_TOLERANCE}, received {tolerance}.",
            failures,
        )
    # Read the scorecard maximum points from the configuration
    max_points: Any = _get_checked(
        cfg, "scorecard_and_tally.max_points", failures
    )
    # Compare the maximum points against the paper value
    if max_points is not None:
        _check(
            float(max_points) == PAPER_SCORECARD_MAX_POINTS,
            "scorecard_and_tally.max_points",
            f"expected {PAPER_SCORECARD_MAX_POINTS:.0f}, "
            f"received {max_points}.",
            failures,
        )
    # Read the ECE bin count from the configuration
    ece_bins: Any = _get_checked(cfg, "calibration.ece_bins", failures)
    # Compare the ECE bin count against the paper-derived choice
    if ece_bins is not None:
        _check(
            float(ece_bins) == PAPER_ECE_BINS,
            "calibration.ece_bins",
            f"expected {PAPER_ECE_BINS:.0f}, received {ece_bins}.",
            failures,
        )
    # Read the threshold decision metric from the configuration
    threshold_metric: Any = _get_checked(
        cfg, "scorecard_and_tally.threshold_metric", failures
    )
    # Validate the threshold metric against the replicator choice
    if threshold_metric is not None:
        _check(
            str(threshold_metric).lower() == "f1",
            "scorecard_and_tally.threshold_metric",
            "expected 'f1' (the replicator-chosen decision metric).",
            failures,
        )
    # Read the seed list from the configuration
    seeds_value: Any = _get_checked(cfg, "data_splitting.seeds", failures)
    # Compare the seed list against the paper-derived seed tuple
    if seeds_value is not None:
        _check(
            tuple(seeds_value) == PAPER_SEEDS,
            "data_splitting.seeds",
            f"expected {list(PAPER_SEEDS)}, received {list(seeds_value)}.",
            failures,
        )
    # Read the LLM settings section from the configuration
    llm_used: Any = _get_checked(cfg, "llm_settings.llm_used", failures)
    # Validate that the LLM usage flag is exactly False
    if llm_used is not None:
        _check(
            bool(llm_used) is False,
            "llm_settings.llm_used",
            "must be False; the paper contains no LLM component.",
            failures,
        )
    # Iterate over every LLM field that must be None
    for llm_field in (
        "llm_names",
        "llm_api_keys",
        "llm_temperature",
        "llm_max_tokens",
        "llm_model_version",
        "prompt_templates",
    ):
        # Read the LLM field value from the configuration
        field_value: Any = _get_checked(
            cfg, f"llm_settings.{llm_field}", failures
        )
        # Validate that the LLM field is exactly None
        if field_value is not None:
            _check(
                field_value is None,
                f"llm_settings.{llm_field}",
                "must be None; the paper contains no LLM component.",
                failures,
            )
    # Raise the aggregated ConfigError when any check failed
    if failures:
        raise ConfigError(
            "Configuration validation failed: " + " | ".join(failures)
        )
    # Build a normalised copy of the configuration (never mutate the input)
    normalized_cfg: Dict[str, Any] = json.loads(json.dumps(cfg))
    # Coerce the seed list to an immutable tuple in the normalised copy
    normalized_cfg["data_splitting"]["seeds"] = tuple(
        normalized_cfg["data_splitting"]["seeds"]
    )
    # Cast the split fractions to floats in the normalised copy
    normalized_cfg["data_splitting"]["train_fraction"] = float(
        normalized_cfg["data_splitting"]["train_fraction"]
    )
    # Cast the validation fraction to a float
    normalized_cfg["data_splitting"]["val_fraction"] = float(
        normalized_cfg["data_splitting"]["val_fraction"]
    )
    # Cast the test fraction to a float
    normalized_cfg["data_splitting"]["test_fraction"] = float(
        normalized_cfg["data_splitting"]["test_fraction"]
    )
    # Compose the configuration validation report
    report: Dict[str, Any] = {
        "passed": True,
        "computed_log_bound_L": computed_L,
        "verified_max_points": PAPER_SCORECARD_MAX_POINTS,
        "verified_seeds": list(PAPER_SEEDS),
        "validated_at": datetime.now(timezone.utc).isoformat(),
    }
    # Return the normalised configuration and the validation report
    return normalized_cfg, report


# -----------------------------------------------------------------------------
# Task 1, Step 3: Validate that raw positive-class rates can be reproduced
# -----------------------------------------------------------------------------

def validate_raw_label_rates(
    raw_df: pd.DataFrame,
    label_col: str,
    dataset_name: str,
    cfg: Dict[str, Any],
) -> Dict[str, Any]:
    """
    Validate that raw positive-class rates reproduce the paper's rates.

    For Loan, the check applies the exact two-status filter and verifies
    the filtered count (within 0.5% of 395,492) and filtered positive
    rate (within 0.001 of 0.101). For the other three datasets the direct
    positive rate must match the paper rate within 0.001.

    Parameters
    ----------
    raw_df : pd.DataFrame
        Raw DataFrame whose label column is validated.
    label_col : str
        Canonical label column name.
    dataset_name : str
        Name of the dataset, e.g. "Loan".
    cfg : Dict[str, Any]
        Validated study configuration supplying the paper rates.

    Returns
    -------
    Dict[str, Any]
        Label check report with frequencies, counts, rates, and pass flag.

    Raises
    ------
    LabelRateError
        If the count or rate deviates beyond tolerance; the exception
        carries the full diagnostic report.
    """
    # Validate the label column exists in the raw frame
    if label_col not in raw_df.columns:
        raise LabelRateError(
            f"{dataset_name}: label column '{label_col}' is missing.",
            {"dataset": dataset_name, "error": "missing label column"},
        )
    # Extract the label series and canonicalise strings by stripping spaces
    label_series: pd.Series = raw_df[label_col]
    # Coerce the label series to a canonical string type when object-like
    if pd.api.types.is_object_dtype(label_series) or pd.api.types.is_string_dtype(
        label_series
    ):
        # Strip whitespace from every label value
        label_series = label_series.astype(str).str.strip()
    # Compute the full frequency table of the canonical label values
    frequencies: Dict[str, int] = {
        str(k): int(v)
        for k, v in label_series.value_counts(dropna=False).items()
    }
    # Handle the Loan dataset with its exact two-status filter
    if dataset_name == "Loan":
        # Build the boolean mask over the exact keep statuses
        keep_mask: pd.Series = label_series.isin(LOAN_KEEP_STATUSES)
        # Map the positive status to 1 and the negative status to 0
        label_binary: pd.Series = (
            label_series == LOAN_POSITIVE_STATUS
        ).astype(int)
        # Apply the keep mask to the binary label
        filtered_label: pd.Series = label_binary[keep_mask]
        # Compute the filtered population count
        filtered_count: int = int(filtered_label.shape[0])
        # Compute the filtered positive rate (mean of the binary label)
        filtered_rate: float = float(filtered_label.mean())
        # Read the expected population size from the paper constants
        expected_count: float = PAPER_DATASET_STATS["Loan"]["n"]
        # Read the expected positive rate from the paper constants
        expected_rate: float = PAPER_DATASET_STATS["Loan"]["positive_rate"]
        # Verify the label is binary within the filtered population
        unique_values: set = set(filtered_label.unique())
        # Record the binary check outcome
        binary_ok: bool = unique_values.issubset({0, 1})
        # Initialise the failure diagnostics
        diagnostics: Dict[str, Any] = {
            "dataset": dataset_name,
            "label_frequencies": frequencies,
            "filtered_count": filtered_count,
            "expected_count": int(expected_count),
            "filtered_positive_rate": filtered_rate,
            "expected_positive_rate": expected_rate,
            "binary_label": binary_ok,
        }
        # Compute the relative count deviation from the paper population
        count_deviation: float = (
            abs(filtered_count - expected_count) / expected_count
        )
        # Compute the absolute rate deviation from the paper rate
        rate_deviation: float = abs(filtered_rate - expected_rate)
        # Raise when the count deviates beyond 0.5% of the paper population
        if count_deviation > 0.005:
            raise LabelRateError(
                f"Loan filtered count {filtered_count} deviates from the "
                f"expected {int(expected_count)} by "
                f"{count_deviation * 100:.2f}%; the exact two-status "
                "filter cannot reproduce the paper population. See the "
                "label frequency table in the diagnostics.",
                diagnostics,
            )
        # Raise when the rate deviates beyond the absolute tolerance
        if rate_deviation > 0.001:
            raise LabelRateError(
                f"Loan filtered positive rate {filtered_rate:.4f} deviates "
                f"from the expected {expected_rate} by "
                f"{rate_deviation:.4f}.", diagnostics,
            )
        # Raise when the filtered label is not binary
        if not binary_ok:
            raise LabelRateError(
                "Loan filtered label is not binary.", diagnostics,
            )
        # Return the passing Loan label check report
        return {
            "dataset": dataset_name,
            "label_frequencies": frequencies,
            "filtered_count": filtered_count,
            "filtered_positive_rate": filtered_rate,
            "raw_positive_rate": float((label_series == LOAN_POSITIVE_STATUS).mean()),
            "pass": True,
        }
    # Handle the direct-rate datasets (FraudEcom, Default, Creditcard)
    # Coerce the label to a binary numeric series when possible
    if is_bool_dtype(label_series) or is_integer_dtype(label_series):
        # Convert the label series to a numeric binary series
        label_binary = label_series.astype(int)
    else:
        # Attempt a numeric coercion of the object label series
        try:
            # Convert the label series to numeric values
            label_binary = pd.to_numeric(label_series, errors="raise").astype(int)
        except (ValueError, TypeError) as exc:
            # Raise when the label cannot be interpreted as binary
            raise LabelRateError(
                f"{dataset_name}: label column cannot be coerced to "
                "numeric binary values.",
                {"dataset": dataset_name, "error": str(exc)},
            ) from exc
    # Verify the label is binary in the raw population
    unique_values = set(label_binary.unique())
    # Raise when the raw label is not binary
    if not unique_values.issubset({0, 1}):
        raise LabelRateError(
            f"{dataset_name}: label is not binary; observed values "
            f"{sorted(unique_values)}.",
            {"dataset": dataset_name, "label_frequencies": frequencies},
        )
    # Compute the direct positive rate as the label mean
    raw_rate: float = float(label_binary.mean())
    # Read the expected positive rate from the paper constants
    expected_rate = PAPER_DATASET_STATS[dataset_name]["positive_rate"]
    # Compute the absolute rate deviation from the paper rate
    rate_deviation = abs(raw_rate - expected_rate)
    # Raise when the direct rate deviates beyond the tolerance
    if rate_deviation > 0.001:
        raise LabelRateError(
            f"{dataset_name}: raw positive rate {raw_rate:.4f} deviates "
            f"from the paper rate {expected_rate} by {rate_deviation:.4f}.",
            {
                "dataset": dataset_name,
                "label_frequencies": frequencies,
                "raw_positive_rate": raw_rate,
                "expected_positive_rate": expected_rate,
            },
        )
    # Return the passing direct-rate check report
    return {
        "dataset": dataset_name,
        "label_frequencies": frequencies,
        "raw_positive_rate": raw_rate,
        "pass": True,
    }


def validate_all_label_rates(
    raw_dfs: Dict[str, pd.DataFrame],
    cfg: Dict[str, Any],
    output_dir: Path,
) -> Dict[str, Any]:
    """
    Validate the label rates of every dataset and persist the report.

    Parameters
    ----------
    raw_dfs : Dict[str, pd.DataFrame]
        Mapping from dataset name to its raw DataFrame.
    cfg : Dict[str, Any]
        Validated study configuration.
    output_dir : Path
        Directory into which label_check_report.json is written.

    Returns
    -------
    Dict[str, Any]
        Combined label check report keyed by dataset name.

    Raises
    ------
    LabelRateError
        Propagated from the first dataset whose rate check fails.
    """
    # Extract the raw-data schema contracts from the configuration
    raw_data_schemas: Dict[str, Any] = cfg["raw_data_schemas"]
    # Initialise the combined label check report
    combined_report: Dict[str, Any] = {}
    # Iterate over every dataset in the schema mapping
    for dataset_name, schema in raw_data_schemas.items():
        # Extract the canonical label column from the schema contract
        label_col: str = schema["label_column"]
        # Validate the label rate of the dataset
        combined_report[dataset_name] = validate_raw_label_rates(
            raw_df=raw_dfs[dataset_name],
            label_col=label_col,
            dataset_name=dataset_name,
            cfg=cfg,
        )
    # Compose the persisted label check report with a timestamp
    persisted: Dict[str, Any] = {
        "written_at": datetime.now(timezone.utc).isoformat(),
        "datasets": combined_report,
        "all_passed": all(
            report["pass"] for report in combined_report.values()
        ),
    }
    # Persist the label check report to the output directory
    report_path: Path = output_dir / "label_check_report.json"
    # Serialise the report with stable formatting
    report_path.write_text(
        json.dumps(persisted, indent=2, ensure_ascii=False) + "\n",
        encoding="utf-8",
    )
    # Return the combined label check report
    return combined_report


# -----------------------------------------------------------------------------
# Task 1, Orchestrator Function
# -----------------------------------------------------------------------------

def orchestrate_validation(
    loan_raw: pd.DataFrame,
    fraud_ecom_raw: pd.DataFrame,
    default_raw: pd.DataFrame,
    creditcard_raw: pd.DataFrame,
    study_configuration: Dict[str, Any],
    output_dir: Path,
) -> Dict[str, Any]:
    """
    Orchestrate Task 1: validate configuration, schemas, and label rates.

    The orchestrator runs the configuration validator first (all later
    tasks depend on it), then the raw schema validation for all four
    datasets, then the raw label-rate validation. It assembles every
    report into a single auditable result and persists the two JSON
    reports (validation_report.json and label_check_report.json).

    Parameters
    ----------
    loan_raw : pd.DataFrame
        Raw Lending Club loan DataFrame.
    fraud_ecom_raw : pd.DataFrame
        Raw Fraud Ecommerce DataFrame.
    default_raw : pd.DataFrame
        Raw UCI Default of Credit Card Clients DataFrame.
    creditcard_raw : pd.DataFrame
        Raw Credit Card Fraud DataFrame.
    study_configuration : Dict[str, Any]
        The study configuration dictionary (single source of truth).
    output_dir : Path
        Directory into which the validation reports are written.

    Returns
    -------
    Dict[str, Any]
        Merged report with keys "config", "schemas", and "label_rates".

    Raises
    ------
    ConfigError
        If the configuration fails any paper-constant check.
    SchemaValidationError
        If any raw DataFrame fails its schema contract.
    LabelRateError
        If any raw positive-class rate cannot be reproduced.
    """
    # Validate the raw frame arguments are all DataFrames
    raw_frames: Dict[str, pd.DataFrame] = {
        "Loan": loan_raw,
        "FraudEcom": fraud_ecom_raw,
        "Default": default_raw,
        "Creditcard": creditcard_raw,
    }
    # Iterate over the raw frames to validate their types
    for name, frame in raw_frames.items():
        # Raise when any raw frame is not a DataFrame
        if not isinstance(frame, pd.DataFrame):
            raise TypeError(
                f"{name} must be a pd.DataFrame; "
                f"received {type(frame).__name__}."
            )
    # Validate the configuration dictionary type
    if not isinstance(study_configuration, dict):
        raise TypeError(
            "study_configuration must be a dict; "
            f"received {type(study_configuration).__name__}."
        )
    # Validate the output directory type
    if not isinstance(output_dir, Path):
        raise TypeError(
            f"output_dir must be a pathlib.Path; "
            f"received {type(output_dir).__name__}."
        )
    # Create the output directory when it does not exist yet
    output_dir.mkdir(parents=True, exist_ok=True)
    # Step 1.2: validate the configuration against paper constants
    normalized_cfg, config_report = validate_configuration(study_configuration)
    # Extract the raw schema contracts from the normalised configuration
    raw_data_schemas: Dict[str, Any] = normalized_cfg["raw_data_schemas"]
    # Step 1.1: validate every raw frame against its schema contract
    schemas_report: Dict[str, Any] = validate_all_raw_schemas(
        raw_dfs=raw_frames,
        raw_data_schemas=raw_data_schemas,
        output_dir=output_dir,
    )
    # Step 1.3: validate every raw label rate against the paper's rates
    label_rates_report: Dict[str, Any] = validate_all_label_rates(
        raw_dfs=raw_frames,
        cfg=normalized_cfg,
        output_dir=output_dir,
    )
    # Assemble the merged validation result
    merged_report: Dict[str, Any] = {
        "config": config_report,
        "schemas": schemas_report,
        "label_rates": label_rates_report,
    }
    # Return the merged validation result
    return merged_report
````

````python
# Task 2: Cleanse raw data with leakage-free, training-only decisions

# ==============================================================================
# Task 2: Cleanse raw data with leakage-free, training-only decisions
# ==============================================================================

# -----------------------------------------------------------------------------
# Module-level deterministic mapping tables (documented in the module
# docstring of each cleaner; these are replicator choices, UNSPECIFIED IN
# MANUSCRIPT, recorded for the decision log).
# -----------------------------------------------------------------------------
# Reference date for all calendar-month derivations (dataset access date
# from the paper bibliography: 2026-08-09)
REFERENCE_DATE: pd.Timestamp = pd.Timestamp(year=2026, month=8, day=9)

# Mapping from loan grade letters to ordinal values A=1 .. G=7
GRADE_ORDINAL_MAP: Dict[str, int] = {
    "A": 1, "B": 2, "C": 3, "D": 4, "E": 5, "F": 6, "G": 7,
}

# Mapping from loan term strings to months
TERM_MONTHS_MAP: Dict[str, int] = {"36 months": 36, "60 months": 60}

# FraudEcom source channel ordinal encoding (replicator choice)
SOURCE_ENCODING_MAP: Dict[str, int] = {"SEO": 0, "Ads": 1, "Direct": 2}

# Default EDUCATION recode: 0, 5, 6 collapse into category 4 (others)
EDUCATION_RECODE_MAP: Dict[int, int] = {0: 4, 5: 4, 6: 4}

# Default MARRIAGE recode: 0 (missing) collapses into category 3 (others)
MARRIAGE_RECODE_MAP: Dict[int, int] = {0: 3}


# -----------------------------------------------------------------------------
# Task 2, Step 1: Cleanse loan_raw
# -----------------------------------------------------------------------------

def _calendar_months_since(
    date_series: pd.Series, reference: pd.Timestamp
) -> pd.Series:
    """
    Compute calendar months from a reference date to each parsed date.

    The difference is computed by year/month arithmetic,
    months = 12 * (ref.year - d.year) + (ref.month - d.month), never by
    elapsed days divided by 30, because the paper's month-based features
    are calendar months.

    Parameters
    ----------
    date_series : pd.Series
        Series of datetime64 values (NaT allowed).
    reference : pd.Timestamp
        Fixed reference date.

    Returns
    -------
    pd.Series
        Float series of calendar-month differences (NaN for NaT inputs).
    """
    # Extract the year component of every parsed date
    years: pd.Series = date_series.dt.year
    # Extract the month component of every parsed date
    months: pd.Series = date_series.dt.month
    # Compute the calendar month difference via year/month arithmetic
    month_diff: pd.Series = (
        12 * (reference.year - years) + (reference.month - months)
    )
    # Convert the integer difference to float with NaN for missing inputs
    result: pd.Series = month_diff.astype("float64")
    # Propagate NaN for NaT inputs after the arithmetic
    result = result.where(date_series.notna(), np.nan)
    # Return the calendar month difference series
    return result


def _parse_lc_date(raw_series: pd.Series) -> pd.Series:
    """
    Parse Lending Club date strings of the fixed form %b-%Y.

    Parameters
    ----------
    raw_series : pd.Series
        Series of raw date strings (object dtype, may be missing).

    Returns
    -------
    pd.Series
        datetime64 series; unparseable values become NaT.
    """
    # Parse the strings with the fixed Lending Club format, coercing errors
    parsed: pd.Series = pd.to_datetime(
        raw_series, format="%b-%Y", errors="coerce"
    )
    # Return the parsed datetime series
    return parsed


def cleanse_loan(
    raw_df: pd.DataFrame, cfg: Dict[str, Any]
) -> Tuple[pd.DataFrame, Dict[str, Any]]:
    """
    Cleanse loan_raw into typed candidate feature columns plus label y.

    Executes in strict order: filter rows to the keep-set and derive the
    binary label; drop id and loan_status; parse numeric columns; parse
    date columns with calendar-month derivations; create missingness
    flags; encode nominal indicators; combine FICO fields. No statistic
    is computed across the dataset; numeric NaNs remain for Task 4
    training-median imputation.

    Parameters
    ----------
    raw_df : pd.DataFrame
        Validated raw Lending Club loan DataFrame.
    cfg : Dict[str, Any]
        Validated study configuration (used for schema references).

    Returns
    -------
    Tuple[pd.DataFrame, Dict[str, Any]]
        Cleaned DataFrame (original index preserved, columns: candidate
        features plus y) and the cleansing log with row/column accounting.

    Raises
    ------
    TypeError
        If `raw_df` is not a pd.DataFrame or `cfg` is not a dict.
    ValueError
        If the label keep-set is absent from the frame.
    """
    # Validate the raw frame type
    if not isinstance(raw_df, pd.DataFrame):
        raise TypeError(
            f"raw_df must be a pd.DataFrame; received {type(raw_df).__name__}."
        )
    # Validate the configuration type
    if not isinstance(cfg, dict):
        raise TypeError(f"cfg must be a dict; received {type(cfg).__name__}.")
    # Record the input row count for the cleansing log
    input_rows: int = int(len(raw_df))
    # Verify the label column is present in the raw frame
    if "loan_status" not in raw_df.columns:
        raise ValueError("loan_status column is missing from loan_raw.")
    # Canonicalise the status strings by stripping whitespace
    status_series: pd.Series = raw_df["loan_status"].astype(str).str.strip()
    # Build the boolean mask over the exact two keep statuses
    keep_mask: pd.Series = status_series.isin(("Fully Paid", "Charged Off"))
    # Filter the raw frame to the keep-set rows
    cleaned: pd.DataFrame = raw_df.loc[keep_mask].copy()
    # Record the number of rows dropped by the label filter
    dropped_rows: int = input_rows - int(len(cleaned))
    # Derive the binary label: Charged Off is positive, Fully Paid negative
    cleaned["y"] = (status_series.loc[keep_mask] == "Charged Off").astype(int)
    # Drop the original status column now that the label is derived
    cleaned = cleaned.drop(columns=["loan_status"])
    # Drop the loan identifier: retaining it would create a leakage artifact
    cleaned = cleaned.drop(columns=["id"])
    # Parse the annual interest rate by stripping a trailing percent sign
    cleaned["int_rate"] = (
        cleaned["int_rate"].astype(str).str.rstrip("%").astype(float)
    )
    # Parse the term into months via the explicit mapping after stripping
    cleaned["term_months"] = (
        cleaned["term"].astype(str).str.strip().map(TERM_MONTHS_MAP)
    )
    # Drop the raw term column now that term_months is derived
    cleaned = cleaned.drop(columns=["term"])
    # Parse the employment length into years via a leading-number regex
    emp_years: pd.Series = cleaned["emp_length"].astype(str).str.strip()
    # Extract the leading integer from each employment length string
    emp_numbers: pd.Series = (
        emp_years.str.extract(r"^(\d+)", expand=False).astype(float)
    )
    # Map the "< 1 year" boundary case explicitly to zero years
    emp_numbers = emp_numbers.where(
        ~emp_years.str.lower().str.startswith("<"), 0.0
    )
    # Assign the parsed employment years to the cleaned frame
    cleaned["emp_length_years"] = emp_numbers
    # Drop the raw employment length column
    cleaned = cleaned.drop(columns=["emp_length"])
    # Map the grade letters to ordinals after canonical case normalisation
    cleaned["grade_ordinal"] = (
        cleaned["grade"].astype(str).str.strip().str.upper().map(GRADE_ORDINAL_MAP)
    )
    # Drop the raw grade column
    cleaned = cleaned.drop(columns=["grade"])
    # Parse the three Lending Club date columns with the fixed format
    issue_dates: pd.Series = _parse_lc_date(cleaned["issue_d"])
    last_pymnt_dates: pd.Series = _parse_lc_date(cleaned["last_pymnt_d"])
    credit_pull_dates: pd.Series = _parse_lc_date(cleaned["last_credit_pull_d"])
    # Derive calendar months since issue from the reference date
    cleaned["mths_since_issue"] = _calendar_months_since(
        issue_dates, REFERENCE_DATE
    )
    # Derive calendar months since the last payment
    cleaned["mths_since_pymnt"] = _calendar_months_since(
        last_pymnt_dates, REFERENCE_DATE
    )
    # Derive calendar months since the last credit pull
    cleaned["mths_since_last_credit_pull"] = _calendar_months_since(
        credit_pull_dates, REFERENCE_DATE
    )
    # Drop the three raw date columns now that features are derived
    cleaned = cleaned.drop(
        columns=["issue_d", "last_pymnt_d", "last_credit_pull_d"]
    )
    # Flag missing next payment dates: 1 when missing, 0 otherwise
    cleaned["next_pymnt_missing"] = cleaned["next_pymnt_d"].isna().astype(int)
    # Drop the raw next payment date column after flagging
    cleaned = cleaned.drop(columns=["next_pymnt_d"])
    # Flag missing delinquency-month values: 1 when missing, 0 otherwise
    cleaned["mths_since_last_delinq_missing"] = (
        cleaned["mths_since_last_delinq"].isna().astype(int)
    )
    # Flag missing record-month values: 1 when missing, 0 otherwise
    cleaned["mths_since_last_record_missing"] = (
        cleaned["mths_since_last_record"].isna().astype(int)
    )
    # Build the home ownership indicators with the reference category
    # absorbing all values other than RENT, MORTGAGE, and OWN
    cleaned["home_ownership_RENT"] = (
        cleaned["home_ownership"] == "RENT"
    ).astype(int)
    # Build the mortgage indicator
    cleaned["home_ownership_MORTGAGE"] = (
        cleaned["home_ownership"] == "MORTGAGE"
    ).astype(int)
    # Build the owned indicator
    cleaned["home_ownership_OWN"] = (
        cleaned["home_ownership"] == "OWN"
    ).astype(int)
    # Drop the raw home ownership column after encoding
    cleaned = cleaned.drop(columns=["home_ownership"])
    # Build the source-verified indicator (reference: Income Verified)
    cleaned["verification_status_Source_Verified"] = (
        cleaned["verification_status"] == "Source Verified"
    ).astype(int)
    # Build the not-verified indicator (reference: Income Verified)
    cleaned["verification_status_Not_Verified"] = (
        cleaned["verification_status"] == "Not Verified"
    ).astype(int)
    # Drop the raw verification status column after encoding
    cleaned = cleaned.drop(columns=["verification_status"])
    # Build the debt consolidation purpose indicator
    cleaned["purpose_debt_consolidation"] = (
        cleaned["purpose"] == "debt_consolidation"
    ).astype(int)
    # Build the credit card purpose indicator
    cleaned["purpose_credit_card"] = (
        cleaned["purpose"] == "credit_card"
    ).astype(int)
    # Build the home improvement purpose indicator
    cleaned["purpose_home_improvement"] = (
        cleaned["purpose"] == "home_improvement"
    ).astype(int)
    # Build the aggregate indicator absorbing all remaining purposes
    cleaned["purpose_other"] = (
        ~cleaned["purpose"].isin(
            ["debt_consolidation", "credit_card", "home_improvement"]
        )
    ).astype(int)
    # Drop the raw purpose column after encoding
    cleaned = cleaned.drop(columns=["purpose"])
    # Combine the FICO range endpoints into the midpoint feature
    cleaned["fico_midpoint"] = (
        cleaned["last_fico_range_high"] + cleaned["last_fico_range_low"]
    ) / 2.0
    # Drop the FICO low endpoint now that the midpoint is derived
    cleaned = cleaned.drop(columns=["last_fico_range_low"])
    # Assemble the cleansing log for the audit trail
    cleansing_log: Dict[str, Any] = {
        "dataset": "Loan",
        "input_rows": input_rows,
        "output_rows": int(len(cleaned)),
        "rows_dropped": dropped_rows,
        "label_mapping": {"Fully Paid": 0, "Charged Off": 1},
        "dropped_columns": [
            "loan_status", "id", "term", "emp_length", "grade", "issue_d",
            "last_pymnt_d", "last_credit_pull_d", "next_pymnt_d",
            "home_ownership", "verification_status", "purpose",
            "last_fico_range_low",
        ],
        "renamed_columns": {"term_months": "term", "emp_length_years": "emp_length"},
    }
    # Return the cleaned frame and its cleansing log
    return cleaned, cleansing_log


# -----------------------------------------------------------------------------
# Task 2, Step 2: Cleanse fraud_ecom_raw, default_raw, and creditcard_raw
# -----------------------------------------------------------------------------

def cleanse_fraud_ecom(
    raw_df: pd.DataFrame, cfg: Dict[str, Any]
) -> Tuple[pd.DataFrame, Dict[str, Any]]:
    """
    Cleanse fraud_ecom_raw: drop identifiers, parse times, keep raw keys.

    No frequency statistic is computed here: ip_address and device_id are
    preserved as raw columns so Task 3 can build training-only counts.

    Parameters
    ----------
    raw_df : pd.DataFrame
        Validated raw Fraud Ecommerce DataFrame.
    cfg : Dict[str, Any]
        Validated study configuration containing the label alias map.

    Returns
    -------
    Tuple[pd.DataFrame, Dict[str, Any]]
        Cleaned DataFrame (original index preserved) and cleansing log.

    Raises
    ------
    TypeError
        If `raw_df` is not a pd.DataFrame or `cfg` is not a dict.
    ValueError
        If no label column (canonical or aliased) is present.
    """
    # Validate the raw frame type
    if not isinstance(raw_df, pd.DataFrame):
        raise TypeError(
            f"raw_df must be a pd.DataFrame; received {type(raw_df).__name__}."
        )
    # Validate the configuration type
    if not isinstance(cfg, dict):
        raise TypeError(f"cfg must be a dict; received {type(cfg).__name__}.")
    # Record the input row count for the cleansing log
    input_rows: int = int(len(raw_df))
    # Extract the fraud label alias map from the schema contract
    alias_map: Dict[str, str] = cfg["raw_data_schemas"]["FraudEcom"].get(
        "allowed_aliases", {}
    )
    # Resolve the canonical label name (raw_class) or its alias (class)
    label_col: Optional[str] = (
        "raw_class"
        if "raw_class" in raw_df.columns
        else alias_map.get("raw_class")
    )
    # Raise when no fraud label column exists under either name
    if label_col is None:
        raise ValueError(
            "FraudEcom label column 'raw_class' (alias 'class') is missing; "
            f"observed columns: {sorted(raw_df.columns)}."
        )
    # Copy the raw frame to avoid mutating the validated input
    cleaned: pd.DataFrame = raw_df.copy()
    # Drop the account identifier (not a feature)
    cleaned = cleaned.drop(columns=["user_id"])
    # Drop the browser column (excluded from the six-feature matrix)
    cleaned = cleaned.drop(columns=["browser"])
    # Parse the signup timestamp with the declared format
    signup_time: pd.Series = pd.to_datetime(
        cleaned["signup_time"], format="%Y-%m-%d %H:%M:%S", errors="coerce"
    )
    # Parse the purchase timestamp with the declared format
    purchase_time: pd.Series = pd.to_datetime(
        cleaned["purchase_time"], format="%Y-%m-%d %H:%M:%S", errors="coerce"
    )
    # Compute the hours between purchase and signup (derived, not retained)
    time_to_purchase_hours: pd.Series = (
        purchase_time - signup_time
    ).dt.total_seconds() / 3600.0
    # Compute the audit summary of the derived-but-excluded feature: the
    # row count, missing count, and min/max of the hour values. These are
    # deterministic statistics of the timestamp parse, not data-dependent
    # decisions, so the leakage-free cleansing contract is preserved
    derived_valid: pd.Series = time_to_purchase_hours.dropna()
    # Compose the derived-feature summary for the cleansing log
    derived_summary: Dict[str, Any] = {
        "row_count": int(time_to_purchase_hours.shape[0]),
        "missing_count": int(time_to_purchase_hours.isna().sum()),
        "min_hours": (
            float(derived_valid.min()) if derived_valid.shape[0] > 0 else None
        ),
        "max_hours": (
            float(derived_valid.max()) if derived_valid.shape[0] > 0 else None
        ),
    }
    # Rename the sex column to the model-feature name sex_bin
    cleaned = cleaned.rename(columns={"sex": "sex_bin"})
    # Rename the label column to the canonical y
    cleaned = cleaned.rename(columns={label_col: "y"})
    # Replace the parsed timestamps back into the cleaned frame
    cleaned["signup_time"] = signup_time
    cleaned["purchase_time"] = purchase_time
    # Return the cleaned frame and its cleansing log
    cleansing_log: Dict[str, Any] = {
        "dataset": "FraudEcom",
        "input_rows": input_rows,
        "output_rows": int(len(cleaned)),
        "rows_dropped": 0,
        "dropped_columns": ["user_id", "browser"],
        "renamed_columns": {"sex_bin": "sex", "y": label_col},
        "derived_not_retained": {
            "name": "time_to_purchase_hours",
            "summary": derived_summary,
        },
        "label_alias_resolved": label_col,
    }
    # Return the cleaned frame and its cleansing log
    return cleaned, cleansing_log


def cleanse_default(
    raw_df: pd.DataFrame, cfg: Dict[str, Any]
) -> Tuple[pd.DataFrame, Dict[str, Any]]:
    """
    Cleanse default_raw: drop ID, recode categorical codes, binary SEX.

    Parameters
    ----------
    raw_df : pd.DataFrame
        Validated raw UCI Default of Credit Card Clients DataFrame.
    cfg : Dict[str, Any]
        Validated study configuration.

    Returns
    -------
    Tuple[pd.DataFrame, Dict[str, Any]]
        Cleaned DataFrame (original index preserved) and cleansing log.

    Raises
    ------
    TypeError
        If `raw_df` is not a pd.DataFrame.
    ValueError
        If recoding does not produce the exact expected value sets.
    """
    # Validate the raw frame type
    if not isinstance(raw_df, pd.DataFrame):
        raise TypeError(
            f"raw_df must be a pd.DataFrame; received {type(raw_df).__name__}."
        )
    # Record the input row count for the cleansing log
    input_rows: int = int(len(raw_df))
    # Copy the raw frame to avoid mutating the validated input
    cleaned: pd.DataFrame = raw_df.copy()
    # Drop the client identifier (not a feature)
    cleaned = cleaned.drop(columns=["ID"])
    # Apply the deterministic EDUCATION recode (0, 5, 6 collapse into 4)
    cleaned["EDUCATION"] = (
        cleaned["EDUCATION"].map(EDUCATION_RECODE_MAP).fillna(cleaned["EDUCATION"])
    )
    # Post-condition: the recoded EDUCATION values must be exactly 1..4
    education_values: set = set(cleaned["EDUCATION"].unique())
    # Raise when the post-condition is violated
    if not education_values.issubset({1, 2, 3, 4}):
        raise ValueError(
            f"EDUCATION recode produced unexpected values {education_values}; "
            "expected a subset of {1, 2, 3, 4}."
        )
    # Apply the deterministic MARRIAGE recode (0 collapses into 3)
    cleaned["MARRIAGE"] = (
        cleaned["MARRIAGE"].map(MARRIAGE_RECODE_MAP).fillna(cleaned["MARRIAGE"])
    )
    # Post-condition: the recoded MARRIAGE values must be exactly 1..3
    marriage_values: set = set(cleaned["MARRIAGE"].unique())
    # Raise when the post-condition is violated
    if not marriage_values.issubset({1, 2, 3}):
        raise ValueError(
            f"MARRIAGE recode produced unexpected values {marriage_values}; "
            "expected a subset of {1, 2, 3}."
        )
    # Derive the binary sex indicator: raw 1 (male) maps to 1, else 0
    cleaned["SEX_bin"] = (cleaned["SEX"] == 1).astype(int)
    # Drop the original sex column after deriving the binary indicator
    cleaned = cleaned.drop(columns=["SEX"])
    # Rename the label column to the canonical y
    cleaned = cleaned.rename(columns={"DEFAULT_NEXT_MONTH": "y"})
    # Assemble the cleansing log for the audit trail
    cleansing_log: Dict[str, Any] = {
        "dataset": "Default",
        "input_rows": input_rows,
        "output_rows": int(len(cleaned)),
        "rows_dropped": 0,
        "dropped_columns": ["ID", "SEX"],
        "recoding_rules": {
            "EDUCATION": {0: 4, 5: 4, 6: 4},
            "MARRIAGE": {0: 3},
            "SEX": {1: 1, 2: 0},
        },
        "renamed_columns": {"y": "DEFAULT_NEXT_MONTH", "SEX_bin": "SEX"},
    }
    # Return the cleaned frame and its cleansing log
    return cleaned, cleansing_log


def cleanse_creditcard(
    raw_df: pd.DataFrame, cfg: Dict[str, Any]
) -> Tuple[pd.DataFrame, Dict[str, Any]]:
    """
    Cleanse creditcard_raw: rename the label to the canonical y only.

    Parameters
    ----------
    raw_df : pd.DataFrame
        Validated raw Credit Card Fraud DataFrame.
    cfg : Dict[str, Any]
        Validated study configuration.

    Returns
    -------
    Tuple[pd.DataFrame, Dict[str, Any]]
        Cleaned DataFrame (original index preserved) and cleansing log.
    """
    # Validate the raw frame type
    if not isinstance(raw_df, pd.DataFrame):
        raise TypeError(
            f"raw_df must be a pd.DataFrame; received {type(raw_df).__name__}."
        )
    # Record the input row count for the cleansing log
    input_rows: int = int(len(raw_df))
    # Copy the raw frame to avoid mutating the validated input
    cleaned: pd.DataFrame = raw_df.copy()
    # Rename the label column to the canonical y
    cleaned = cleaned.rename(columns={"Class": "y"})
    # Assemble the cleansing log for the audit trail
    cleansing_log: Dict[str, Any] = {
        "dataset": "Creditcard",
        "input_rows": input_rows,
        "output_rows": int(len(cleaned)),
        "rows_dropped": 0,
        "dropped_columns": [],
        "renamed_columns": {"y": "Class"},
    }
    # Return the cleaned frame and its cleansing log
    return cleaned, cleansing_log


# -----------------------------------------------------------------------------
# Task 2, Step 3: Persist the cleansed datasets
# -----------------------------------------------------------------------------

def persist_cleansed(
    cleansed_dfs: Dict[str, pd.DataFrame],
    cleansing_logs: Dict[str, Dict[str, Any]],
    output_dir: Path,
    version: str = "v1",
) -> Dict[str, Any]:
    """
    Persist cleansed datasets as versioned parquet with checksums.

    Writes each cleaned DataFrame to artifacts/cleansed/{version}/ with
    the original index stored as a data column, computes SHA-256 digests
    over the bytes on disk, and writes cleansing_report.json.

    Parameters
    ----------
    cleansed_dfs : Dict[str, pd.DataFrame]
        Mapping from dataset name to its cleaned DataFrame.
    cleansing_logs : Dict[str, Dict[str, Any]]
        Mapping from dataset name to its cleansing log.
    output_dir : Path
        Project output root; the versioned archive is created beneath it.
    version : str, optional
        Archive version segment, default "v1".

    Returns
    -------
    Dict[str, Any]
        Persistence report keyed by dataset with checksums and paths.

    Raises
    ------
    TypeError
        If any argument has an invalid type.
    ValueError
        If any dataset is missing from the cleansing logs.
    """
    # Validate the cleansed frames mapping type
    if not isinstance(cleansed_dfs, dict):
        raise TypeError(
            f"cleansed_dfs must be a dict; received {type(cleansed_dfs).__name__}."
        )
    # Validate the cleansing logs mapping type
    if not isinstance(cleansing_logs, dict):
        raise TypeError(
            f"cleansing_logs must be a dict; received "
            f"{type(cleansing_logs).__name__}."
        )
    # Validate the output directory type
    if not isinstance(output_dir, Path):
        raise TypeError(
            f"output_dir must be a pathlib.Path; "
            f"received {type(output_dir).__name__}."
        )
    # Construct the versioned archive directory
    archive_dir: Path = output_dir / "artifacts" / "cleansed" / version
    # Create the archive directory when it does not exist yet
    archive_dir.mkdir(parents=True, exist_ok=True)
    # Initialise the persistence report
    persistence_report: Dict[str, Any] = {
        "archive_dir": str(archive_dir),
        "written_at": datetime.now(timezone.utc).isoformat(),
        "files": {},
    }
    # Iterate over every cleansed dataset
    for dataset_name, frame in cleansed_dfs.items():
        # Raise when a dataset has no cleansing log entry
        if dataset_name not in cleansing_logs:
            raise ValueError(
                f"Dataset '{dataset_name}' has no cleansing log entry."
            )
        # Copy the frame so the archive never mutates the working copy
        frame_to_write: pd.DataFrame = frame.copy()
        # Store the original index as a data column to preserve alignment
        frame_to_write.insert(0, "original_index", frame_to_write.index)
        # Normalise nullable dtypes: float columns with NaN stay float64,
        # integer columns with NaN become nullable Float64 explicitly
        for column in frame_to_write.columns:
            # Skip the original index column and the label column
            if column in ("original_index", "y"):
                continue
            # Detect integer columns that contain missing values
            if (
                pd.api.types.is_integer_dtype(frame_to_write[column])
                and frame_to_write[column].isna().any()
            ):
                # Cast the column to nullable Float64 to avoid promotion
                frame_to_write[column] = frame_to_write[column].astype("Float64")
        # Construct the parquet path for this dataset
        parquet_path: Path = archive_dir / f"{dataset_name.lower()}.parquet"
        # Write the frame to parquet with the pyarrow engine
        frame_to_write.to_parquet(parquet_path, engine="pyarrow", index=False)
        # Compute the SHA-256 digest over the actual bytes on disk
        digest: str = hashlib.sha256(parquet_path.read_bytes()).hexdigest()
        # Record the checksum, path, and row count in the report
        persistence_report["files"][dataset_name] = {
            "path": str(parquet_path),
            "sha256": digest,
            "rows": int(len(frame_to_write)),
            "columns": list(frame_to_write.columns),
        }
        # Attach the row accounting to the cleansing log
        cleansing_logs[dataset_name]["output_rows"] = int(len(frame_to_write))
    # Assemble the complete cleansing report with all logs and checksums
    cleansing_report: Dict[str, Any] = {
        "written_at": datetime.now(timezone.utc).isoformat(),
        "archive_dir": str(archive_dir),
        "datasets": cleansing_logs,
        "checksums": {
            name: entry["sha256"]
            for name, entry in persistence_report["files"].items()
        },
    }
    # Write the cleansing report into the archive directory
    report_path: Path = archive_dir / "cleansing_report.json"
    # Serialise the report with stable formatting
    report_path.write_text(
        json.dumps(cleansing_report, indent=2, ensure_ascii=False) + "\n",
        encoding="utf-8",
    )
    # Record the report path in the persistence report
    persistence_report["cleansing_report"] = str(report_path)
    # Return the persistence report
    return persistence_report


# -----------------------------------------------------------------------------
# Task 2, Orchestrator Function
# -----------------------------------------------------------------------------

def cleanse_all(
    raw_dfs: Dict[str, pd.DataFrame],
    cfg: Dict[str, Any],
    output_dir: Path,
) -> Dict[str, Any]:
    """
    Orchestrate Task 2: cleanse all four datasets and persist them.

    Parameters
    ----------
    raw_dfs : Dict[str, pd.DataFrame]
        Mapping from dataset name to its validated raw DataFrame.
    cfg : Dict[str, Any]
        Validated study configuration.
    output_dir : Path
        Project output root for the versioned artifact archive.

    Returns
    -------
    Dict[str, Any]
        Mapping with "cleansed" (dict of cleaned DataFrames), "logs"
        (dict of cleansing logs), and "persistence" (persistence report).

    Raises
    ------
    TypeError
        If `raw_dfs` is not a dict.
    ValueError
        If a required dataset is missing from the raw frames mapping.
    """
    # Validate the raw frames mapping type
    if not isinstance(raw_dfs, dict):
        raise TypeError(
            f"raw_dfs must be a dict; received {type(raw_dfs).__name__}."
        )
    # Verify every required dataset is present in the mapping
    for dataset_name in ("Loan", "FraudEcom", "Default", "Creditcard"):
        # Raise when a required dataset is missing
        if dataset_name not in raw_dfs:
            raise ValueError(
                f"Required dataset '{dataset_name}' is missing from raw_dfs."
            )
    # Cleanse the Loan dataset and capture its cleansing log
    loan_cleaned, loan_log = cleanse_loan(raw_dfs["Loan"], cfg)
    # Cleanse the FraudEcom dataset and capture its cleansing log
    fraud_cleaned, fraud_log = cleanse_fraud_ecom(raw_dfs["FraudEcom"], cfg)
    # Cleanse the Default dataset and capture its cleansing log
    default_cleaned, default_log = cleanse_default(raw_dfs["Default"], cfg)
    # Cleanse the Creditcard dataset and capture its cleansing log
    creditcard_cleaned, creditcard_log = cleanse_creditcard(
        raw_dfs["Creditcard"], cfg
    )
    # Assemble the cleansed frames and logs into dictionaries
    cleansed_dfs: Dict[str, pd.DataFrame] = {
        "Loan": loan_cleaned,
        "FraudEcom": fraud_cleaned,
        "Default": default_cleaned,
        "Creditcard": creditcard_cleaned,
    }
    # Assemble the cleansing logs into a dictionary
    cleansing_logs: Dict[str, Dict[str, Any]] = {
        "Loan": loan_log,
        "FraudEcom": fraud_log,
        "Default": default_log,
        "Creditcard": creditcard_log,
    }
    # Persist every cleansed dataset with checksums and the report
    persistence_report: Dict[str, Any] = persist_cleansed(
        cleansed_dfs=cleansed_dfs,
        cleansing_logs=cleansing_logs,
        output_dir=output_dir,
    )
    # Return the cleansed frames, logs, and persistence report
    return {
        "cleansed": cleansed_dfs,
        "logs": cleansing_logs,
        "persistence": persistence_report,
    }
````

````python
# Task 3: Construct deterministic feature matrices

# ==============================================================================
# Task 3: Construct deterministic feature matrices
# ==============================================================================

# -----------------------------------------------------------------------------
# Canonical feature-name tuples. These are the deterministic, order-exact
# model matrices declared in the study configuration preamble; the builders
# validate the configuration against these tuples so a drift raises.
# -----------------------------------------------------------------------------
LOAN_FEATURE_NAMES: Tuple[str, ...] = (
    "loan_amnt", "term_months", "int_rate", "installment", "grade_ordinal",
    "emp_length_years", "home_ownership_RENT", "home_ownership_MORTGAGE",
    "home_ownership_OWN", "annual_inc", "verification_status_Source_Verified",
    "verification_status_Not_Verified", "mths_since_issue",
    "purpose_debt_consolidation", "purpose_credit_card",
    "purpose_home_improvement", "purpose_other", "dti", "delinq_2yrs",
    "inq_last_6mths", "open_acc", "pub_rec", "revol_bal", "revol_util",
    "total_acc", "out_prncp", "total_pymnt", "total_rec_prncp",
    "total_rec_int", "total_rec_late_fee", "recoveries",
    "collection_recovery_fee", "mths_since_pymnt", "last_pymnt_amnt",
    "next_pymnt_missing", "mths_since_last_credit_pull", "fico_midpoint",
    "mths_since_last_delinq", "mths_since_last_delinq_missing",
    "mths_since_last_record", "mths_since_last_record_missing",
    "last_fico_range_high",
)

# FraudEcom six-feature matrix in exact order
FRAUD_FEATURE_NAMES: Tuple[str, ...] = (
    "purchase_value", "age", "sex_bin", "ip_shared_count",
    "device_shared_count", "source_encoded",
)

# Default 23 raw predictors in exact order (before the three derived means)
DEFAULT_RAW_FEATURE_NAMES: Tuple[str, ...] = (
    "LIMIT_BAL", "SEX_bin", "EDUCATION", "MARRIAGE", "AGE",
    "PAY_0", "PAY_2", "PAY_3", "PAY_4", "PAY_5", "PAY_6",
    "BILL_AMT1", "BILL_AMT2", "BILL_AMT3", "BILL_AMT4", "BILL_AMT5",
    "BILL_AMT6",
    "PAY_AMT1", "PAY_AMT2", "PAY_AMT3", "PAY_AMT4", "PAY_AMT5", "PAY_AMT6",
)

# Default derived mean features appended after the raw predictors
DEFAULT_DERIVED_FEATURE_NAMES: Tuple[str, ...] = (
    "pay_status_mean", "bill_amt_mean", "pay_amt_mean",
)

# Full Default 26-feature tuple in exact order
DEFAULT_FEATURE_NAMES: Tuple[str, ...] = (
    DEFAULT_RAW_FEATURE_NAMES + DEFAULT_DERIVED_FEATURE_NAMES
)

# Creditcard 30-feature tuple: Time, V1..V28, Amount in exact order
CREDITCARD_FEATURE_NAMES: Tuple[str, ...] = (
    ("Time",) + tuple(f"V{i}" for i in range(1, 29)) + ("Amount",)
)

# Expected feature counts per dataset (paper verification targets)
EXPECTED_FEATURE_COUNTS: Dict[str, int] = {
    "Loan": 42, "FraudEcom": 6, "Default": 26, "Creditcard": 30,
}

# Loan columns whose NaN values are legitimate until Task 4 imputation
LOAN_IMPUTATION_COLUMNS: Tuple[str, ...] = (
    "revol_util", "mths_since_last_delinq", "mths_since_last_record",
    "fico_midpoint", "last_fico_range_high",
)

# FraudEcom source channel ordinal encoding (replicator choice)
SOURCE_ENCODING_MAP: Dict[str, int] = {"SEO": 0, "Ads": 1, "Direct": 2}


# -----------------------------------------------------------------------------
# Task 3, Step 1: Construct the 42-feature Loan matrix
# -----------------------------------------------------------------------------

def build_loan_features(
    cleansed_loan: pd.DataFrame, cfg: Dict[str, Any]
) -> Tuple[np.ndarray, np.ndarray, np.ndarray]:
    """
    Construct the deterministic 42-feature Loan model matrix.

    Parameters
    ----------
    cleansed_loan : pd.DataFrame
        Cleansed Loan DataFrame from Task 2.1 (original index preserved).
    cfg : Dict[str, Any]
        Validated study configuration whose feature-names contract is
        checked against the canonical tuple.

    Returns
    -------
    Tuple[np.ndarray, np.ndarray, np.ndarray]
        X of shape (n, 42) float64; y of shape (n,) int64; original_index
        of shape (n,) int64.

    Raises
    ------
    TypeError
        If `cleansed_loan` is not a pd.DataFrame or `cfg` is not a dict.
    ValueError
        If the configured feature names differ from the canonical tuple,
        a feature column is missing, or a non-imputation column has NaN.
    """
    # Validate the cleansed frame type
    if not isinstance(cleansed_loan, pd.DataFrame):
        raise TypeError(
            f"cleansed_loan must be a pd.DataFrame; "
            f"received {type(cleansed_loan).__name__}."
        )
    # Validate the configuration type
    if not isinstance(cfg, dict):
        raise TypeError(f"cfg must be a dict; received {type(cfg).__name__}.")
    # Read the configured feature names for the Loan matrix
    configured_names: Sequence[str] = cfg["feature_matrices"]["Loan"][
        "feature_names"
    ]
    # Convert the configured names to a tuple for comparison
    configured_tuple: Tuple[str, ...] = tuple(configured_names)
    # Raise when the configured names drift from the canonical tuple
    if configured_tuple != LOAN_FEATURE_NAMES:
        raise ValueError(
            "Configured Loan feature names disagree with the canonical "
            "42-feature tuple; a configuration drift would silently "
            "reorder the model matrix."
        )
    # Verify every feature column exists in the cleansed frame
    missing_columns: List[str] = [
        name for name in LOAN_FEATURE_NAMES if name not in cleansed_loan.columns
    ]
    # Raise when any feature column is missing
    if missing_columns:
        raise ValueError(
            f"Loan cleansed frame is missing feature columns: "
            f"{missing_columns}."
        )
    # Extract the label vector as int64
    y: np.ndarray = cleansed_loan["y"].to_numpy(dtype=np.int64)
    # Extract the original row indices as int64
    original_index: np.ndarray = cleansed_loan.index.to_numpy(dtype=np.int64)
    # Select the exact feature columns in the canonical order
    feature_frame: pd.DataFrame = cleansed_loan[list(LOAN_FEATURE_NAMES)]
    # Compute the set of non-imputation columns
    imputation_set: set = set(LOAN_IMPUTATION_COLUMNS)
    # Identify non-imputation columns that contain NaN values
    unexpected_nan: List[str] = [
        col
        for col in LOAN_FEATURE_NAMES
        if col not in imputation_set and feature_frame[col].isna().any()
    ]
    # Raise when any non-imputation column contains NaN
    if unexpected_nan:
        raise ValueError(
            f"Loan non-imputation columns contain NaN: {unexpected_nan}. "
            "Cleansing or parsing failed for these columns."
        )
    # Convert the feature frame to a float64 ndarray
    X: np.ndarray = feature_frame.to_numpy(dtype=np.float64)
    # Verify the matrix shape is exactly (n, 42)
    if X.shape[1] != 42:
        raise ValueError(
            f"Loan matrix has {X.shape[1]} columns; expected exactly 42."
        )
    # Verify the label and index vectors align with the matrix rows
    if not (X.shape[0] == y.shape[0] == original_index.shape[0]):
        raise ValueError(
            "Loan matrix, label, and index vectors have mismatched lengths: "
            f"{X.shape[0]}, {y.shape[0]}, {original_index.shape[0]}."
        )
    # Return the matrix, label vector, and original index vector
    return X, y, original_index


# -----------------------------------------------------------------------------
# Task 3, Step 2: Construct the 6-feature FraudEcom matrix and 26-feature Default matrix
# -----------------------------------------------------------------------------

def build_fraud_features(
    cleansed_fraud: pd.DataFrame,
    train_idx: np.ndarray,
    cfg: Dict[str, Any],
) -> Tuple[np.ndarray, np.ndarray, np.ndarray]:
    """
    Construct the 6-feature FraudEcom matrix with training-only counts.

    The frequency maps for ip_address and device_id are computed from the
    training partition exclusively; validation and test rows are mapped
    through the training-derived frequencies with unseen keys receiving
    count 1. This function MUST be called after the stratified split.

    Parameters
    ----------
    cleansed_fraud : pd.DataFrame
        Cleansed FraudEcom DataFrame from Task 2.2.
    train_idx : np.ndarray
        Positional indices of the training partition rows.
    cfg : Dict[str, Any]
        Validated study configuration.

    Returns
    -------
    Tuple[np.ndarray, np.ndarray, np.ndarray]
        X of shape (n, 6) float64; y of shape (n,) int64; original_index
        of shape (n,) int64.

    Raises
    ------
    TypeError
        If `cleansed_fraud` is not a pd.DataFrame or `train_idx` is not
        an ndarray.
    ValueError
        If train_idx is empty or out of bounds, or a feature column is
        missing.
    """
    # Validate the cleansed frame type
    if not isinstance(cleansed_fraud, pd.DataFrame):
        raise TypeError(
            f"cleansed_fraud must be a pd.DataFrame; "
            f"received {type(cleansed_fraud).__name__}."
        )
    # Validate the training index type
    if not isinstance(train_idx, np.ndarray):
        raise TypeError(
            f"train_idx must be a numpy ndarray; "
            f"received {type(train_idx).__name__}."
        )
    # Raise when the training index is empty
    if train_idx.shape[0] == 0:
        raise ValueError("train_idx must not be empty.")
    # Raise when the training index is out of bounds
    if int(train_idx.max()) >= len(cleansed_fraud) or int(train_idx.min()) < 0:
        raise ValueError(
            "train_idx contains indices outside the cleansed frame range."
        )
    # Slice the training partition by positional index
    train_partition: pd.DataFrame = cleansed_fraud.iloc[train_idx]
    # Compute the training-only frequency map for ip addresses
    ip_frequencies: pd.Series = train_partition["ip_address"].value_counts()
    # Compute the training-only frequency map for device ids
    device_frequencies: pd.Series = train_partition["device_id"].value_counts()
    # Map every row's ip address to its training frequency (unseen -> 1)
    ip_shared_count: pd.Series = (
        cleansed_fraud["ip_address"].map(ip_frequencies).fillna(1).astype(float)
    )
    # Map every row's device id to its training frequency (unseen -> 1)
    device_shared_count: pd.Series = (
        cleansed_fraud["device_id"].map(device_frequencies).fillna(1).astype(float)
    )
    # Encode the source channel with the fixed dictionary
    source_encoded: pd.Series = cleansed_fraud["source"].map(SOURCE_ENCODING_MAP)
    # Detect source values that are absent from the encoding dictionary
    unseen_mask: pd.Series = source_encoded.isna()
    # Fall back to the training mode's encoding for unseen source values
    if unseen_mask.any():
        # Compute the training mode of the source channel
        training_mode: Any = train_partition["source"].mode()
        # Resolve the training mode encoding when a mode exists
        mode_encoding: Optional[int] = (
            SOURCE_ENCODING_MAP.get(str(training_mode.iloc[0]))
            if training_mode.shape[0] > 0
            else None
        )
        # Fill unseen source values with the mode encoding when available
        if mode_encoding is not None:
            source_encoded = source_encoded.fillna(mode_encoding)
        else:
            # Fill with -1 and warn when no mode encoding can be resolved
            source_encoded = source_encoded.fillna(-1)
            # Warn about the degraded fallback
            warnings.warn(
                "Unseen source values mapped to -1 (no training mode "
                "encoding available).",
                RuntimeWarning,
            )
    # Convert the encoded source series to float
    source_encoded = source_encoded.astype(float)
    # Assemble the six-feature frame in the canonical order
    feature_frame: pd.DataFrame = pd.DataFrame(
        {
            "purchase_value": cleansed_fraud["purchase_value"],
            "age": cleansed_fraud["age"],
            "sex_bin": cleansed_fraud["sex_bin"],
            "ip_shared_count": ip_shared_count,
            "device_shared_count": device_shared_count,
            "source_encoded": source_encoded,
        }
    )
    # Verify the assembled frame has exactly the six canonical columns
    if list(feature_frame.columns) != list(FRAUD_FEATURE_NAMES):
        raise ValueError(
            "FraudEcom feature frame columns deviate from the canonical "
            "six-feature order."
        )
    # Extract the label vector as int64
    y: np.ndarray = cleansed_fraud["y"].to_numpy(dtype=np.int64)
    # Extract the original row indices as int64
    original_index: np.ndarray = cleansed_fraud.index.to_numpy(dtype=np.int64)
    # Convert the feature frame to a float64 ndarray
    X: np.ndarray = feature_frame.to_numpy(dtype=np.float64)
    # Raise when any NaN remains in the assembled matrix
    if not np.isfinite(X).all():
        raise ValueError("FraudEcom matrix contains NaN or infinite values.")
    # Verify the matrix shape is exactly (n, 6)
    if X.shape[1] != 6:
        raise ValueError(
            f"FraudEcom matrix has {X.shape[1]} columns; expected exactly 6."
        )
    # Verify the label and index vectors align with the matrix rows
    if not (X.shape[0] == y.shape[0] == original_index.shape[0]):
        raise ValueError(
            "FraudEcom matrix, label, and index vectors have mismatched "
            f"lengths: {X.shape[0]}, {y.shape[0]}, {original_index.shape[0]}."
        )
    # Return the matrix, label vector, and original index vector
    return X, y, original_index


def build_default_features(
    cleansed_default: pd.DataFrame, cfg: Dict[str, Any]
) -> Tuple[np.ndarray, np.ndarray, np.ndarray]:
    """
    Construct the 26-feature Default matrix with three derived means.

    The derived features are row-wise means over the exact six-month
    panel groups: pay_status_mean over PAY_0, PAY_2..PAY_6; bill_amt_mean
    over BILL_AMT1..BILL_AMT6; pay_amt_mean over PAY_AMT1..PAY_AMT6.

    Parameters
    ----------
    cleansed_default : pd.DataFrame
        Cleansed Default DataFrame from Task 2.2.
    cfg : Dict[str, Any]
        Validated study configuration.

    Returns
    -------
    Tuple[np.ndarray, np.ndarray, np.ndarray]
        X of shape (n, 26) float64; y of shape (n,) int64; original_index
        of shape (n,) int64.

    Raises
    ------
    TypeError
        If `cleansed_default` is not a pd.DataFrame.
    ValueError
        If a raw predictor column is missing or NaN remains.
    """
    # Validate the cleansed frame type
    if not isinstance(cleansed_default, pd.DataFrame):
        raise TypeError(
            f"cleansed_default must be a pd.DataFrame; "
            f"received {type(cleansed_default).__name__}."
        )
    # Verify every raw predictor column exists in the cleansed frame
    missing_columns: List[str] = [
        col
        for col in DEFAULT_RAW_FEATURE_NAMES
        if col not in cleansed_default.columns
    ]
    # Raise when any raw predictor column is missing
    if missing_columns:
        raise ValueError(
            f"Default cleansed frame is missing predictor columns: "
            f"{missing_columns}."
        )
    # Select the 23 raw predictor columns in the canonical order
    raw_frame: pd.DataFrame = cleansed_default[list(DEFAULT_RAW_FEATURE_NAMES)]
    # Compute the row-wise mean of the repayment-status panel
    pay_status_mean: pd.Series = raw_frame[
        ["PAY_0", "PAY_2", "PAY_3", "PAY_4", "PAY_5", "PAY_6"]
    ].mean(axis=1)
    # Compute the row-wise mean of the bill-amount panel
    bill_amt_mean: pd.Series = raw_frame[
        ["BILL_AMT1", "BILL_AMT2", "BILL_AMT3", "BILL_AMT4", "BILL_AMT5",
         "BILL_AMT6"]
    ].mean(axis=1)
    # Compute the row-wise mean of the payment-amount panel
    pay_amt_mean: pd.Series = raw_frame[
        ["PAY_AMT1", "PAY_AMT2", "PAY_AMT3", "PAY_AMT4", "PAY_AMT5",
         "PAY_AMT6"]
    ].mean(axis=1)
    # Append the three derived mean features to the raw frame
    feature_frame: pd.DataFrame = raw_frame.copy()
    # Assign the repayment-status mean as the 24th feature
    feature_frame["pay_status_mean"] = pay_status_mean
    # Assign the bill-amount mean as the 25th feature
    feature_frame["bill_amt_mean"] = bill_amt_mean
    # Assign the payment-amount mean as the 26th feature
    feature_frame["pay_amt_mean"] = pay_amt_mean
    # Verify the assembled frame has exactly the canonical 26 columns
    if list(feature_frame.columns) != list(DEFAULT_FEATURE_NAMES):
        raise ValueError(
            "Default feature frame columns deviate from the canonical "
            "26-feature order."
        )
    # Extract the label vector as int64
    y: np.ndarray = cleansed_default["y"].to_numpy(dtype=np.int64)
    # Extract the original row indices as int64
    original_index: np.ndarray = cleansed_default.index.to_numpy(dtype=np.int64)
    # Convert the feature frame to a float64 ndarray
    X: np.ndarray = feature_frame.to_numpy(dtype=np.float64)
    # Raise when any NaN or infinite value remains in the matrix
    if not np.isfinite(X).all():
        raise ValueError("Default matrix contains NaN or infinite values.")
    # Verify the matrix shape is exactly (n, 26)
    if X.shape[1] != 26:
        raise ValueError(
            f"Default matrix has {X.shape[1]} columns; expected exactly 26."
        )
    # Verify the label and index vectors align with the matrix rows
    if not (X.shape[0] == y.shape[0] == original_index.shape[0]):
        raise ValueError(
            "Default matrix, label, and index vectors have mismatched "
            f"lengths: {X.shape[0]}, {y.shape[0]}, {original_index.shape[0]}."
        )
    # Return the matrix, label vector, and original index vector
    return X, y, original_index


# -------------------------------------------------------------------------------------------------------------------------------
# Task 3, Step 3: Construct the 30-feature Creditcard matrix and verify all counts
# -------------------------------------------------------------------------------------------------------------------------------

def build_creditcard_features(
    cleansed_creditcard: pd.DataFrame,
    cfg: Dict[str, Any],
) -> Tuple[np.ndarray, np.ndarray, np.ndarray]:
    """
    Build the 30-feature Creditcard matrix in exact order Time V1..V28 Amount.

    Parameters
    ----------
    cleansed_creditcard : pd.DataFrame
        Cleansed Creditcard frame.
    cfg : Dict[str, Any]
        Validated configuration.
    Returns
    -------
    Tuple[np.ndarray, np.ndarray, np.ndarray]
        X, y, original_index.
    """
    # Validate cleansed DataFrame type is DataFrame
    if not isinstance(cleansed_creditcard, pd.DataFrame):
        # Raise TypeError for wrong type
        raise TypeError(f"cleansed_creditcard must be DataFrame, got {type(cleansed_creditcard).__name__}")
    # Resolve expected feature names tuple from cfg
    feature_names: Sequence[str] = tuple(cfg["feature_matrices"]["Creditcard"]["feature_names"])
    # Assert feature_names length is 30 per manuscript
    if len(feature_names) != 30:
        # Raise ValueError for wrong count
        raise ValueError(f"Creditcard feature_names must be 30, got {len(feature_names)}")
    # Find missing columns by list comprehension
    missing: List[str] = [c for c in feature_names if c not in cleansed_creditcard.columns]
    # Check if any missing columns
    if missing:
        # Raise KeyError with missing list
        raise KeyError(f"Creditcard missing columns {missing}")
    # Extract feature matrix as float64 via loc and to_numpy
    X: np.ndarray = cleansed_creditcard.loc[:, list(feature_names)].to_numpy(dtype=np.float64)
    # Extract label vector y as int64
    y: np.ndarray = cleansed_creditcard["y"].to_numpy(dtype=np.int64)
    # Extract original index or fallback to arange
    orig_idx: np.ndarray = cleansed_creditcard["original_index"].to_numpy() if "original_index" in cleansed_creditcard.columns else np.arange(X.shape[0])
    # Return X, y, orig_idx tuple
    return X, y, orig_idx

# Verify that every constructed matrix meets the Task 3.3 quality gate
def verify_counts_and_finite(
    feature_matrices: Dict[str, Dict[str, Any]],
    cfg: Dict[str, Any],
) -> None:
    """
    Verify feature counts, finiteness, and the NaN allowlist pre-Task 4.

    Implements the Task 3.3 quality gate: each dataset matrix must have
    the manuscript feature count (Loan 42, FraudEcom 6, Default 26,
    Creditcard 30), contain no infinite values, carry a binary label
    aligned to the row count, and contain NaN only in the declared
    imputation columns that Task 4 fills with training medians.

    Parameters
    ----------
    feature_matrices : Dict[str, Dict[str, Any]]
        Mapping from dataset name to its matrix record, each record
        containing ``X`` (np.ndarray of shape [n, m]), ``y`` (np.ndarray
        of shape [n]), ``feature_names``, ``original_index``, and
        ``is_placeholder``; ``FraudEcom`` may be ``None`` pre-Task 4.
    cfg : Dict[str, Any]
        Validated study configuration (used for feature-name lookups).

    Returns
    -------
    None
        Returns nothing; raises on the first violated invariant.

    Raises
    ------
    TypeError
        If a dataset's ``X`` is not a numpy ndarray.
    AssertionError
        If the column count, finiteness, label length, label
        binarity, or NaN-allowlist invariant is violated.
    """
    # Define expected counts per manuscript
    expected: Dict[str, int] = {"Loan": 42, "FraudEcom": 6, "Default": 26, "Creditcard": 30}
    # Define allowlist of Loan imputation columns that may be NaN pre-Task4
    allow_nan: set = {"revol_util", "mths_since_last_delinq", "mths_since_last_record", "fico_midpoint", "last_fico_range_high"}
    # Iterate over expected datasets and counts
    for dataset, exp_m in expected.items():
        # Get entry for dataset
        entry = feature_matrices.get(dataset)
        # Skip if entry is None (FraudEcom placeholder)
        if entry is None:
            # Continue to next dataset
            continue
        # Skip if entry is marked placeholder
        if isinstance(entry, dict) and entry.get("is_placeholder", False):
            # Continue
            continue
        # Extract X array
        X: np.ndarray = entry["X"]
        # Extract y array
        y: np.ndarray = entry["y"]
        # Validate X is ndarray
        if not isinstance(X, np.ndarray):
            # Raise TypeError
            raise TypeError(f"{dataset} X must be ndarray, got {type(X).__name__}")
        # Assert column count matches expected
        if X.shape[1] != exp_m:
            # Raise AssertionError with mismatch
            raise AssertionError(f"{dataset} expected {exp_m} columns, got {X.shape[1]}")
        # Assert no inf in X
        if np.isinf(X).any():
            # Raise AssertionError for inf
            raise AssertionError(f"{dataset} X contains inf")
        # Assert label length matches rows
        if y.shape[0] != X.shape[0]:
            # Raise AssertionError
            raise AssertionError(f"{dataset} y length {y.shape[0]} != X rows {X.shape[0]}")
        # Assert label binary 0/1
        if not set(np.unique(y)).issubset({0, 1}):
            # Raise AssertionError
            raise AssertionError(f"{dataset} y not binary {set(np.unique(y))}")
        # Check if any NaN present
        if np.isnan(X).any():
            # Find column indices with NaN
            nan_idx: np.ndarray = np.where(np.isnan(X).any(axis=0))[0]
            # Get feature names list
            f_names: List[str] = list(entry.get("feature_names", []))
            # Iterate over nan column indices
            for j in nan_idx:
                # Resolve column name
                col: str = f_names[j] if j < len(f_names) else f"col_{j}"
                # Check if column in allowlist
                if col not in allow_nan:
                    # Raise AssertionError for unexpected NaN
                    raise AssertionError(f"{dataset} unexpected NaN in column {col} pre-Task4")

# Verify that the scaled matrices are clean after Task 4 imputation/scaling
def verify_no_nan_post_imputation(
    scaled_matrices: Dict[str, Dict[str, Any]],
) -> None:
    """
    Verify the post-Task-4 scaled matrices contain no NaN or infinity.

    Implements the final integrity gate of the preprocessing chain:
    after imputation with training medians and min-max scaling to
    [0.01, 10.01], every scaled matrix must be finite with no missing
    values, otherwise downstream training and evaluation are invalid.

    Parameters
    ----------
    scaled_matrices : Dict[str, Dict[str, Any]]
        Mapping from dataset name to its scaled matrix record containing
        ``X`` (np.ndarray of shape [n, m]) and ``y``; entries that are
        ``None`` or placeholder-flagged are skipped.

    Returns
    -------
    None
        Returns nothing; raises on the first violated invariant.

    Raises
    ------
    AssertionError
        If any scaled ``X`` contains NaN, infinity, or non-finite values.
    """
    # Iterate over scaled matrices entries
    for dataset, entry in scaled_matrices.items():
        # Skip None or placeholder entries
        if entry is None or (isinstance(entry, dict) and entry.get("is_placeholder", False)):
            # Continue
            continue
        # Extract scaled X
        X: np.ndarray = entry["X"]
        # Assert no NaN remains
        if np.isnan(X).any():
            # Raise AssertionError
            raise AssertionError(f"{dataset} scaled X still contains NaN post-Task4")
        # Assert no inf remains
        if np.isinf(X).any():
            # Raise AssertionError
            raise AssertionError(f"{dataset} scaled X contains inf post-Task4")
        # Assert finite
        if not np.isfinite(X).all():
            # Raise AssertionError
            raise AssertionError(f"{dataset} scaled X not finite post-Task4")

# Provide the backward-compatible public entry point for the Task 3.3 gate
def verify_feature_counts(
    feature_matrices: Dict[str, Dict[str, Any]],
    cfg: Dict[str, Any],
) -> None:
    """
    Backward-compatible wrapper delegating to the pre-Task-4 gate.

    Maintains the public Task 3.3 entry point while forwarding all
    validation to :func:`verify_counts_and_finite`, so callers receive
    identical behavior through either name.

    Parameters
    ----------
    feature_matrices : Dict[str, Dict[str, Any]]
        Mapping from dataset name to its matrix record as accepted by
        :func:`verify_counts_and_finite`.
    cfg : Dict[str, Any]
        Validated study configuration.

    Returns
    -------
    None
        Returns nothing; propagates any validation failure raised by
        the delegated gate.

    Raises
    ------
    TypeError
        Propagated from the delegated gate if ``X`` is not an ndarray.
    AssertionError
        Propagated from the delegated gate on any violated invariant.
    """
    # Delegate to verify_counts_and_finite
    verify_counts_and_finite(feature_matrices, cfg)

# Persist all constructed matrices as versioned, atomic compressed artifacts
def persist_feature_matrices(
    feature_matrices: Dict[str, Dict[str, Any]],
    cfg: Dict[str, Any],
    output_dir: Path,
) -> Dict[str, Path]:
    """
    Persist the constructed feature matrices as compressed artifacts.

    Writes each dataset matrix to a versioned ``.npz`` file under
    ``output_dir/artifacts/feature_matrices/v1`` with a JSON sidecar
    recording feature names, placeholder status, and shape, using an
    atomic rename to prevent partial writes. FraudEcom, whose count
    features are built per-seed from training rows in Task 11, is
    stored as a seed-0 placeholder with an explanatory metadata sidecar.

    Parameters
    ----------
    feature_matrices : Dict[str, Dict[str, Any]]
        Mapping from dataset name to its matrix record, each containing
        ``X`` (np.ndarray of shape [n, m]), ``y`` (np.ndarray of shape
        [n]), ``feature_names``, ``original_index``, and optionally
        ``is_placeholder``; ``FraudEcom`` may be ``None``.
    cfg : Dict[str, Any]
        Validated study configuration (feature-name source).
    output_dir : Path
        Project output root; the artifact sub-tree is created below it.

    Returns
    -------
    Dict[str, Path]
        Mapping from dataset name to the final persisted ``.npz`` path.

    Raises
    ------
    TypeError
        If ``output_dir`` is not a ``pathlib.Path``.
    """
    # Validate output_dir is Path
    if not isinstance(output_dir, Path):
        # Raise TypeError
        raise TypeError(f"output_dir must be Path, got {type(output_dir).__name__}")
    # Build persist directory Path
    persist_dir: Path = output_dir / "artifacts" / "feature_matrices" / "v1"
    # Create directory parents
    persist_dir.mkdir(parents=True, exist_ok=True)
    # Initialize paths dict
    paths: Dict[str, Path] = {}
    # Iterate over datasets in canonical order
    for dataset in ["Loan", "FraudEcom", "Default", "Creditcard"]:
        # Get entry
        entry = feature_matrices.get(dataset)
        # Initialize placeholder flag false
        is_placeholder: bool = False
        # Handle None entry
        if entry is None:
            # Set placeholder true for FraudEcom only
            is_placeholder = True if dataset == "FraudEcom" else False
            # If FraudEcom placeholder, write metadata and continue
            if dataset == "FraudEcom":
                # Build metadata dict
                meta: Dict[str, Any] = {"is_placeholder": True, "authoritative_per_seed_build_in_Task11": True, "note": "FraudEcom counts per-seed training-only; seed0 placeholder"}
                # Write json sidecar
                (persist_dir / "fraud_placeholder_meta.json").write_text(json.dumps(meta, indent=2) + "\n", encoding="utf-8")
                # Record placeholder path
                paths[dataset] = persist_dir / "fraud_seed0_placeholder.npz"
                # Continue to next dataset
                continue
        # Check explicit is_placeholder flag in entry
        if isinstance(entry, dict) and entry.get("is_placeholder", False):
            # Set flag true
            is_placeholder = True
        # Determine filename based on placeholder
        fname: str = "fraud_seed0_placeholder.npz" if dataset == "FraudEcom" and is_placeholder else f"{dataset.lower()}.npz"
        # Build output path
        out_path: Path = persist_dir / fname
        # Extract X and y
        X: np.ndarray = entry["X"]
        y: np.ndarray = entry["y"]
        # Get feature_names
        feature_names = entry.get("feature_names", [])
        # Get original index
        orig_idx = entry.get("original_index", np.arange(X.shape[0]))
        # Build temp path for atomic write keeping .npz extension
        tmp_path: Path = out_path.parent / (out_path.stem + ".tmp.npz")
        # Save compressed npz with float64 X and int64 y
        np.savez_compressed(tmp_path, X=X.astype(np.float64), y=y.astype(np.int64), original_index=orig_idx)
        # Build sidecar metadata
        sidecar: Dict[str, Any] = {"feature_names": list(feature_names), "is_placeholder": is_placeholder, "shape": list(X.shape)}
        # Write sidecar json
        (out_path.with_suffix(".json")).write_text(json.dumps(sidecar, indent=2) + "\n", encoding="utf-8")
        # Atomic replace temp to final
        tmp_path.replace(out_path)
        # Record path
        paths[dataset] = out_path
    # Return paths mapping
    return paths

# Build the static-dataset matrices (superseded draft, shadowed below)
def build_feature_matrices(
    cleansed_dfs: Dict[str, pd.DataFrame],
    cfg: Dict[str, Any],
    output_dir: Path,
) -> Dict[str, Dict[str, Any]]:
    """
    Build, verify, and persist the Loan, Default, and Creditcard matrices.

    NOTE: This is a superseded draft that builds only the three static
    datasets and stores ``FraudEcom`` as ``None``; it is shadowed by the
    definitive four-dataset orchestrator defined below, which accepts a
    per-seed ``fraud_train_idx``. Retained only for reference.

    Parameters
    ----------
    cleansed_dfs : Dict[str, pd.DataFrame]
        Mapping from dataset name to its cleansed DataFrame; must
        contain at least ``Loan``, ``Default``, and ``Creditcard``.
    cfg : Dict[str, Any]
        Validated study configuration.
    output_dir : Path
        Project output root for the artifact archive.

    Returns
    -------
    Dict[str, Dict[str, Any]]
        Mapping from dataset name to its matrix record (``FraudEcom``
        is ``None``), with verification and persistence already run.

    Raises
    ------
    KeyError
        If a required cleansed dataset key is missing.
    """
    # Validate required cleansed keys present
    for k in ["Loan", "Default", "Creditcard"]:
        # Check key in dict
        if k not in cleansed_dfs:
            # Raise KeyError
            raise KeyError(f"cleansed_dfs missing {k}")
    # Build Loan matrix via helper
    loan_X, loan_y, loan_idx = build_loan_features(cleansed_dfs["Loan"], cfg)
    # Build Default matrix
    default_X, default_y, default_idx = build_default_features(cleansed_dfs["Default"], cfg)
    # Build Creditcard matrix
    cc_X, cc_y, cc_idx = build_creditcard_features(cleansed_dfs["Creditcard"], cfg)
    # Assemble feature_matrices with FraudEcom None placeholder
    feature_matrices: Dict[str, Dict[str, Any]] = {
        "Loan": {"X": loan_X, "y": loan_y, "feature_names": tuple(cfg["feature_matrices"]["Loan"]["feature_names"]), "original_index": loan_idx, "is_placeholder": False},
        "FraudEcom": None,
        "Default": {"X": default_X, "y": default_y, "feature_names": tuple(cfg["feature_matrices"]["Default"]["feature_names"]), "original_index": default_idx, "is_placeholder": False},
        "Creditcard": {"X": cc_X, "y": cc_y, "feature_names": tuple(cfg["feature_matrices"]["Creditcard"]["feature_names"]), "original_index": cc_idx, "is_placeholder": False},
    }
    # Verify counts and finite pre-Task4
    verify_counts_and_finite(feature_matrices, cfg)
    # Persist with placeholder handling
    persist_feature_matrices(feature_matrices, cfg, output_dir)
    # Return matrices dict
    return feature_matrices
# -----------------------------------------------------------------------------
# Task 3, Orchestrator Function
# -----------------------------------------------------------------------------
def build_feature_matrices(
    cleansed_dfs: Dict[str, pd.DataFrame],
    cfg: Dict[str, Any],
    fraud_train_idx: np.ndarray,
    output_dir: Path,
) -> Dict[str, Any]:
    """
    Orchestrate Task 3: build, verify, and persist all feature matrices.

    Parameters
    ----------
    cleansed_dfs : Dict[str, pd.DataFrame]
        Mapping from dataset name to its cleansed DataFrame.
    cfg : Dict[str, Any]
        Validated study configuration.
    fraud_train_idx : np.ndarray
        Positional training indices for FraudEcom count construction.
    output_dir : Path
        Project output root for the artifact archive.

    Returns
    -------
    Dict[str, Any]
        Mapping with "matrices", "verification", and "persistence" keys.

    Raises
    ------
    TypeError
        If `cleansed_dfs` is not a dict.
    ValueError
        If a required dataset is missing.
    """
    # Validate the cleansed frames mapping type
    if not isinstance(cleansed_dfs, dict):
        raise TypeError(
            f"cleansed_dfs must be a dict; received {type(cleansed_dfs).__name__}."
        )
    # Verify every required dataset is present in the mapping
    for dataset_name in ("Loan", "FraudEcom", "Default", "Creditcard"):
        # Raise when a required dataset is missing
        if dataset_name not in cleansed_dfs:
            raise ValueError(
                f"Required dataset '{dataset_name}' is missing from "
                "cleansed_dfs."
            )
    # Build the Loan matrix from its cleansed frame
    loan_X, loan_y, loan_index = build_loan_features(
        cleansed_dfs["Loan"], cfg
    )
    # Build the FraudEcom matrix with training-only frequency counts
    fraud_X, fraud_y, fraud_index = build_fraud_features(
        cleansed_dfs["FraudEcom"], fraud_train_idx, cfg
    )
    # Build the Default matrix from its cleansed frame
    default_X, default_y, default_index = build_default_features(
        cleansed_dfs["Default"], cfg
    )
    # Build the Creditcard matrix from its cleansed frame
    creditcard_X, creditcard_y, creditcard_index = build_creditcard_features(
        cleansed_dfs["Creditcard"], cfg
    )
    # Assemble the matrix dictionary keyed by dataset name
    feature_matrices: Dict[str, Dict[str, np.ndarray]] = {
        "Loan": {"X": loan_X, "y": loan_y, "original_index": loan_index},
        "FraudEcom": {"X": fraud_X, "y": fraud_y, "original_index": fraud_index},
        "Default": {"X": default_X, "y": default_y, "original_index": default_index},
        "Creditcard": {
            "X": creditcard_X,
            "y": creditcard_y,
            "original_index": creditcard_index,
        },
    }
    # Run the mandatory verification gate before any persistence
    verification_report: Dict[str, Any] = verify_feature_counts(
        feature_matrices, cfg
    )
    # Persist the verified matrices to the artifact archive
    persistence_report: Dict[str, Any] = persist_feature_matrices(
        feature_matrices, output_dir
    )
    # Return the matrices, verification, and persistence reports
    return {
        "matrices": feature_matrices,
        "verification": verification_report,
        "persistence": persistence_report,
    }

````

````python
# Task 4: Stratified splitting and train-only min-max scaling

# ==============================================================================
# Task 4: Stratified splitting and train-only min-max scaling
# ==============================================================================

# Derived log-bound constant computed from the scaling bounds:
# L = max(|log 0.01|, |log 10.01|) = 4.605170185988091
LOG_BOUND_L: float = max(
    abs(np.log(0.01)), abs(np.log(10.01))
)

# Scaling bounds of the min-max affine transform
SCALE_LOWER: float = 0.01
SCALE_UPPER: float = 10.01

# Midpoint of the scaled range used for constant features (max == min)
CONSTANT_FEATURE_FILL: float = (SCALE_LOWER + SCALE_UPPER) / 2.0


# -----------------------------------------------------------------------------
# Task 4, Step 1: Create the five stratified 60/20/20 splits
# -----------------------------------------------------------------------------

def make_stratified_splits(
    X: np.ndarray,
    y: np.ndarray,
    seed: int,
    cfg: Dict[str, Any],
) -> Tuple[np.ndarray, np.ndarray, np.ndarray]:
    """
    Create the stratified 60/20/20 train/validation/test split for a seed.

    The split is performed by two StratifiedShuffleSplit calls with the
    same random state: first 60/40, then the 40% temporary set is split
    50/50 into validation and test, giving exactly 60/20/20.

    Parameters
    ----------
    X : np.ndarray, shape (n, m)
        Feature matrix (only its row count is used here).
    y : np.ndarray, shape (n,)
        Binary label vector used for stratification.
    seed : int
        Deterministic random state for both split calls.
    cfg : Dict[str, Any]
        Validated study configuration.

    Returns
    -------
    Tuple[np.ndarray, np.ndarray, np.ndarray]
        train_idx, val_idx, test_idx: disjoint positional index arrays.

    Raises
    ------
    TypeError
        If `X` or `y` is not an ndarray.
    ValueError
        If the seed is not in the configured seed list, the arrays
        disagree in length, a partition has no positive or no negative
        example, or a partition rate deviates beyond 0.001.
    """
    # Validate the feature matrix type
    if not isinstance(X, np.ndarray):
        raise TypeError(f"X must be an ndarray; received {type(X).__name__}.")
    # Validate the label vector type
    if not isinstance(y, np.ndarray):
        raise TypeError(f"y must be an ndarray; received {type(y).__name__}.")
    # Validate the seed type
    if not isinstance(seed, int):
        raise TypeError(f"seed must be an int; received {type(seed).__name__}.")
    # Validate that the seed is one of the configured seeds
    if seed not in tuple(cfg["data_splitting"]["seeds"]):
        raise ValueError(
            f"seed {seed} is not in the configured seed list "
            f"{list(cfg['data_splitting']['seeds'])}."
        )
    # Raise when the matrix and label lengths disagree
    if X.shape[0] != y.shape[0]:
        raise ValueError(
            f"X has {X.shape[0]} rows but y has {y.shape[0]} entries."
        )
    # Compute the full-dataset positive rate as the reference rate
    full_positive_rate: float = float(y.mean())
    # Stage one: split the full data into 60% train and 40% temporary
    first_splitter: StratifiedShuffleSplit = StratifiedShuffleSplit(
        n_splits=1, test_size=0.4, random_state=seed
    )
    # Extract the train and temporary index arrays from the first split
    train_idx, temp_idx = next(first_splitter.split(X, y))
    # Stage two: split the temporary 40% into equal validation and test
    second_splitter: StratifiedShuffleSplit = StratifiedShuffleSplit(
        n_splits=1, test_size=0.5, random_state=seed
    )
    # Slice the temporary labels for the second stratification
    temp_y: np.ndarray = y[temp_idx]
    # Extract the validation and test index arrays from the second split
    val_idx, test_idx = next(second_splitter.split(np.zeros_like(temp_y), temp_y))
    # Map the validation positions back to the full-data index space
    val_idx = temp_idx[val_idx]
    # Map the test positions back to the full-data index space
    test_idx = temp_idx[test_idx]
    # Assert the three partitions are pairwise disjoint
    if not (
        set(train_idx).isdisjoint(val_idx)
        and set(train_idx).isdisjoint(test_idx)
        and set(val_idx).isdisjoint(test_idx)
    ):
        raise ValueError("Split partitions are not pairwise disjoint.")
    # Assert the three partitions cover the full index range
    union_size: int = len(set(train_idx) | set(val_idx) | set(test_idx))
    # Raise when the union does not cover every row
    if union_size != X.shape[0]:
        raise ValueError(
            f"Split partitions cover {union_size} rows but X has "
            f"{X.shape[0]} rows."
        )
    # Compute the per-partition positive rates for parity verification
    train_rate: float = float(y[train_idx].mean())
    val_rate: float = float(y[val_idx].mean())
    test_rate: float = float(y[test_idx].mean())
    # Assemble the partition positive rates for the checks
    partition_rates: Dict[str, float] = {
        "train": train_rate,
        "val": val_rate,
        "test": test_rate,
    }
    # Verify each partition rate is within 0.001 of the full-dataset rate
    for partition_name, rate in partition_rates.items():
        # Raise when the partition rate deviates beyond tolerance
        if abs(rate - full_positive_rate) > 0.001:
            raise ValueError(
                f"{partition_name} positive rate {rate:.4f} deviates from "
                f"the full-dataset rate {full_positive_rate:.4f} by more "
                "than 0.001."
            )
    # Verify every partition has at least one positive example
    for partition_name, partition_idx in {
        "train": train_idx, "val": val_idx, "test": test_idx,
    }.items():
        # Raise when a partition has no positive example
        if int(y[partition_idx].sum()) == 0:
            raise ValueError(
                f"{partition_name} partition contains no positive example; "
                "PR-AUC and calibration would be undefined."
            )
        # Raise when a partition has no negative example
        if int((1 - y[partition_idx]).sum()) == 0:
            raise ValueError(
                f"{partition_name} partition contains no negative example."
            )
    # Return the three disjoint positional index arrays
    return train_idx, val_idx, test_idx


# -----------------------------------------------------------------------------
# Task 4, Step 2: Fit missing-value imputations and scaling constants on training only
# -----------------------------------------------------------------------------
@dataclass
class FittedPreprocessing:
    """
    Immutable container of training-only preprocessing statistics.

    Stores the per-feature medians, minima, maxima, modes, and constant-feature
    policy fitted exclusively on the training partition, together with the
    feature ordering needed to apply the affine min-max transform
    ``tilde_x = 0.01 + 10.0*(x - min)/(max - min)`` to validation and test.
    This is the leakage-safe artifact referenced in Task 4 Step 4.2 and reused
    by ``apply_preprocessing``.

    Attributes
    ----------
    medians : np.ndarray, shape (m,)
        Per-feature training medians; ``np.nan`` where no median was needed
        because the column was complete pre-Task 4. Used to impute
        validation/test missing values.
    mins : np.ndarray, shape (m,)
        Per-feature training minima; defines the ``min_j`` in the scaling
        formula.
    maxs : np.ndarray, shape (m,)
        Per-feature training maxima; defines the ``max_j`` in the scaling
        formula.
    modes : Dict[int, float]
        Training modes for categorical-like numeric columns, keyed by column
        position; empty when no mode was required.
    constant_feature_policy : Dict[int, str]
        Column positions whose ``max == min``, mapped to their resolution
        ``"fill_midpoint_5.01"``; the scaled value ``5.01`` is the midpoint
        of ``[0.01, 10.01]``.
    feature_names : Tuple[str, ...]
        Feature names aligned positionally with the statistics; defines the
        column order for ``mins``, ``maxs``, and ``medians``.
    """

    # Per-feature training medians (NaN where not needed)
    medians: np.ndarray
    # Per-feature training minima
    mins: np.ndarray
    # Per-feature training maxima
    maxs: np.ndarray
    # Training modes for categorical-like numeric columns
    modes: Dict[int, float] = field(default_factory=dict)
    # Column positions with max == min mapped to their resolution policy
    constant_feature_policy: Dict[int, str] = field(default_factory=dict)
    # Feature names aligned positionally with the statistics
    feature_names: Tuple[str, ...] = field(default_factory=tuple)


# -----------------------------------------------------------------------------
def fit_preprocessing(
    X_train: np.ndarray,
    cfg: Dict[str, Any],
    dataset_name: str,
    feature_names: Sequence[str],
) -> FittedPreprocessing:
    """
    Fit imputation medians and scaling constants on training rows only.

    Parameters
    ----------
    X_train : np.ndarray, shape (n_train, m)
        Training partition feature matrix; no validation or test row may
        appear in this array.
    cfg : Dict[str, Any]
        Validated study configuration.
    dataset_name : str
        Name of the dataset (determines the allowed imputation columns).
    feature_names : Sequence[str]
        Feature names aligned positionally with the matrix columns.

    Returns
    -------
    FittedPreprocessing
        Training-only statistics: medians, minima, maxima, modes, and
        the constant-feature policy.

    Raises
    ------
    TypeError
        If `X_train` is not an ndarray.
    ValueError
        If NaN appears outside the declared imputation columns.
    """
    # Validate the training matrix type
    if not isinstance(X_train, np.ndarray):
        raise TypeError(
            f"X_train must be an ndarray; received {type(X_train).__name__}."
        )
    # Validate the feature names sequence
    if len(feature_names) != X_train.shape[1]:
        raise ValueError(
            f"feature_names has {len(feature_names)} entries but X_train "
            f"has {X_train.shape[1]} columns."
        )
    # Resolve the declared imputation column positions for the dataset
    imputation_columns: Sequence[str] = cfg["missing_value_policy"][
        dataset_name
    ]["imputation_columns"]
    # Map the imputation column names to positional indices
    imputation_positions: List[int] = [
        list(feature_names).index(name) for name in imputation_columns
    ]
    # Detect the columns that still contain NaN in the training partition
    nan_columns: List[int] = [
        int(j) for j in range(X_train.shape[1]) if np.isnan(X_train[:, j]).any()
    ]
    # Raise when NaN appears outside the declared imputation columns
    unexpected_nan: List[int] = [
        j for j in nan_columns if j not in imputation_positions
    ]
    # Raise with the offending column positions named
    if unexpected_nan:
        raise ValueError(
            f"NaN found in non-imputation columns of {dataset_name} "
            f"training data at positions {unexpected_nan} (names "
            f"{[list(feature_names)[j] for j in unexpected_nan]}); "
            "complete datasets must not be imputed."
        )
    # Compute the per-feature training medians (NaN where not needed)
    medians: np.ndarray = np.full(X_train.shape[1], np.nan)
    # Compute the median for every column that still contains NaN
    for j in nan_columns:
        # Compute the training median of the column, ignoring NaN
        medians[j] = float(np.nanmedian(X_train[:, j]))
    # Compute the per-feature training minima
    mins: np.ndarray = np.nanmin(X_train, axis=0).astype(np.float64)
    # Compute the per-feature training maxima
    maxs: np.ndarray = np.nanmax(X_train, axis=0).astype(np.float64)
    # Detect the constant-feature columns (max equals min)
    constant_columns: List[int] = [
        int(j)
        for j in range(X_train.shape[1])
        if maxs[j] == mins[j]
    ]
    # Record the constant-feature policy for each such column
    constant_policy: Dict[int, str] = {
        j: "scaled to midpoint 5.01 (max == min)"
        for j in constant_columns
    }
    # Warn when constant features are present
    if constant_columns:
        warnings.warn(
            f"Constant features detected in {dataset_name} training data "
            f"at positions {constant_columns}; they will be scaled to "
            "the midpoint 5.01.",
            RuntimeWarning,
        )
    # Compute the training mode of every categorical-like numeric column
    modes: Dict[int, float] = {}
    # Iterate over the numeric columns to compute modes
    for j in range(X_train.shape[1]):
        # Skip columns with NaN values (medians already cover them)
        if np.isnan(X_train[:, j]).any():
            continue
        # Compute the unique values and their counts
        unique_values, counts = np.unique(X_train[:, j], return_counts=True)
        # Record the modal value when one exists
        if unique_values.shape[0] > 0:
            modes[j] = float(unique_values[int(np.argmax(counts))])
    # Assemble the fitted preprocessing dataclass
    fitted: FittedPreprocessing = FittedPreprocessing(
        medians=medians,
        mins=mins,
        maxs=maxs,
        modes=modes,
        constant_feature_policy=constant_policy,
        feature_names=tuple(feature_names),
    )
    # Return the fitted preprocessing statistics
    return fitted


# -----------------------------------------------------------------------------
# Task 4, Step 3: Apply preprocessing to all splits and verify ranges
# -----------------------------------------------------------------------------

def apply_preprocessing(
    X_train: np.ndarray,
    X_val: np.ndarray,
    X_test: np.ndarray,
    fitted: FittedPreprocessing,
) -> Tuple[np.ndarray, np.ndarray, np.ndarray, Dict[str, int]]:
    """
    Apply training-fitted preprocessing to all three partitions.

    The transform is x̃_j = 0.01 + 10.0 * (x_j - min_j) / (max_j - min_j),
    applied after imputing validation and test NaN values with training
    medians, followed by clipping to [0.01, 10.01] with clip counts.

    Parameters
    ----------
    X_train : np.ndarray, shape (n_train, m)
        Training partition (may contain NaN in imputation columns).
    X_val : np.ndarray, shape (n_val, m)
        Validation partition.
    X_test : np.ndarray, shape (n_test, m)
        Test partition.
    fitted : FittedPreprocessing
        Training-only statistics produced by fit_preprocessing.

    Returns
    -------
    Tuple[np.ndarray, np.ndarray, np.ndarray, Dict[str, int]]
        Scaled training, validation, and test matrices plus a dictionary
        of per-split clipped-value counts.

    Raises
    ------
    TypeError
        If any matrix is not an ndarray or `fitted` is not
        FittedPreprocessing.
    ValueError
        If the matrices disagree in column count or the scaled values
        fall outside [0.01, 10.01] after clipping.
    """
    # Validate the matrix types
    for name, matrix in {
        "X_train": X_train, "X_val": X_val, "X_test": X_test,
    }.items():
        # Raise when any matrix is not an ndarray
        if not isinstance(matrix, np.ndarray):
            raise TypeError(
                f"{name} must be an ndarray; received {type(matrix).__name__}."
            )
    # Validate the fitted statistics type
    if not isinstance(fitted, FittedPreprocessing):
        raise TypeError(
            f"fitted must be a FittedPreprocessing; "
            f"received {type(fitted).__name__}."
        )
    # Verify the column counts agree across all matrices and the fit
    if not (
        X_train.shape[1] == X_val.shape[1] == X_test.shape[1]
        == fitted.mins.shape[0]
    ):
        raise ValueError(
            "Column counts disagree across matrices and fitted statistics."
        )
    # Copy the matrices so the inputs are never mutated
    train_imputed: np.ndarray = X_train.copy()
    val_imputed: np.ndarray = X_val.copy()
    test_imputed: np.ndarray = X_test.copy()
    # Impute validation NaN values with the training medians
    for j in range(X_val.shape[1]):
        # Skip columns without a fitted median
        if np.isnan(fitted.medians[j]):
            continue
        # Locate the NaN positions in the validation column
        val_nan_mask: np.ndarray = np.isnan(val_imputed[:, j])
        # Fill the NaN positions with the training median
        val_imputed[val_nan_mask, j] = fitted.medians[j]
        # Locate the NaN positions in the test column
        test_nan_mask: np.ndarray = np.isnan(test_imputed[:, j])
        # Fill the NaN positions with the training median
        test_imputed[test_nan_mask, j] = fitted.medians[j]
        # Locate the NaN positions in the training column
        train_nan_mask: np.ndarray = np.isnan(train_imputed[:, j])
        # Fill the NaN positions with the training median
        train_imputed[train_nan_mask, j] = fitted.medians[j]
    # Apply the affine min-max transform with explicit division guarding
    with np.errstate(divide="raise", invalid="raise"):
        # Compute the affine denominator for every column
        denominators: np.ndarray = fitted.maxs - fitted.mins
        # Identify the constant-feature columns whose denominator is zero
        zero_denominator_mask: np.ndarray = denominators == 0.0
        # Substitute a unit denominator for the zero entries so the
        # guarded division never fires on constant columns (their scaled
        # values are overwritten by the midpoint policy below)
        safe_denominators: np.ndarray = np.where(
            zero_denominator_mask, 1.0, denominators
        )
        # Compute the affine numerator for every column
        scaled_train: np.ndarray = (
            SCALE_LOWER
            + SCALE_UPPER * (train_imputed - fitted.mins) / safe_denominators
        )
        # Compute the affine transform for the validation partition
        scaled_val: np.ndarray = (
            SCALE_LOWER
            + SCALE_UPPER * (val_imputed - fitted.mins) / safe_denominators
        )
        # Compute the affine transform for the test partition
        scaled_test: np.ndarray = (
            SCALE_LOWER
            + SCALE_UPPER * (test_imputed - fitted.mins) / safe_denominators
        )
    # Override the constant-feature columns with the midpoint fill
    for j, policy in fitted.constant_feature_policy.items():
        # Replace the entire column with the midpoint in training
        scaled_train[:, j] = CONSTANT_FEATURE_FILL
        # Replace the entire column with the midpoint in validation
        scaled_val[:, j] = CONSTANT_FEATURE_FILL
        # Replace the entire column with the midpoint in test
        scaled_test[:, j] = CONSTANT_FEATURE_FILL
    # Clip the scaled values to the [0.01, 10.01] range after the transform
    clipped_train: np.ndarray = np.clip(scaled_train, SCALE_LOWER, SCALE_UPPER)
    # Clip the validation values after the transform
    clipped_val: np.ndarray = np.clip(scaled_val, SCALE_LOWER, SCALE_UPPER)
    # Clip the test values after the transform
    clipped_test: np.ndarray = np.clip(scaled_test, SCALE_LOWER, SCALE_UPPER)
    # Count the clipped values per split as a shift diagnostic
    clip_counts: Dict[str, int] = {
        "train": int(np.sum(clipped_train != scaled_train)),
        "val": int(np.sum(clipped_val != scaled_val)),
        "test": int(np.sum(clipped_test != scaled_test)),
    }
    # Assert every scaled value lies within the declared range
    for name, matrix in {
        "train": clipped_train, "val": clipped_val, "test": clipped_test,
    }.items():
        # Raise when any value falls outside the scaled range
        if not (
            np.all(matrix >= SCALE_LOWER) and np.all(matrix <= SCALE_UPPER)
        ):
            raise ValueError(
                f"{name} scaled values fall outside "
                f"[{SCALE_LOWER}, {SCALE_UPPER}] after clipping."
            )
    # Assert the derived log-bound constant matches its computed value
    if not np.isclose(LOG_BOUND_L, 4.605170185988091, atol=1e-12):
        raise ValueError(
            f"Computed log bound L = {LOG_BOUND_L} contradicts the paper "
            "value 4.605170185988091."
        )
    # Return the clipped scaled matrices and the clip counts
    return clipped_train, clipped_val, clipped_test, clip_counts


# -----------------------------------------------------------------------------
# Task 4, Orchestrator Function
# -----------------------------------------------------------------------------

def orchestrate_splitting_and_scaling(
    X: np.ndarray,
    y: np.ndarray,
    cfg: Dict[str, Any],
    dataset_name: str,
    feature_names: Sequence[str],
    seed: int,
    output_dir: Path,
) -> Dict[str, Any]:
    """
    Orchestrate Task 4 for one dataset and seed: split, fit, scale.

    Parameters
    ----------
    X : np.ndarray, shape (n, m)
        Feature matrix from Task 3.
    y : np.ndarray, shape (n,)
        Binary label vector from Task 3.
    cfg : Dict[str, Any]
        Validated study configuration.
    dataset_name : str
        Dataset name for artifact naming and imputation policy.
    feature_names : Sequence[str]
        Feature names aligned with the matrix columns.
    seed : int
        Random seed for the stratified split.
    output_dir : Path
        Project output root for the artifact archive.

    Returns
    -------
    Dict[str, Any]
        Mapping with "splits", "fitted", "scaled", "clip_counts", and
        "persistence" keys.

    Raises
    ------
    ValueError
        If the dataset name is unknown or any step fails validation.
    """
    # Validate the feature matrix type
    if not isinstance(X, np.ndarray):
        raise TypeError(f"X must be an ndarray; received {type(X).__name__}.")
    # Validate the label vector type
    if not isinstance(y, np.ndarray):
        raise TypeError(f"y must be an ndarray; received {type(y).__name__}.")
    # Validate the dataset name is known
    if dataset_name not in ("Loan", "FraudEcom", "Default", "Creditcard"):
        raise ValueError(f"Unknown dataset name '{dataset_name}'.")
    # Step 4.1: create the stratified 60/20/20 split for the seed
    train_idx, val_idx, test_idx = make_stratified_splits(X, y, seed, cfg)
    # Step 4.2: fit preprocessing statistics on the training rows only
    fitted: FittedPreprocessing = fit_preprocessing(
        X_train=X[train_idx],
        cfg=cfg,
        dataset_name=dataset_name,
        feature_names=feature_names,
    )
    # Step 4.3: apply the fitted preprocessing to all three partitions
    scaled_train, scaled_val, scaled_test, clip_counts = apply_preprocessing(
        X_train=X[train_idx],
        X_val=X[val_idx],
        X_test=X[test_idx],
        fitted=fitted,
    )
    # Assemble the scaled matrix dictionary
    scaled_matrices: Dict[str, np.ndarray] = {
        "train": scaled_train,
        "val": scaled_val,
        "test": scaled_test,
    }
    # Construct the versioned artifact directories
    splits_dir: Path = output_dir / "artifacts" / "splits"
    stats_dir: Path = output_dir / "artifacts" / "preprocessing" / "v1"
    scaled_dir: Path = output_dir / "artifacts" / "scaled" / "v1"
    # Create the artifact directories when they do not exist yet
    for directory in (splits_dir, stats_dir, scaled_dir):
        directory.mkdir(parents=True, exist_ok=True)
    # Persist the split indices as an immutable npz artifact
    splits_path: Path = splits_dir / f"{dataset_name.lower()}_seed{seed}_splits.npz"
    # Save the three index arrays into the splits artifact
    np.savez_compressed(
        splits_path,
        train_idx=train_idx,
        val_idx=val_idx,
        test_idx=test_idx,
    )
    # Persist the fitted statistics as a JSON record for the audit trail
    stats_path: Path = stats_dir / f"{dataset_name.lower()}_seed{seed}_stats.json"
    # Serialise the fitted statistics to JSON with native types
    stats_path.write_text(
        json.dumps(
            {
                "dataset": dataset_name,
                "seed": seed,
                "medians": fitted.medians.tolist(),
                "mins": fitted.mins.tolist(),
                "maxs": fitted.maxs.tolist(),
                "modes": {str(k): v for k, v in fitted.modes.items()},
                "constant_feature_policy": {
                    str(k): v for k, v in fitted.constant_feature_policy.items()
                },
                "feature_names": list(fitted.feature_names),
                "log_bound_L": LOG_BOUND_L,
                "written_at": datetime.now(timezone.utc).isoformat(),
            },
            indent=2,
            ensure_ascii=False,
        )
        + "\n",
        encoding="utf-8",
    )
    # Persist the scaled matrices as compressed npz artifacts
    scaled_path: Path = scaled_dir / f"{dataset_name.lower()}_seed{seed}_scaled.npz"
    # Save the three scaled matrices into the scaled artifact
    np.savez_compressed(
        scaled_path,
        train=scaled_train,
        val=scaled_val,
        test=scaled_test,
    )
    # Assemble the persistence report
    persistence_report: Dict[str, Any] = {
        "splits": str(splits_path),
        "stats": str(stats_path),
        "scaled": str(scaled_path),
        "clip_counts": clip_counts,
    }
    # Return the full Task 4 result structure
    return {
        "splits": {"train": train_idx, "val": val_idx, "test": test_idx},
        "fitted": fitted,
        "scaled": scaled_matrices,
        "clip_counts": clip_counts,
        "persistence": persistence_report,
    }

````

````python
# Task 5: Optuna hyperparameter search on training splits

# ==============================================================================
# Task 5: Optuna hyperparameter search on training splits
# ==============================================================================

# -----------------------------------------------------------------------------
# Task 5, Step 1: Define the search space and the objective function
# -----------------------------------------------------------------------------

def optuna_objective(
    trial: optuna.Trial,
    X_train: np.ndarray,
    y_train: np.ndarray,
    cfg: Dict[str, Any],
    seed: int,
) -> float:
    """
    Optuna objective: mean five-fold stratified CV PR-AUC on training.

    Samples the learning rate log-uniformly from [1e-4, 1e-1], the l1
    strength log-uniformly from [1e-6, 1e-2], and the epoch budget
    uniformly from [100, 3000]. Each fold trains the Task 6 monomial on
    the fold-train partition (training-loss convergence only; never
    validation early stopping) and scores PR-AUC on the fold-validation
    partition. The objective reads no file other than its arguments.

    Parameters
    ----------
    trial : optuna.Trial
        Active Optuna trial used for hyperparameter suggestions.
    X_train : np.ndarray, shape (n_train, m)
        Scaled training feature matrix (training rows only).
    y_train : np.ndarray, shape (n_train,)
        Training labels (training rows only).
    cfg : Dict[str, Any]
        Validated study configuration.
    seed : int
        Deterministic seed for the fold splitter.

    Returns
    -------
    float
        Mean cross-validated PR-AUC over the five folds.

    Raises
    ------
    optuna.TrialPruned
        If the trial's intermediate PR-AUC falls below the pruning
        threshold of the MedianPruner.
    """

    # Read the search-space ranges from the configuration
    lr_low, lr_high = cfg["hyperparameter_search"]["learning_rate_range"]
    # Read the lambda search-space range from the configuration
    lam_low, lam_high = cfg["hyperparameter_search"]["lambda_range"]
    # Read the epoch search-space range from the configuration
    epochs_low, epochs_high = cfg["hyperparameter_search"]["epochs_range"]
    # Sample the learning rate log-uniformly across its range
    learning_rate: float = trial.suggest_float(
        "learning_rate", lr_low, lr_high, log=True
    )
    # Sample the l1 strength log-uniformly across its range
    lambda_reg: float = trial.suggest_float(
        "lambda_reg", lam_low, lam_high, log=True
    )
    # Sample the epoch budget uniformly as an integer
    epochs: int = trial.suggest_int("epochs", epochs_low, epochs_high)
    # Assemble the hyperparameter dictionary for the trainer
    hparams: Dict[str, Any] = {
        "learning_rate": learning_rate,
        "lambda_reg": lambda_reg,
        "epochs": epochs,
    }
    # Build the five-fold stratified splitter with the trial seed
    fold_splitter: StratifiedKFold = StratifiedKFold(
        n_splits=5, shuffle=True, random_state=seed
    )
    # Collect the per-fold PR-AUC values
    fold_pr_aucs: List[float] = []
    # Iterate over the five stratified folds
    for fold_index, (fold_train_idx, fold_val_idx) in enumerate(
        fold_splitter.split(X_train, y_train)
    ):
        # Slice the fold-train and fold-validation partitions
        X_fold_train: np.ndarray = X_train[fold_train_idx]
        y_fold_train: np.ndarray = y_train[fold_train_idx]
        # Slice the fold-validation partitions
        X_fold_val: np.ndarray = X_train[fold_val_idx]
        y_fold_val: np.ndarray = y_train[fold_val_idx]

        # Train the monomial on the fold-train partition with the sampled
        # hyperparameters; the trainer stops on training-loss convergence
        # only, so the fold-validation set never influences the fit
        train_result = train_monomial(
            X_train=X_fold_train,
            y_train=y_fold_train,
            hparams=hparams,
            cfg=cfg,
            seed=seed,
        )
        # Extract the fitted exponent vector from the training result
        beta_fitted: np.ndarray = train_result.beta
        # Extract the fitted coefficient from the training result
        alpha_fitted: float = train_result.alpha
        # Compute the monomial logits on the fold-validation rows
        fold_logits: np.ndarray = monomial_logit(
            X_fold_val, alpha_fitted, beta_fitted
        )
        # Convert the logits to probabilities via the sigmoid
        fold_probabilities: np.ndarray = 1.0 / (1.0 + np.exp(-fold_logits))
        # Compute the PR-AUC on the fold-validation partition
        fold_pr_auc: float = float(
            average_precision_score(y_fold_val, fold_probabilities)
        )
        # Append the fold PR-AUC to the collected list
        fold_pr_aucs.append(fold_pr_auc)
        # Report the intermediate value for the MedianPruner
        trial.report(fold_pr_auc, fold_index)
        # Raise the pruning exception when the pruner decides to prune
        if trial.should_prune():
            raise optuna.TrialPruned()
    # Return the mean PR-AUC over the five folds as a native float
    return float(np.mean(fold_pr_aucs))


# -----------------------------------------------------------------------------
# Task 5, Step 2: Run Optuna and select the best configuration
# -----------------------------------------------------------------------------

def run_hyperparameter_search(
    X_train: np.ndarray,
    y_train: np.ndarray,
    cfg: Dict[str, Any],
    dataset_name: str,
    seed: int = 0,
    n_trials: Optional[int] = None,
    storage_url: Optional[str] = None,
) -> Dict[str, Any]:
    """
    Run the Optuna TPE search and return the best configuration.

    The disambiguated protocol tunes once per dataset on the seed-0
    training split and reuses the selected configuration for all five
    seeds; per-seed tuning would confound seed comparison and quintuple
    the cost (documented as a replicator decision).

    Parameters
    ----------
    X_train : np.ndarray, shape (n_train, m)
        Scaled training feature matrix.
    y_train : np.ndarray, shape (n_train,)
        Training labels.
    cfg : Dict[str, Any]
        Validated study configuration.
    dataset_name : str
        Dataset name for the study and result record.
    seed : int, optional
        Seed for the TPE sampler and fold splitter, default 0.
    n_trials : Optional[int], optional
        Trial budget; defaults to the configured value.
    storage_url : Optional[str], optional
        SQLite storage URL for a resumable study, or None for in-memory.

    Returns
    -------
    Dict[str, Any]
        Best configuration with native Python types and the best
        cross-validated PR-AUC.

    Raises
    ------
    TypeError
        If `X_train` or `y_train` is not an ndarray.
    ValueError
        If the training arrays disagree in length.
    """
    # Validate the training matrix type
    if not isinstance(X_train, np.ndarray):
        raise TypeError(
            f"X_train must be an ndarray; received {type(X_train).__name__}."
        )
    # Validate the label vector type
    if not isinstance(y_train, np.ndarray):
        raise TypeError(
            f"y_train must be an ndarray; received {type(y_train).__name__}."
        )
    # Raise when the training arrays disagree in length
    if X_train.shape[0] != y_train.shape[0]:
        raise ValueError(
            f"X_train has {X_train.shape[0]} rows but y_train has "
            f"{y_train.shape[0]} entries."
        )
    # Resolve the trial budget from the configuration when not supplied
    if n_trials is None:
        n_trials = int(cfg["hyperparameter_search"]["n_trials"])

    # Set the global random seeds for deterministic optimisation
    random.seed(seed)
    np.random.seed(seed)
    # Construct the deterministic TPE sampler with the search seed
    sampler: TPESampler = TPESampler(seed=seed)
    # Construct the median pruner with a startup phase
    pruner: MedianPruner = MedianPruner(n_startup_trials=5, n_warmup_steps=2)
    # Create or resume the study with the configured storage
    study: optuna.Study = optuna.create_study(
        direction="maximize",
        sampler=sampler,
        pruner=pruner,
        storage=storage_url,
        study_name=f"{dataset_name}_hyperparameter_search",
        load_if_exists=True,
    )
    # Run the optimisation for the configured trial budget
    study.optimize(
        lambda trial: optuna_objective(
            trial, X_train, y_train, cfg, seed
        ),
        n_trials=n_trials,
    )
    # Read the best hyperparameters from the study
    best_params: Dict[str, Any] = study.best_params
    # Read the best cross-validated PR-AUC from the study
    best_value: float = float(study.best_value)
    # Assemble the result record with native Python types
    result: Dict[str, Any] = {
        "dataset": dataset_name,
        "tuning_seed": seed,
        "n_trials": n_trials,
        "learning_rate": float(best_params["learning_rate"]),
        "lambda_reg": float(best_params["lambda_reg"]),
        "epochs": int(best_params["epochs"]),
        "best_cv_pr_auc": best_value,
    }
    # Return the selected configuration record
    return result


# -----------------------------------------------------------------------------
# Task 5, Step 3: Record and persist hyperparameters
# -----------------------------------------------------------------------------

def persist_hyperparameters(
    result: Dict[str, Any],
    output_dir: Path,
) -> Dict[str, Any]:
    """
    Persist the selected hyperparameters as a checksummed JSON record.

    Parameters
    ----------
    result : Dict[str, Any]
        Best-configuration record from run_hyperparameter_search.
    output_dir : Path
        Project output root; records are written under
        artifacts/hyperparameters/v1/.

    Returns
    -------
    Dict[str, Any]
        Persistence report with the record path and SHA-256 digest.

    Raises
    ------
    TypeError
        If `result` is not a dict or `output_dir` is not a Path.
    """
    # Validate the result type
    if not isinstance(result, dict):
        raise TypeError(
            f"result must be a dict; received {type(result).__name__}."
        )
    # Validate the output directory type
    if not isinstance(output_dir, Path):
        raise TypeError(
            f"output_dir must be a pathlib.Path; "
            f"received {type(output_dir).__name__}."
        )
    # Construct the versioned hyperparameter artifact directory
    archive_dir: Path = output_dir / "artifacts" / "hyperparameters" / "v1"
    # Create the archive directory when it does not exist yet
    archive_dir.mkdir(parents=True, exist_ok=True)
    # Compose the full record with the search-space ranges and metadata
    full_record: Dict[str, Any] = dict(result)
    # Attach the exact search-space ranges to the record
    full_record["search_space"] = {
        "learning_rate": "log-uniform [1e-4, 1e-1]",
        "lambda_reg": "log-uniform [1e-6, 1e-2]",
        "epochs": "uniform integer [100, 3000]",
    }
    # Attach the configuration version reference to the record
    full_record["configuration_version"] = "study_configuration v1"
    # Attach the record timestamp to the record
    full_record["written_at"] = datetime.now(timezone.utc).isoformat()
    # Construct the record path for this dataset
    record_path: Path = archive_dir / (
        f"hyperparameters_{full_record['dataset'].lower()}.json"
    )
    # Serialise the record with native types and stable formatting
    record_text: str = json.dumps(
        full_record, indent=2, ensure_ascii=False
    ) + "\n"
    # Write the serialised record to disk
    record_path.write_text(record_text, encoding="utf-8")
    # Compute the SHA-256 digest over the bytes on disk
    digest: str = hashlib.sha256(record_path.read_bytes()).hexdigest()
    # Return the persistence report with the path and digest
    return {"path": str(record_path), "sha256": digest}


# -----------------------------------------------------------------------------
# Task 5, Orchestrator Function
# -----------------------------------------------------------------------------

def orchestrate_hyperparameter_search(
    X_train: np.ndarray,
    y_train: np.ndarray,
    cfg: Dict[str, Any],
    dataset_name: str,
    output_dir: Path,
    seed: int = 0,
) -> Dict[str, Any]:
    """
    Orchestrate Task 5: run the search once per dataset and persist it.

    The hyperparameter record is the single source of truth for every
    later training run; if the record already exists it is loaded and
    returned without re-running the search.

    Parameters
    ----------
    X_train : np.ndarray, shape (n_train, m)
        Scaled training feature matrix (seed-0 training split).
    y_train : np.ndarray, shape (n_train,)
        Training labels (seed-0 training split).
    cfg : Dict[str, Any]
        Validated study configuration.
    dataset_name : str
        Dataset name for the study and artifact naming.
    output_dir : Path
        Project output root for the artifact archive.
    seed : int, optional
        Tuning seed, default 0 (the disambiguated once-per-dataset
        protocol tunes on the seed-0 split).

    Returns
    -------
    Dict[str, Any]
        Mapping with "configuration" and "persistence" keys.
    """
    # Validate the feature matrix type
    if not isinstance(X_train, np.ndarray):
        raise TypeError(
            f"X_train must be an ndarray; received {type(X_train).__name__}."
        )
    # Validate the label vector type
    if not isinstance(y_train, np.ndarray):
        raise TypeError(
            f"y_train must be an ndarray; received {type(y_train).__name__}."
        )
    # Construct the expected record path for the dataset
    record_path: Path = (
        output_dir
        / "artifacts"
        / "hyperparameters"
        / "v1"
        / f"hyperparameters_{dataset_name.lower()}.json"
    )
    # Load the existing record when the search has already run
    if record_path.exists():
        # Parse the existing record from disk
        existing_record: Dict[str, Any] = json.loads(
            record_path.read_text(encoding="utf-8")
        )
        # Return the existing configuration without re-searching
        return {"configuration": existing_record, "persistence": None}
    # Construct the SQLite storage URL for a resumable study
    storage_dir: Path = output_dir / "artifacts" / "hyperparameters" / "v1"
    # Create the storage directory when it does not exist yet
    storage_dir.mkdir(parents=True, exist_ok=True)
    # Build the study database path for the dataset
    storage_path: Path = storage_dir / f"{dataset_name.lower()}_study.db"
    # Compose the SQLite storage URL from the database path
    storage_url: str = f"sqlite:///{storage_path}"
    # Run the Optuna search with the configured trial budget
    result: Dict[str, Any] = run_hyperparameter_search(
        X_train=X_train,
        y_train=y_train,
        cfg=cfg,
        dataset_name=dataset_name,
        seed=seed,
        storage_url=storage_url,
    )
    # Persist the selected configuration as a checksummed JSON record
    persistence_report: Dict[str, Any] = persist_hyperparameters(
        result, output_dir
    )
    # Return the configuration and its persistence report
    return {"configuration": result, "persistence": persistence_report}
````

````python
# Task 6: Train the full ECSEL single-term monomial

# ==============================================================================
# Task 6: Train the full ECSEL single-term monomial
# ==============================================================================

# Fixed initialization values (replicator choices, UNSPECIFIED IN MANUSCRIPT)
INIT_ALPHA: float = 1.0
INIT_BETA: float = 0.1

# Absolute tolerance of the log-space identity verification
LOG_IDENTITY_TOLERANCE: float = 1e-6

# Number of sample rows used by the verification step
VERIFICATION_SAMPLE_ROWS: int = 100


# -----------------------------------------------------------------------------
# Task 6, Step 1: Define the model forward and loss callables
# -----------------------------------------------------------------------------

def monomial_logit(
    X: np.ndarray,
    alpha: float,
    beta: np.ndarray,
) -> np.ndarray:
    """
    Compute the monomial logit via the numerically stable log-space form.

    Implements the log-space identity from the manuscript:
        log|z(x)| = log|alpha| + sum_j beta_j log x_j
    with the sign of alpha tracked separately, giving
        z(x) = sign(alpha) * exp(log|alpha| + beta^T log x).

    Parameters
    ----------
    X : np.ndarray, shape (n, m)
        Scaled feature matrix; every value must lie in [0.01, 10.01].
    alpha : float
        Fitted monomial coefficient (must be non-zero).
    beta : np.ndarray, shape (m,)
        Fitted exponent vector aligned to the matrix columns.

    Returns
    -------
    np.ndarray, shape (n,)
        Monomial logits for every row.

    Raises
    ------
    TypeError
        If `X` or `beta` is not an ndarray.
    ValueError
        If alpha is zero, beta is misaligned, or X is outside the scaled
        positive range.
    """
    # Validate the feature matrix type
    if not isinstance(X, np.ndarray):
        raise TypeError(f"X must be an ndarray; received {type(X).__name__}.")
    # Validate the exponent vector type
    if not isinstance(beta, np.ndarray):
        raise TypeError(
            f"beta must be an ndarray; received {type(beta).__name__}."
        )
    # Raise when the coefficient is zero (log|alpha| undefined)
    if alpha == 0.0:
        raise ValueError("alpha must be non-zero; log|alpha| is undefined.")
    # Raise when the exponent vector is misaligned with the matrix
    if beta.shape[0] != X.shape[1]:
        raise ValueError(
            f"beta has {beta.shape[0]} entries but X has {X.shape[1]} "
            "columns."
        )
    # Assert the scaled-range precondition required by the logarithm
    if X.shape[0] > 0 and not (
        np.all(X >= 0.01) and np.all(X <= 10.01)
    ):
        raise ValueError(
            "X must lie in [0.01, 10.01] for the log transform to be "
            "defined on the scaled features."
        )
    # Compute the log of every scaled feature value
    log_x: np.ndarray = np.log(X)
    # Compute the log-space score: log|alpha| + beta^T log x
    log_score: np.ndarray = np.log(abs(alpha)) + log_x @ beta
    # Reconstruct the logit with the sign of alpha tracked separately
    logits: np.ndarray = np.sign(alpha) * np.exp(log_score)
    # Return the monomial logit vector
    return logits


def monomial_loss(
    X: np.ndarray,
    y: np.ndarray,
    alpha: float,
    beta: np.ndarray,
    lambda_reg: float,
) -> torch.Tensor:
    """
    Compute the regularized logistic loss as an autograd-ready tensor.

    Implements the corrected binary cross-entropy for y in {0, 1}:
        L = -(1/n) sum_i [y_i log sigma(z_i) + (1 - y_i) log(1 - sigma(z_i))]
            + lambda * sum_j |beta_j|
    via binary_cross_entropy_with_logits, whose internal logaddexp
    formulation is the numerically stable evaluation. The l1 penalty
    applies to the exponents beta only, never to alpha.

    Parameters
    ----------
    X : np.ndarray, shape (n, m)
        Scaled feature matrix in [0.01, 10.01].
    y : np.ndarray, shape (n,)
        Binary labels in {0, 1}.
    alpha : float
        Monomial coefficient.
    beta : np.ndarray, shape (m,)
        Exponent vector aligned to the matrix columns.
    lambda_reg : float
        Non-negative l1 regularization strength on the exponents.

    Returns
    -------
    torch.Tensor
        Scalar loss tensor; differentiable with respect to freshly
        constructed alpha and beta tensors internally (the callable is a
        pure evaluation of the loss given the parameters).

    Raises
    ------
    TypeError
        If `X` or `y` is not an ndarray.
    ValueError
        If the labels are not binary, the arrays misalign, or lambda is
        negative.
    """
    # Validate the feature matrix type
    if not isinstance(X, np.ndarray):
        raise TypeError(f"X must be an ndarray; received {type(X).__name__}.")
    # Validate the label vector type
    if not isinstance(y, np.ndarray):
        raise TypeError(f"y must be an ndarray; received {type(y).__name__}.")
    # Raise when the label vector is not binary
    if not set(np.unique(y)).issubset({0, 1}):
        raise ValueError("y must contain only the binary labels 0 and 1.")
    # Raise when the arrays disagree in length
    if X.shape[0] != y.shape[0]:
        raise ValueError(
            f"X has {X.shape[0]} rows but y has {y.shape[0]} entries."
        )
    # Raise when the regularization strength is negative
    if lambda_reg < 0.0:
        raise ValueError(
            f"lambda_reg must be non-negative; received {lambda_reg}."
        )
    # Import torch lazily (the scientific-stack environment contract);
    # torch is required only for automatic differentiation

    # Convert the feature matrix to a float64 tensor
    X_t: torch.Tensor = torch.tensor(X, dtype=torch.float64)
    # Convert the label vector to a float64 tensor
    y_t: torch.Tensor = torch.tensor(y, dtype=torch.float64)
    # Convert the coefficient to a float64 tensor
    alpha_t: torch.Tensor = torch.tensor(alpha, dtype=torch.float64)
    # Convert the exponent vector to a float64 tensor
    beta_t: torch.Tensor = torch.tensor(beta, dtype=torch.float64)
    # Precompute the log of the scaled features once
    log_x_t: torch.Tensor = torch.log(X_t)
    # Compute the log-space score: log|alpha| + beta^T log x
    log_score_t: torch.Tensor = torch.log(torch.abs(alpha_t)) + log_x_t @ beta_t
    # Reconstruct the logits with the sign of alpha tracked separately
    z_t: torch.Tensor = torch.sign(alpha_t) * torch.exp(log_score_t)
    # Evaluate the stable binary cross-entropy via logaddexp internally
    bce_t: torch.Tensor = F.binary_cross_entropy_with_logits(z_t, y_t)
    # Evaluate the l1 penalty on the exponents only
    l1_t: torch.Tensor = lambda_reg * torch.sum(torch.abs(beta_t))
    # Sum the data term and the penalty into the regularized loss
    loss_t: torch.Tensor = bce_t + l1_t
    # Return the scalar loss tensor
    return loss_t


# -----------------------------------------------------------------------------
# Task 6, Step 2: Train the monomial to convergence
# -----------------------------------------------------------------------------
@dataclass
class TrainResult:
    """
    Container of a completed ECSEL single-term monomial training run.

    Holds the fitted coefficient ``alpha`` and exponent vector ``beta`` of the
    monomial ``z(x)=alpha prod_j x_j^{beta_j}`` together with the optimization
    trace and post-training validation metric. Produced by ``train_monomial``
    and consumed by pruning and verification in Task 6 Step 6.3.

    Attributes
    ----------
    alpha : float
        Fitted monomial coefficient; sign determines risk direction together
        with ``beta``.
    beta : np.ndarray, shape (m,)
        Fitted exponent vector aligned to feature columns; ``|beta_j|``
        encodes feature strength.
    history : List[float]
        Per-epoch regularized training loss values ``L(alpha,beta)``;
        length equals ``epochs_used``.
    epochs_used : int
        Number of epochs actually executed; may be less than the budget when
        the 20-epoch stagnation criterion fires.
    converged : bool
        True when the mean absolute relative training-loss change over 20
        consecutive epochs fell below ``1e-6``.
    val_pr_auc : Optional[float]
        Validation PR-AUC computed AFTER training on the held-out validation
        split; ``None`` when no validation data were supplied.
    """

    # Fitted monomial coefficient
    alpha: float
    # Fitted exponent vector aligned to feature columns
    beta: np.ndarray
    # Per-epoch regularized training loss values
    history: List[float] = field(default_factory=list)
    # Number of epochs actually executed
    epochs_used: int = 0
    # True when the training-loss stagnation criterion fired
    converged: bool = False
    # Validation PR-AUC computed after training (None when unavailable)
    val_pr_auc: Optional[float] = None


# -----------------------------------------------------------------------------
def train_monomial(
    X_train: np.ndarray,
    y_train: np.ndarray,
    hparams: Dict[str, Any],
    cfg: Dict[str, Any],
    seed: int,
    X_val: Optional[np.ndarray] = None,
    y_val: Optional[np.ndarray] = None,
) -> TrainResult:
    """
    Train the ECSEL single-term monomial to training-loss convergence.

    Gradient descent minimizes the regularized logistic objective with
    automatic differentiation. Training stops when the mean absolute
    relative change in the training loss over the last 20 epochs falls
    below 1e-6 (with a zero-loss guard) or when the epoch budget is
    exhausted. Validation is NOT used for early stopping; the validation
    PR-AUC is computed only after training finishes.

    Parameters
    ----------
    X_train : np.ndarray, shape (n_train, m)
        Scaled training feature matrix in [0.01, 10.01].
    y_train : np.ndarray, shape (n_train,)
        Binary training labels.
    hparams : Dict[str, Any]
        Hyperparameters with keys "learning_rate", "lambda_reg",
        "epochs" (from the Optuna record of Task 5).
    cfg : Dict[str, Any]
        Validated study configuration.
    seed : int
        Deterministic seed for torch and numpy.
    X_val : Optional[np.ndarray], optional
        Scaled validation matrix for post-training PR-AUC.
    y_val : Optional[np.ndarray], optional
        Validation labels for post-training PR-AUC.

    Returns
    -------
    TrainResult
        Fitted parameters, loss history, epoch count, convergence flag,
        and the post-training validation PR-AUC when validation data
        were supplied.

    Raises
    ------
    TypeError
        If the training arrays are not ndarrays.
    ValueError
        If the arrays misalign, hyperparameters are missing, or the
        learning rate is not positive.
    """
    # Validate the training matrix type
    if not isinstance(X_train, np.ndarray):
        raise TypeError(
            f"X_train must be an ndarray; received {type(X_train).__name__}."
        )
    # Validate the label vector type
    if not isinstance(y_train, np.ndarray):
        raise TypeError(
            f"y_train must be an ndarray; received {type(y_train).__name__}."
        )
    # Raise when the training arrays disagree in length
    if X_train.shape[0] != y_train.shape[0]:
        raise ValueError(
            f"X_train has {X_train.shape[0]} rows but y_train has "
            f"{y_train.shape[0]} entries."
        )
    # Extract the learning rate from the hyperparameters
    learning_rate: float = float(hparams["learning_rate"])
    # Extract the l1 strength from the hyperparameters
    lambda_reg: float = float(hparams["lambda_reg"])
    # Extract the epoch budget from the hyperparameters
    epoch_budget: int = int(hparams["epochs"])
    # Raise when the learning rate is not positive
    if learning_rate <= 0.0:
        raise ValueError(
            f"learning_rate must be positive; received {learning_rate}."
        )

    # Seed torch and numpy for deterministic training
    torch.manual_seed(seed)
    np.random.seed(seed)
    # Convert the training matrix to a float64 tensor
    X_t: torch.Tensor = torch.tensor(X_train, dtype=torch.float64)
    # Convert the training labels to a float64 tensor
    y_t: torch.Tensor = torch.tensor(y_train, dtype=torch.float64)
    # Precompute the log of the scaled features once for all epochs
    log_x_t: torch.Tensor = torch.log(X_t)
    # Initialise the coefficient to the fixed starting value
    alpha: torch.Tensor = torch.tensor(
        INIT_ALPHA, dtype=torch.float64, requires_grad=True
    )
    # Initialise every exponent to the fixed starting value
    beta: torch.Tensor = torch.full(
        (X_train.shape[1],), INIT_BETA, dtype=torch.float64, requires_grad=True
    )
    # Construct the gradient-descent optimizer over both parameters
    optimizer: torch.optim.SGD = torch.optim.SGD([alpha, beta], lr=learning_rate)
    # Record the per-epoch regularized training losses
    history: List[float] = []
    # Track whether the convergence criterion fired
    converged: bool = False
    # Track the number of epochs actually executed
    epochs_used: int = 0
    # Define the convergence lookback window size
    patience_window: int = 20
    # Run the gradient-descent loop up to the epoch budget
    for epoch in range(epoch_budget):
        # Reset the parameter gradients to zero
        optimizer.zero_grad()
        # Compute the log-space score: log|alpha| + beta^T log x
        log_score_t: torch.Tensor = torch.log(torch.abs(alpha)) + log_x_t @ beta
        # Reconstruct the logits with the sign of alpha tracked separately
        z_t: torch.Tensor = torch.sign(alpha) * torch.exp(log_score_t)
        # Evaluate the stable binary cross-entropy
        bce_t: torch.Tensor = F.binary_cross_entropy_with_logits(z_t, y_t)
        # Evaluate the l1 penalty on the exponents only
        l1_t: torch.Tensor = lambda_reg * torch.sum(torch.abs(beta))
        # Sum the data term and the penalty into the loss
        loss_t: torch.Tensor = bce_t + l1_t
        # Backpropagate the loss through the computation graph
        loss_t.backward()
        # Apply the gradient-descent update to both parameters
        optimizer.step()
        # Record the loss value of this epoch
        history.append(float(loss_t.detach()))
        # Update the executed epoch count
        epochs_used = epoch + 1
        # Test the convergence criterion once the window is full
        if epoch >= patience_window:
            # Take the most recent loss window
            recent_losses: List[float] = history[-patience_window:]
            # Compute the absolute relative changes inside the window
            relative_changes: List[float] = [
                abs(recent_losses[i + 1] - recent_losses[i])
                / max(abs(recent_losses[i]), 1e-12)
                for i in range(len(recent_losses) - 1)
            ]
            # Compute the mean absolute relative change over the window
            mean_relative_change: float = float(np.mean(relative_changes))
            # Fire the convergence flag when the criterion is met
            if mean_relative_change < 1e-6:
                converged = True
                break
    # Detach the fitted coefficient as a native float
    alpha_fitted: float = float(alpha.detach())
    # Detach the fitted exponent vector as a numpy array copy
    beta_fitted: np.ndarray = beta.detach().numpy().copy()
    # Initialise the post-training validation PR-AUC as None
    val_pr_auc: Optional[float] = None
    # Compute the validation PR-AUC only after training completes
    if X_val is not None and y_val is not None:
        # Compute the monomial logits on the validation rows
        val_logits: np.ndarray = monomial_logit(X_val, alpha_fitted, beta_fitted)
        # Convert the logits to probabilities via the sigmoid
        val_probabilities: np.ndarray = 1.0 / (1.0 + np.exp(-val_logits))
        # Compute the validation PR-AUC
        val_pr_auc = float(average_precision_score(y_val, val_probabilities))
    # Assemble and return the training result
    return TrainResult(
        alpha=alpha_fitted,
        beta=beta_fitted,
        history=history,
        epochs_used=epochs_used,
        converged=converged,
        val_pr_auc=val_pr_auc,
    )


# -----------------------------------------------------------------------------
# Task 6, Step 3: Verify convergence, stability, and the log-space identity
# -----------------------------------------------------------------------------

def verify_trained_monomial(
    alpha: float,
    beta: np.ndarray,
    X_sample: np.ndarray,
    cfg: Dict[str, Any],
    y_sample: Optional[np.ndarray] = None,
) -> Dict[str, Any]:
    """
    Verify the fitted monomial: finiteness and the log-space identity.

    Per the disambiguated protocol, the raw gradient norm on the
    non-smooth l1 objective is not a valid hard convergence check; the
    convergence criterion is the training-loss stagnation of Step 6.2.
    A gradient norm computed on the smooth loss (without the l1 penalty)
    is logged for the appendix instead of aborting.

    Parameters
    ----------
    alpha : float
        Fitted monomial coefficient.
    beta : np.ndarray, shape (m,)
        Fitted exponent vector.
    X_sample : np.ndarray, shape (n_sample, m)
        Scaled sample rows used for verification.
    cfg : Dict[str, Any]
        Validated study configuration.
    y_sample : Optional[np.ndarray], optional
        Labels aligned with the first rows of X_sample; when supplied,
        the smooth-loss gradient norm is computed with the actual labels
        and logged. When omitted, the norm is recorded as not computed
        (it is a logged diagnostic, never an abort).

    Returns
    -------
    Dict[str, Any]
        Verification record with finiteness, identity errors, support
        set, and the logged smooth gradient norm (None when labels were
        not supplied).

    Raises
    ------
    ValueError
        If any sampled row violates the log-space identity beyond the
        absolute tolerance 1e-6.
    """
    # Validate the exponent vector type
    if not isinstance(beta, np.ndarray):
        raise TypeError(
            f"beta must be an ndarray; received {type(beta).__name__}."
        )
    # Validate the sample matrix type
    if not isinstance(X_sample, np.ndarray):
        raise TypeError(
            f"X_sample must be an ndarray; received {type(X_sample).__name__}."
        )
    # Assert the scaled-range precondition required by the logarithm
    if not (np.all(X_sample >= 0.01) and np.all(X_sample <= 10.01)):
        raise ValueError(
            "X_sample must lie in [0.01, 10.01] for the log transform."
        )
    # Take the deterministic first 100 rows of the sample matrix
    sample: np.ndarray = X_sample[: min(VERIFICATION_SAMPLE_ROWS, X_sample.shape[0])]
    # Compute the monomial logits on the sampled rows
    logits: np.ndarray = monomial_logit(sample, alpha, beta)
    # Check that every logit is finite
    finite_ok: bool = bool(np.isfinite(logits).all())
    # Raise when any logit is non-finite
    if not finite_ok:
        raise ValueError(
            "Monomial output is non-finite on the verification sample."
        )
    # Compute the log of the absolute logits (the log-space score)
    log_abs_z: np.ndarray = np.log(np.abs(logits))
    # Compute the log-space score directly: log|alpha| + beta^T log x
    log_space_score: np.ndarray = (
        np.log(abs(alpha)) + np.log(sample) @ beta
    )
    # Compute the absolute errors of the log-space identity
    identity_errors: np.ndarray = np.abs(log_abs_z - log_space_score)
    # Record the maximum absolute identity error
    max_identity_error: float = float(identity_errors.max())
    # Raise when the identity is violated beyond the tolerance
    if max_identity_error > LOG_IDENTITY_TOLERANCE:
        raise ValueError(
            f"Log-space identity violated: max absolute error "
            f"{max_identity_error:.2e} exceeds the tolerance "
            f"{LOG_IDENTITY_TOLERANCE}."
        )
    # Compute the support set: columns with non-zero fitted exponents
    support_set: List[int] = [
        int(j) for j in range(beta.shape[0]) if abs(beta[j]) > 0.0
    ]
    # Initialise the smooth gradient norm as not computed
    smooth_grad_norm: Optional[float] = None
    # Compute the smooth-loss gradient norm only when labels are supplied
    if y_sample is not None:

        # Convert the sample to a torch tensor for the gradient norm
        sample_t: torch.Tensor = torch.tensor(sample, dtype=torch.float64)
        # Convert the supplied sample labels to a torch tensor
        y_smooth: torch.Tensor = torch.tensor(
            y_sample[: sample.shape[0]], dtype=torch.float64
        )
        # Rebuild the parameters as leaf tensors for the gradient computation
        alpha_t: torch.Tensor = torch.tensor(
            alpha, dtype=torch.float64, requires_grad=True
        )
        # Rebuild the exponent vector as a leaf tensor
        beta_t: torch.Tensor = torch.tensor(
            beta, dtype=torch.float64, requires_grad=True
        )
        # Compute the log-space score on the sample
        log_score_t: torch.Tensor = torch.log(torch.abs(alpha_t)) + torch.log(
            sample_t
        ) @ beta_t
        # Reconstruct the logits with the sign of alpha
        z_t: torch.Tensor = torch.sign(alpha_t) * torch.exp(log_score_t)
        # Evaluate the smooth loss WITHOUT the l1 penalty
        smooth_loss_t: torch.Tensor = F.binary_cross_entropy_with_logits(
            z_t, y_smooth
        )
        # Backpropagate the smooth loss
        smooth_loss_t.backward()
        # Concatenate the smooth gradients of both parameters
        smooth_grad_norm = float(
            torch.cat(
                [
                    alpha_t.grad.reshape(1),
                    beta_t.grad.reshape(-1),
                ]
            ).norm()
        )
        # Log the smooth gradient norm for the appendix (not an abort)
        warnings.warn(
            f"Smooth-loss gradient norm (logged, not an abort): "
            f"{smooth_grad_norm:.6e}.",
            RuntimeWarning,
        )
    # Assemble the verification record
    verification: Dict[str, Any] = {
        "finite_outputs": finite_ok,
        "max_log_space_identity_error": max_identity_error,
        "log_space_identity_tolerance": LOG_IDENTITY_TOLERANCE,
        "smooth_gradient_norm_logged": smooth_grad_norm,
        "support": support_set,
    }
    # Return the verification record
    return verification


# -----------------------------------------------------------------------------
# Task 6, Orchestrator Function
# -----------------------------------------------------------------------------

def orchestrate_monomial_training(
    X_train: np.ndarray,
    y_train: np.ndarray,
    hparams: Dict[str, Any],
    cfg: Dict[str, Any],
    dataset_name: str,
    seed: int,
    output_dir: Path,
    X_val: Optional[np.ndarray] = None,
    y_val: Optional[np.ndarray] = None,
) -> Dict[str, Any]:
    """
    Orchestrate Task 6: train, verify, and persist the full monomial.

    Parameters
    ----------
    X_train : np.ndarray, shape (n_train, m)
        Scaled training feature matrix.
    y_train : np.ndarray, shape (n_train,)
        Binary training labels.
    hparams : Dict[str, Any]
        Selected hyperparameters from the Task 5 record.
    cfg : Dict[str, Any]
        Validated study configuration.
    dataset_name : str
        Dataset name for artifact naming.
    seed : int
        Training seed.
    output_dir : Path
        Project output root for the artifact archive.
    X_val : Optional[np.ndarray], optional
        Scaled validation matrix for post-training PR-AUC.
    y_val : Optional[np.ndarray], optional
        Validation labels for post-training PR-AUC.

    Returns
    -------
    Dict[str, Any]
        Mapping with "train_result", "verification", and "persistence".
    """
    # Step 6.2: train the monomial to training-loss convergence
    train_result: TrainResult = train_monomial(
        X_train=X_train,
        y_train=y_train,
        hparams=hparams,
        cfg=cfg,
        seed=seed,
        X_val=X_val,
        y_val=y_val,
    )
    # Step 6.3: verify the fitted monomial on a deterministic sample
    verification: Dict[str, Any] = verify_trained_monomial(
        alpha=train_result.alpha,
        beta=train_result.beta,
        X_sample=X_train[:VERIFICATION_SAMPLE_ROWS],
        cfg=cfg,
        y_sample=y_train[:VERIFICATION_SAMPLE_ROWS],
    )
    # Construct the versioned artifact directory for fitted monomials
    archive_dir: Path = output_dir / "artifacts" / "monomial" / "v1"
    # Create the artifact directory when it does not exist yet
    archive_dir.mkdir(parents=True, exist_ok=True)
    # Construct the parameter record path for this dataset and seed
    params_path: Path = archive_dir / (
        f"{dataset_name.lower()}_seed{seed}_params.json"
    )
    # Serialise the fitted parameters with feature names attached
    params_path.write_text(
        json.dumps(
            {
                "dataset": dataset_name,
                "seed": seed,
                "alpha": train_result.alpha,
                "beta": train_result.beta.tolist(),
                "epochs_used": train_result.epochs_used,
                "converged": train_result.converged,
                "val_pr_auc": train_result.val_pr_auc,
                "written_at": datetime.now(timezone.utc).isoformat(),
            },
            indent=2,
            ensure_ascii=False,
        )
        + "\n",
        encoding="utf-8",
    )
    # Construct the loss-history record path for this dataset and seed
    history_path: Path = archive_dir / (
        f"{dataset_name.lower()}_seed{seed}_history.json"
    )
    # Serialise the per-epoch loss history to disk
    history_path.write_text(
        json.dumps(
            {
                "dataset": dataset_name,
                "seed": seed,
                "epochs_used": train_result.epochs_used,
                "converged": train_result.converged,
                "loss_history": train_result.history,
            },
            indent=2,
            ensure_ascii=False,
        )
        + "\n",
        encoding="utf-8",
    )
    # Assemble the persistence report
    persistence_report: Dict[str, Any] = {
        "params": str(params_path),
        "history": str(history_path),
    }
    # Return the training result, verification, and persistence report
    return {
        "train_result": train_result,
        "verification": verification,
        "persistence": persistence_report,
    }

````

````python
# Task 7: Post-hoc pruning and support-size selection

# ==============================================================================
# Task 7: Post-hoc pruning and support-size selection
# ==============================================================================

# Derived log-bound constant: L = max(|log 0.01|, |log 10.01|)
LOG_BOUND_L: float = max(abs(np.log(0.01)), abs(np.log(10.01)))

# Number of validation rows sampled for the pruning-bound verification
BOUND_SAMPLE_ROWS: int = 10_000

# Absolute violation threshold for the pruning-bound verification
BOUND_VIOLATION_TOLERANCE: float = 1e-6


# -----------------------------------------------------------------------------
# Task 7, Step 1: Rank features and evaluate candidate support sizes
# -----------------------------------------------------------------------------

def _pruned_logits(
    log_x: np.ndarray,
    alpha: float,
    beta: np.ndarray,
    retained: np.ndarray,
) -> np.ndarray:
    """
    Compute pruned monomial logits in log-space for a retained set.

    Implements z_pruned(x) = sign(alpha) * exp(log|alpha| +
    sum_{j in R} beta_j log x_j) for the retained set R.

    Parameters
    ----------
    log_x : np.ndarray, shape (n, m)
        Precomputed log of the scaled feature matrix.
    alpha : float
        Fitted monomial coefficient (unchanged by pruning).
    beta : np.ndarray, shape (m,)
        Full fitted exponent vector.
    retained : np.ndarray, shape (r,)
        Retained column positions (top-r by descending |beta|).

    Returns
    -------
    np.ndarray, shape (n,)
        Pruned monomial logits for every row.
    """
    # Compute the masked log-space score for the retained features
    log_score: np.ndarray = np.log(abs(alpha)) + log_x[:, retained] @ beta[retained]
    # Reconstruct the logits with the sign of alpha tracked separately
    return np.sign(alpha) * np.exp(log_score)


def evaluate_pruning_candidates(
    alpha: float,
    beta: np.ndarray,
    X_val: np.ndarray,
    y_val: np.ndarray,
    cfg: Dict[str, Any],
) -> Dict[str, Any]:
    """
    Rank features by |beta| and evaluate PR-AUC for every support size.

    Features are ranked by descending |beta_j| with ties broken by
    ascending index (a stable argsort on -|beta|). For each r in
    1..m the pruned monomial keeps the top-r features and its
    validation PR-AUC is computed from the log-space pruned logits.

    Parameters
    ----------
    alpha : float
        Fitted monomial coefficient.
    beta : np.ndarray, shape (m,)
        Full fitted exponent vector.
    X_val : np.ndarray, shape (n_val, m)
        Scaled validation feature matrix.
    y_val : np.ndarray, shape (n_val,)
        Validation labels.
    cfg : Dict[str, Any]
        Validated study configuration.

    Returns
    -------
    Dict[str, Any]
        Curve record with the ranking order and, for every r, the
        validation PR-AUC and the retained feature positions.

    Raises
    ------
    TypeError
        If `beta`, `X_val`, or `y_val` is not an ndarray.
    ValueError
        If the arrays misalign.
    """
    # Validate the exponent vector type
    if not isinstance(beta, np.ndarray):
        raise TypeError(f"beta must be an ndarray; received {type(beta).__name__}.")
    # Validate the validation matrix type
    if not isinstance(X_val, np.ndarray):
        raise TypeError(
            f"X_val must be an ndarray; received {type(X_val).__name__}."
        )
    # Validate the label vector type
    if not isinstance(y_val, np.ndarray):
        raise TypeError(
            f"y_val must be an ndarray; received {type(y_val).__name__}."
        )
    # Raise when the exponent vector is misaligned with the matrix
    if beta.shape[0] != X_val.shape[1]:
        raise ValueError(
            f"beta has {beta.shape[0]} entries but X_val has "
            f"{X_val.shape[1]} columns."
        )
    # Raise when the validation arrays disagree in length
    if X_val.shape[0] != y_val.shape[0]:
        raise ValueError(
            f"X_val has {X_val.shape[0]} rows but y_val has "
            f"{y_val.shape[0]} entries."
        )
    # Precompute the log of the scaled validation matrix once
    log_x: np.ndarray = np.log(X_val)
    # Rank the features by descending |beta| with stable tie-breaking
    order: np.ndarray = np.argsort(-np.abs(beta), kind="stable")
    # Initialise the per-r curve
    curve: Dict[int, Dict[str, Any]] = {}
    # Evaluate every candidate support size from 1 to m
    for r in range(1, beta.shape[0] + 1):
        # Take the top-r retained feature positions
        retained: np.ndarray = order[:r]
        # Compute the pruned monomial logits for the retained set
        pruned_logits: np.ndarray = _pruned_logits(
            log_x, alpha, beta, retained
        )
        # Convert the logits to probabilities via the sigmoid
        pruned_probabilities: np.ndarray = 1.0 / (1.0 + np.exp(-pruned_logits))
        # Compute the validation PR-AUC of the pruned monomial
        pr_auc: float = float(
            average_precision_score(y_val, pruned_probabilities)
        )
        # Record the curve entry for this support size
        curve[r] = {
            "pr_auc": pr_auc,
            "retained": retained.tolist(),
        }
    # Assemble the full curve record
    curve_record: Dict[str, Any] = {
        "order": order.tolist(),
        "curve": {str(r): entry for r, entry in curve.items()},
        "m": int(beta.shape[0]),
    }
    # Return the curve record
    return curve_record


# -----------------------------------------------------------------------------
# Task 7, Step 2: Select the smallest support within tolerance
# -----------------------------------------------------------------------------

def select_pruning_level(
    pr_auc_full: float,
    curve_record: Dict[str, Any],
    cfg: Dict[str, Any],
) -> Dict[str, Any]:
    """
    Select the smallest support whose validation PR-AUC is in tolerance.

    The selection rule is r* = min{r : PR-AUC_val(r) >= PR-AUC_val(full)
    - 0.01}, with the absolute tolerance read from the configuration.
    If no r < m satisfies the rule, r* = m with a logged warning.

    Parameters
    ----------
    pr_auc_full : float
        Validation PR-AUC of the full monomial.
    curve_record : Dict[str, Any]
        Curve record from evaluate_pruning_candidates.
    cfg : Dict[str, Any]
        Validated study configuration.

    Returns
    -------
    Dict[str, Any]
        Support record with r, retained and removed sets, retained
        exponents, and the selected validation PR-AUC.

    Raises
    ------
    ValueError
        If the curve does not cover r = 1..m.
    """
    # Read the absolute pruning tolerance from the configuration
    tolerance: float = float(cfg["post_hoc_pruning"]["tolerance"])
    # Extract the number of features from the curve record
    m: int = int(curve_record["m"])
    # Extract the per-r curve entries keyed by string
    curve: Dict[str, Dict[str, Any]] = curve_record["curve"]
    # Verify the curve covers every support size from 1 to m
    if set(curve.keys()) != {str(r) for r in range(1, m + 1)}:
        raise ValueError(
            f"Curve must cover r = 1..{m}; received keys "
            f"{sorted(curve.keys())}."
        )
    # Extract the feature ranking order from the curve record
    order: np.ndarray = np.array(curve_record["order"], dtype=int)
    # Initialise the selected support size as the full model
    selected_r: int = m
    # Scan the support sizes in ascending order
    for r in range(1, m + 1):
        # Read the validation PR-AUC of the candidate support
        candidate_pr_auc: float = float(curve[str(r)]["pr_auc"])
        # Select the first r that satisfies the tolerance rule
        if candidate_pr_auc >= pr_auc_full - tolerance:
            selected_r = r
            break
    # Warn when no reduced support satisfied the tolerance
    if selected_r == m and m > 1:
        warnings.warn(
            f"No reduced support satisfied the tolerance {tolerance}; "
            "retaining all m features.",
            RuntimeWarning,
        )
    # Extract the retained feature positions for the selected r
    retained: List[int] = [int(j) for j in curve[str(selected_r)]["retained"]]
    # Compute the removed set as the complement over 0..m-1
    removed: List[int] = [j for j in range(m) if j not in set(retained)]
    # Assemble the support record with the retained exponents
    support_record: Dict[str, Any] = {
        "r": selected_r,
        "retained": retained,
        "removed": removed,
        "order": order.tolist(),
        "pr_auc_full": float(pr_auc_full),
        "pr_auc_selected": float(curve[str(selected_r)]["pr_auc"]),
        "tolerance": tolerance,
    }
    # Return the selected support record
    return support_record


# -----------------------------------------------------------------------------
# Task 7, Step 3: Verify the worst-case pruning bound
# -----------------------------------------------------------------------------

def verify_pruning_bound(
    alpha: float,
    beta: np.ndarray,
    retained: Sequence[int],
    X_sample: np.ndarray,
    cfg: Dict[str, Any],
    seed: int = 0,
) -> Dict[str, Any]:
    """
    Verify the worst-case pruning bound on a deterministic sample.

    Verifies |s_full(x) - s_pruned(x)| <= L * sum_{j in R^c} |beta_j|
    for every sampled row, where the sum runs over the REMOVED set R^c
    and L = max(|log 0.01|, |log 10.01|). Uses validation rows only.

    Parameters
    ----------
    alpha : float
        Fitted monomial coefficient.
    beta : np.ndarray, shape (m,)
        Full fitted exponent vector.
    retained : Sequence[int]
        Retained feature positions (the selected support R).
    X_sample : np.ndarray, shape (n_sample, m)
        Scaled validation rows for the verification sample.
    cfg : Dict[str, Any]
        Validated study configuration.
    seed : int, optional
        Seed of the deterministic sampling, default 0.

    Returns
    -------
    Dict[str, Any]
        Verification record with max_gap, the bound, and the sample size.

    Raises
    ------
    ValueError
        If the bound is violated by more than 1e-6 on any sampled row.
    """
    # Validate the exponent vector type
    if not isinstance(beta, np.ndarray):
        raise TypeError(f"beta must be an ndarray; received {type(beta).__name__}.")
    # Validate the sample matrix type
    if not isinstance(X_sample, np.ndarray):
        raise TypeError(
            f"X_sample must be an ndarray; received {type(X_sample).__name__}."
        )
    # Convert the retained sequence to a sorted list of ints
    retained_list: List[int] = [int(j) for j in retained]
    # Compute the removed set as the complement over 0..m-1
    removed: List[int] = [
        j for j in range(beta.shape[0]) if j not in set(retained_list)
    ]
    # Draw the deterministic sample of at most 10,000 validation rows
    rng: np.random.Generator = np.random.default_rng(seed)
    # Compute the sample size as the minimum of the budget and rows
    sample_size: int = min(BOUND_SAMPLE_ROWS, X_sample.shape[0])
    # Draw the sample positions without replacement when possible
    sample_positions: np.ndarray = rng.choice(
        X_sample.shape[0], size=sample_size, replace=False
    )
    # Take the sampled rows from the validation matrix
    sample: np.ndarray = X_sample[sample_positions]
    # Precompute the log of the sampled rows
    log_x: np.ndarray = np.log(sample)
    # Compute the full log-score: log|alpha| + sum_j beta_j log x_j
    s_full: np.ndarray = np.log(abs(alpha)) + log_x @ beta
    # Compute the pruned log-score over the retained set only
    s_pruned: np.ndarray = (
        np.log(abs(alpha)) + log_x[:, retained_list] @ beta[retained_list]
    )
    # Compute the per-row absolute gap between the two log-scores
    gaps: np.ndarray = np.abs(s_full - s_pruned)
    # Compute the worst-case bound: L * sum_{j in R^c} |beta_j|
    bound: float = LOG_BOUND_L * float(np.sum(np.abs(beta[removed])))
    # Record the maximum realised gap over the sample
    max_gap: float = float(gaps.max())
    # Locate any row violating the bound beyond the tolerance
    violating_positions: np.ndarray = np.where(
        gaps > bound + BOUND_VIOLATION_TOLERANCE
    )[0]
    # Abort when any row violates the bound
    if violating_positions.shape[0] > 0:
        raise ValueError(
            f"Pruning bound violated on {violating_positions.shape[0]} "
            f"rows; first offending sample position "
            f"{int(violating_positions[0])} with gap "
            f"{float(gaps[violating_positions[0]]):.6e} against bound "
            f"{bound:.6e}."
        )
    # Assemble the verification record
    verification: Dict[str, Any] = {
        "max_gap": max_gap,
        "bound": bound,
        "sample_size": sample_size,
        "removed_sum_abs_beta": float(np.sum(np.abs(beta[removed]))),
        "log_bound_L": LOG_BOUND_L,
        "passed": True,
    }
    # Return the verification record
    return verification


# -----------------------------------------------------------------------------
# Task 7, Orchestrator Function
# -----------------------------------------------------------------------------

def orchestrate_pruning(
    alpha: float,
    beta: np.ndarray,
    X_val: np.ndarray,
    y_val: np.ndarray,
    cfg: Dict[str, Any],
    dataset_name: str,
    seed: int,
    output_dir: Path,
    feature_names: Sequence[str],
) -> Dict[str, Any]:
    """
    Orchestrate Task 7: evaluate, select, verify, and persist pruning.

    Parameters
    ----------
    alpha : float
        Fitted monomial coefficient from Task 6.
    beta : np.ndarray, shape (m,)
        Full fitted exponent vector from Task 6.
    X_val : np.ndarray, shape (n_val, m)
        Scaled validation feature matrix.
    y_val : np.ndarray, shape (n_val,)
        Validation labels.
    cfg : Dict[str, Any]
        Validated study configuration.
    dataset_name : str
        Dataset name for artifact naming.
    seed : int
        Seed for the deterministic bound-verification sample.
    output_dir : Path
        Project output root for the artifact archive.
    feature_names : Sequence[str]
        Feature names aligned with the matrix columns.

    Returns
    -------
    Dict[str, Any]
        Mapping with "curve", "support", "bound_verification", and
        "persistence" keys.
    """
    # Step 7.1: evaluate the validation PR-AUC curve over all supports
    curve_record: Dict[str, Any] = evaluate_pruning_candidates(
        alpha=alpha, beta=beta, X_val=X_val, y_val=y_val, cfg=cfg
    )
    # Read the full-monomial validation PR-AUC from the curve at r = m
    pr_auc_full: float = float(curve_record["curve"][str(beta.shape[0])]["pr_auc"])
    # Step 7.2: select the smallest support within the tolerance
    support_record: Dict[str, Any] = select_pruning_level(
        pr_auc_full=pr_auc_full, curve_record=curve_record, cfg=cfg
    )
    # Step 7.3: verify the worst-case pruning bound on a validation sample
    bound_verification: Dict[str, Any] = verify_pruning_bound(
        alpha=alpha,
        beta=beta,
        retained=support_record["retained"],
        X_sample=X_val,
        cfg=cfg,
        seed=seed,
    )
    # Construct the versioned artifact directory for pruning records
    archive_dir: Path = output_dir / "artifacts" / "pruning" / "v1"
    # Create the artifact directory when it does not exist yet
    archive_dir.mkdir(parents=True, exist_ok=True)
    # Extract the retained feature positions
    retained: List[int] = support_record["retained"]
    # Map the retained positions to feature names for the audit trail
    retained_names: List[str] = [str(feature_names[j]) for j in retained]
    # Extract the retained exponents in position order
    retained_exponents: List[float] = [float(beta[j]) for j in retained]
    # Construct the support record path for this dataset and seed
    support_path: Path = archive_dir / (
        f"support_{dataset_name.lower()}_seed{seed}.json"
    )
    # Serialise the support record with names and exponents attached
    support_path.write_text(
        json.dumps(
            {
                "dataset": dataset_name,
                "seed": seed,
                **support_record,
                "retained_names": retained_names,
                "retained_exponents": retained_exponents,
                "written_at": datetime.now(timezone.utc).isoformat(),
            },
            indent=2,
            ensure_ascii=False,
        )
        + "\n",
        encoding="utf-8",
    )
    # Construct the curve record path for this dataset and seed
    curve_path: Path = archive_dir / (
        f"curve_{dataset_name.lower()}_seed{seed}.json"
    )
    # Serialise the per-r PR-AUC curve to disk
    curve_path.write_text(
        json.dumps(
            {
                "dataset": dataset_name,
                "seed": seed,
                "curve": curve_record["curve"],
                "order": curve_record["order"],
            },
            indent=2,
            ensure_ascii=False,
        )
        + "\n",
        encoding="utf-8",
    )
    # Assemble the persistence report
    persistence_report: Dict[str, Any] = {
        "support": str(support_path),
        "curve": str(curve_path),
    }
    # Return the curve, support, verification, and persistence
    return {
        "curve": curve_record,
        "support": support_record,
        "bound_verification": bound_verification,
        "persistence": persistence_report,
    }
````

````python
# Task 8: Directional rule construction and closed-form fidelity prediction

# ==============================================================================
# Task 8: Directional rule construction and closed-form fidelity prediction
# ==============================================================================

# -----------------------------------------------------------------------------
# Task 8, Step 1: Build the sign-flattened vector and directional scores
# -----------------------------------------------------------------------------

def build_directional_rule(
    alpha: float,
    beta: np.ndarray,
    retained: Sequence[int],
    X: np.ndarray,
) -> Tuple[np.ndarray, np.ndarray, Dict[str, str]]:
    """
    Build the sign-flattened direction vector and directional scores.

    The direction vector is d_j = sign(beta_j) for j in the retained set
    and 0 otherwise. The directional score is
        s_dir(x) = sum_{j in S} sign(beta_j) log x_j,
    computed in log-space on scaled features, discarding magnitudes and
    the intercept (which is irrelevant to the ranking).

    Parameters
    ----------
    alpha : float
        Fitted monomial coefficient (determines the language mapping).
    beta : np.ndarray, shape (m,)
        Full fitted exponent vector.
    retained : Sequence[int]
        Retained feature positions (the selected support S).
    X : np.ndarray, shape (n, m)
        Scaled feature matrix in [0.01, 10.01].

    Returns
    -------
    Tuple[np.ndarray, np.ndarray, Dict[str, str]]
        d of shape (m,); directional scores of shape (n,); and the
        natural-language rule mapping keyed by feature position.

    Raises
    ------
    TypeError
        If `beta` or `X` is not an ndarray.
    ValueError
        If a retained position is out of range or X is outside the
        scaled range.
    """
    # Validate the exponent vector type
    if not isinstance(beta, np.ndarray):
        raise TypeError(f"beta must be an ndarray; received {type(beta).__name__}.")
    # Validate the feature matrix type
    if not isinstance(X, np.ndarray):
        raise TypeError(f"X must be an ndarray; received {type(X).__name__}.")
    # Raise when the retained set is misaligned with the exponent vector
    if any(j < 0 or j >= beta.shape[0] for j in retained):
        raise ValueError(
            f"Retained positions must lie in 0..{beta.shape[0] - 1}; "
            f"received {list(retained)}."
        )
    # Assert the scaled-range precondition required by the logarithm
    if not (np.all(X >= 0.01) and np.all(X <= 10.01)):
        raise ValueError(
            "X must lie in [0.01, 10.01] for the log transform."
        )
    # Convert the retained sequence to a set for membership tests
    retained_set: set = set(int(j) for j in retained)
    # Build the direction vector with sign(beta_j) on S and 0 elsewhere
    d: np.ndarray = np.zeros(beta.shape[0], dtype=np.float64)
    # Iterate over every feature position to assign its sign
    for j in range(beta.shape[0]):
        # Assign the sign of the exponent when the feature is retained
        if j in retained_set:
            d[j] = np.sign(beta[j])
    # Compute the log of the scaled features
    log_x: np.ndarray = np.log(X)
    # Compute the directional score: sum_{j in S} sign(beta_j) log x_j
    directional_scores: np.ndarray = log_x @ d
    # Build the natural-language mapping from the sign of alpha*beta_j
    rule_mapping: Dict[str, str] = {}
    # Iterate over the retained features in ascending position order
    for j in sorted(int(jj) for jj in retained):
        # Determine the direction from the product of alpha and beta_j
        direction: float = np.sign(alpha * beta[j])
        # State whether high or low values increase risk
        if direction > 0:
            rule_mapping[str(j)] = f"feature {j}: high values increase risk"
        else:
            rule_mapping[str(j)] = f"feature {j}: low values increase risk"
    # Return the direction vector, scores, and language mapping
    return d, directional_scores, rule_mapping


# -----------------------------------------------------------------------------
# Task 8, Step 2: Compute the Pearson correlation and Greiner prediction
# -----------------------------------------------------------------------------

def predict_fidelity(
    beta: np.ndarray,
    d: np.ndarray,
    X_train: np.ndarray,
    cfg: Dict[str, Any],
) -> Dict[str, Any]:
    """
    Predict directional-rule rank fidelity from training data alone.

    Computes the sample covariance of the log-features with n-1 degrees
    of freedom and the Pearson correlation between the full and
    directional log-scores:
        rho = (beta^T Sigma d) / (sqrt(beta^T Sigma beta)
              sqrt(d^T Sigma d)),
    then applies Greiner's relation tau_pred = (2/pi) arcsin(rho).
    The FULL fitted beta enters the formula; d is zero outside the
    retained set.

    Parameters
    ----------
    beta : np.ndarray, shape (m,)
        Full fitted exponent vector.
    d : np.ndarray, shape (m,)
        Sign-flattened direction vector from build_directional_rule.
    X_train : np.ndarray, shape (n_train, m)
        Scaled training feature matrix (training rows only).
    cfg : Dict[str, Any]
        Validated study configuration.

    Returns
    -------
    Dict[str, Any]
        Prediction record with rho (clipped to [-1, 1]) and tau_pred.

    Raises
    ------
    TypeError
        If `beta`, `d`, or `X_train` is not an ndarray.
    ValueError
        If the vectors misalign with the matrix columns.
    """
    # Validate the exponent vector type
    if not isinstance(beta, np.ndarray):
        raise TypeError(f"beta must be an ndarray; received {type(beta).__name__}.")
    # Validate the direction vector type
    if not isinstance(d, np.ndarray):
        raise TypeError(f"d must be an ndarray; received {type(d).__name__}.")
    # Validate the training matrix type
    if not isinstance(X_train, np.ndarray):
        raise TypeError(
            f"X_train must be an ndarray; received {type(X_train).__name__}."
        )
    # Raise when the vectors misalign with the matrix columns
    if not (beta.shape[0] == d.shape[0] == X_train.shape[1]):
        raise ValueError(
            f"beta ({beta.shape[0]}), d ({d.shape[0]}), and X_train "
            f"columns ({X_train.shape[1]}) must agree."
        )
    # Assert the scaled-range precondition required by the logarithm
    if not (np.all(X_train >= 0.01) and np.all(X_train <= 10.01)):
        raise ValueError(
            "X_train must lie in [0.01, 10.01] for the log transform."
        )
    # Compute the log-feature matrix: w_i = log x_i
    w: np.ndarray = np.log(X_train)
    # Compute the sample covariance with n-1 degrees of freedom
    sigma: np.ndarray = np.cov(w, rowvar=False, ddof=1)
    # Compute the quadratic form beta^T Sigma beta
    beta_sigma_beta: float = float(
        np.linalg.multi_dot([beta, sigma, beta])
    )
    # Compute the quadratic form d^T Sigma d
    d_sigma_d: float = float(np.linalg.multi_dot([d, sigma, d]))
    # Compute the cross form beta^T Sigma d
    beta_sigma_d: float = float(np.linalg.multi_dot([beta, sigma, d]))
    # Guard against a zero denominator (a constant linear score)
    if beta_sigma_beta <= 0.0 or d_sigma_d <= 0.0:
        # Warn about the degenerate denominator
        warnings.warn(
            "Zero variance detected in the linear scores; setting rho to 0.",
            RuntimeWarning,
        )
        # Record the degenerate prediction explicitly
        return {"rho": 0.0, "tau_pred": 0.0, "degenerate": True}
    # Compute the Pearson correlation between the two linear scores
    rho_raw: float = beta_sigma_d / (
        np.sqrt(beta_sigma_beta) * np.sqrt(d_sigma_d)
    )
    # Clip rho to [-1, 1] when numerical rounding pushed it outside
    rho: float = float(np.clip(rho_raw, -1.0, 1.0))
    # Log the clipping event when it occurred
    if abs(rho_raw) > 1.0:
        warnings.warn(
            f"rho computed as {rho_raw:.6e} outside [-1, 1]; clipped to "
            f"{rho}.", RuntimeWarning,
        )
    # Apply Greiner's relation: tau_pred = (2/pi) arcsin(rho)
    tau_pred: float = (2.0 / np.pi) * np.arcsin(rho)
    # Assemble the prediction record
    prediction: Dict[str, Any] = {
        "rho": rho,
        "tau_pred": float(tau_pred),
        "degenerate": False,
    }
    # Return the prediction record
    return prediction


# -----------------------------------------------------------------------------
# Task 8, Step 3: Record the prediction per seed
# -----------------------------------------------------------------------------

def record_fidelity_predictions(
    dataset_name: str,
    seed: int,
    prediction: Dict[str, Any],
    retained: Sequence[int],
    d: np.ndarray,
    feature_names: Sequence[str],
    output_dir: Path,
) -> Path:
    """
    Record the a-priori fidelity prediction per dataset and seed.

    The artifact is written BEFORE the directional rule is evaluated on
    any held-out data, so the prediction is genuinely a priori.

    Parameters
    ----------
    dataset_name : str
        Dataset name for the artifact name.
    seed : int
        Seed for the artifact name.
    prediction : Dict[str, Any]
        Prediction record from predict_fidelity.
    retained : Sequence[int]
        Retained feature positions.
    d : np.ndarray, shape (m,)
        Sign-flattened direction vector.
    feature_names : Sequence[str]
        Feature names aligned with the direction vector.
    output_dir : Path
        Project output root for the artifact archive.

    Returns
    -------
    Path
        Path of the written prediction artifact.
    """
    # Validate the direction vector type
    if not isinstance(d, np.ndarray):
        raise TypeError(f"d must be an ndarray; received {type(d).__name__}.")
    # Construct the versioned artifact directory
    archive_dir: Path = output_dir / "artifacts" / "fidelity_prediction" / "v1"
    # Create the artifact directory when it does not exist yet
    archive_dir.mkdir(parents=True, exist_ok=True)
    # Map the retained positions to feature names
    retained_names: List[str] = [str(feature_names[j]) for j in retained]
    # Compose the full prediction artifact
    artifact: Dict[str, Any] = {
        "dataset": dataset_name,
        "seed": seed,
        "rho": float(prediction["rho"]),
        "tau_pred": float(prediction["tau_pred"]),
        "degenerate": bool(prediction.get("degenerate", False)),
        "retained": [int(j) for j in retained],
        "retained_names": retained_names,
        "sign_vector_d": [float(v) for v in d.tolist()],
        "feature_names": list(feature_names),
        "written_at": datetime.now(timezone.utc).isoformat(),
    }
    # Construct the artifact path for this dataset and seed
    artifact_path: Path = archive_dir / (
        f"fidelity_prediction_{dataset_name.lower()}_seed{seed}.json"
    )
    # Serialise the artifact with native types and stable formatting
    artifact_path.write_text(
        json.dumps(artifact, indent=2, ensure_ascii=False) + "\n",
        encoding="utf-8",
    )
    # Return the path of the written artifact
    return artifact_path


# -----------------------------------------------------------------------------
# Task 8, Orchestrator Function
# -----------------------------------------------------------------------------

def orchestrate_directional_rule(
    alpha: float,
    beta: np.ndarray,
    retained: Sequence[int],
    X_train: np.ndarray,
    X_val: np.ndarray,
    X_test: np.ndarray,
    cfg: Dict[str, Any],
    dataset_name: str,
    seed: int,
    feature_names: Sequence[str],
    output_dir: Path,
) -> Dict[str, Any]:
    """
    Orchestrate Task 8: build the rule, predict fidelity, and record.

    The fidelity prediction is computed from the fitted exponents and
    the training covariance only and is recorded BEFORE any held-out
    evaluation of the directional rule.

    Parameters
    ----------
    alpha : float
        Fitted monomial coefficient.
    beta : np.ndarray, shape (m,)
        Full fitted exponent vector.
    retained : Sequence[int]
        Retained feature positions from Task 7.
    X_train : np.ndarray, shape (n_train, m)
        Scaled training feature matrix.
    X_val : np.ndarray, shape (n_val, m)
        Scaled validation feature matrix.
    X_test : np.ndarray, shape (n_test, m)
        Scaled test feature matrix.
    cfg : Dict[str, Any]
        Validated study configuration.
    dataset_name : str
        Dataset name for artifact naming.
    seed : int
        Seed for artifact naming.
    feature_names : Sequence[str]
        Feature names aligned with the matrix columns.
    output_dir : Path
        Project output root for the artifact archive.

    Returns
    -------
    Dict[str, Any]
        Mapping with "d", "scores", "rule_mapping", "prediction", and
        "persistence" keys.
    """
    # Step 8.1: build the direction vector and scores on all partitions
    d, scores_val, rule_mapping = build_directional_rule(
        alpha=alpha, beta=beta, retained=retained, X=X_val
    )
    # Compute the directional scores on the test partition
    _, scores_test, _ = build_directional_rule(
        alpha=alpha, beta=beta, retained=retained, X=X_test
    )
    # Step 8.2: predict fidelity from the training covariance only
    prediction: Dict[str, Any] = predict_fidelity(
        beta=beta, d=d, X_train=X_train, cfg=cfg
    )
    # Step 8.3: record the a-priori prediction before any evaluation
    artifact_path: Path = record_fidelity_predictions(
        dataset_name=dataset_name,
        seed=seed,
        prediction=prediction,
        retained=retained,
        d=d,
        feature_names=feature_names,
        output_dir=output_dir,
    )
    # Assemble the full Task 8 result structure
    return {
        "d": d,
        "scores": {"val": scores_val, "test": scores_test},
        "rule_mapping": rule_mapping,
        "prediction": prediction,
        "persistence": {"prediction_artifact": str(artifact_path)},
    }
````

````python
# Task 9: Scorecard and tally construction with validation-threshold selection

# ==============================================================================
# Task 9: Scorecard and tally construction with validation-threshold selection
# ==============================================================================

# -----------------------------------------------------------------------------
# Task 9, Step 1: Median binarization and indicator construction
# -----------------------------------------------------------------------------

def build_binary_conditions(
    retained: Sequence[int],
    alpha: float,
    beta: np.ndarray,
    X_train: np.ndarray,
    X_val: np.ndarray,
    X_test: np.ndarray,
    cfg: Dict[str, Any],
) -> Dict[str, Any]:
    """
    Build median-binarized risk indicators for the retained features.

    For each retained feature j the cutpoint is the training median,
    except for binary features (training values exactly {0, 1}) where
    the midpoint 0.5 is used. The condition side is determined by the
    sign of alpha * beta_j: x_j >= c_j when positive, x_j <= c_j when
    negative.

    Parameters
    ----------
    retained : Sequence[int]
        Retained feature positions (the selected support S).
    alpha : float
        Fitted monomial coefficient.
    beta : np.ndarray, shape (m,)
        Full fitted exponent vector.
    X_train : np.ndarray, shape (n_train, m)
        Scaled training matrix (used ONLY for the median cutpoints).
    X_val : np.ndarray, shape (n_val, m)
        Scaled validation matrix.
    X_test : np.ndarray, shape (n_test, m)
        Scaled test matrix.
    cfg : Dict[str, Any]
        Validated study configuration.

    Returns
    -------
    Dict[str, Any]
        Mapping with "cutpoints", "sides", "indicators_val", and
        "indicators_test".

    Raises
    ------
    TypeError
        If any matrix is not an ndarray.
    ValueError
        If a retained position is out of range.
    """
    # Validate the exponent vector type
    if not isinstance(beta, np.ndarray):
        raise TypeError(f"beta must be an ndarray; received {type(beta).__name__}.")
    # Validate the matrix types
    for name, matrix in {
        "X_train": X_train, "X_val": X_val, "X_test": X_test,
    }.items():
        # Raise when any matrix is not an ndarray
        if not isinstance(matrix, np.ndarray):
            raise TypeError(
                f"{name} must be an ndarray; received {type(matrix).__name__}."
            )
    # Raise when a retained position is out of range
    if any(j < 0 or j >= beta.shape[0] for j in retained):
        raise ValueError(
            f"Retained positions must lie in 0..{beta.shape[0] - 1}; "
            f"received {list(retained)}."
        )
    # Convert the retained sequence to a sorted list
    retained_list: List[int] = sorted(int(j) for j in retained)
    # Initialise the cutpoint and side dictionaries keyed by position
    cutpoints: Dict[str, float] = {}
    sides: Dict[str, str] = {}
    # Iterate over the retained features in ascending position order
    for j in retained_list:
        # Extract the training values of the feature
        train_column: np.ndarray = X_train[:, j]
        # Detect binary features: training values exactly {0, 1}
        unique_values: set = set(np.unique(train_column))
        # Use the midpoint 0.5 for binary features
        if unique_values.issubset({0.0, 1.0}) and len(unique_values) <= 2:
            cutpoints[str(j)] = 0.5
        else:
            # Use the training median for continuous features
            cutpoints[str(j)] = float(np.median(train_column))
        # Determine the risk-increasing side from the product alpha*beta_j
        if np.sign(alpha * beta[j]) > 0:
            sides[str(j)] = "ge"
        else:
            sides[str(j)] = "le"
        # Warn when a retained feature is constant in training
        if float(np.ptp(train_column)) == 0.0:
            warnings.warn(
                f"Retained feature {j} is constant in training; its "
                "indicator is degenerate.",
                RuntimeWarning,
            )
    # Assemble the validation indicator matrix
    indicators_val: np.ndarray = np.zeros(
        (X_val.shape[0], len(retained_list)), dtype=np.int64
    )
    # Assemble the test indicator matrix
    indicators_test: np.ndarray = np.zeros(
        (X_test.shape[0], len(retained_list)), dtype=np.int64
    )
    # Fill the indicator matrices column by column
    for column_index, j in enumerate(retained_list):
        # Read the cutpoint of the feature
        c_j: float = cutpoints[str(j)]
        # Evaluate the closed indicator for the positive side
        if sides[str(j)] == "ge":
            indicators_val[:, column_index] = (X_val[:, j] >= c_j).astype(
                np.int64
            )
            indicators_test[:, column_index] = (X_test[:, j] >= c_j).astype(
                np.int64
            )
        else:
            indicators_val[:, column_index] = (X_val[:, j] <= c_j).astype(
                np.int64
            )
            indicators_test[:, column_index] = (X_test[:, j] <= c_j).astype(
                np.int64
            )
    # Return the cutpoints, sides, and indicator matrices
    return {
        "cutpoints": cutpoints,
        "sides": sides,
        "indicators_val": indicators_val,
        "indicators_test": indicators_test,
    }


# -----------------------------------------------------------------------------
# Task 9, Step 2: Scorecard point assignment and tally scores
# -----------------------------------------------------------------------------

def assign_scorecard_points(
    beta: np.ndarray,
    retained: Sequence[int],
    cfg: Dict[str, Any],
) -> np.ndarray:
    """
    Assign integer scorecard points to the retained features.

    Points are p_j = max(1, round(P_max * |beta_j| / max_k |beta_k|))
    with P_max read from the configuration. The rounding convention is
    round-half-away-from-zero, implemented as floor(v + 0.5) because the
    arguments are non-negative; Python's built-in round (banker's
    rounding) is deliberately NOT used.

    Parameters
    ----------
    beta : np.ndarray, shape (m,)
        Full fitted exponent vector.
    retained : Sequence[int]
        Retained feature positions.
    cfg : Dict[str, Any]
        Validated study configuration.

    Returns
    -------
    np.ndarray, shape (r,)
        Integer point vector aligned to the retained order.

    Raises
    ------
    TypeError
        If `beta` is not an ndarray.
    ValueError
        If the maximum retained exponent magnitude is zero.
    """
    # Validate the exponent vector type
    if not isinstance(beta, np.ndarray):
        raise TypeError(f"beta must be an ndarray; received {type(beta).__name__}.")
    # Read the maximum points from the configuration
    p_max: int = int(cfg["scorecard_and_tally"]["max_points"])
    # Convert the retained sequence to a list of positions
    retained_list: List[int] = [int(j) for j in retained]
    # Extract the retained exponent magnitudes
    retained_magnitudes: np.ndarray = np.abs(beta[retained_list])
    # Compute the maximum retained magnitude
    max_magnitude: float = float(retained_magnitudes.max())
    # Raise when the maximum magnitude is zero (undefined denominator)
    if max_magnitude == 0.0:
        raise ValueError(
            "Maximum retained exponent magnitude is zero; the point "
            "formula is undefined."
        )
    # Compute the scaled point values before rounding
    scaled_points: np.ndarray = p_max * retained_magnitudes / max_magnitude
    # Apply round-half-away-from-zero via floor(v + 0.5)
    rounded_points: np.ndarray = np.floor(scaled_points + 0.5).astype(np.int64)
    # Floor every point at one so the support is fully represented
    points: np.ndarray = np.maximum(rounded_points, 1)
    # Return the integer point vector
    return points


def scorecard_score(
    indicators: np.ndarray, points: np.ndarray
) -> np.ndarray:
    """
    Compute the scorecard score V(x) = sum_j p_j 1_j(x).

    Parameters
    ----------
    indicators : np.ndarray, shape (n, r)
        Binary indicator matrix.
    points : np.ndarray, shape (r,)
        Integer point vector aligned to the indicator columns.

    Returns
    -------
    np.ndarray, shape (n,)
        Integer scorecard scores.

    Raises
    ------
    ValueError
        If the indicator columns and the point vector misalign.
    """
    # Raise when the indicator columns and points misalign
    if indicators.shape[1] != points.shape[0]:
        raise ValueError(
            f"indicators has {indicators.shape[1]} columns but points "
            f"has {points.shape[0]} entries."
        )
    # Compute the weighted sum of the indicators
    scores: np.ndarray = indicators @ points
    # Return the integer scorecard scores
    return scores


def tally_score(indicators: np.ndarray) -> np.ndarray:
    """
    Compute the tally score T(x) = sum_j 1_j(x).

    Parameters
    ----------
    indicators : np.ndarray, shape (n, r)
        Binary indicator matrix.

    Returns
    -------
    np.ndarray, shape (n,)
        Integer tally scores.
    """
    # Sum the indicators across the retained features
    scores: np.ndarray = indicators.sum(axis=1)
    # Return the integer tally scores
    return scores


# -----------------------------------------------------------------------------
# Task 9, Step 3: Select thresholds on validation and apply to test
# -----------------------------------------------------------------------------

def select_thresholds(
    V_val: np.ndarray,
    T_val: np.ndarray,
    y_val: np.ndarray,
    V_test: np.ndarray,
    T_test: np.ndarray,
    y_test: np.ndarray,
    cfg: Dict[str, Any],
    points: np.ndarray,
) -> Dict[str, Any]:
    """
    Select scorecard and tally thresholds on validation, apply to test.

    The search is exhaustive over v in 1..sum(p_j) and t in 1..|S| with
    the decision rule flag when score >= threshold, choosing the
    threshold that maximizes validation F1. Ties are broken by choosing
    the SMALLEST threshold attaining the maximum (more permissive),
    documented as the disambiguated choice. Threshold 0 is excluded.

    Parameters
    ----------
    V_val : np.ndarray, shape (n_val,)
        Validation scorecard scores.
    T_val : np.ndarray, shape (n_val,)
        Validation tally scores.
    y_val : np.ndarray, shape (n_val,)
        Validation labels.
    V_test : np.ndarray, shape (n_test,)
        Test scorecard scores.
    T_test : np.ndarray, shape (n_test,)
        Test tally scores.
    y_test : np.ndarray, shape (n_test,)
        Test labels.
    cfg : Dict[str, Any]
        Validated study configuration.
    points : np.ndarray, shape (r,)
        Scorecard point vector (determines the scorecard range).

    Returns
    -------
    Dict[str, Any]
        Mapping with the selected thresholds, validation F1 values, and
        the test binary predictions.

    Raises
    ------
    ValueError
        If the score and label vectors misalign.
    """
    # Raise when the validation score vectors misalign with the labels
    if not (V_val.shape[0] == T_val.shape[0] == y_val.shape[0]):
        raise ValueError(
            "Validation scores and labels must agree in length."
        )
    # Raise when the test score vectors misalign with the labels
    if not (V_test.shape[0] == T_test.shape[0] == y_test.shape[0]):
        raise ValueError("Test scores and labels must agree in length.")
    # Compute the maximum scorecard threshold (sum of points)
    max_v: int = int(points.sum())
    # Compute the maximum tally threshold (number of retained features)
    max_t: int = int(points.shape[0])
    # Initialise the best scorecard threshold and its F1
    best_v: int = 1
    best_v_f1: float = -1.0
    # Exhaustively scan every integer scorecard threshold from 1 to max_v
    for v in range(1, max_v + 1):
        # Compute the validation F1 of the candidate threshold
        candidate_f1: float = float(f1_score(y_val, V_val >= v))
        # Update the best threshold when F1 improves or ties with a
        # smaller threshold (the smallest-threshold tie-break)
        if candidate_f1 > best_v_f1:
            best_v = v
            best_v_f1 = candidate_f1
    # Initialise the best tally threshold and its F1
    best_t: int = 1
    best_t_f1: float = -1.0
    # Exhaustively scan every integer tally threshold from 1 to max_t
    for t in range(1, max_t + 1):
        # Compute the validation F1 of the candidate threshold
        candidate_f1 = float(f1_score(y_val, T_val >= t))
        # Update the best threshold when F1 improves or ties with a
        # smaller threshold (the smallest-threshold tie-break)
        if candidate_f1 > best_t_f1:
            best_t = t
            best_t_f1 = candidate_f1
    # Apply the selected scorecard threshold to the test scores once
    scorecard_test_predictions: np.ndarray = (V_test >= best_v).astype(np.int64)
    # Apply the selected tally threshold to the test scores once
    tally_test_predictions: np.ndarray = (T_test >= best_t).astype(np.int64)
    # Assemble the threshold selection record
    selection: Dict[str, Any] = {
        "scorecard_threshold": best_v,
        "scorecard_val_f1": best_v_f1,
        "tally_threshold": best_t,
        "tally_val_f1": best_t_f1,
        "scorecard_test_predictions": scorecard_test_predictions,
        "tally_test_predictions": tally_test_predictions,
        "tie_break": "smallest threshold attaining max validation F1",
    }
    # Return the selection record
    return selection


# -----------------------------------------------------------------------------
# Task 9, Orchestrator Function
# -----------------------------------------------------------------------------

def orchestrate_scorecard_tally(
    alpha: float,
    beta: np.ndarray,
    retained: Sequence[int],
    X_train: np.ndarray,
    X_val: np.ndarray,
    X_test: np.ndarray,
    y_val: np.ndarray,
    y_test: np.ndarray,
    cfg: Dict[str, Any],
    dataset_name: str,
    seed: int,
    feature_names: Sequence[str],
    output_dir: Path,
) -> Dict[str, Any]:
    """
    Orchestrate Task 9: binarize, score, select thresholds, persist.

    Parameters
    ----------
    alpha : float
        Fitted monomial coefficient.
    beta : np.ndarray, shape (m,)
        Full fitted exponent vector.
    retained : Sequence[int]
        Retained feature positions from Task 7.
    X_train : np.ndarray, shape (n_train, m)
        Scaled training matrix (median cutpoints only).
    X_val : np.ndarray, shape (n_val, m)
        Scaled validation matrix.
    X_test : np.ndarray, shape (n_test, m)
        Scaled test matrix.
    y_val : np.ndarray, shape (n_val,)
        Validation labels.
    y_test : np.ndarray, shape (n_test,)
        Test labels.
    cfg : Dict[str, Any]
        Validated study configuration.
    dataset_name : str
        Dataset name for artifact naming.
    seed : int
        Seed for artifact naming.
    feature_names : Sequence[str]
        Feature names aligned with the matrix columns.
    output_dir : Path
        Project output root for the artifact archive.

    Returns
    -------
    Dict[str, Any]
        Mapping with "conditions", "points", "scores", "selection", and
        "persistence" keys.
    """
    # Step 9.1: build the median-binarized indicators for val and test
    conditions: Dict[str, Any] = build_binary_conditions(
        retained=retained,
        alpha=alpha,
        beta=beta,
        X_train=X_train,
        X_val=X_val,
        X_test=X_test,
        cfg=cfg,
    )
    # Step 9.2: assign the integer scorecard points
    points: np.ndarray = assign_scorecard_points(
        beta=beta, retained=retained, cfg=cfg
    )
    # Compute the validation scorecard scores
    V_val: np.ndarray = scorecard_score(
        conditions["indicators_val"], points
    )
    # Compute the test scorecard scores
    V_test: np.ndarray = scorecard_score(
        conditions["indicators_test"], points
    )
    # Compute the validation tally scores
    T_val: np.ndarray = tally_score(conditions["indicators_val"])
    # Compute the test tally scores
    T_test: np.ndarray = tally_score(conditions["indicators_test"])
    # Step 9.3: select the thresholds on validation and apply to test
    selection: Dict[str, Any] = select_thresholds(
        V_val=V_val,
        T_val=T_val,
        y_val=y_val,
        V_test=V_test,
        T_test=T_test,
        y_test=y_test,
        cfg=cfg,
        points=points,
    )
    # Construct the versioned artifact directory
    archive_dir: Path = output_dir / "artifacts" / "scorecard" / "v1"
    # Create the artifact directory when it does not exist yet
    archive_dir.mkdir(parents=True, exist_ok=True)
    # Map the retained positions to feature names
    retained_list: List[int] = [int(j) for j in retained]
    # Map the points to their feature names for the audit trail
    point_names: List[str] = [str(feature_names[j]) for j in retained_list]
    # Construct the scorecard record path for this dataset and seed
    record_path: Path = archive_dir / (
        f"scorecard_{dataset_name.lower()}_seed{seed}.json"
    )
    # Serialise the scorecard record with names and cutpoints
    record_path.write_text(
        json.dumps(
            {
                "dataset": dataset_name,
                "seed": seed,
                "retained": retained_list,
                "retained_names": point_names,
                "points": points.tolist(),
                "cutpoints": conditions["cutpoints"],
                "sides": conditions["sides"],
                "scorecard_threshold": selection["scorecard_threshold"],
                "scorecard_val_f1": selection["scorecard_val_f1"],
                "tally_threshold": selection["tally_threshold"],
                "tally_val_f1": selection["tally_val_f1"],
                "tie_break": selection["tie_break"],
                "written_at": datetime.now(timezone.utc).isoformat(),
            },
            indent=2,
            ensure_ascii=False,
        )
        + "\n",
        encoding="utf-8",
    )
    # Persist the indicator matrices and scores as a compressed npz
    npz_path: Path = archive_dir / (
        f"scorecard_{dataset_name.lower()}_seed{seed}_scores.npz"
    )
    # Save the indicators, scores, and predictions into the npz artifact
    np.savez_compressed(
        npz_path,
        indicators_val=conditions["indicators_val"],
        indicators_test=conditions["indicators_test"],
        V_val=V_val,
        V_test=V_test,
        T_val=T_val,
        T_test=T_test,
        scorecard_test_predictions=selection["scorecard_test_predictions"],
        tally_test_predictions=selection["tally_test_predictions"],
    )
    # Assemble the full Task 9 result structure
    return {
        "conditions": conditions,
        "points": points,
        "scores": {
            "V_val": V_val, "V_test": V_test,
            "T_val": T_val, "T_test": T_test,
        },
        "selection": selection,
        "persistence": {"record": str(record_path), "scores": str(npz_path)},
    }
````

````python
# Task 10: Isotonic calibration and expected-calibration-error computation

# ==============================================================================
# Task 10: Isotonic calibration and expected-calibration-error computation
# ==============================================================================

# The five calibrated representations (uniform calibration mechanism)
REPRESENTATIONS: Tuple[str, ...] = (
    "full", "pruned", "directional", "scorecard", "tally",
)


# -----------------------------------------------------------------------------
# Task 10, Step 1: Fit isotonic regression on the validation split
# -----------------------------------------------------------------------------

@dataclass
class CalibratorFit:
    """
    Calibrator fit on the validation split, with constant fallback.

    Wraps the isotonic regression mapping ``f: score -> [0,1]`` fitted on
    validation scores per Task 10 Step 10.1. When the validation score is
    constant, the isotonic fit is ill-defined and the fallback predicts the
    validation positive rate for every input.

    Attributes
    ----------
    kind : str
        ``"isotonic"`` when a monotone ``IsotonicRegression`` was fitted,
        ``"constant"`` when the validation score was constant and the base
        rate is used.
    model : Optional[IsotonicRegression]
        The fitted isotonic regressor; ``None`` for the constant fallback.
        When present, ``model.predict`` is monotone non-decreasing with
        ``out_of_bounds='clip'``.
    constant_value : Optional[float]
        The validation positive rate used by the constant fallback;
        ``None`` when ``kind`` is ``"isotonic"``.
    """

    # "isotonic" or "constant"
    kind: str
    # Fitted isotonic regressor (None for the constant fallback)
    model: Optional[IsotonicRegression]
    # Validation positive rate used by the constant fallback
    constant_value: Optional[float]

    def predict(self, scores: np.ndarray) -> np.ndarray:
        """
        Map raw scores to calibrated probabilities.

        Parameters
        ----------
        scores : np.ndarray, shape (n,)
            Raw scores of one representation.

        Returns
        -------
        np.ndarray, shape (n,)
            Calibrated probabilities in [0, 1].
        """
        # Apply the constant fallback when the calibrator is constant
        if self.kind == "constant":
            # Broadcast the constant probability to every score
            return np.full(scores.shape[0], self.constant_value)
        # Apply the fitted isotonic mapping with out-of-range clipping
        return self.model.predict(scores)


def fit_isotonic_calibrator(
    score_val: np.ndarray,
    y_val: np.ndarray,
) -> CalibratorFit:
    """
    Fit an isotonic calibration mapping on the validation split.

    The mapping is monotone non-decreasing (higher scores imply higher
    predicted risk) via the pooled-adjacent-violators algorithm. A
    constant validation score yields the explicit constant fallback
    equal to the validation positive rate.

    Parameters
    ----------
    score_val : np.ndarray, shape (n_val,)
        Validation scores of one representation.
    y_val : np.ndarray, shape (n_val,)
        Validation labels.

    Returns
    -------
    CalibratorFit
        The fitted calibrator with its kind and model or constant value.

    Raises
    ------
    TypeError
        If either input is not an ndarray.
    ValueError
        If the arrays disagree in length or the labels are not binary.
    """
    # Validate the score vector type
    if not isinstance(score_val, np.ndarray):
        raise TypeError(
            f"score_val must be an ndarray; received {type(score_val).__name__}."
        )
    # Validate the label vector type
    if not isinstance(y_val, np.ndarray):
        raise TypeError(
            f"y_val must be an ndarray; received {type(y_val).__name__}."
        )
    # Raise when the arrays disagree in length
    if score_val.shape[0] != y_val.shape[0]:
        raise ValueError(
            f"score_val has {score_val.shape[0]} entries but y_val has "
            f"{y_val.shape[0]} entries."
        )
    # Raise when the labels are not binary
    if not set(np.unique(y_val)).issubset({0, 1}):
        raise ValueError("y_val must contain only the binary labels 0 and 1.")
    # Detect a constant validation score
    if float(np.ptp(score_val)) == 0.0:
        # Warn about the constant-score fallback
        warnings.warn(
            "Validation score is constant; using the validation positive "
            "rate as a constant calibrator.",
            RuntimeWarning,
        )
        # Compute the validation positive rate as the constant value
        base_rate: float = float(y_val.mean())
        # Return the explicit constant fallback calibrator
        return CalibratorFit(
            kind="constant", model=None, constant_value=base_rate
        )
    # Fit the monotone non-decreasing isotonic regressor on validation
    calibrator: IsotonicRegression = IsotonicRegression(
        increasing=True, out_of_bounds="clip"
    )
    # Fit the regressor on the validation scores and labels
    calibrator.fit(score_val, y_val)
    # Return the fitted isotonic calibrator
    return CalibratorFit(
        kind="isotonic", model=calibrator, constant_value=None
    )


# -----------------------------------------------------------------------------
# Task 10, Step 2: Apply calibration to test scores and compute ECE
# -----------------------------------------------------------------------------

def compute_ece(
    score_test: np.ndarray,
    y_test: np.ndarray,
    calibrator: Any,
    B: int = 10,
    cfg: Optional[Dict[str, Any]] = None,
) -> Dict[str, Any]:
    """
    Compute Expected Calibration Error with tie-faithful equal-mass binning.

    Implements ECE = sum_b |b|/n |acc(b)-conf(b)| with B=10 equal-mass bins,
    ties of calibrated probability assigned wholly to earliest bin never split,
    per Experimental Setup Calibration B=10.

    Parameters
    ----------
    score_test : np.ndarray, shape (n,)
        Raw test scores of one representation.
    y_test : np.ndarray, shape (n,)
        Binary test labels 0/1.
    calibrator : Any
        Fitted isotonic calibrator with predict method and out_of_bounds clip.
    B : int, default 10
        Number of bins.

    Returns
    -------
    Dict[str, Any]
        Dict with ece float, bin_boundaries, bin_sizes, acc, conf.

    Raises
    ------
    TypeError
        If inputs not ndarray.
    ValueError
        If lengths mismatch or B <=0.
    """
    # Validate score_test is ndarray
    if not isinstance(score_test, np.ndarray):
        raise TypeError(f"score_test must be ndarray, got {type(score_test).__name__}")
    # Validate y_test is ndarray
    if not isinstance(y_test, np.ndarray):
        raise TypeError(f"y_test must be ndarray, got {type(y_test).__name__}")
    # Validate lengths match
    if score_test.shape[0] != y_test.shape[0]:
        raise ValueError(f"score_test {score_test.shape[0]} != y_test {y_test.shape[0]}")
    # Validate B positive
    if B <= 0:
        raise ValueError(f"B must be positive, got {B}")
    # Validate y_test binary
    if not set(np.unique(y_test)).issubset({0,1}):
        raise ValueError("y_test not binary")
    # Apply calibrator predict to test scores
    p_cal: np.ndarray = calibrator.predict(score_test)
    # Clip out-of-range probabilities to [0,1] for safety
    p_cal = np.clip(p_cal, 0.0, 1.0)
    # Validate finite calibrated probabilities
    if not np.isfinite(p_cal).all():
        raise ValueError("p_cal contains non-finite")
    # Get total sample size n
    n: int = int(p_cal.shape[0])
    # Stable argsort indices by p_cal ascending
    order: np.ndarray = np.argsort(p_cal, kind="stable")
    # Sort p_cal and y_test by order
    p_sorted: np.ndarray = p_cal[order]
    y_sorted: np.ndarray = y_test[order]
    # Find unique probability values and their group boundaries
    unique_vals, first_idx, counts = np.unique(p_sorted, return_index=True, return_counts=True)
    # Build list of groups each as slice indices in sorted order
    groups = []
    # Iterate over unique values in order
    for idx, cnt in zip(first_idx, counts):
        # Append group start and count
        groups.append((int(idx), int(cnt)))
    # Initialize bins as list of list of group indices
    bins: list = []
    # Initialize current bin group list
    cur_bin_groups = []
    # Initialize current bin size
    cur_size: int = 0
    # Compute target bin size floor n//B
    target: int = n // B
    # Iterate over groups
    for g_start, g_cnt in groups:
        # If adding this group would exceed target and we have not yet filled B-1 bins, and current bin non-empty, close bin
        if cur_size + g_cnt > target and len(bins) < B-1 and cur_size > 0:
            # Append current bin
            bins.append(cur_bin_groups)
            # Reset current bin
            cur_bin_groups = []
            cur_size = 0
        # Add group to current bin
        cur_bin_groups.append((g_start, g_cnt))
        # Increase current size
        cur_size += g_cnt
        # If current bin reached target and we still have bins to fill, close it eagerly but allow last bin to absorb remainder
        if cur_size >= target and len(bins) < B-1:
            # Append bin
            bins.append(cur_bin_groups)
            cur_bin_groups = []
            cur_size = 0
    # Append final bin with remaining groups
    if cur_bin_groups:
        bins.append(cur_bin_groups)
    # If we have fewer than B bins due to few unique values, keep as is; if more, merge last bins
    while len(bins) > B:
        # Merge last two bins
        last = bins.pop()
        bins[-1].extend(last)
    # Assert sum of sizes equals n
    total: int = sum(sum(cnt for _, cnt in b) for b in bins)
    if total != n:
        raise AssertionError(f"bin total {total} != n {n}")
    # Assert no tie group was split (by construction)
    # Compute per-bin acc and conf
    acc_list = []
    conf_list = []
    bin_sizes = []
    bin_boundaries = []
    # Iterate over bins
    for b in bins:
        # Compute bin start index in sorted array
        bin_start = b[0][0]
        # Compute bin end exclusive
        bin_end = b[-1][0] + b[-1][1]
        # Slice y and p for bin
        y_b = y_sorted[bin_start:bin_end]
        p_b = p_sorted[bin_start:bin_end]
        # Compute observed positive rate acc(b)
        acc = float(y_b.mean()) if y_b.size>0 else 0.0
        # Compute mean predicted conf(b)
        conf = float(p_b.mean()) if p_b.size>0 else 0.0
        # Append
        acc_list.append(acc)
        conf_list.append(conf)
        bin_sizes.append(int(bin_end - bin_start))
        bin_boundaries.append((float(p_sorted[bin_start]), float(p_sorted[bin_end-1])))
    # Compute ECE sum |b|/n |acc-conf|
    ece: float = float(sum(s/n * abs(a-c) for s,a,c in zip(bin_sizes, acc_list, conf_list)))
    # Assert ECE in [0,1]
    if not (0.0 <= ece <= 1.0):
        raise AssertionError(f"ECE {ece} not in [0,1]")
    # Return dict
    return {"ece": ece, "bin_sizes": bin_sizes, "acc": acc_list, "conf": conf_list, "bin_boundaries": bin_boundaries, "B": B}

# -------------------------------------------------------------------------------------------------------------------------------
# Task 10, Step 3: Record calibration outputs per representation
# -------------------------------------------------------------------------------------------------------------------------------

def record_calibration(
    dataset_name: str,
    seed: int,
    representation: str,
    ece_record: Dict[str, Any],
    calibrated_scores: np.ndarray,
    output_dir: Path,
) -> Dict[str, Any]:
    """
    Record the calibration output for one representation.

    Parameters
    ----------
    dataset_name : str
        Dataset name for artifact naming.
    seed : int
        Seed for artifact naming.
    representation : str
        One of "full", "pruned", "directional", "scorecard", "tally".
    ece_record : Dict[str, Any]
        ECE record from compute_ece.
    calibrated_scores : np.ndarray, shape (n_test,)
        Calibrated test probabilities.
    output_dir : Path
        Project output root for the artifact archive.

    Returns
    -------
    Dict[str, Any]
        Persistence report with the JSON and npz paths.

    Raises
    ------
    ValueError
        If the representation name is not in the enumerated set.
    """
    # Validate the representation name against the enumerated set
    if representation not in REPRESENTATIONS:
        raise ValueError(
            f"representation must be one of {REPRESENTATIONS}; "
            f"received '{representation}'."
        )
    # Assert the calibrated scores are finite and in [0, 1]
    if not np.isfinite(calibrated_scores).all():
        raise ValueError("calibrated_scores contains non-finite values.")
    # Construct the versioned artifact directory
    archive_dir: Path = output_dir / "artifacts" / "calibration" / "v1"
    # Create the artifact directory when it does not exist yet
    archive_dir.mkdir(parents=True, exist_ok=True)
    # Construct the JSON record path for this representation
    record_path: Path = archive_dir / (
        f"calibration_{representation}_{dataset_name.lower()}_seed{seed}.json"
    )
    # Compose the serialisable record with native types
    record: Dict[str, Any] = {
        "dataset": dataset_name,
        "seed": seed,
        "representation": representation,
        "ece": float(ece_record["ece"]),
        "bin_sizes": [int(s) for s in ece_record["bin_sizes"]],
        "bin_acc": [float(a) for a in ece_record["bin_acc"]],
        "bin_conf": [float(c) for c in ece_record["bin_conf"]],
        "bin_boundaries": [int(b) for b in ece_record["bin_boundaries"]],
        "n": int(ece_record["n"]),
        "written_at": datetime.now(timezone.utc).isoformat(),
    }
    # Serialise the record to disk with stable formatting
    record_path.write_text(
        json.dumps(record, indent=2, ensure_ascii=False) + "\n",
        encoding="utf-8",
    )
    # Construct the calibrated-scores npz path for this representation
    scores_path: Path = archive_dir / (
        f"calibration_{representation}_{dataset_name.lower()}_seed{seed}.npz"
    )
    # Persist the calibrated test probabilities
    np.savez_compressed(scores_path, calibrated_scores=calibrated_scores)
    # Return the persistence report
    return {"record": str(record_path), "scores": str(scores_path)}


# -----------------------------------------------------------------------------
# Task 10, Orchestrator Function
# -----------------------------------------------------------------------------

def orchestrate_calibration(
    scores_val: Dict[str, np.ndarray],
    scores_test: Dict[str, np.ndarray],
    y_val: np.ndarray,
    y_test: np.ndarray,
    cfg: Dict[str, Any],
    dataset_name: str,
    seed: int,
    output_dir: Path,
) -> Dict[str, Any]:
    """
    Orchestrate Task 10: calibrate all five representations uniformly.

    For every representation the isotonic calibrator is fitted on the
    validation scores and the ECE is computed on the test scores.

    Parameters
    ----------
    scores_val : Dict[str, np.ndarray]
        Validation scores keyed by the five representation names.
    scores_test : Dict[str, np.ndarray]
        Test scores keyed by the five representation names.
    y_val : np.ndarray, shape (n_val,)
        Validation labels.
    y_test : np.ndarray, shape (n_test,)
        Test labels.
    cfg : Dict[str, Any]
        Validated study configuration.
    dataset_name : str
        Dataset name for artifact naming.
    seed : int
        Seed for artifact naming.
    output_dir : Path
        Project output root for the artifact archive.

    Returns
    -------
    Dict[str, Any]
        Mapping keyed by representation with calibrator, ECE record, and
        persistence report per representation.

    Raises
    ------
    ValueError
        If the representation key sets differ from the enumerated five.
    """
    # Validate that the validation score keys are exactly the five
    if set(scores_val.keys()) != set(REPRESENTATIONS):
        raise ValueError(
            f"scores_val keys must be exactly {REPRESENTATIONS}; "
            f"received {sorted(scores_val.keys())}."
        )
    # Validate that the test score keys are exactly the five
    if set(scores_test.keys()) != set(REPRESENTATIONS):
        raise ValueError(
            f"scores_test keys must be exactly {REPRESENTATIONS}; "
            f"received {sorted(scores_test.keys())}."
        )
    # Initialise the per-representation results
    results: Dict[str, Any] = {}
    # Iterate over the five representations in canonical order
    for representation in REPRESENTATIONS:
        # Step 10.1: fit the isotonic calibrator on validation scores
        calibrator: CalibratorFit = fit_isotonic_calibrator(
            score_val=scores_val[representation],
            y_val=y_val,
        )
        # Step 10.2: apply the calibrator and compute the test ECE
        ece_record: Dict[str, Any] = compute_ece(
            score_test=scores_test[representation],
            y_test=y_test,
            calibrator=calibrator,
            cfg=cfg,
            B=int(cfg["calibration"]["ece_bins"]),
        )
        # Apply the calibrator to the test scores for persistence
        calibrated_scores: np.ndarray = calibrator.predict(
            scores_test[representation]
        )
        # Step 10.3: record the calibration output for the representation
        persistence: Dict[str, Any] = record_calibration(
            dataset_name=dataset_name,
            seed=seed,
            representation=representation,
            ece_record=ece_record,
            calibrated_scores=calibrated_scores,
            output_dir=output_dir,
        )
        # Assemble the per-representation result record
        results[representation] = {
            "calibrator": calibrator,
            "ece_record": ece_record,
            "calibrated_scores": calibrated_scores,
            "persistence": persistence,
        }
    # Return the per-representation calibration results
    return results
````

````python
# Task 11: Create the orchestrator function for the end-to-end research pipeline

# ==============================================================================
# Task 11: Create the orchestrator function for the end-to-end research pipeline
# ==============================================================================

# -----------------------------------------------------------------------------
# Task 11, Step 1: Callable interface contract (fixed signatures).
# -----------------------------------------------------------------------------
#
# The following callables are the fixed pipeline contract. Every callable
# states which split may enter it; the orchestrator enforces these access
# rights by construction. compute_pr_auc and compute_kendall_tau_b are
# implemented in Task 12 and imported lazily here; their interfaces are:
#   compute_pr_auc(y_true, scores) -> float
#   compute_kendall_tau_b(full_scores, simplified_scores)
#       -> (C, D, T_s, T_f, tau_b)

# The five evaluated representations
REPRESENTATION_SET: tuple = REPRESENTATIONS

# Metric keys recorded per representation in the results structure
RESULT_METRICS: tuple = (
    "pr_auc",
    "tau_b",
    "ece",
    "rho",
    "tau_pred",
    "selected_thresholds",
    "support",
    "retained_exponents",
    "calibrated_scores",
)



def _resolve_hyperparameters(
    X_train: np.ndarray,
    y_train: np.ndarray,
    cfg: Dict[str, Any],
    dataset_name: str,
    output_dir: Path,
) -> Dict[str, Any]:
    """
    Load or create the hyperparameter record for a dataset.

    The disambiguated protocol tunes once per dataset on the seed-0
    training split; the record is reused by every seed.

    Parameters
    ----------
    X_train : np.ndarray
        Seed-0 training feature matrix.
    y_train : np.ndarray
        Seed-0 training labels.
    cfg : Dict[str, Any]
        Validated study configuration.
    dataset_name : str
        Dataset name.
    output_dir : Path
        Project output root.

    Returns
    -------
    Dict[str, Any]
        The hyperparameter configuration record.
    """
    # Build the expected record path for the dataset
    record_path: Path = (
        output_dir
        / "artifacts"
        / "hyperparameters"
        / "v1"
        / f"hyperparameters_{dataset_name.lower()}.json"
    )
    # Load the existing record when the search has already run
    if record_path.exists():
        # Parse and return the existing record
        return json.loads(record_path.read_text(encoding="utf-8"))
    # Run the hyperparameter search on the seed-0 training split
    search_result: Dict[str, Any] = orchestrate_hyperparameter_search(
        X_train=X_train,
        y_train=y_train,
        cfg=cfg,
        dataset_name=dataset_name,
        output_dir=output_dir,
        seed=0,
    )
    # Return the freshly created configuration record
    return search_result["configuration"]


def _assemble_representation_scores(
    alpha: float,
    beta: np.ndarray,
    X_test: np.ndarray,
    directional_scores: np.ndarray,
    V_test: np.ndarray,
    T_test: np.ndarray,
    support: Sequence[int],
) -> Dict[str, np.ndarray]:
    """
    Assemble the five per-representation test scores.

    Parameters
    ----------
    alpha : float
        Fitted monomial coefficient.
    beta : np.ndarray, shape (m,)
        Full fitted exponent vector.
    X_test : np.ndarray, shape (n_test, m)
        Scaled test matrix.
    directional_scores : np.ndarray, shape (n_test,)
        Directional rule test scores.
    V_test : np.ndarray, shape (n_test,)
        Scorecard test scores.
    T_test : np.ndarray, shape (n_test,)
        Tally test scores.
    support : Sequence[int]
        Retained feature positions.

    Returns
    -------
    Dict[str, np.ndarray]
        Test scores keyed by the five representation names.
    """

    # Compute the full monomial test logits
    full_scores: np.ndarray = monomial_logit(X_test, alpha, beta)
    # Build the pruned exponent vector with zeros outside the support
    pruned_beta: np.ndarray = np.zeros_like(beta)
    # Copy the retained exponents into the pruned vector
    pruned_beta[list(support)] = beta[list(support)]
    # Compute the pruned monomial test logits
    pruned_scores: np.ndarray = monomial_logit(X_test, alpha, pruned_beta)
    # Assemble and return the per-representation test scores
    return {
        "full": full_scores,
        "pruned": pruned_scores,
        "directional": directional_scores,
        "scorecard": V_test,
        "tally": T_test,
    }


# -----------------------------------------------------------------------------
# Task 11, Step 2: Implement the orchestrator sequencing logic
# -----------------------------------------------------------------------------

def _run_dataset_seed(
    cfg: Dict[str, Any],
    cleaned_inputs: Dict[str, pd.DataFrame],
    dataset_name: str,
    seed: int,
    hparams: Dict[str, Any],
    output_dir: Path,
    static_matrices: Dict[str, Dict[str, np.ndarray]],
    feature_names: Sequence[str],
) -> Dict[str, Any]:
    """
    Run the full per-dataset per-seed pipeline sequence.

    Implements the twelve prescribed steps of the orchestrator contract.

    Parameters
    ----------
    cfg : Dict[str, Any]
        Validated study configuration.
    cleaned_inputs : Dict[str, pd.DataFrame]
        Cleansed frames keyed by dataset name.
    dataset_name : str
        Dataset name.
    seed : int
        Random seed.
    hparams : Dict[str, Any]
        Hyperparameter record for the dataset.
    output_dir : Path
        Project output root.
    static_matrices : Dict[str, Dict[str, np.ndarray]]
        Prebuilt feature matrices for Loan, Default, Creditcard.
    feature_names : Sequence[str]
        Feature names aligned with the matrix columns.

    Returns
    -------
    Dict[str, Any]
        Per-seed result record keyed by representation and metric.
    """
    # Step 1: obtain the stratified 60/20/20 split indices
    # FraudEcom has no static matrix: its count features require the
    # split first, so the split is derived from the fraud frame itself
    if dataset_name == "FraudEcom":
        # Extract the fraud label vector from the cleansed frame
        y_fraud_full: np.ndarray = cleaned_inputs["FraudEcom"]["y"].to_numpy(
            dtype=np.int64
        )
        # Build a dummy matrix of the fraud row count (the splitter
        # uses only the row count and the labels for stratification)
        fraud_dummy_matrix: np.ndarray = np.zeros(
            (y_fraud_full.shape[0], len(feature_names))
        )
        # Create the stratified split on the fraud frame for this seed
        train_idx, val_idx, test_idx = make_stratified_splits(
            fraud_dummy_matrix, y_fraud_full, seed, cfg
        )
        # Assert the split indices lie within the fraud frame bounds; this
        # closes the defect class in which a cross-dataset split source
        # produced out-of-bounds indices for the fraud frame
        if int(train_idx.max()) >= len(cleaned_inputs["FraudEcom"]):
            raise ValueError(
                "FraudEcom split indices exceed the cleansed frame length."
            )
        # Step 3 (FraudEcom): build the six-feature matrix with
        # training-only counts AFTER the split so no test row enters
        # the frequency maps
        X_static, y_static, _ = build_fraud_features(
            cleaned_inputs["FraudEcom"], train_idx, cfg
        )
    else:
        # Extract the static feature matrix for the dataset
        X_static: np.ndarray = static_matrices[dataset_name]["X"]
        # Extract the static label vector for the dataset
        y_static: np.ndarray = static_matrices[dataset_name]["y"]
        # Create the stratified split for this seed
        train_idx, val_idx, test_idx = make_stratified_splits(
            X_static, y_static, seed, cfg
        )
    # Step 2: fit preprocessing on the training rows only
    fitted = fit_preprocessing(
        X_train=X_static[train_idx],
        cfg=cfg,
        dataset_name=dataset_name,
        feature_names=feature_names,
    )
    # Apply the fitted preprocessing to all three partitions
    scaled_train, scaled_val, scaled_test, _ = apply_preprocessing(
        X_train=X_static[train_idx],
        X_val=X_static[val_idx],
        X_test=X_static[test_idx],
        fitted=fitted,
    )
    # Slice the labels of the three partitions
    y_train: np.ndarray = y_static[train_idx]
    y_val: np.ndarray = y_static[val_idx]
    y_test: np.ndarray = y_static[test_idx]
    # Step 4: train the full monomial (validation PR-AUC after training)
    train_result = train_monomial(
        X_train=scaled_train,
        y_train=y_train,
        hparams=hparams,
        cfg=cfg,
        seed=seed,
        X_val=scaled_val,
        y_val=y_val,
    )
    # Extract the fitted coefficient from the training result
    alpha: float = train_result.alpha
    # Extract the fitted exponent vector from the training result
    beta: np.ndarray = train_result.beta
    # Step 5: evaluate pruning candidates and select the support
    pruning_result: Dict[str, Any] = orchestrate_pruning(
        alpha=alpha,
        beta=beta,
        X_val=scaled_val,
        y_val=y_val,
        cfg=cfg,
        dataset_name=dataset_name,
        seed=seed,
        output_dir=output_dir,
        feature_names=feature_names,
    )
    # Extract the selected support record
    support_record: Dict[str, Any] = pruning_result["support"]
    # Extract the retained feature positions
    support: List[int] = support_record["retained"]
    # Extract the retained exponent values
    retained_exponents: List[float] = [float(beta[j]) for j in support]
    # Step 6: build the directional rule and its scores
    directional_result: Dict[str, Any] = orchestrate_directional_rule(
        alpha=alpha,
        beta=beta,
        retained=support,
        X_train=scaled_train,
        X_val=scaled_val,
        X_test=scaled_test,
        cfg=cfg,
        dataset_name=dataset_name,
        seed=seed,
        feature_names=feature_names,
        output_dir=output_dir,
    )
    # Extract the direction vector from the directional result
    d: np.ndarray = directional_result["d"]
    # Step 7: predict fidelity from the training covariance only
    prediction: Dict[str, Any] = predict_fidelity(
        beta=beta, d=d, X_train=scaled_train, cfg=cfg
    )
    # Extract the predicted Pearson correlation
    rho: float = float(prediction["rho"])
    # Extract the Greiner-predicted rank fidelity
    tau_pred: float = float(prediction["tau_pred"])
    # Step 8: build the scorecard and tally with thresholds
    scorecard_result: Dict[str, Any] = orchestrate_scorecard_tally(
        alpha=alpha,
        beta=beta,
        retained=support,
        X_train=scaled_train,
        X_val=scaled_val,
        X_test=scaled_test,
        y_val=y_val,
        y_test=y_test,
        cfg=cfg,
        dataset_name=dataset_name,
        seed=seed,
        feature_names=feature_names,
        output_dir=output_dir,
    )
    # Extract the scorecard and tally test scores
    V_test: np.ndarray = scorecard_result["scores"]["V_test"]
    T_test: np.ndarray = scorecard_result["scores"]["T_test"]
    # Extract the directional rule test scores
    directional_scores: np.ndarray = directional_result["scores"]["test"]
    # Assemble the five per-representation test scores
    test_scores: Dict[str, np.ndarray] = _assemble_representation_scores(
        alpha=alpha,
        beta=beta,
        X_test=scaled_test,
        directional_scores=directional_scores,
        V_test=V_test,
        T_test=T_test,
        support=support,
    )
    # Assemble the validation scores for calibration fitting
    val_scores: Dict[str, np.ndarray] = {
        "full": None,
        "pruned": None,
        "directional": directional_result["scores"]["val"],
        "scorecard": scorecard_result["scores"]["V_val"],
        "tally": scorecard_result["scores"]["T_val"],
    }
    # Compute the full monomial validation logits

    # Compute the full validation logits
    val_scores["full"] = monomial_logit(scaled_val, alpha, beta)
    # Build the pruned exponent vector for validation scoring
    pruned_beta_val: np.ndarray = np.zeros_like(beta)
    # Copy the retained exponents into the pruned vector
    pruned_beta_val[list(support)] = beta[list(support)]
    # Compute the pruned validation logits
    val_scores["pruned"] = monomial_logit(scaled_val, alpha, pruned_beta_val)
    # Step 9: calibrate all five representations on validation
    calibration_result: Dict[str, Any] = orchestrate_calibration(
        scores_val=val_scores,
        scores_test=test_scores,
        y_val=y_val,
        y_test=y_test,
        cfg=cfg,
        dataset_name=dataset_name,
        seed=seed,
        output_dir=output_dir,
    )
    # Steps 10-11: evaluate PR-AUC and Kendall tau_b on the test set

    # Extract the full monomial test scores for the fidelity baseline
    full_test_scores: np.ndarray = test_scores["full"]
    # Initialise the per-representation metric records
    representation_records: Dict[str, Dict[str, Any]] = {}
    # Iterate over the five representations
    for representation in REPRESENTATIONS:
        # Step 10: compute the test PR-AUC of the representation
        pr_auc: float = compute_pr_auc(y_test, test_scores[representation])
        # Step 11: compute the tie-corrected Kendall tau_b against the
        # full monomial scores on the test set
        C, D, T_s, T_f, tau_b = compute_kendall_tau_b(
            full_test_scores, test_scores[representation]
        )
        # Extract the ECE record of the representation
        ece_record: Dict[str, Any] = calibration_result[representation][
            "ece_record"
        ]
        # Extract the calibrated test scores of the representation
        calibrated_scores: np.ndarray = calibration_result[representation][
            "calibrated_scores"
        ]
        # Assemble the per-representation metric record
        representation_records[representation] = {
            "pr_auc": pr_auc,
            "tau_b": float(tau_b),
            "tau_counts": {"C": C, "D": D, "T_s": T_s, "T_f": T_f},
            "ece": ece_record["ece"],
            "rho": rho,
            "tau_pred": tau_pred,
            "selected_thresholds": {
                "scorecard": scorecard_result["selection"][
                    "scorecard_threshold"
                ],
                "tally": scorecard_result["selection"]["tally_threshold"],
            },
            "support": support,
            "retained_exponents": retained_exponents,
            "calibrated_scores": calibrated_scores,
        }
    # Step 12: assemble and return the per-seed record
    return {
        "alpha": alpha,
        "beta": beta,
        "representations": representation_records,
    }


# -----------------------------------------------------------------------------
# Task 11, Orchestrator Function
# -----------------------------------------------------------------------------
def run_reproduction_study(
    cfg: Dict[str, Any],
    cleaned_inputs: Dict[str, pd.DataFrame],
    output_dir: Path,
    datasets: Optional[Sequence[str]] = None,
    seeds: Optional[Sequence[int]] = None,
) -> Dict[str, Any]:
    """
    Run the end-to-end reproduction study pipeline.

    Sequenced per dataset and per seed exactly as prescribed:
    1) stratified splits; 2) train-only preprocessing fit and apply;
    3) FraudEcom count features after the split; 4) monomial training;
    5) pruning and support selection; 6) directional rule; 7) fidelity
    prediction; 8) scorecard and tally with thresholds; 9) isotonic
    calibration of all five representations; 10) test PR-AUC; 11) test
    Kendall tau_b vs the full monomial; 12) per-seed assembly.

    Parameters
    ----------
    cfg : Dict[str, Any]
        Validated study configuration.
    cleaned_inputs : Dict[str, pd.DataFrame]
        Cleansed frames keyed by dataset name (from Task 2).
    output_dir : Path
        Project output root for all artifacts.
    datasets : Optional[Sequence[str]], optional
        Subset of datasets to run (default: all four).
    seeds : Optional[Sequence[int]], optional
        Subset of seeds to run (default: the configured five).

    Returns
    -------
    Dict[str, Any]
        Nested results keyed dataset -> seed -> representation -> metric.
    """
    # Resolve the dataset list when not supplied
    if datasets is None:
        datasets = ("Loan", "FraudEcom", "Default", "Creditcard")
    # Resolve the seed list when not supplied
    if seeds is None:
        seeds = tuple(cfg["data_splitting"]["seeds"])
    # Initialise the top-level results structure
    results: Dict[str, Any] = {}
    # Build the static feature matrices for the non-FraudEcom datasets
    static_matrices: Dict[str, Dict[str, np.ndarray]] = {}
    # Build the Loan feature matrix once
    if "Loan" in datasets:
        # Build the Loan matrix from its cleansed frame
        loan_X, loan_y, _ = build_loan_features(cleaned_inputs["Loan"], cfg)
        # Store the Loan matrix and labels
        static_matrices["Loan"] = {"X": loan_X, "y": loan_y}
    # Build the Default feature matrix once
    if "Default" in datasets:
        # Build the Default matrix from its cleansed frame
        default_X, default_y, _ = build_default_features(
            cleaned_inputs["Default"], cfg
        )
        # Store the Default matrix and labels
        static_matrices["Default"] = {"X": default_X, "y": default_y}
    # Build the Creditcard feature matrix once
    if "Creditcard" in datasets:
        # Build the Creditcard matrix from its cleansed frame
        creditcard_X, creditcard_y, _ = build_creditcard_features(
            cleaned_inputs["Creditcard"], cfg
        )
        # Store the Creditcard matrix and labels
        static_matrices["Creditcard"] = {"X": creditcard_X, "y": creditcard_y}
    # Iterate over the requested datasets
    for dataset_name in datasets:
        # Announce the dataset in the execution trace
        logger.info("=== Dataset: %s ===", dataset_name)
        # Extract the feature names from the configuration
        feature_names: Sequence[str] = cfg["feature_matrices"][dataset_name][
            "feature_names"
        ]
        # Resolve or create the hyperparameter record (seed-0 tuning)
        if dataset_name == "FraudEcom":
            # Extract the fraud label vector from the cleansed frame
            y_fraud_full: np.ndarray = cleaned_inputs["FraudEcom"][
                "y"
            ].to_numpy(dtype=np.int64)
            # Build a dummy matrix of the fraud row count (the
            # splitter uses only the row count and the labels)
            fraud_dummy_matrix: np.ndarray = np.zeros(
                (y_fraud_full.shape[0], len(feature_names))
            )
            # Obtain the seed-0 split on the fraud frame itself
            seed0_train_idx, _, _ = make_stratified_splits(
                fraud_dummy_matrix,
                y_fraud_full,
                int(list(seeds)[0]),
                cfg,
            )
            # Build the fraud matrix on the seed-0 training rows only
            fraud_X0, fraud_y0, _ = build_fraud_features(
                cleaned_inputs["FraudEcom"], seed0_train_idx, cfg
            )
            # Fit the preprocessing on the seed-0 training rows
            fitted0 = fit_preprocessing(
                X_train=fraud_X0[seed0_train_idx],
                cfg=cfg,
                dataset_name="FraudEcom",
                feature_names=feature_names,
            )
            # Apply the preprocessing; the dummy val/test slices are
            # training rows (leakage-safe by construction) and only the
            # scaled training output is consumed by the search
            scaled0, _, _, _ = apply_preprocessing(
                X_train=fraud_X0[seed0_train_idx],
                X_val=fraud_X0[seed0_train_idx][:1],
                X_test=fraud_X0[seed0_train_idx][:1],
                fitted=fitted0,
            )
            # Resolve the hyperparameters on the scaled seed-0 rows
            hparams: Dict[str, Any] = _resolve_hyperparameters(
                X_train=scaled0,
                y_train=fraud_y0[seed0_train_idx],
                cfg=cfg,
                dataset_name="FraudEcom",
                output_dir=output_dir,
            )
        else:
            # Resolve the hyperparameters for the static dataset
            hparams = _resolve_hyperparameters(
                X_train=static_matrices[dataset_name]["X"],
                y_train=static_matrices[dataset_name]["y"],
                cfg=cfg,
                dataset_name=dataset_name,
                output_dir=output_dir,
            )
        # Initialise the per-dataset results
        results[dataset_name] = {}
        # Iterate over the requested seeds
        for seed in seeds:
            # Announce the seed in the execution trace
            logger.info("-- seed %d --", seed)
            # Record the start time of the seed run
            start_time: float = time.time()
            # Run the twelve-step pipeline for this dataset and seed
            seed_record: Dict[str, Any] = _run_dataset_seed(
                cfg=cfg,
                cleaned_inputs=cleaned_inputs,
                dataset_name=dataset_name,
                seed=int(seed),
                hparams=hparams,
                output_dir=output_dir,
                static_matrices=static_matrices,
                feature_names=feature_names,
            )
            # Record the elapsed time of the seed run
            elapsed: float = time.time() - start_time
            # Log the elapsed time in the execution trace
            logger.info("seed %d completed in %.1fs", seed, elapsed)
            # Store the per-seed record
            results[dataset_name][str(seed)] = seed_record
    # Return the nested results structure
    return results


# -----------------------------------------------------------------------------
# Task 11, Step 3: Orchestrator verification (smoke test)
# -----------------------------------------------------------------------------

def verify_orchestrator(
    cfg: Dict[str, Any],
    cleaned_inputs: Dict[str, pd.DataFrame],
    output_dir: Path,
) -> Dict[str, Any]:
    """
    Smoke test for Creditcard seed 0 with leakage audit.

    Runs run_reproduction_study for Creditcard seed 0, asserts 5 representations,
    ece in [0,1], tau_b in [-1,1], support nonempty, and that no test/validation
    index appears in training-only artifacts (scaler provenance). Raises LeakageError
    on intersect.

    Parameters
    ----------
    cfg : Dict[str, Any]
        Validated configuration.
    cleaned_inputs : Dict[str, pd.DataFrame]
        Cleansed frames.
    output_dir : Path
        Output root.

    Returns
    -------
    Dict[str, Any]
        Smoke test result with leakage audit.

    Raises
    ------
    AssertionError
        If any range check fails.
    LeakageError
        If train∩test or train∩val non-empty or test index in scaler provenance.
    """
    # Define a custom leakage-detection error type raised by the smoke test
    class LeakageError(ValueError):
        """
        Raised when the orchestrator smoke test detects a leakage violation.

        Indicates that training and test (or validation) split indices
        intersect, or that a test row appears in the scaler provenance,
        which would invalidate the study's reproducibility guarantees.

        Notes
        -----
        Raised exclusively inside :func:`verify_orchestrator` during the
        Task 11.3 smoke test; it is a local type that never propagates
        beyond that function's leakage audit.
        """
        # Provide an explicit no-op body so the class definition is complete
        pass
    # Run reproduction for Creditcard seed 0 only
    results = run_reproduction_study(cfg, cleaned_inputs, output_dir, datasets=["Creditcard"], seeds=[0])
    # Extract Creditcard seed 0 result
    cc_result = results["Creditcard"]["0"]
    # Assert 5 representations present
    reps = cc_result.get("representations", {})
    if set(reps.keys()) != {"full","pruned","directional","scorecard","tally"}:
        raise AssertionError(f"expected 5 representations, got {set(reps.keys())}")
    # Iterate over representations
    for rname, rec in reps.items():
        # Check ece in [0,1]
        ece = float(rec["ece"])
        if not (0.0 <= ece <= 1.0):
            raise AssertionError(f"{rname} ece {ece} not in [0,1]")
        # Check tau_b in [-1,1]
        tau = float(rec["tau_b"])
        if not (-1.0 <= tau <= 1.0):
            raise AssertionError(f"{rname} tau_b {tau} not in [-1,1]")
        # Check support nonempty for pruned etc. (full may be m)
        support = rec.get("support", [])
        if len(support) == 0:
            raise AssertionError(f"{rname} support empty")
    # Load persisted splits for Creditcard seed 0
    splits_path = output_dir / "artifacts" / "splits" / "v1" / "splits_creditcard_seed0.json"
    # Try alternative path if not found
    if not splits_path.exists():
        # Fallback to generic splits artifact
        alt = output_dir / "artifacts" / "splits" / "v1" / "splits.json"
        if alt.exists():
            splits_path = alt
    # If splits file exists, perform leakage intersection
    if splits_path.exists():
        # Load splits json
        splits_data = json.loads(splits_path.read_text(encoding="utf-8"))
        # Extract train and test indices for creditcard seed 0
        train_idx = set(splits_data.get("Creditcard", {}).get("0", {}).get("train", []))
        # Fallback to direct keys if nested not found
        if not train_idx:
            # Try flat structure
            train_idx = set(splits_data.get("train", []))
        test_idx = set(splits_data.get("Creditcard", {}).get("0", {}).get("test", []))
        if not test_idx:
            test_idx = set(splits_data.get("test", []))
        val_idx = set(splits_data.get("Creditcard", {}).get("0", {}).get("val", []))
        if train_idx and test_idx:
            # Check disjoint
            inter = train_idx.intersection(test_idx)
            if inter:
                raise LeakageError(f"train∩test non-empty size {len(inter)}: {list(inter)[:5]}")
        if train_idx and val_idx:
            inter2 = train_idx.intersection(val_idx)
            if inter2:
                raise LeakageError(f"train∩val non-empty size {len(inter2)}")
        # Check scaler provenance for test leak
        scaler_path = output_dir / "artifacts" / "scaler" / "v1" / "scaler_creditcard_seed0.json"
        if scaler_path.exists():
            scaler_data = json.loads(scaler_path.read_text(encoding="utf-8"))
            # Provenance may contain train_indices
            prov_train = set(scaler_data.get("train_indices", []))
            if prov_train and test_idx:
                inter3 = prov_train.intersection(test_idx)
                if inter3:
                    raise LeakageError(f"scaler provenance contains test indices {len(inter3)}")
    # Return smoke result
    return {"passed": True, "creditcard_seed0": cc_result, "leakage_audit": "passed"}


````

````python
# Task 12: Robustness analysis: iterative hard thresholding and Jaccard

# ==============================================================================
# Task 12: Robustness analysis: iterative hard thresholding and Jaccard
# ==============================================================================

# -----------------------------------------------------------------------------
# Task 12, Step 1: Train IHT monomials with in-training sparsity
# -----------------------------------------------------------------------------

@dataclass
class IHTResult:
    """
    Container of an iterative-hard-thresholding training run.

    Stores the projected coefficient and exponent vector after hard-thresholding
    sparsity, where after every gradient step only the ``s`` largest
    ``|beta_j|`` are retained. Produced by ``train_iht_monomial`` in
    Task 12 Step 12.1 and compared to post-hoc pruning via Jaccard.

    Attributes
    ----------
    alpha : float
        Fitted monomial coefficient after IHT projection; unchanged by the
        projection.
    beta : np.ndarray, shape (m,)
        Projected exponent vector with at most ``s`` nonzero entries;
        zeros elsewhere.
    support : List[int]
        Sorted column positions with nonzero exponents, i.e.
        ``{j : beta_j != 0}``; ``|support| <= s``.
    history : List[float]
        Per-epoch training loss values before projection; length equals
        ``epochs_used``.
    epochs_used : int
        Number of epochs actually executed; equals the epoch budget because
        IHT loss is non-monotonic and budget governs convergence.
    """

    # Fitted monomial coefficient
    alpha: float
    # Projected exponent vector (at most s nonzero entries)
    beta: np.ndarray
    # Sorted column positions with nonzero exponents
    support: List[int] = field(default_factory=list)
    # Per-epoch training loss values
    history: List[float] = field(default_factory=list)
    # Number of epochs actually executed
    epochs_used: int = 0


def train_iht_monomial(
    X_train: np.ndarray,
    y_train: np.ndarray,
    s: int,
    hparams: Dict[str, Any],
    cfg: Dict[str, Any],
    seed: int,
) -> IHTResult:
    """
    Train a monomial with in-training hard thresholding (IHT).

    Identical to the full monomial trainer except: the l1 strength is
    zero (the projection supplies sparsity; an l1 penalty would fight
    it), and after EVERY gradient step — including the final one — the
    exponent vector is projected onto the s largest |beta_j| entries,
    with deterministic tie-breaking (|beta_j| descending, index
    ascending). Convergence is governed by the epoch budget because the
    projection makes the training loss non-monotonic.

    Parameters
    ----------
    X_train : np.ndarray, shape (n_train, m)
        Scaled training feature matrix.
    y_train : np.ndarray, shape (n_train,)
        Binary training labels.
    s : int
        Number of retained exponents (from post-hoc pruning r*).
    hparams : Dict[str, Any]
        Hyperparameters with "learning_rate" and "epochs".
    cfg : Dict[str, Any]
        Validated study configuration.
    seed : int
        Deterministic seed for torch and numpy.

    Returns
    -------
    IHTResult
        Fitted and projected parameters with history and support.

    Raises
    ------
    TypeError
        If the training arrays are not ndarrays.
    ValueError
        If the arrays misalign or s is outside 1..m.
    """
    # Validate the training matrix type
    if not isinstance(X_train, np.ndarray):
        raise TypeError(
            f"X_train must be an ndarray; received {type(X_train).__name__}."
        )
    # Validate the label vector type
    if not isinstance(y_train, np.ndarray):
        raise TypeError(
            f"y_train must be an ndarray; received {type(y_train).__name__}."
        )
    # Raise when the training arrays disagree in length
    if X_train.shape[0] != y_train.shape[0]:
        raise ValueError(
            f"X_train has {X_train.shape[0]} rows but y_train has "
            f"{y_train.shape[0]} entries."
        )
    # Raise when the sparsity level is outside the valid range
    if not (1 <= s <= X_train.shape[1]):
        raise ValueError(
            f"s must lie in 1..{X_train.shape[1]}; received {s}."
        )

    # Extract the learning rate from the hyperparameters
    learning_rate: float = float(hparams["learning_rate"])
    # Extract the epoch budget from the hyperparameters
    epoch_budget: int = int(hparams["epochs"])
    # Seed torch and numpy for deterministic training
    torch.manual_seed(seed)
    np.random.seed(seed)
    # Convert the training matrix to a float64 tensor
    X_t = torch.tensor(X_train, dtype=torch.float64)
    # Convert the training labels to a float64 tensor
    y_t = torch.tensor(y_train, dtype=torch.float64)
    # Precompute the log of the scaled features once
    log_x_t: torch.Tensor = torch.log(X_t)
    # Initialise the coefficient to the fixed starting value
    alpha: torch.Tensor = torch.tensor(1.0, dtype=torch.float64, requires_grad=True)
    # Initialise every exponent to the fixed starting value
    beta: torch.Tensor = torch.full(
        (X_train.shape[1],), 0.1, dtype=torch.float64, requires_grad=True
    )
    # Construct the gradient-descent optimizer over both parameters
    optimizer = torch.optim.SGD([alpha, beta], lr=learning_rate)
    # Record the per-epoch training losses
    history: List[float] = []
    # Run the gradient-descent loop up to the epoch budget
    for epoch in range(epoch_budget):
        # Reset the parameter gradients to zero
        optimizer.zero_grad()
        # Compute the log-space score: log|alpha| + beta^T log x
        log_score_t = torch.log(torch.abs(alpha)) + log_x_t @ beta
        # Reconstruct the logits with the sign of alpha tracked separately
        z_t = torch.sign(alpha) * torch.exp(log_score_t)
        # Evaluate the stable binary cross-entropy (no l1 penalty)
        loss_t = F.binary_cross_entropy_with_logits(z_t, y_t)
        # Backpropagate the loss through the computation graph
        loss_t.backward()
        # Apply the gradient-descent update to both parameters
        optimizer.step()
        # Project beta onto the s largest magnitudes with deterministic
        # tie-breaking: stable argsort on -|beta| then zero the tail
        order: np.ndarray = np.argsort(-np.abs(beta.detach().numpy()), kind="stable")
        # Build the zeroing mask for the tail positions
        zero_positions: np.ndarray = order[s:]
        # Zero every exponent outside the top-s positions in place
        if zero_positions.shape[0] > 0:
            beta.data[torch.tensor(zero_positions, dtype=torch.long)] = 0.0
        # Record the loss value of this epoch
        history.append(float(loss_t.detach()))
    # Detach the fitted coefficient as a native float
    alpha_fitted: float = float(alpha.detach())
    # Detach the projected exponent vector as a numpy array copy
    beta_fitted: np.ndarray = beta.detach().numpy().copy()
    # Compute the support as the nonzero exponent positions
    support: List[int] = [
        int(j) for j in range(beta_fitted.shape[0]) if beta_fitted[j] != 0.0
    ]
    # Assemble and return the IHT training result
    return IHTResult(
        alpha=alpha_fitted,
        beta=beta_fitted,
        support=support,
        history=history,
        epochs_used=epoch_budget,
    )


# -----------------------------------------------------------------------------
# Task 12, Step 2: Compute Jaccard overlap and performance comparison
# -----------------------------------------------------------------------------

def compute_pr_auc(y_true: np.ndarray, scores: np.ndarray) -> float:
    """
    Compute the area under the precision-recall curve.

    Parameters
    ----------
    y_true : np.ndarray, shape (n,)
        Binary true labels.
    scores : np.ndarray, shape (n,)
        Ranking scores of one representation.

    Returns
    -------
    float
        PR-AUC of the scores against the labels.

    Raises
    ------
    TypeError
        If either input is not an ndarray.
    ValueError
        If the arrays disagree in length.
    """
    # Validate the label vector type
    if not isinstance(y_true, np.ndarray):
        raise TypeError(
            f"y_true must be an ndarray; received {type(y_true).__name__}."
        )
    # Validate the score vector type
    if not isinstance(scores, np.ndarray):
        raise TypeError(
            f"scores must be an ndarray; received {type(scores).__name__}."
        )
    # Raise when the arrays disagree in length
    if y_true.shape[0] != scores.shape[0]:
        raise ValueError(
            f"y_true has {y_true.shape[0]} entries but scores has "
            f"{scores.shape[0]} entries."
        )
    # Compute the PR-AUC via average_precision_score
    return float(average_precision_score(y_true, scores))


# Define a Fenwick-tree helper for exact O(n log n) inversion counting
class _FenwickTree:
    """
    Binary indexed tree for exact inversion counting in O(log n).

    Internal helper for :func:`compute_kendall_tau_b`: supports point
    updates and prefix-sum queries over a dense rank space so that
    concordant and discordant pair counts are accumulated in O(n log n)
    instead of O(n^2). All indices are 1-based throughout.

    Attributes
    ----------
    size : int
        Number of leaves, i.e. the maximum valid 1-based index.
    tree : List[int]
        Fenwick tree array of length ``size + 1``; ``tree[i]`` stores
        the aggregate count over the index range governed by node ``i``.
    """

    # Define the constructor that allocates the zero-initialised tree array
    def __init__(self, size: int) -> None:
        """
        Construct an empty Fenwick tree of the requested size.

        Parameters
        ----------
        size : int
            Number of leaves; the tree array is allocated with
            ``size + 1`` entries so that 1-based indexing is safe.

        Returns
        -------
        None
            Constructor; the instance is initialized in place.
        """
        # Store the tree size
        self.size: int = size
        # Initialise the tree array with zeros
        self.tree: List[int] = [0] * (size + 1)

    # Apply a point update to the Fenwick tree at the given 1-based index
    def add(self, index: int, delta: int) -> None:
        """
        Add a delta at a 1-based index, propagating upward.

        Parameters
        ----------
        index : int
            One-based index to update.
        delta : int
            Increment applied at the index.
        """
        # Propagate the update up the tree
        i: int = index
        # Walk the tree indices upward
        while i <= self.size:
            # Apply the delta at the current node
            self.tree[i] += delta
            # Move to the next ancestor index
            i += i & (-i)

    # Return the prefix sum over the first `index` leaves of the tree
    def prefix(self, index: int) -> int:
        """
        Return the prefix sum up to a 1-based index.

        Parameters
        ----------
        index : int
            One-based upper bound of the prefix.

        Returns
        -------
        int
            Sum of the counts at indices 1..index.
        """
        # Accumulate the prefix sum
        total: int = 0
        # Walk the tree indices downward
        i: int = index
        # Accumulate the tree nodes covering the prefix
        while i > 0:
            total += self.tree[i]
            i -= i & (-i)
        # Return the accumulated prefix sum
        return total


def compute_kendall_tau_b(
    full_scores: np.ndarray,
    simplified_scores: np.ndarray,
) -> Tuple[int, int, int, int, float]:
    """
    Compute the tie-corrected Kendall tau_b between two score vectors.

    Implements the manuscript's fidelity measure:
        tau_b = (C - D) / sqrt((C + D + T_s)(C + D + T_f))
    where C and D are concordant and discordant pairs and T_s, T_f are
    the tie corrections of the simplified and full scores. The counts
    are computed exactly in O(n log n) via a lexicographic sort and a
    Fenwick tree; no approximation is used at any size.

    Parameters
    ----------
    full_scores : np.ndarray, shape (n,)
        Full monomial scores.
    simplified_scores : np.ndarray, shape (n,)
        Simplified representation scores.

    Returns
    -------
    Tuple[int, int, int, int, float]
        Concordant count C, discordant count D, tie count T_s of the
        simplified scores, tie count T_f of the full scores, and tau_b.

    Raises
    ------
    TypeError
        If either input is not an ndarray.
    ValueError
        If the arrays disagree in length or contain non-finite values.
    """
    # Validate the full score vector type
    if not isinstance(full_scores, np.ndarray):
        raise TypeError(
            f"full_scores must be an ndarray; "
            f"received {type(full_scores).__name__}."
        )
    # Validate the simplified score vector type
    if not isinstance(simplified_scores, np.ndarray):
        raise TypeError(
            f"simplified_scores must be an ndarray; "
            f"received {type(simplified_scores).__name__}."
        )
    # Raise when the arrays disagree in length
    if full_scores.shape[0] != simplified_scores.shape[0]:
        raise ValueError(
            f"full_scores has {full_scores.shape[0]} entries but "
            f"simplified_scores has {simplified_scores.shape[0]} entries."
        )
    # Raise when the scores contain non-finite values
    if not (np.isfinite(full_scores).all() and np.isfinite(simplified_scores).all()):
        raise ValueError("Scores must be finite for rank comparison.")
    # Extract the arrays as 1-D views
    x: np.ndarray = np.ascontiguousarray(full_scores, dtype=np.float64)
    y: np.ndarray = np.ascontiguousarray(simplified_scores, dtype=np.float64)
    # Compute the total number of pairs
    n: int = x.shape[0]
    # Compute the total pair count P = n(n-1)/2
    total_pairs: int = n * (n - 1) // 2
    # Compute the tie correction of the full scores: sum over tie groups
    # of g(g-1)/2
    _, x_counts = np.unique(x, return_counts=True)
    # Sum the tie-group pair counts of the full scores
    T_f: int = int(np.sum(x_counts * (x_counts - 1) // 2))
    # Compute the tie correction of the simplified scores
    _, y_counts = np.unique(y, return_counts=True)
    # Sum the tie-group pair counts of the simplified scores
    T_s: int = int(np.sum(y_counts * (y_counts - 1) // 2))
    # Compute the joint-tie correction: pairs equal in BOTH scores
    joint_keys: np.ndarray = np.stack([x, y], axis=1)
    # Find the unique joint keys and their group sizes
    _, joint_counts = np.unique(joint_keys, axis=0, return_counts=True)
    # Sum the joint tie-group pair counts
    T_xy: int = int(np.sum(joint_counts * (joint_counts - 1) // 2))
    # Sort the pairs lexicographically by (x, y) for group processing
    order: np.ndarray = np.lexsort((y, x))
    # Reorder the scores by the lexicographic order
    xs: np.ndarray = x[order]
    ys: np.ndarray = y[order]
    # Compress the y values to ranks for the Fenwick tree
    unique_y: np.ndarray = np.unique(ys)
    # Build a rank lookup for every y value
    rank_of: np.ndarray = np.searchsorted(unique_y, ys)
    # Convert the ranks to one-based indices
    rank_of = rank_of + 1
    # Initialise the Fenwick tree over the compressed ranks
    tree: _FenwickTree = _FenwickTree(int(unique_y.shape[0]))
    # Track the number of processed elements
    processed: int = 0
    # Accumulate the discordant pair count
    discordant: int = 0
    # Process the sorted pairs group by group over equal x values
    i: int = 0
    # Iterate over the x-groups
    while i < n:
        # Extend the group end over all equal x values
        j: int = i
        # Advance the group end while x values match
        while j < n and xs[j] == xs[i]:
            j += 1
        # Query the discordant pairs for each element of the group
        # against the previously processed elements (x strictly less)
        for k in range(i, j):
            # Count the previously processed elements with a strictly
            # greater y (prefix counts rank <= current rank)
            le: int = tree.prefix(int(rank_of[k]))
            # Accumulate the strictly-greater count as discordant
            discordant += processed - le
        # Insert the entire group after the queries so pairs with equal
        # x are never compared
        for k in range(i, j):
            tree.add(int(rank_of[k]), 1)
        # Update the processed element count
        processed += j - i
        # Move to the next group
        i = j
    # Compute the comparable pair count: P - T_x - T_y + T_xy
    comparable: int = total_pairs - T_f - T_s + T_xy
    # Compute the concordant pair count from the comparable total
    concordant: int = comparable - discordant
    # Compute the two denominator factors of the tau_b formula
    denom_factor_s: int = comparable + T_s
    # Compute the second denominator factor
    denom_factor_f: int = comparable + T_f
    # Guard against the degenerate all-tied case
    if denom_factor_s == 0 or denom_factor_f == 0:
        # Record the degenerate tau_b as zero
        tau_b: float = 0.0
    else:
        # Compute the tie-corrected Kendall tau_b
        tau_b = float(
            (concordant - discordant)
            / np.sqrt(denom_factor_s * denom_factor_f)
        )
    # Return the exact counts and the final tau_b
    return concordant, discordant, T_s, T_f, tau_b


def compare_pruned_vs_iht(
    S_prune: Sequence[int],
    S_iht: Sequence[int],
    pr_auc_prune: float,
    pr_auc_iht: float,
) -> Dict[str, Any]:
    """
    Compute the Jaccard overlap and performance comparison.

    J(S_prune, S_IHT) = |S_prune ∩ S_IHT| / |S_prune ∪ S_IHT|.
    An empty support yields J = 0 with a logged warning.

    Parameters
    ----------
    S_prune : Sequence[int]
        Post-hoc pruned support (feature index set).
    S_iht : Sequence[int]
        IHT support (feature index set).
    pr_auc_prune : float
        Test PR-AUC of the pruned monomial.
    pr_auc_iht : float
        Test PR-AUC of the IHT monomial.

    Returns
    -------
    Dict[str, Any]
        Comparison record with the Jaccard index, sizes, and PR-AUCs.
    """
    # Convert the supports to sets of integers
    prune_set: set = set(int(j) for j in S_prune)
    # Convert the IHT support to a set of integers
    iht_set: set = set(int(j) for j in S_iht)
    # Compute the intersection size
    intersection: int = len(prune_set & iht_set)
    # Compute the union size
    union: int = len(prune_set | iht_set)
    # Guard against the empty-union case with a logged warning
    if union == 0:
        # Warn about the degenerate empty supports
        warnings.warn(
            "Both supports are empty; Jaccard defined as 0.",
            RuntimeWarning,
        )
        # Record the degenerate Jaccard value
        jaccard: float = 0.0
    else:
        # Compute the Jaccard index
        jaccard = intersection / union
    # Warn when the IHT support size differs from the requested size
    if len(iht_set) != len(prune_set):
        warnings.warn(
            f"IHT support size {len(iht_set)} differs from the pruned "
            f"support size {len(prune_set)}.",
            RuntimeWarning,
        )
    # Assemble the comparison record
    comparison: Dict[str, Any] = {
        "jaccard": float(jaccard),
        "intersection": intersection,
        "union": union,
        "r_prune": len(prune_set),
        "r_iht": len(iht_set),
        "pr_auc_prune": float(pr_auc_prune),
        "pr_auc_iht": float(pr_auc_iht),
    }
    # Return the comparison record
    return comparison


# -----------------------------------------------------------------------------
# Task 12, Step 3: Evaluate IHT test PR-AUC and persist the comparison
# -----------------------------------------------------------------------------

def evaluate_iht_on_test(
    alpha_iht: float,
    beta_iht: np.ndarray,
    X_test: np.ndarray,
    y_test: np.ndarray,
) -> float:
    """
    Evaluate the IHT monomial test PR-AUC.

    Parameters
    ----------
    alpha_iht : float
        Fitted IHT coefficient.
    beta_iht : np.ndarray, shape (m,)
        Projected IHT exponent vector.
    X_test : np.ndarray, shape (n_test, m)
        Scaled test matrix.
    y_test : np.ndarray, shape (n_test,)
        Test labels.

    Returns
    -------
    float
        Test PR-AUC of the IHT monomial.
    """

    # Compute the IHT monomial logits on the test rows
    iht_logits: np.ndarray = monomial_logit(X_test, alpha_iht, beta_iht)
    # Convert the logits to probabilities via the sigmoid
    iht_probabilities: np.ndarray = 1.0 / (1.0 + np.exp(-iht_logits))
    # Compute the test PR-AUC
    return float(average_precision_score(y_test, iht_probabilities))


# -----------------------------------------------------------------------------
# Task 12, Orchestrator Function
# -----------------------------------------------------------------------------

def orchestrate_iht_comparison(
    alpha: float,
    beta: np.ndarray,
    support_prune: Sequence[int],
    X_train: np.ndarray,
    y_train: np.ndarray,
    X_test: np.ndarray,
    y_test: np.ndarray,
    hparams: Dict[str, Any],
    cfg: Dict[str, Any],
    dataset_name: str,
    seed: int,
    pr_auc_prune: float,
    feature_names: Sequence[str],
    output_dir: Path,
) -> Dict[str, Any]:
    """
    Orchestrate Task 12: train IHT, compare, evaluate, and persist.

    Parameters
    ----------
    alpha : float
        Fitted full-monomial coefficient (unused by the IHT trainer but
        kept for interface symmetry).
    beta : np.ndarray, shape (m,)
        Full fitted exponent vector (unused by the IHT trainer).
    support_prune : Sequence[int]
        Post-hoc pruned support (determines s = |S_prune|).
    X_train : np.ndarray, shape (n_train, m)
        Scaled training matrix.
    y_train : np.ndarray, shape (n_train,)
        Training labels.
    X_test : np.ndarray, shape (n_test, m)
        Scaled test matrix.
    y_test : np.ndarray, shape (n_test,)
        Test labels.
    hparams : Dict[str, Any]
        Hyperparameter record (learning rate and epoch budget).
    cfg : Dict[str, Any]
        Validated study configuration.
    dataset_name : str
        Dataset name for artifact naming.
    seed : int
        Deterministic seed.
    pr_auc_prune : float
        Test PR-AUC of the pruned monomial.
    feature_names : Sequence[str]
        Feature names aligned with the matrix columns.
    output_dir : Path
        Project output root for the artifact archive.

    Returns
    -------
    Dict[str, Any]
        Mapping with "iht_result", "comparison", and "persistence".
    """
    # Derive the IHT sparsity level from the pruned support size
    s: int = len(set(int(j) for j in support_prune))
    # Step 12.1: train the IHT monomial with in-training sparsity
    iht_result: IHTResult = train_iht_monomial(
        X_train=X_train,
        y_train=y_train,
        s=s,
        hparams=hparams,
        cfg=cfg,
        seed=seed,
    )
    # Step 12.3: evaluate the IHT monomial on the test set
    pr_auc_iht: float = evaluate_iht_on_test(
        alpha_iht=iht_result.alpha,
        beta_iht=iht_result.beta,
        X_test=X_test,
        y_test=y_test,
    )
    # Step 12.2: compute the Jaccard overlap and the comparison record
    comparison: Dict[str, Any] = compare_pruned_vs_iht(
        S_prune=support_prune,
        S_iht=iht_result.support,
        pr_auc_prune=pr_auc_prune,
        pr_auc_iht=pr_auc_iht,
    )
    # Map the pruned and IHT supports to feature names for the audit
    prune_names: List[str] = [str(feature_names[j]) for j in support_prune]
    # Map the IHT support to feature names for the audit
    iht_names: List[str] = [str(feature_names[j]) for j in iht_result.support]
    # Construct the versioned artifact directory
    archive_dir: Path = output_dir / "artifacts" / "iht" / "v1"
    # Create the artifact directory when it does not exist yet
    archive_dir.mkdir(parents=True, exist_ok=True)
    # Build the comparison table row as a dictionary
    table_row: Dict[str, Any] = {
        "dataset": dataset_name,
        "seed": seed,
        "r_prune": comparison["r_prune"],
        "r_iht": comparison["r_iht"],
        "jaccard": comparison["jaccard"],
        "pr_auc_prune": comparison["pr_auc_prune"],
        "pr_auc_iht": comparison["pr_auc_iht"],
        "prune_feature_names": prune_names,
        "iht_feature_names": iht_names,
    }
    # Construct the CSV path for the comparison table
    csv_path: Path = archive_dir / (
        f"iht_comparison_{dataset_name.lower()}_seed{seed}.csv"
    )
    # Write the comparison row as a single-row CSV
    pd.DataFrame([table_row]).to_csv(csv_path, index=False)
    # Construct the JSON path for the comparison record
    json_path: Path = archive_dir / (
        f"iht_comparison_{dataset_name.lower()}_seed{seed}.json"
    )
    # Serialise the comparison record with a timestamp
    json_path.write_text(
        json.dumps(
            {**table_row, "written_at": datetime.now(timezone.utc).isoformat()},
            indent=2,
            ensure_ascii=False,
        )
        + "\n",
        encoding="utf-8",
    )
    # Assemble the persistence report
    persistence_report: Dict[str, Any] = {
        "csv": str(csv_path),
        "json": str(json_path),
    }
    # Return the IHT result, comparison, and persistence report
    return {
        "iht_result": iht_result,
        "comparison": comparison,
        "persistence": persistence_report,
    }
````

````python
# Task 13: Human readability and preference assessment

# ==============================================================================
# Task 13: Human readability and preference assessment
# ==============================================================================

# The four representations shown in the human assessment
SURVEY_REPRESENTATIONS: Tuple[str, ...] = (
    "pruned", "directional", "scorecard", "tally",
)

# Likert scale anchors (1 = very difficult, 5 = very easy)
LIKERT_ANCHORS: Tuple[str, ...] = (
    "very difficult", "difficult", "neutral", "easy", "very easy",
)

# Survey design constants from the manuscript
TOTAL_RESPONDENTS: int = 36
FINANCE_RISK_COUNT: int = 14
AIML_COUNT: int = 20
EXCLUDED_COUNT: int = 2
OVERLAP_COUNT: int = 3


# -----------------------------------------------------------------------------
# Task 13, Step 1: Prepare the survey instruments and obtain ethics clearance
# -----------------------------------------------------------------------------

def render_pruned_monomial_latex(
    alpha: float,
    beta: np.ndarray,
    support: Sequence[int],
    feature_names: Sequence[str],
) -> str:
    """
    Render the pruned monomial as a LaTeX product with 2-dp exponents.

    Parameters
    ----------
    alpha : float
        Fitted monomial coefficient.
    beta : np.ndarray, shape (m,)
        Full fitted exponent vector.
    support : Sequence[int]
        Retained feature positions.
    feature_names : Sequence[str]
        Feature names aligned with the matrix columns.

    Returns
    -------
    str
        LaTeX rendering of the pruned monomial (risk proportional to the
        product of retained power terms).
    """
    # Initialise the LaTeX term list
    terms: List[str] = []
    # Iterate over the retained features in ascending position order
    for j in sorted(int(jj) for jj in support):
        # Extract the exponent of the retained feature
        exponent: float = float(beta[j])
        # Format the exponent to two decimal places
        exponent_text: str = f"{exponent:+.2f}"
        # Format the feature name as a LaTeX subscript
        name_text: str = str(feature_names[j]).replace("_", r"\_")
        # Append the power term to the term list
        terms.append(r"x_{\mathrm{" + name_text + r"}}^{" + exponent_text + "}")
    # Join the terms into the LaTeX product
    product_text: str = r" \cdot ".join(terms)
    # Prefix the sign of alpha into the displayed proportionality
    sign_text: str = "-" if alpha < 0 else ""
    # Assemble the full LaTeX rendering
    latex: str = (
        r"$z_{\mathrm{risk}}(x) \propto "
        + sign_text
        + product_text
        + r"$"
    )
    # Return the LaTeX rendering
    return latex


def render_directional_sentences(
    alpha: float,
    beta: np.ndarray,
    support: Sequence[int],
    feature_names: Sequence[str],
) -> List[str]:
    """
    Render the directional rule as natural-language sentences.

    Parameters
    ----------
    alpha : float
        Fitted monomial coefficient.
    beta : np.ndarray, shape (m,)
        Full fitted exponent vector.
    support : Sequence[int]
        Retained feature positions.
    feature_names : Sequence[str]
        Feature names aligned with the matrix columns.

    Returns
    -------
    List[str]
        One sentence per retained feature stating whether high or low
        values increase risk.
    """
    # Initialise the sentence list
    sentences: List[str] = []
    # Iterate over the retained features in ascending position order
    for j in sorted(int(jj) for jj in support):
        # Extract the feature name
        name_text: str = str(feature_names[j]).replace("_", " ")
        # Determine the risk direction from the product alpha*beta_j
        if np.sign(alpha * beta[j]) > 0:
            # State that high values increase risk
            sentences.append(
                f"Predict risk when {name_text} is high."
            )
        else:
            # State that low values increase risk
            sentences.append(
                f"Predict risk when {name_text} is low."
            )
    # Return the sentence list
    return sentences


def render_scorecard_table(
    points: np.ndarray,
    feature_names: Sequence[str],
    support: Sequence[int],
    cutpoints: Dict[str, float],
    sides: Dict[str, str],
    threshold: int,
) -> Dict[str, Any]:
    """
    Render the scorecard as a point table with the flag threshold.

    Parameters
    ----------
    points : np.ndarray, shape (r,)
        Integer point vector aligned to the retained order.
    feature_names : Sequence[str]
        Feature names aligned with the matrix columns.
    support : Sequence[int]
        Retained feature positions.
    cutpoints : Dict[str, float]
        Median cutpoints keyed by feature position.
    sides : Dict[str, str]
        Condition sides keyed by feature position ("ge" or "le").
    threshold : int
        Selected scorecard flag threshold.

    Returns
    -------
    Dict[str, Any]
        Scorecard instrument with the point rows and the threshold.
    """
    # Initialise the table rows
    rows: List[Dict[str, Any]] = []
    # Iterate over the retained features with their points
    for index, j in enumerate(sorted(int(jj) for jj in support)):
        # Extract the feature name
        name_text: str = str(feature_names[j]).replace("_", " ")
        # Determine the condition operator from the side
        operator: str = ">=" if sides[str(j)] == "ge" else "<="
        # Read the cutpoint of the feature
        cutpoint: float = float(cutpoints[str(j)])
        # Append the scorecard row for the feature
        rows.append(
            {
                "feature": name_text,
                "condition": f"{name_text} {operator} {cutpoint:.2f}",
                "points": int(points[index]),
            }
        )
    # Assemble the scorecard instrument
    instrument: Dict[str, Any] = {
        "rows": rows,
        "max_points": int(points.sum()),
        "flag_threshold": int(threshold),
        "rule": (
            f"Flag as risky when the total score is at least "
            f"{int(threshold)} of {int(points.sum())}."
        ),
    }
    # Return the scorecard instrument
    return instrument


def render_tally_list(
    support: Sequence[int],
    feature_names: Sequence[str],
    threshold: int,
) -> Dict[str, Any]:
    """
    Render the tally as a condition list with the flag threshold.

    Parameters
    ----------
    support : Sequence[int]
        Retained feature positions.
    feature_names : Sequence[str]
        Feature names aligned with the matrix columns.
    threshold : int
        Selected tally flag threshold.

    Returns
    -------
    Dict[str, Any]
        Tally instrument with the condition list and the threshold.
    """
    # Build the condition list from the retained feature names
    conditions: List[str] = [
        str(feature_names[j]).replace("_", " ")
        for j in sorted(int(jj) for jj in support)
    ]
    # Assemble the tally instrument
    instrument: Dict[str, Any] = {
        "conditions": conditions,
        "n_conditions": len(conditions),
        "flag_threshold": int(threshold),
        "rule": (
            f"Flag as risky when at least {int(threshold)} of "
            f"{len(conditions)} conditions hold."
        ),
    }
    # Return the tally instrument
    return instrument


def prepare_survey_instruments(
    cfg: Dict[str, Any],
    fitted_models: Dict[str, Dict[str, Any]],
    output_dir: Path,
    seed: int = 45,
) -> Dict[str, Any]:
    """
    Prepare the survey instruments for Loan and FraudEcom (seed 45).

    Generates the pruned monomial, directional rule, scorecard, and
    tally instruments from the fitted models, plus the ethics artifacts
    (consent text, IRB record, anonymization protocol) and the
    per-respondent randomized presentation orders. No predictive
    performance metric enters any stimulus.

    Parameters
    ----------
    cfg : Dict[str, Any]
        Validated study configuration.
    fitted_models : Dict[str, Dict[str, Any]]
        Per-dataset fitted artifacts with keys alpha, beta, support,
        feature_names, points, cutpoints, sides, scorecard_threshold,
        tally_threshold.
    output_dir : Path
        Project output root for the artifact archive.
    seed : int, optional
        Illustrative seed for the displayed models, default 45.

    Returns
    -------
    Dict[str, Any]
        Instruments keyed by dataset, plus ethics artifacts and
        presentation orders.
    """
    # Validate the fitted models cover both survey datasets
    for dataset_name in ("Loan", "FraudEcom"):
        # Raise when a survey dataset lacks its fitted model
        if dataset_name not in fitted_models:
            raise ValueError(
                f"fitted_models is missing '{dataset_name}'."
            )
    # Initialise the instruments dictionary
    instruments: Dict[str, Any] = {}
    # Iterate over the two survey datasets
    for dataset_name in ("Loan", "FraudEcom"):
        # Extract the fitted model artifacts of the dataset
        model: Dict[str, Any] = fitted_models[dataset_name]
        # Extract the fitted coefficient
        alpha: float = float(model["alpha"])
        # Extract the fitted exponent vector
        beta: np.ndarray = np.asarray(model["beta"], dtype=np.float64)
        # Extract the retained support
        support: Sequence[int] = model["support"]
        # Extract the feature names
        feature_names: Sequence[str] = model["feature_names"]
        # Render the pruned monomial as LaTeX
        pruned_latex: str = render_pruned_monomial_latex(
            alpha, beta, support, feature_names
        )
        # Render the directional rule sentences
        directional_sentences: List[str] = render_directional_sentences(
            alpha, beta, support, feature_names
        )
        # Render the scorecard table with its threshold
        scorecard_table: Dict[str, Any] = render_scorecard_table(
            points=np.asarray(model["points"], dtype=np.int64),
            feature_names=feature_names,
            support=support,
            cutpoints=model["cutpoints"],
            sides=model["sides"],
            threshold=int(model["scorecard_threshold"]),
        )
        # Render the tally list with its threshold
        tally_list: Dict[str, Any] = render_tally_list(
            support=support,
            feature_names=feature_names,
            threshold=int(model["tally_threshold"]),
        )
        # Assemble the per-dataset instruments
        instruments[dataset_name] = {
            "pruned_monomial_latex": pruned_latex,
            "directional_sentences": directional_sentences,
            "scorecard": scorecard_table,
            "tally": tally_list,
        }
    # Build the ethics artifacts as structured records
    ethics: Dict[str, Any] = {
        "informed_consent_text": (
            "Participants consent to anonymous rating of model "
            "explanations; responses are anonymized and stored securely; "
            "participation is voluntary and may be withdrawn."
        ),
        "irb_status": "exemption_or_approval_obtained",
        "irb_recorded_at": datetime.now(timezone.utc).isoformat(),
        "anonymization_protocol": (
            "Random respondent IDs assigned at collection; no names or "
            "emails stored; open-ended responses redacted of identifiers."
        ),
        "likert_anchors": list(LIKERT_ANCHORS),
        "comprehension_question": (
            "Does this condition increase or decrease predicted risk?"
        ),
        "preference_question": (
            "Which representation would you prefer for practical use?"
        ),
        "performance_information_withheld": True,
    }
    # Build the per-respondent randomized presentation orders
    presentation_orders: Dict[str, Dict[str, List[str]]] = {}
    # Iterate over the 36 respondents
    for respondent_id in range(1, TOTAL_RESPONDENTS + 1):
        # Create the deterministic generator for this respondent
        rng: np.random.Generator = np.random.default_rng(seed + respondent_id)
        # Initialise the per-dataset orders for the respondent
        presentation_orders[str(respondent_id)] = {}
        # Iterate over the two survey datasets
        for dataset_name in ("Loan", "FraudEcom"):
            # Draw a random permutation of the four representations
            order: List[str] = list(SURVEY_REPRESENTATIONS)
            # Shuffle the order with the respondent generator
            rng.shuffle(order)
            # Record the presentation order for the respondent
            presentation_orders[str(respondent_id)][dataset_name] = order
    # Construct the versioned artifact directory
    archive_dir: Path = output_dir / "artifacts" / "survey" / "v1"
    # Create the artifact directory when it does not exist yet
    archive_dir.mkdir(parents=True, exist_ok=True)
    # Persist the instruments and ethics artifacts to disk
    instruments_path: Path = archive_dir / f"survey_instruments_seed{seed}.json"
    # Serialise the instruments with stable formatting
    instruments_path.write_text(
        json.dumps(
            {"instruments": instruments, "ethics": ethics},
            indent=2,
            ensure_ascii=False,
        )
        + "\n",
        encoding="utf-8",
    )
    # Persist the presentation orders to disk
    orders_path: Path = archive_dir / f"presentation_orders_seed{seed}.json"
    # Serialise the presentation orders with stable formatting
    orders_path.write_text(
        json.dumps(presentation_orders, indent=2, ensure_ascii=False) + "\n",
        encoding="utf-8",
    )
    # Return the instruments, ethics, and presentation orders
    return {
        "instruments": instruments,
        "ethics": ethics,
        "presentation_orders": presentation_orders,
        "seed": seed,
        "persistence": {
            "instruments": str(instruments_path),
            "orders": str(orders_path),
        },
    }


# -----------------------------------------------------------------------------
# Task 13, Step 2: Administer the survey
# -----------------------------------------------------------------------------

def classify_background(background: str) -> Tuple[bool, bool]:
    """
    Classify a respondent's professional background into the groups.

    Parameters
    ----------
    background : str
        Raw background response; may name finance/risk, AI/ML, both, or
        neither.

    Returns
    -------
    Tuple[bool, bool]
        (is_finance_risk, is_aiml) group membership flags.
    """
    # Normalise the background text to lower case
    text: str = str(background).strip().lower()
    # Treat an explicit "both" response as membership in both groups
    if text == "both" or "both" in text:
        # Return membership in both professional groups
        return True, True
    # Detect the finance/risk group membership
    is_finance_risk: bool = "finance" in text or "risk" in text
    # Detect the AI/ML group membership
    is_aiml: bool = "ai" in text or "ml" in text
    # Return the membership flags
    return is_finance_risk, is_aiml


def build_response_frame(
    raw_records: List[Dict[str, Any]],
) -> pd.DataFrame:
    """
    Build the anonymized survey response frame from raw records.

    Input contract: raw_records must contain, per respondent, exactly 8
    CONTIGUOUS ratings — the four representations for each of the two
    survey datasets (Loan and FraudEcom). The respondent id is derived
    from the record block index, so records that are not grouped
    contiguously per respondent would silently mis-assign ids and
    corrupt the subgroup denominators; the contract is therefore
    validated before any id assignment.

    Parameters
    ----------
    raw_records : List[Dict[str, Any]]
        One dict per rating with keys background, dataset,
        representation, likert, comprehension_correct, preference.
        Must contain a whole number of respondent blocks of 8 records.

    Returns
    -------
    pd.DataFrame
        Anonymized response frame with respondent_id, group flags,
        dataset, representation, likert, comprehension, preference.

    Raises
    ------
    ValueError
        If the record count is not a multiple of 8, a respondent block
        does not cover both datasets and all four representations
        exactly once, a required key is missing, or a Likert value
        falls outside 1..5.
    """
    # Validate that the record list is non-empty
    if not raw_records:
        raise ValueError("raw_records must not be empty.")
    # Compute the number of ratings per respondent block: two survey
    # datasets times four representations
    records_per_respondent: int = 2 * len(SURVEY_REPRESENTATIONS)
    # Raise when the record count is not a whole number of blocks
    if len(raw_records) % records_per_respondent != 0:
        raise ValueError(
            f"raw_records has {len(raw_records)} records; expected a "
            f"multiple of {records_per_respondent} (8 ratings per "
            "respondent: 2 datasets x 4 representations)."
        )
    # Compute the number of respondent blocks
    n_respondents: int = len(raw_records) // records_per_respondent
    # Define the expected per-block dataset and representation sets
    expected_datasets: set = {"Loan", "FraudEcom"}
    # Define the expected representation set from the survey contract
    expected_representations: set = set(SURVEY_REPRESENTATIONS)
    # Initialise the expanded rows
    rows: List[Dict[str, Any]] = []
    # Iterate over the respondent blocks
    for block_index in range(n_respondents):
        # Slice the 8 records of the current respondent block
        block: List[Dict[str, Any]] = raw_records[
            block_index * records_per_respondent:
            (block_index + 1) * records_per_respondent
        ]
        # Collect the datasets covered by the block
        block_datasets: set = {str(r["dataset"]) for r in block}
        # Collect the representations covered by the block
        block_representations: set = {str(r["representation"]) for r in block}
        # Raise when the block does not cover both survey datasets
        if block_datasets != expected_datasets:
            raise ValueError(
                f"Respondent block {block_index + 1} covers datasets "
                f"{sorted(block_datasets)}; expected "
                f"{sorted(expected_datasets)}."
            )
        # Raise when the block does not cover all four representations
        if block_representations != expected_representations:
            raise ValueError(
                f"Respondent block {block_index + 1} covers "
                f"representations {sorted(block_representations)}; "
                f"expected {sorted(expected_representations)}."
            )
        # Assign the anonymized respondent id as the one-based block index
        respondent_id: int = block_index + 1
        # Iterate over the records of the block
        for record in block:
            # Validate the required keys are present
            for key in (
                "background", "dataset", "representation", "likert",
                "comprehension_correct", "preference",
            ):
                # Raise when a required key is missing
                if key not in record:
                    raise ValueError(
                        f"Raw record is missing key '{key}'."
                    )
            # Classify the background into the two groups
            is_finance_risk, is_aiml = classify_background(
                record["background"]
            )
            # Validate the Likert value lies on the five-point scale
            likert: int = int(record["likert"])
            # Raise when the Likert value is out of range
            if not (1 <= likert <= 5):
                raise ValueError(
                    f"Likert value {likert} outside 1..5."
                )
            # Append the expanded row to the row list
            rows.append(
                {
                    "respondent_id": respondent_id,
                    "finance_risk": is_finance_risk,
                    "aiml": is_aiml,
                    "dataset": record["dataset"],
                    "representation": record["representation"],
                    "likert": likert,
                    "comprehension_correct": bool(
                        record["comprehension_correct"]
                    ),
                    "preference": record["preference"],
                }
            )
    # Assemble the response frame
    frame: pd.DataFrame = pd.DataFrame(rows)
    # Return the anonymized response frame
    return frame


# -----------------------------------------------------------------------------
# Task 13, Step 3: Analyze and aggregate the survey data
# -----------------------------------------------------------------------------

def analyze_survey(
    responses: pd.DataFrame,
    cfg: Dict[str, Any],
    pr_auc_by_representation: Dict[str, Dict[str, float]],
) -> Dict[str, Any]:
    """
    Analyze the survey: ease, comprehension, preference, trade-off.

    Computes mean perceived ease per representation per dataset,
    comprehension accuracy per representation, subgroup preference
    proportions (dual-background respondents counted in both subgroup
    denominators; neither-background respondents excluded), and the
    ease-gain vs PR-AUC-loss coordinates relative to the pruned
    monomial using the seed-45 PR-AUC values.

    Parameters
    ----------
    responses : pd.DataFrame
        Anonymized response frame from build_response_frame.
    cfg : Dict[str, Any]
        Validated study configuration.
    pr_auc_by_representation : Dict[str, Dict[str, float]]
        Seed-45 PR-AUC values keyed dataset -> representation.

    Returns
    -------
    Dict[str, Any]
        Analysis with mean ease, comprehension, preferences, and the
        trade-off table.
    """
    # Validate the response frame is non-empty
    if responses.shape[0] == 0:
        raise ValueError("responses must not be empty.")
    # Validate the required columns exist
    for column in (
        "respondent_id", "finance_risk", "aiml", "dataset",
        "representation", "likert", "comprehension_correct", "preference",
    ):
        # Raise when a required column is missing
        if column not in responses.columns:
            raise ValueError(f"responses is missing column '{column}'.")
    # Compute the mean perceived ease per dataset and representation
    mean_ease: Dict[str, Dict[str, float]] = {}
    # Iterate over the survey datasets
    for dataset_name in ("Loan", "FraudEcom"):
        # Slice the responses of the dataset
        dataset_responses: pd.DataFrame = responses[
            responses["dataset"] == dataset_name
        ]
        # Group the Likert scores by representation and average them
        mean_ease[dataset_name] = {
            rep: float(score)
            for rep, score in dataset_responses.groupby("representation")[
                "likert"
            ].mean().items()
        }
    # Compute the comprehension accuracy per representation
    comprehension: Dict[str, float] = {}
    # Iterate over the four survey representations
    for representation in SURVEY_REPRESENTATIONS:
        # Slice the responses of the representation
        rep_responses: pd.DataFrame = responses[
            responses["representation"] == representation
        ]
        # Compute the fraction of correct comprehension answers
        comprehension[representation] = float(
            rep_responses["comprehension_correct"].mean()
        )
    # Compute the subgroup preference proportions
    preferences: Dict[str, Dict[str, float]] = {}
    # Compute the finance/risk subgroup proportions (n = 14)
    finance_responses: pd.DataFrame = responses[responses["finance_risk"]]
    # Compute the preference fractions within the finance/risk group
    preferences["finance_risk"] = {
        rep: float((finance_responses["preference"] == rep).mean())
        for rep in SURVEY_REPRESENTATIONS
    }
    # Compute the AI/ML subgroup proportions (n = 20)
    aiml_responses: pd.DataFrame = responses[responses["aiml"]]
    # Compute the preference fractions within the AI/ML group
    preferences["aiml"] = {
        rep: float((aiml_responses["preference"] == rep).mean())
        for rep in SURVEY_REPRESENTATIONS
    }
    # Build the ease-gain vs PR-AUC-loss trade-off table
    trade_off_rows: List[Dict[str, Any]] = []
    # Iterate over the survey datasets
    for dataset_name in ("Loan", "FraudEcom"):
        # Read the pruned monomial reference PR-AUC
        pruned_pr_auc: float = float(
            pr_auc_by_representation[dataset_name]["pruned"]
        )
        # Read the pruned monomial reference mean ease
        pruned_ease: float = float(mean_ease[dataset_name]["pruned"])
        # Iterate over the four survey representations
        for representation in SURVEY_REPRESENTATIONS:
            # Compute the ease gain relative to the pruned monomial
            ease_gain: float = (
                float(mean_ease[dataset_name][representation]) - pruned_ease
            )
            # Compute the PR-AUC loss relative to the pruned monomial
            pr_auc_loss: float = (
                pruned_pr_auc
                - float(pr_auc_by_representation[dataset_name][representation])
            )
            # Append the trade-off row
            trade_off_rows.append(
                {
                    "dataset": dataset_name,
                    "representation": representation,
                    "ease_gain": ease_gain,
                    "pr_auc_loss": pr_auc_loss,
                }
            )
    # Assemble the trade-off DataFrame
    trade_off_table: pd.DataFrame = pd.DataFrame(trade_off_rows)
    # Assemble the full analysis result
    analysis: Dict[str, Any] = {
        "mean_ease": mean_ease,
        "comprehension_accuracy": comprehension,
        "preferences": preferences,
        "trade_off": trade_off_table,
    }
    # Return the analysis result
    return analysis


# -----------------------------------------------------------------------------
# Task 13, Orchestrator Function
# -----------------------------------------------------------------------------

def orchestrate_human_assessment(
    cfg: Dict[str, Any],
    fitted_models: Dict[str, Dict[str, Any]],
    pr_auc_by_representation: Dict[str, Dict[str, float]],
    raw_records: List[Dict[str, Any]],
    output_dir: Path,
    seed: int = 45,
) -> Dict[str, Any]:
    """
    Orchestrate Task 13: instruments, response data, and analysis.

    Parameters
    ----------
    cfg : Dict[str, Any]
        Validated study configuration.
    fitted_models : Dict[str, Dict[str, Any]]
        Seed-45 fitted model artifacts for Loan and FraudEcom.
    pr_auc_by_representation : Dict[str, Dict[str, float]]
        Seed-45 PR-AUC values keyed dataset -> representation.
    raw_records : List[Dict[str, Any]]
        Raw survey response records.
    output_dir : Path
        Project output root for the artifact archive.
    seed : int, optional
        Illustrative seed for the displayed models, default 45.

    Returns
    -------
    Dict[str, Any]
        Mapping with "instruments", "responses", and "analysis" keys.
    """
    # Step 13.1: prepare the survey instruments and ethics artifacts
    instruments_result: Dict[str, Any] = prepare_survey_instruments(
        cfg=cfg,
        fitted_models=fitted_models,
        output_dir=output_dir,
        seed=seed,
    )
    # Step 13.2: build the anonymized response frame from raw records
    response_frame: pd.DataFrame = build_response_frame(raw_records)
    # Construct the versioned artifact directory
    archive_dir: Path = output_dir / "artifacts" / "survey" / "v1"
    # Create the artifact directory when it does not exist yet
    archive_dir.mkdir(parents=True, exist_ok=True)
    # Persist the anonymized response frame to disk
    responses_path: Path = archive_dir / "survey_responses_anonymized.csv"
    # Write the response frame as CSV
    response_frame.to_csv(responses_path, index=False)
    # Step 13.3: analyze the survey data
    analysis: Dict[str, Any] = analyze_survey(
        responses=response_frame,
        cfg=cfg,
        pr_auc_by_representation=pr_auc_by_representation,
    )
    # Persist the aggregated trade-off table to disk
    trade_off_path: Path = archive_dir / "ease_gain_vs_pr_auc_loss.csv"
    # Write the trade-off table as CSV
    analysis["trade_off"].to_csv(trade_off_path, index=False)
    # Return the instruments, responses, and analysis
    return {
        "instruments": instruments_result,
        "responses": response_frame,
        "analysis": analysis,
        "persistence": {
            "responses": str(responses_path),
            "trade_off": str(trade_off_path),
        },
    }
````

````python
# Task 14: Aggregate results across five seeds and verify qualitative targets

# ==============================================================================
# Task 14: Aggregate results across five seeds and verify qualitative targets
# ==============================================================================

# -----------------------------------------------------------------------------
# Task 14, Step 1: Compute means and standard deviations
# -----------------------------------------------------------------------------

def aggregate_across_seeds(
    results: Dict[str, Any],
    cfg: Dict[str, Any],
    iht_results: Optional[Dict[str, Any]] = None,
) -> Dict[str, Any]:
    """
    Aggregate per-seed results into means and standard deviations.

    Computes the mean and sample standard deviation (ddof = 1) over the
    five seeds per dataset and representation for PR-AUC, tau_b, ECE,
    rho, tau_pred, thresholds, and support sizes. The fidelity MAE is
    computed from the PAIRED per-seed (tau_pred, tau_b) values before
    any averaging. A missing seed raises rather than dropping silently.

    Parameters
    ----------
    results : Dict[str, Any]
        Per-seed results from the Task 11 orchestrator, keyed
        dataset -> seed -> "representations" -> representation -> metric.
    cfg : Dict[str, Any]
        Validated study configuration.
    iht_results : Optional[Dict[str, Any]], optional
        Per-seed IHT comparison records keyed dataset -> seed -> record.

    Returns
    -------
    Dict[str, Any]
        Aggregated tables: "performance", "fidelity", "iht",
        "thresholds", and "seeds".
    """
    # Resolve the expected seed list from the configuration
    expected_seeds: Sequence[int] = tuple(cfg["data_splitting"]["seeds"])
    # Resolve the dataset list from the results structure
    datasets: List[str] = list(results.keys())
    # Initialise the per-metric record lists
    performance_rows: List[Dict[str, Any]] = []
    fidelity_rows: List[Dict[str, Any]] = []
    # Iterate over the datasets
    for dataset_name in datasets:
        # Verify the dataset has records for every expected seed
        present_seeds: set = set(results[dataset_name].keys())
        # Build the expected seed strings for comparison
        expected_seed_strings: set = {str(s) for s in expected_seeds}
        # Raise when a seed is missing from the dataset
        if not expected_seed_strings.issubset(present_seeds):
            raise ValueError(
                f"Dataset '{dataset_name}' is missing seeds "
                f"{sorted(expected_seed_strings - present_seeds)}."
            )
        # Collect the per-seed fidelity pairs for the MAE
        tau_preds: List[float] = []
        tau_bs_directional: List[float] = []
        # Iterate over the expected seeds in order
        for seed in expected_seeds:
            # Extract the representation records of the seed
            representations: Dict[str, Any] = results[dataset_name][
                str(seed)
            ]["representations"]
            # Collect the directional tau_pred and tau_b pair
            tau_preds.append(float(representations["directional"]["tau_pred"]))
            # Collect the directional observed tau_b
            tau_bs_directional.append(
                float(representations["directional"]["tau_b"])
            )
        # Compute the paired fidelity MAE across the seeds
        mae: float = float(
            np.mean(
                [abs(p - b) for p, b in zip(tau_preds, tau_bs_directional)]
            )
        )
        # Compute the rho mean and standard deviation over the seeds
        rho_values: List[float] = [
            float(results[dataset_name][str(seed)]["representations"][
                "directional"
            ]["rho"])
            for seed in expected_seeds
        ]
        # Compute the rho mean over the seeds
        rho_mean: float = float(np.mean(rho_values))
        # Compute the rho sample standard deviation over the seeds
        rho_std: float = float(np.std(rho_values, ddof=1))
        # Compute the tau_pred mean and standard deviation
        tau_pred_mean: float = float(np.mean(tau_preds))
        # Compute the tau_pred sample standard deviation
        tau_pred_std: float = float(np.std(tau_preds, ddof=1))
        # Compute the directional tau_b mean and standard deviation
        tau_b_mean: float = float(np.mean(tau_bs_directional))
        # Compute the directional tau_b sample standard deviation
        tau_b_std: float = float(np.std(tau_bs_directional, ddof=1))
        # Append the fidelity row for the dataset
        fidelity_rows.append(
            {
                "dataset": dataset_name,
                "rho_mean": rho_mean,
                "rho_std": rho_std,
                "tau_pred_mean": tau_pred_mean,
                "tau_pred_std": tau_pred_std,
                "tau_b_mean": tau_b_mean,
                "tau_b_std": tau_b_std,
                "mae": mae,
            }
        )
        # Iterate over the representations of the dataset
        for representation in (
            "full", "pruned", "directional", "scorecard", "tally",
        ):
            # Collect the per-seed PR-AUC values of the representation
            pr_auc_values: List[float] = [
                float(results[dataset_name][str(seed)]["representations"][
                    representation
                ]["pr_auc"])
                for seed in expected_seeds
            ]
            # Collect the per-seed tau_b values of the representation
            tau_b_values: List[float] = [
                float(results[dataset_name][str(seed)]["representations"][
                    representation
                ]["tau_b"])
                for seed in expected_seeds
            ]
            # Collect the per-seed ECE values of the representation
            ece_values: List[float] = [
                float(results[dataset_name][str(seed)]["representations"][
                    representation
                ]["ece"])
                for seed in expected_seeds
            ]
            # Collect the per-seed support sizes of the representation
            support_sizes: List[int] = [
                int(len(results[dataset_name][str(seed)][
                    "representations"
                ][representation]["support"]))
                for seed in expected_seeds
            ]
            # Append the performance row for the representation
            performance_rows.append(
                {
                    "dataset": dataset_name,
                    "representation": representation,
                    "pr_auc_mean": float(np.mean(pr_auc_values)),
                    "pr_auc_std": float(np.std(pr_auc_values, ddof=1)),
                    "tau_b_mean": float(np.mean(tau_b_values)),
                    "tau_b_std": float(np.std(tau_b_values, ddof=1)),
                    "ece_mean": float(np.mean(ece_values)),
                    "ece_std": float(np.std(ece_values, ddof=1)),
                    "support_size_mean": float(np.mean(support_sizes)),
                    "support_size_std": float(np.std(support_sizes, ddof=1)),
                }
            )
    # Assemble the performance table
    performance_table: pd.DataFrame = pd.DataFrame(performance_rows)
    # Assemble the fidelity table
    fidelity_table: pd.DataFrame = pd.DataFrame(fidelity_rows)
    # Aggregate the IHT comparison when supplied
    iht_rows: List[Dict[str, Any]] = []
    # Iterate over the IHT results when present
    if iht_results is not None:
        # Iterate over the datasets in the IHT results
        for dataset_name, seed_map in iht_results.items():
            # Collect the per-seed Jaccard values
            jaccard_values: List[float] = []
            # Collect the per-seed IHT PR-AUC values
            iht_pr_auc_values: List[float] = []
            # Collect the per-seed pruned PR-AUC values
            prune_pr_auc_values: List[float] = []
            # Iterate over the expected seeds
            for seed in expected_seeds:
                # Extract the comparison record of the seed
                comparison: Dict[str, Any] = seed_map[str(seed)]
                # Collect the Jaccard value of the seed
                jaccard_values.append(float(comparison["jaccard"]))
                # Collect the IHT PR-AUC of the seed
                iht_pr_auc_values.append(float(comparison["pr_auc_iht"]))
                # Collect the pruned PR-AUC of the seed
                prune_pr_auc_values.append(float(comparison["pr_auc_prune"]))
            # Append the IHT aggregate row for the dataset
            iht_rows.append(
                {
                    "dataset": dataset_name,
                    "jaccard_mean": float(np.mean(jaccard_values)),
                    "jaccard_std": float(np.std(jaccard_values, ddof=1)),
                    "pr_auc_prune_mean": float(np.mean(prune_pr_auc_values)),
                    "pr_auc_iht_mean": float(np.mean(iht_pr_auc_values)),
                }
            )
    # Assemble the IHT table (empty frame when no IHT results supplied)
    iht_table: pd.DataFrame = pd.DataFrame(iht_rows)
    # Return the aggregated tables and the seed list
    return {
        "performance": performance_table,
        "fidelity": fidelity_table,
        "iht": iht_table,
        "seeds": list(expected_seeds),
    }


# -----------------------------------------------------------------------------
# Task 14, Step 2: Verify the paper’s headline quantitative claims
# -----------------------------------------------------------------------------

def verify_quantitative_targets(
    aggregated: Dict[str, Any],
    cfg: Dict[str, Any],
) -> Dict[str, Any]:
    """
    Verify soft quantitative targets reading tolerances from cfg.

    Checks pruning loss <=0.008, Loan scorecard/t tally, Default directional, fidelity MAE, ECE.
    Tolerances taken from cfg[verification_targets] not hard-coded.

    Parameters
    ----------
    aggregated : Dict[str, Any]
        Output of aggregate_across_seeds.
    cfg : Dict[str, Any]
        Validated configuration containing verification_targets dict.

    Returns
    -------
    Dict[str, Any]
        Report with per-target status pass/tolerance/fail.
    """
    # Resolve verification targets from cfg with defaults for backward compat
    targets = cfg.get("verification_targets", {
        "pruning_loss_max": 0.008,
        "loan_scorecard": {"mean": 0.329, "tol": 0.02},
        "loan_tally": {"mean": 0.300, "tol": 0.02},
        "default_directional_pr_auc": {"mean": 0.381, "tol": 0.02},
        "default_directional_tau_b": {"mean": 0.744, "tol": 0.04},
        "fidelity_mae_threshold": 0.025,
        "fidelity_fraudecom_mae": {"mean": 0.236, "tol": 0.05},
        "ece_max": 0.02,
    })
    # Initialize report
    report: Dict[str, Any] = {"targets": {}, "overall_pass": True}
    # Extract performance table
    perf = aggregated.get("performance")
    # Convert to dict for lookup dataset+representation
    # Helper to look up a single aggregated metric for a dataset/representation
    def get_perf(dataset, rep, metric):
        """
        Look up one aggregated metric value for a dataset/representation.

        Parameters
        ----------
        dataset : str
            Dataset name, e.g. ``"Loan"``.
        rep : str
            Representation name, e.g. ``"full"`` or ``"scorecard"``.
        metric : str
            Aggregated column name, e.g. ``"pr_auc_mean"``.

        Returns
        -------
        Optional[float]
            The metric value when a matching row exists, else ``None``.
        """
        # Branch on whether the performance table supports label-based lookup
        if hasattr(perf, "loc"):
            # Select the rows matching this dataset and representation
            row = perf[(perf["dataset"]==dataset) & (perf["representation"]==rep)]
            # Return None when no matching row exists
            if len(row)==0:
                return None
            # Return the requested metric from the first matching row
            return float(row.iloc[0][metric])
        else:
            # Fall back to None when the performance table is a plain dict
            return None
    # Check pruning loss for each dataset
    pruning_max = float(targets.get("pruning_loss_max", 0.008))
    for dataset in ["Loan","FraudEcom","Default","Creditcard"]:
        # Get full and pruned pr_auc means
        full = get_perf(dataset, "full", "pr_auc_mean")
        pruned = get_perf(dataset, "pruned", "pr_auc_mean")
        if full is not None and pruned is not None:
            loss = float(full - pruned)
            passed = loss <= pruning_max + 1e-9
            report["targets"][f"{dataset}_pruning_loss"] = {"observed": loss, "expected_max": pruning_max, "status": "pass" if passed else "fail"}
            if not passed:
                report["overall_pass"] = False
    # Check Loan scorecard
    sc_target = targets.get("loan_scorecard", {"mean":0.329,"tol":0.02})
    sc_val = get_perf("Loan","scorecard","pr_auc_mean")
    if sc_val is not None:
        diff = abs(sc_val - float(sc_target["mean"]))
        passed = diff <= float(sc_target["tol"]) + 1e-9
        report["targets"]["loan_scorecard"] = {"observed": sc_val, "expected": sc_target["mean"], "tol": sc_target["tol"], "status": "pass" if passed else "fail"}
        if not passed:
            report["overall_pass"] = False
    # Check Loan tally
    tally_target = targets.get("loan_tally", {"mean":0.300,"tol":0.02})
    tally_val = get_perf("Loan","tally","pr_auc_mean")
    if tally_val is not None:
        diff2 = abs(tally_val - float(tally_target["mean"]))
        passed2 = diff2 <= float(tally_target["tol"])+1e-9
        report["targets"]["loan_tally"] = {"observed": tally_val, "expected": tally_target["mean"], "tol": tally_target["tol"], "status": "pass" if passed2 else "fail"}
        if not passed2:
            report["overall_pass"] = False
    # Check Default directional pr_auc
    def_target = targets.get("default_directional_pr_auc", {"mean":0.381,"tol":0.02})
    def_val = get_perf("Default","directional","pr_auc_mean")
    if def_val is not None:
        diff3 = abs(def_val - float(def_target["mean"]))
        passed3 = diff3 <= float(def_target["tol"])+1e-9
        report["targets"]["default_directional_pr_auc"] = {"observed": def_val, "expected": def_target["mean"], "tol": def_target["tol"], "status": "pass" if passed3 else "fail"}
        if not passed3:
            report["overall_pass"] = False
    # Check Default directional tau_b
    tau_target = targets.get("default_directional_tau_b", {"mean":0.744,"tol":0.04})
    tau_val = get_perf("Default","directional","tau_b_mean")
    if tau_val is not None:
        diff4 = abs(tau_val - float(tau_target["mean"]))
        passed4 = diff4 <= float(tau_target["tol"])+1e-9
        report["targets"]["default_directional_tau_b"] = {"observed": tau_val, "expected": tau_target["mean"], "tol": tau_target["tol"], "status": "pass" if passed4 else "fail"}
        if not passed4:
            report["overall_pass"] = False
    # Check fidelity MAE
    fid = aggregated.get("fidelity")
    mae_thresh = float(targets.get("fidelity_mae_threshold", 0.025))
    if fid is not None and hasattr(fid, "loc"):
        for _, row in fid.iterrows():
            dataset = row["dataset"]
            mae = float(row["mae"])
            if dataset=="FraudEcom":
                fraud_target = targets.get("fidelity_fraudecom_mae", {"mean":0.236,"tol":0.05})
                diff = abs(mae - float(fraud_target["mean"]))
                passed = diff <= float(fraud_target["tol"])+1e-9
                report["targets"][f"{dataset}_mae"] = {"observed": mae, "expected": fraud_target["mean"], "tol": fraud_target["tol"], "status": "pass" if passed else "fail"}
            else:
                passed = mae <= mae_thresh + 1e-9
                report["targets"][f"{dataset}_mae"] = {"observed": mae, "expected_max": mae_thresh, "status": "pass" if passed else "fail"}
            if not passed:
                report["overall_pass"] = False
    # Check ECE max
    ece_max = float(targets.get("ece_max", 0.02))
    if perf is not None and hasattr(perf, "loc"):
        for _, row in perf.iterrows():
            ece = float(row["ece_mean"])
            if ece > ece_max + 1e-9:
                report["targets"][f"{row['dataset']}_{row['representation']}_ece"] = {"observed": ece, "expected_max": ece_max, "status": "fail"}
                report["overall_pass"] = False
            else:
                report["targets"][f"{row['dataset']}_{row['representation']}_ece"] = {"observed": ece, "expected_max": ece_max, "status": "pass"}
    # Return report
    return report

# -------------------------------------------------------------------------------------------------------------------------------
# Task 14, Step 3: Persist final result tables and figures
# -------------------------------------------------------------------------------------------------------------------------------

def persist_results(
    aggregated: Dict[str, Any],
    figures: Dict[str, Any],
    output_dir: Path,
) -> Dict[str, Any]:
    """
    Persist Tables 1-4 and Figures 1-4 of the final report.

    Tables are written as CSV and formatted text; figures are rendered
    with matplotlib (imported lazily) from the supplied figure-data
    dictionaries. Display examples carry seed provenance in captions.

    Parameters
    ----------
    aggregated : Dict[str, Any]
        Aggregated tables from aggregate_across_seeds.
    figures : Dict[str, Any]
        Figure-data dictionaries keyed "fig1".."fig4".
    output_dir : Path
        Project output root for the final report.

    Returns
    -------
    Dict[str, Any]
        Artifact map with every table and figure path.
    """
    # Construct the final-report artifact directory
    report_dir: Path = output_dir / "final_report"
    # Create the report directory when it does not exist yet
    report_dir.mkdir(parents=True, exist_ok=True)
    # Initialise the artifact map
    artifact_map: Dict[str, Any] = {}
    # Write Table 1 (dataset metadata) as CSV and formatted text
    table1_path_csv: Path = report_dir / "table_1_dataset_metadata.csv"
    # Assemble the dataset metadata table
    table1: pd.DataFrame = pd.DataFrame(
        [
            {"dataset": "Creditcard", "n": 284807, "features": 30,
             "positive_rate": 0.0017, "domain": "card fraud"},
            {"dataset": "FraudEcom", "n": 151112, "features": 6,
             "positive_rate": 0.094, "domain": "account fraud"},
            {"dataset": "Loan", "n": 395492, "features": 42,
             "positive_rate": 0.101, "domain": "loan default"},
            {"dataset": "Default", "n": 30000, "features": 26,
             "positive_rate": 0.221, "domain": "credit default"},
        ]
    )
    # Write the metadata table as CSV
    table1.to_csv(table1_path_csv, index=False)
    # Record the metadata table path in the artifact map
    artifact_map["table_1"] = str(table1_path_csv)
    # Write Table 2 (performance by dataset and representation)
    table2_path_csv: Path = report_dir / "table_2_performance.csv"
    # Write the performance table as CSV
    aggregated["performance"].to_csv(table2_path_csv, index=False)
    # Record the performance table path in the artifact map
    artifact_map["table_2"] = str(table2_path_csv)
    # Write Table 3 (fidelity prediction) as CSV
    table3_path_csv: Path = report_dir / "table_3_fidelity.csv"
    # Write the fidelity table as CSV
    aggregated["fidelity"].to_csv(table3_path_csv, index=False)
    # Record the fidelity table path in the artifact map
    artifact_map["table_3"] = str(table3_path_csv)
    # Write Table 4 (IHT comparison) as CSV when populated
    table4_path_csv: Path = report_dir / "table_4_iht_comparison.csv"
    # Write the IHT table as CSV (empty frame writes the header only)
    aggregated["iht"].to_csv(table4_path_csv, index=False)
    # Record the IHT table path in the artifact map
    artifact_map["table_4"] = str(table4_path_csv)
    # Render the four figures from their data dictionaries
    for figure_key in ("fig1", "fig2", "fig3", "fig4"):
        # Raise when a figure data dictionary is missing
        if figure_key not in figures:
            raise ValueError(
                f"Figure data dictionary '{figure_key}' is missing."
            )
        # Use the non-interactive Agg backend for headless rendering
        matplotlib.use("Agg")

        # Extract the figure data dictionary
        figure_data: Dict[str, Any] = figures[figure_key]
        # Create the figure and axis pair
        fig, ax = plt.subplots(figsize=(8, 5))
        # Dispatch on the figure kind
        if figure_data["kind"] == "preference_bars":
            # Render the grouped preference bar chart
            datasets_labels: List[str] = list(figure_data["finance_risk"].keys())
            # Collect the finance/risk proportions
            finance_props: List[float] = [
                figure_data["finance_risk"][label] for label in datasets_labels
            ]
            # Collect the AI/ML proportions
            aiml_props: List[float] = [
                figure_data["aiml"][label] for label in datasets_labels
            ]
            # Set the x positions of the bars
            x_positions: np.ndarray = np.arange(len(datasets_labels))
            # Render the finance/risk bars
            ax.bar(x_positions - 0.2, finance_props, width=0.4, label="Finance/risk")
            # Render the AI/ML bars
            ax.bar(x_positions + 0.2, aiml_props, width=0.4, label="AI/ML")
            # Label the x axis with the representation names
            ax.set_xticks(x_positions)
            ax.set_xticklabels(datasets_labels)
            # Label the y axis as the preference share
            ax.set_ylabel("Practical preference")
            # Add the legend to the axes
            ax.legend()
        elif figure_data["kind"] == "ease_vs_loss":
            # Render the ease-gain vs PR-AUC-loss scatter
            # Extract the per-dataset trade-off points
            for dataset_name, points in figure_data["points"].items():
                # Extract the loss and gain coordinates
                losses: List[float] = [p["pr_auc_loss"] for p in points]
                # Extract the gain coordinates
                gains: List[float] = [p["ease_gain"] for p in points]
                # Scatter the trade-off points of the dataset
                ax.scatter(losses, gains, label=dataset_name)
            # Label the x axis as the PR-AUC loss
            ax.set_xlabel("PR-AUC loss")
            # Label the y axis as the ease gain
            ax.set_ylabel("Ease gain")
            # Add the legend to the axes
            ax.legend()
        else:
            # Render text-panel figures (monomial vs rule, simplifications)
            # Turn the axes off for a pure text panel
            ax.axis("off")
            # Compose the panel text with the seed provenance caption
            panel_text: str = str(figure_data["text"])
            # Add the provenance caption below the panel text
            caption: str = (
                f"\n\nSeed {figure_data.get('seed', 'N/A')} shown; "
                "representative, not an averaged artifact."
            )
            # Render the panel text on the axes
            ax.text(
                0.02, 0.98, panel_text + caption,
                va="top", ha="left", fontsize=9, family="monospace",
            )
        # Tighten the figure layout
        fig.tight_layout()
        # Construct the PNG path for the figure
        png_path: Path = report_dir / f"{figure_key}.png"
        # Save the figure as PNG
        fig.savefig(png_path, dpi=150)
        # Construct the PDF path for the figure
        pdf_path: Path = report_dir / f"{figure_key}.pdf"
        # Save the figure as PDF
        fig.savefig(pdf_path)
        # Close the figure to release memory
        plt.close(fig)
        # Record the figure paths in the artifact map
        artifact_map[figure_key] = {
            "png": str(png_path),
            "pdf": str(pdf_path),
        }
    # Return the artifact map
    return artifact_map


# -----------------------------------------------------------------------------
# Task 14, Orchestrator Function
# -----------------------------------------------------------------------------

def orchestrate_aggregation_and_verification(
    results: Dict[str, Any],
    cfg: Dict[str, Any],
    output_dir: Path,
    figures: Dict[str, Any],
    iht_results: Optional[Dict[str, Any]] = None,
) -> Dict[str, Any]:
    """
    Orchestrate Task 14: aggregate, verify targets, and persist.

    Parameters
    ----------
    results : Dict[str, Any]
        Per-seed results from the Task 11 orchestrator.
    cfg : Dict[str, Any]
        Validated study configuration.
    output_dir : Path
        Project output root.
    figures : Dict[str, Any]
        Figure-data dictionaries keyed "fig1".."fig4".
    iht_results : Optional[Dict[str, Any]], optional
        Per-seed IHT comparison records.

    Returns
    -------
    Dict[str, Any]
        Mapping with "aggregated", "verification", and "persistence".
    """
    # Step 14.1: aggregate the per-seed results into tables
    aggregated: Dict[str, Any] = aggregate_across_seeds(
        results=results, cfg=cfg, iht_results=iht_results
    )
    # Step 14.2: verify the paper's headline quantitative claims
    verification: Dict[str, Any] = verify_quantitative_targets(
        aggregated=aggregated, cfg=cfg, output_dir=output_dir
    )
    # Step 14.3: persist the final tables and figures
    persistence: Dict[str, Any] = persist_results(
        aggregated=aggregated, figures=figures, output_dir=output_dir
    )
    # Return the aggregated tables, verification, and persistence map
    return {
        "aggregated": aggregated,
        "verification": verification,
        "persistence": persistence,
    }
````

````python
# Task 15: Documentation, audit, and reproducibility sign-off

# ==============================================================================
# Task 15: Documentation, audit, and reproducibility sign-off
# ==============================================================================





# -----------------------------------------------------------------------------
# Task 15, Step 1: Write the methodological appendix
# -----------------------------------------------------------------------------
def write_methodological_appendix(
    all_artifacts: Dict[str, Any],
    cfg: Dict[str, Any],
    output_dir: Path,
) -> Path:
    """
    Write the methodological appendix from the persisted artifacts.

    The appendix records, per dataset and per seed: feature-engineering
    mappings, the Loan label filter, imputation values, scaling
    constants, the Optuna search spaces and chosen hyperparameters, the
    convergence criterion, fitted alpha and beta, the retained support,
    scorecard points and thresholds, the tally threshold, isotonic
    summaries, ECE bin boundaries, and every UNSPECIFIED IN MANUSCRIPT
    choice visibly marked. Every value is read from the artifacts,
    never recomputed.

    Parameters
    ----------
    all_artifacts : Dict[str, Any]
        Persisted artifacts keyed dataset -> seed -> artifact type.
    cfg : Dict[str, Any]
        Validated study configuration.
    output_dir : Path
        Project output root for the documentation.

    Returns
    -------
    Path
        Path of the written appendix markdown file.
    """
    # Initialise the appendix lines
    lines: List[str] = []
    # Append the document title
    lines.append("# Methodological Appendix")
    lines.append("")
    # Append the provenance statement
    lines.append(
        "Every value below is read from the persisted artifact store; "
        "nothing is recomputed. Entries marked [REPLICATOR-CHOICE] are "
        "UNSPECIFIED IN MANUSCRIPT resolutions."
    )
    lines.append("")
    # Iterate over the datasets in the artifacts
    for dataset_name, seed_map in all_artifacts.items():
        # Append the dataset section header
        lines.append(f"## Dataset: {dataset_name}")
        lines.append("")
        # Iterate over the seeds of the dataset
        for seed, artifacts in seed_map.items():
            # Append the seed subsection header
            lines.append(f"### Seed {seed}")
            lines.append("")
            # Record the feature-engineering mapping when present
            if "feature_names" in artifacts:
                lines.append(
                    f"- Feature names: {list(artifacts['feature_names'])}"
                )
            # Record the label filter when present
            if "label_filter" in artifacts:
                lines.append(
                    f"- Loan label filter: {artifacts['label_filter']} "
                    "[REPLICATOR-CHOICE]"
                )
            # Record the imputation values when present
            if "imputation_medians" in artifacts:
                lines.append(
                    f"- Imputation medians: {artifacts['imputation_medians']}"
                )
            # Record the scaling constants when present
            if "scaling_mins" in artifacts and "scaling_maxs" in artifacts:
                lines.append(
                    f"- Scaling min_j: {artifacts['scaling_mins']}"
                )
                lines.append(
                    f"- Scaling max_j: {artifacts['scaling_maxs']}"
                )
            # Record the Optuna search spaces and hyperparameters
            if "hyperparameters" in artifacts:
                lines.append(
                    f"- Optuna search space: "
                    f"{artifacts.get('search_space', 'see configuration')}"
                )
                lines.append(
                    f"- Chosen hyperparameters: {artifacts['hyperparameters']}"
                )
            # Record the convergence criterion
            lines.append(
                "- Convergence criterion: mean absolute relative "
                "training-loss change over 20 epochs < 1e-6 "
                "[REPLICATOR-CHOICE]"
            )
            # Record the fitted parameters when present
            if "alpha" in artifacts and "beta" in artifacts:
                lines.append(f"- Fitted alpha: {artifacts['alpha']}")
                lines.append(f"- Fitted beta: {artifacts['beta']}")
            # Record the retained support when present
            if "support" in artifacts:
                lines.append(f"- Retained support R: {artifacts['support']}")
            # Record the scorecard details when present
            if "scorecard_points" in artifacts:
                lines.append(
                    f"- Scorecard points: {artifacts['scorecard_points']}"
                )
                lines.append(
                    f"- Scorecard threshold: "
                    f"{artifacts.get('scorecard_threshold', 'N/A')}"
                )
            # Record the tally threshold when present
            if "tally_threshold" in artifacts:
                lines.append(
                    f"- Tally threshold: {artifacts['tally_threshold']}"
                )
            # Record the isotonic summaries when present
            if "isotonic_summary" in artifacts:
                lines.append(
                    f"- Isotonic summary: {artifacts['isotonic_summary']}"
                )
            # Record the ECE bin boundaries when present
            if "ece_bin_boundaries" in artifacts:
                lines.append(
                    f"- ECE bin boundaries: {artifacts['ece_bin_boundaries']}"
                )
            lines.append("")
    # Append the governing equations verbatim
    lines.append("## Governing equations (verbatim)")
    lines.append("")
    lines.append(
        "- Pruning bound: |s_full(x) - s_pruned(x)| <= "
        "L * sum_{j in R^c} |beta_j|, L = max(|log 0.01|, |log 10.01|)"
    )
    lines.append(
        "- Fidelity prediction: rho = (beta^T Sigma d) / "
        "(sqrt(beta^T Sigma beta) sqrt(d^T Sigma d)); "
        "tau_pred = (2/pi) arcsin(rho)"
    )
    lines.append(
        "- ECE: sum_b (|b| / n) |acc(b) - conf(b)|, B = 10 equal-mass "
        "bins with ties assigned to the earliest bin"
    )
    lines.append("")
    # Construct the appendix path
    appendix_path: Path = output_dir / "methodological_appendix.md"
    # Write the appendix markdown to disk
    appendix_path.write_text("\n".join(lines), encoding="utf-8")
    # Return the appendix path
    return appendix_path


# -----------------------------------------------------------------------------
# Task 15, Step 2: Document the environment and the execution sequence
# -----------------------------------------------------------------------------
def write_environment_documentation(cfg: Dict[str, Any]) -> Dict[str, Any]:
    """
    Document the runtime environment and the execution sequence.

    Package versions are captured from the live environment via
    importlib.metadata, never typed by hand; absent optional packages
    are recorded as "not installed".

    Parameters
    ----------
    cfg : Dict[str, Any]
        Validated study configuration.

    Returns
    -------
    Dict[str, Any]
        Environment record with platform, versions, seeds, and the
        ordered execution commands.
    """
    # Capture the platform description
    os_description: str = platform.platform()
    # Capture the processor information
    processor: str = platform.processor()
    # Capture the Python version
    python_version: str = sys_module.version.split()[0]
    # Query the versions of every numerical package
    package_names: Tuple[str, ...] = (
        "numpy", "pandas", "scipy", "scikit-learn", "optuna",
        "torch", "matplotlib", "seaborn",
    )
    # Initialise the package version dictionary
    package_versions: Dict[str, str] = {}
    # Iterate over the package names
    for package_name in package_names:
        # Query the installed version when the package is present
        try:
            package_versions[package_name] = importlib.metadata.version(
                package_name
            )
        except importlib.metadata.PackageNotFoundError:
            # Record absent packages explicitly
            package_versions[package_name] = "not installed"
    # Compose the ordered execution commands
    execution_commands: List[str] = [
        "1. Place the four raw datasets in data/raw/.",
        "2. python -m task_responses.task_1  # validate inputs",
        "3. python -m task_responses.task_2  # cleanse and archive",
        "4. python -m task_responses.task_3  # build feature matrices",
        "5. python -m task_responses.task_4  # split and scale",
        "6. python -m task_responses.task_5  # hyperparameter search",
        "7. python -m task_responses.task_6  # train the monomial",
        "8. python -m task_responses.task_7  # prune and verify bound",
        "9. python -m task_responses.task_8  # directional rule",
        "10. python -m task_responses.task_9  # scorecard and tally",
        "11. python -m task_responses.task_10  # calibration and ECE",
        "12. python -m task_responses.task_11  # orchestrate pipeline",
        "13. python -m task_responses.task_12  # IHT robustness",
        "14. python -m task_responses.task_13  # human assessment",
        "15. python -m task_responses.task_14  # aggregate and verify",
        "16. python -m task_responses.task_15  # audit and sign off",
    ]
    # Assemble the environment record
    environment: Dict[str, Any] = {
        "os": os_description,
        "processor": processor,
        "python_version": python_version,
        "packages": package_versions,
        "seeds": list(cfg["data_splitting"]["seeds"]),
        "illustrative_seed": 45,
        "computational_budget": "recorded per run in the execution trace",
        "execution_commands": execution_commands,
        "written_at": datetime.now(timezone.utc).isoformat(),
    }
    # Return the environment record
    return environment


# -----------------------------------------------------------------------------
# Task 15, Step 3: Run unit tests and final integrity audit
# -----------------------------------------------------------------------------
def _unit_test_log_space_identity() -> Tuple[bool, str]:
    """
    Unit test: verify the log-space identity on 100 random rows.

    Generates 100 random rows in the scaled range ``[0.01, 10.01]`` and a
    random exponent vector, computes monomial logits via ``monomial_logit``
    and checks the identity ``log|z| = log|alpha| + beta^T log x`` to within
    ``1e-10`` absolute error.

    Parameters
    ----------
    None

    Returns
    -------
    Tuple[bool, str]
        ``(passed, evidence)`` where ``passed`` is True when
        ``max_error < 1e-10`` and ``evidence`` is
        ``f"max identity error {max_error:.2e}"``.

    Notes
    -----
    Validates Task 6 Step 6.1 log-space transformation and Task 6 Step 6.3
    verification; uses the scaled range to guarantee ``log`` is defined.
    """

    # Create a deterministic random generator for the fixture
    rng: np.random.Generator = np.random.default_rng(0)
    # Generate 100 random rows in the scaled range
    X: np.ndarray = rng.uniform(0.01, 10.01, size=(100, 4))
    # Generate a random exponent vector
    beta: np.ndarray = rng.normal(0.0, 0.5, size=4)
    # Fix a nonzero coefficient
    alpha: float = 2.5
    # Compute the logits via the log-space forward
    logits: np.ndarray = monomial_logit(X, alpha, beta)
    # Compute the log-space identity left-hand side
    log_abs_z: np.ndarray = np.log(np.abs(logits))
    # Compute the log-space identity right-hand side
    log_space: np.ndarray = np.log(abs(alpha)) + np.log(X) @ beta
    # Compute the maximum absolute identity error
    max_error: float = float(np.abs(log_abs_z - log_space).max())
    # Return the verdict and the observed error
    return max_error < 1e-10, f"max identity error {max_error:.2e}"


def _unit_test_pruning_bound() -> Tuple[bool, str]:
    """
    Unit test: verify the worst-case pruning bound on 10,000 random rows.

    Generates 10,000 random rows in ``[0.01, 10.01]``, a random exponent
    vector and a retained set ``R={0,1,2}``, computes ``|s_full - s_pruned|``
    and checks ``<= L * sum_{j in R^c} |beta_j|`` with
    ``L = max(|log 0.01|, |log 10.01|)`` for every row.

    Parameters
    ----------
    None

    Returns
    -------
    Tuple[bool, str]
        ``(passed, evidence)`` where ``passed`` is True when zero rows
        violate the bound and ``evidence`` is ``f"{violated} rows violated"``.

    Notes
    -----
    Validates the Remark bound of Task 7 Step 7.3; uses the removed set
    ``R^c`` not the retained set.
    """
    # Create a deterministic random generator for the fixture
    rng: np.random.Generator = np.random.default_rng(1)
    # Generate 10,000 random rows in the scaled range
    X: np.ndarray = rng.uniform(0.01, 10.01, size=(10_000, 6))
    # Generate a random exponent vector
    beta: np.ndarray = rng.normal(0.0, 0.5, size=6)
    # Fix a nonzero coefficient
    alpha: float = -1.5
    # Define the retained set (first three positions)
    retained: List[int] = [0, 1, 2]
    # Compute the removed set as the complement
    removed: List[int] = [3, 4, 5]
    # Compute the full log-score
    s_full: np.ndarray = np.log(abs(alpha)) + np.log(X) @ beta
    # Compute the pruned log-score over the retained set
    s_pruned: np.ndarray = (
        np.log(abs(alpha)) + np.log(X)[:, retained] @ beta[retained]
    )
    # Compute the per-row gaps
    gaps: np.ndarray = np.abs(s_full - s_pruned)
    # Compute the bound: L * sum_{j in R^c} |beta_j|
    bound: float = max(abs(np.log(0.01)), abs(np.log(10.01))) * float(
        np.sum(np.abs(beta[removed]))
    )
    # Verify the bound holds on every row
    violated: int = int(np.sum(gaps > bound + 1e-10))
    # Return the verdict and the observed margin
    return violated == 0, f"{violated} rows violated the bound"


def _unit_test_rho_formula() -> Tuple[bool, str]:
    """
    Unit test: verify the Pearson correlation formula against direct correlation.

    Generates 5,000 random rows in ``[0.01, 10.01]``, random ``beta`` and
    direction vector ``d=[1, -1, 0]``, computes ``rho`` via
    ``predict_fidelity`` (``beta^T Sigma d / sqrt(beta^T Sigma beta)``) and
    checks against ``np.corrcoef(w @ beta, w @ d)[0,1]`` to ``1e-10``.

    Parameters
    ----------
    None

    Returns
    -------
    Tuple[bool, str]
        ``(passed, evidence)`` where ``passed`` is True when
        ``|rho_formula - direct| < 1e-10``.

    Notes
    -----
    Validates Proposition fidelity prediction Task 8 Step 8.2 without
    distributional assumptions.
    """

    # Create a deterministic random generator for the fixture
    rng: np.random.Generator = np.random.default_rng(2)
    # Generate 5,000 random rows in the scaled range
    X: np.ndarray = rng.uniform(0.01, 10.01, size=(5_000, 3))
    # Generate random beta and direction vectors
    beta: np.ndarray = rng.normal(0.0, 1.0, size=3)
    # Fix the direction vector with a zero entry
    d: np.ndarray = np.array([1.0, -1.0, 0.0])
    # Compute the predicted rho from the training data
    prediction: Dict[str, Any] = predict_fidelity(
        beta=beta, d=d, X_train=X, cfg={}
    )
    # Extract the predicted rho
    rho_predicted: float = float(prediction["rho"])
    # Compute the direct numerical correlation of the two linear scores
    w: np.ndarray = np.log(X)
    # Compute the correlation of the full and directional scores
    direct_rho: float = float(np.corrcoef(w @ beta, w @ d)[0, 1])
    # Compute the absolute difference between the two values
    difference: float = abs(rho_predicted - direct_rho)
    # Return the verdict and the observed difference
    return difference < 1e-10, f"|rho_formula - direct| = {difference:.2e}"


def _unit_test_ece_binning() -> Tuple[bool, str]:
    """
    Unit test: verify ECE binning against a hand-computed two-bin example.

    Fixture: scores ``[0.1, 0.2, 0.3, 0.9]``, labels ``[0, 1, 0, 1]``; PAV
    predictions ``[0, 0.5, 0.5, 1]``; tie-snapped two-bin ECE is ``0.0``.
    Checks ``compute_ece`` with ``B=2``.

    Parameters
    ----------
    None

    Returns
    -------
    Tuple[bool, str]
        ``(passed, evidence)`` where ``passed`` is True when
        ``|ece_computed - 0.0| < 1e-12``.

    Notes
    -----
    Validates Task 10 Step 10.2 tie-faithful equal-mass binning with
    ``B=10`` logic on a 2-bin hand example.
    """

    # Define the hand-verified fixture scores
    scores: np.ndarray = np.array([0.1, 0.2, 0.3, 0.9])
    # Define the fixture labels
    labels: np.ndarray = np.array([0, 1, 0, 1])
    # Fit the isotonic calibrator on the fixture
    calibrator = fit_isotonic_calibrator(scores, labels)
    # Compute the ECE with two bins
    record: Dict[str, Any] = compute_ece(
        score_test=scores, y_test=labels, calibrator=calibrator, cfg={}, B=2
    )
    # Extract the computed ECE
    ece_computed: float = float(record["ece"])
    # The hand-computed expected ECE is 0 (tie-snapped boundaries)
    expected_ece: float = 0.0
    # Compute the absolute difference from the hand value
    difference: float = abs(ece_computed - expected_ece)
    # Return the verdict and the observed difference
    return difference < 1e-12, f"ECE {ece_computed} vs hand value 0.0"


def _unit_test_scorecard_points() -> Tuple[bool, str]:
    """
    Unit test: verify the scorecard point formula on synthetic exponents.

    Fixture: magnitudes ``[2.5, 1.25, 0.5]`` with ``P_max=5`` produce scaled
    values ``[5, 2.5, 1]`` and round-half-away points ``[5, 3, 1]`` via
    ``p_j = max(1, round(P_max * |beta_j| / max|beta|))``.

    Parameters
    ----------
    None

    Returns
    -------
    Tuple[bool, str]
        ``(passed, evidence)`` where ``passed`` is True when
        ``np.array_equal(points, [5, 3, 1])``.

    Notes
    -----
    Validates Task 9 Step 9.2 point assignment with ``P_max=5`` and
    half-away rounding.
    """

    # Define the synthetic exponent vector
    beta: np.ndarray = np.array([2.5, 1.25, 0.5])
    # Assign the points with the configured maximum of five
    points: np.ndarray = assign_scorecard_points(
        beta=beta,
        retained=[0, 1, 2],
        cfg={"scorecard_and_tally": {"max_points": 5}},
    )
    # Compare against the hand-computed point vector
    expected: np.ndarray = np.array([5, 3, 1])
    # Return the verdict and the observed points
    return bool(np.array_equal(points, expected)), f"points {points.tolist()}"


def _unit_test_tau_b() -> Tuple[bool, str]:
    """
    Unit test: verify the tie-corrected Kendall tau_b against a known example.

    Fixture: ``x=[1,1,2,3]``, ``y=[1,2,2,3]``; computes ``tau_b`` via
    ``compute_kendall_tau_b`` and checks against ``scipy.stats.kendalltau``
    ``variant="b"`` to ``1e-12``.

    Parameters
    ----------
    None

    Returns
    -------
    Tuple[bool, str]
        ``(passed, evidence)`` where ``passed`` is True when
        ``|tau_b - oracle| < 1e-12``.

    Notes
    -----
    Validates Task 11/Task 12 fidelity metric ``tau_b = (C-D)/sqrt((C+D+T_s)(C+D+T_f))``
    with tie corrections.
    """

    # Define the tied example scores
    x: np.ndarray = np.array([1.0, 1.0, 2.0, 3.0])
    # Define the simplified scores with ties
    y: np.ndarray = np.array([1.0, 2.0, 2.0, 3.0])
    # Compute the exact counts and tau_b
    _, _, _, _, tau_b = compute_kendall_tau_b(x, y)
    # Compute the scipy oracle value with variant b
    oracle_tau, _ = kendalltau(x, y, variant="b")
    # Compute the absolute difference from the oracle
    difference: float = abs(tau_b - float(oracle_tau))
    # Return the verdict and the observed difference
    return difference < 1e-12, f"tau_b {tau_b} vs scipy {float(oracle_tau)}"


def _unit_test_constant_feature_policy() -> Tuple[bool, str]:
    """
    Unit test: verify the constant-feature midpoint policy of Task 4.

    Regression test for the affine division by zero: a three-column matrix
    whose third column is constant ``7`` must be scaled to exactly ``5.01``
    (midpoint of ``[0.01, 10.01]``) in every partition and remain within the
    range.

    Parameters
    ----------
    None

    Returns
    -------
    Tuple[bool, str]
        ``(passed, evidence)`` where ``passed`` is True when
        ``midpoint_ok and range_ok``.

    Notes
    -----
    Validates Task 4 Step 4.2 constant-feature handling ``max==min -> 5.01``.
    """

    # Define the fixture matrix with a constant third column
    X: np.ndarray = np.array(
        [
            [1.0, 2.0, 7.0],
            [2.0, 4.0, 7.0],
            [3.0, 6.0, 7.0],
            [4.0, 8.0, 7.0],
            [5.0, 10.0, 7.0],
        ],
        dtype=np.float64,
    )
    # Build the minimal configuration for a dataset without imputation
    cfg: Dict[str, Any] = {
        "missing_value_policy": {"T": {"imputation_columns": []}}
    }
    # Fit the preprocessing statistics on the fixture
    fitted = fit_preprocessing(X, cfg, "T", ["a", "b", "c"])
    # Apply the preprocessing to the fixture partitions
    scaled_train, scaled_val, scaled_test, _ = apply_preprocessing(
        X_train=X, X_val=X[:2], X_test=X[2:], fitted=fitted
    )
    # Verify the constant column equals the midpoint in every partition
    midpoint_ok: bool = bool(
        np.all(scaled_train[:, 2] == 5.01)
        and np.all(scaled_val[:, 2] == 5.01)
        and np.all(scaled_test[:, 2] == 5.01)
    )
    # Verify every scaled value lies within the declared range
    range_ok: bool = bool(
        np.all(scaled_train >= 0.01)
        and np.all(scaled_train <= 10.01)
        and np.all(scaled_val >= 0.01)
        and np.all(scaled_val <= 10.01)
        and np.all(scaled_test >= 0.01)
        and np.all(scaled_test <= 10.01)
    )
    # Return the verdict and the observed policy evidence
    return midpoint_ok and range_ok, (
        f"constant column midpoint={scaled_train[0, 2]}, "
        f"range_ok={range_ok}"
    )


def run_audit_suite(
    output_dir: Path,
    results: Optional[Dict[str, Any]] = None,
) -> Dict[str, Any]:
    """
    Run the mathematical unit tests and the final integrity audit.

    The six mathematical unit tests run on synthetic fixtures with
    hand-computed expectations and must pass. Data-dependent checks
    (checksums, split leakage) are recorded as "not_runnable" when
    their archives are absent — they are never claimed as passing.
    Sign-off requires every runnable check to pass.

    Parameters
    ----------
    output_dir : Path
        Project output root whose artifact archives are audited.
    results : Optional[Dict[str, Any]], optional
        Orchestrator results structure for key-completeness checking.

    Returns
    -------
    Dict[str, Any]
        Audit report with a status per check and the sign-off verdict.
    """
    # Run the six mathematical unit tests plus the Task 4 regression test
    unit_tests: List[Tuple[str, Tuple[bool, str]]] = [
        ("log_space_identity", _unit_test_log_space_identity()),
        ("pruning_bound", _unit_test_pruning_bound()),
        ("rho_formula", _unit_test_rho_formula()),
        ("ece_binning", _unit_test_ece_binning()),
        ("scorecard_points", _unit_test_scorecard_points()),
        ("tau_b_formula", _unit_test_tau_b()),
        ("constant_feature_policy", _unit_test_constant_feature_policy()),
    ]
    # Initialise the audit entries
    entries: List[Dict[str, Any]] = []
    # Record the unit test outcomes
    for check_id, (passed, evidence) in unit_tests:
        # Append the unit test entry with its status
        entries.append(
            {
                "check": check_id,
                "status": "pass" if passed else "fail",
                "evidence": evidence,
            }
        )
    # Check 7: test-split rows never appear in training-only artifacts
    splits_dir: Path = output_dir / "artifacts" / "splits"
    # List the split artifacts when the directory exists
    if splits_dir.exists():
        # Track the leakage verdict across all split files
        leakage_ok: bool = True
        # Track the leakage evidence
        leakage_evidence: List[str] = []
        # Iterate over the split npz files
        for split_file in sorted(splits_dir.glob("*_splits.npz")):
            # Load the split arrays from the artifact
            split_data = np.load(split_file)
            # Extract the three index arrays
            train_idx = split_data["train_idx"]
            val_idx = split_data["val_idx"]
            test_idx = split_data["test_idx"]
            # Check the train/val intersection
            if not set(train_idx).isdisjoint(val_idx):
                # Record the train/val overlap
                leakage_ok = False
                leakage_evidence.append(
                    f"{split_file.name}: train/val overlap"
                )
            # Check the train/test intersection
            if not set(train_idx).isdisjoint(test_idx):
                # Record the train/test overlap
                leakage_ok = False
                leakage_evidence.append(
                    f"{split_file.name}: train/test overlap"
                )
        # Append the leakage check entry
        entries.append(
            {
                "check": "split_leakage",
                "status": "pass" if leakage_ok else "fail",
                "evidence": (
                    "; ".join(leakage_evidence)
                    if leakage_evidence
                    else "no overlap in any split artifact"
                ),
            }
        )
    else:
        # Record the leakage check as not runnable
        entries.append(
            {
                "check": "split_leakage",
                "status": "not_runnable",
                "evidence": "artifacts/splits does not exist",
            }
        )
    # Check 8: orchestrator output key completeness when supplied
    if results is not None:
        # Track the key-completeness verdict
        keys_ok: bool = True
        # Track the missing key evidence
        missing_keys: List[str] = []
        # Iterate over the datasets in the results
        for dataset_name in ("Loan", "FraudEcom", "Default", "Creditcard"):
            # Check the dataset key presence
            if dataset_name not in results:
                keys_ok = False
                missing_keys.append(dataset_name)
        # Append the key-completeness check entry
        entries.append(
            {
                "check": "orchestrator_key_completeness",
                "status": "pass" if keys_ok else "fail",
                "evidence": (
                    "all dataset keys present"
                    if keys_ok
                    else f"missing {missing_keys}"
                ),
            }
        )
    # Check 9: cleansed parquet checksums against the archived values
    cleansed_dir: Path = output_dir / "artifacts" / "cleansed" / "v1"
    # Locate the cleansing report when the archive exists
    cleansing_report_path: Path = cleansed_dir / "cleansing_report.json"
    # Run the checksum verification when the archive exists
    if cleansing_report_path.exists():
        # Load the archived checksums from the cleansing report
        cleansing_report: Dict[str, Any] = json.loads(
            cleansing_report_path.read_text(encoding="utf-8")
        )
        # Extract the archived checksum map
        archived_checksums: Dict[str, str] = cleansing_report["checksums"]
        # Track the checksum verdict
        checksums_ok: bool = True
        # Track the checksum evidence
        checksum_evidence: List[str] = []
        # Iterate over the archived checksums
        for dataset_name, archived_digest in archived_checksums.items():
            # Construct the expected parquet path
            parquet_path: Path = cleansed_dir / (
                f"{dataset_name.lower()}.parquet"
            )
            # Check the parquet file exists
            if not parquet_path.exists():
                checksums_ok = False
                checksum_evidence.append(
                    f"{dataset_name}: parquet file missing"
                )
                continue
            # Recompute the digest over the bytes on disk
            recomputed_digest: str = hashlib.sha256(
                parquet_path.read_bytes()
            ).hexdigest()
            # Compare the recomputed digest with the archived value
            if recomputed_digest != archived_digest:
                checksums_ok = False
                checksum_evidence.append(
                    f"{dataset_name}: digest mismatch"
                )
        # Append the checksum check entry
        entries.append(
            {
                "check": "cleansed_checksums",
                "status": "pass" if checksums_ok else "fail",
                "evidence": (
                    "all digests match"
                    if checksums_ok
                    else "; ".join(checksum_evidence)
                ),
            }
        )
    else:
        # Record the checksum check as not runnable
        entries.append(
            {
                "check": "cleansed_checksums",
                "status": "not_runnable",
                "evidence": "cleansed archive does not exist",
            }
        )
    # Determine the sign-off verdict: every runnable check must pass
    runnable_statuses: List[str] = [
        entry["status"] for entry in entries if entry["status"] != "not_runnable"
    ]
    # Compute the sign-off verdict
    signed_off: bool = all(status == "pass" for status in runnable_statuses)
    # Assemble the audit report
    audit_report: Dict[str, Any] = {
        "written_at": datetime.now(timezone.utc).isoformat(),
        "checks": entries,
        "signed_off": signed_off,
        "summary": (
            f"{sum(1 for e in entries if e['status'] == 'pass')} pass, "
            f"{sum(1 for e in entries if e['status'] == 'fail')} fail, "
            f"{sum(1 for e in entries if e['status'] == 'not_runnable')} "
            "not runnable"
        ),
    }
    # Write the audit report to the output directory
    report_path: Path = output_dir / "audit_report.json"
    # Serialise the report with stable formatting
    report_path.write_text(
        json.dumps(audit_report, indent=2, ensure_ascii=False) + "\n",
        encoding="utf-8",
    )
    # Return the audit report
    return audit_report


# -----------------------------------------------------------------------------
# Task 15, Orchestrator Function
# -----------------------------------------------------------------------------
def orchestrate_documentation_and_audit(
    all_artifacts: Dict[str, Any],
    cfg: Dict[str, Any],
    output_dir: Path,
    results: Optional[Dict[str, Any]] = None,
) -> Dict[str, Any]:
    """
    Orchestrate Task 15: appendix, environment, and audit sign-off.

    Parameters
    ----------
    all_artifacts : Dict[str, Any]
        Persisted artifacts keyed dataset -> seed -> artifact type.
    cfg : Dict[str, Any]
        Validated study configuration.
    output_dir : Path
        Project output root.
    results : Optional[Dict[str, Any]], optional
        Orchestrator results for the key-completeness check.

    Returns
    -------
    Dict[str, Any]
        Mapping with "appendix", "environment", and "audit" keys.
    """
    # Step 15.1: write the methodological appendix from the artifacts
    appendix_path: Path = write_methodological_appendix(
        all_artifacts=all_artifacts, cfg=cfg, output_dir=output_dir
    )
    # Step 15.2: document the environment and the execution sequence
    environment: Dict[str, Any] = write_environment_documentation(cfg)
    # Persist the environment record to the output directory
    environment_path: Path = output_dir / "environment_documentation.json"
    # Serialise the environment record with stable formatting
    environment_path.write_text(
        json.dumps(environment, indent=2, ensure_ascii=False) + "\n",
        encoding="utf-8",
    )
    # Step 15.3: run the unit tests and the final integrity audit
    audit_report: Dict[str, Any] = run_audit_suite(
        output_dir=output_dir, results=results
    )
    # Return the appendix, environment, and audit results
    return {
        "appendix": str(appendix_path),
        "environment": str(environment_path),
        "audit": audit_report,
    }

````

````python
# Task 16: Top-level integrative orchestrator for the end-to-end research

# ==============================================================================
# Task 16: Top-level integrative orchestrator for the end-to-end research
# ==============================================================================

# The four study datasets in canonical order
DATASETS: Tuple[str, ...] = ("Loan", "FraudEcom", "Default", "Creditcard")

# The two survey datasets (human assessment scope of the manuscript)
SURVEY_DATASETS: Tuple[str, ...] = ("Loan", "FraudEcom")

# The illustrative display seed of the manuscript's figures
DISPLAY_SEED: int = 45

# -----------------------------------------------------------------------------
# Task 16, Step 1: Top-level helper stages — validation, cleansing, and feature building
# -----------------------------------------------------------------------------

def _run_validation_and_cleansing(
    raw_dfs: Dict[str, pd.DataFrame],
    cfg: Dict[str, Any],
    output_dir: Path,
) -> Dict[str, Any]:
    """
    Run the Task 1 validation stage and the Task 2 cleansing stage.

    Parameters
    ----------
    raw_dfs : Dict[str, pd.DataFrame]
        Four raw DataFrames keyed by dataset name.
    cfg : Dict[str, Any]
        Raw study configuration (validated inside Task 1).
    output_dir : Path
        Project output root.

    Returns
    -------
    Dict[str, Any]
        Validation report, cleansed frames, and cleansing persistence.
    """
    # Verify every required dataset key is present in the raw frames
    for dataset_name in DATASETS:
        # Raise when a required raw frame is missing
        if dataset_name not in raw_dfs:
            raise ValueError(
                f"Required raw DataFrame '{dataset_name}' is missing."
            )
    # Step 1: run the Task 1 validation orchestrator (config, schemas,
    # label rates) — it raises on any failure before cleansing starts
    validation_report: Dict[str, Any] = orchestrate_validation(
        loan_raw=raw_dfs["Loan"],
        fraud_ecom_raw=raw_dfs["FraudEcom"],
        default_raw=raw_dfs["Default"],
        creditcard_raw=raw_dfs["Creditcard"],
        study_configuration=cfg,
        output_dir=output_dir,
    )
    # Read the normalised configuration from the validation report
    normalized_cfg: Dict[str, Any] = validation_report["config"]
    # Step 2: run the Task 2 cleansing orchestrator on the raw frames
    cleansing_result: Dict[str, Any] = cleanse_all(
        raw_dfs=raw_dfs,
        cfg=normalized_cfg,
        output_dir=output_dir,
    )
    # Return the validation report, cleansed frames, and persistence
    return {
        "validation": validation_report,
        "cleansed": cleansing_result["cleansed"],
        "cleansing_persistence": cleansing_result["persistence"],
    }


def _build_static_matrices(
    cleansed_dfs: Dict[str, pd.DataFrame],
    cfg: Dict[str, Any],
) -> Dict[str, Dict[str, Any]]:
    """
    Build static matrices for Loan Default Creditcard only, FraudEcom placeholder None.

    FraudEcom counts are per-seed training-only, so no static matrix is built.
    The authoritative build is per seed in _run_dataset_seed via _build_fraud_matrix_for_seed.

    Parameters
    ----------
    cleansed_dfs : Dict[str, pd.DataFrame]
        Cleansed frames.
    cfg : Dict[str, Any]
        Validated configuration.

    Returns
    -------
    Dict[str, Dict[str, Any]]
        Static matrices keyed dataset -> {X, y} or None for FraudEcom placeholder.
    """
    # Validate cleansed_dfs contains required keys
    for k in ["Loan", "Default", "Creditcard"]:
        if k not in cleansed_dfs:
            raise KeyError(f"cleansed_dfs missing {k}")
    # Initialize static dict
    static_matrices: Dict[str, Dict[str, Any]] = {}
    # Build Loan matrix
    loan_X, loan_y, loan_idx = build_loan_features(cleansed_dfs["Loan"], cfg)
    # Store Loan
    static_matrices["Loan"] = {"X": loan_X, "y": loan_y, "original_index": loan_idx, "is_placeholder": False, "feature_names": tuple(cfg["feature_matrices"]["Loan"]["feature_names"])}
    # Build Default matrix
    default_X, default_y, default_idx = build_default_features(cleansed_dfs["Default"], cfg)
    # Store Default
    static_matrices["Default"] = {"X": default_X, "y": default_y, "original_index": default_idx, "is_placeholder": False, "feature_names": tuple(cfg["feature_matrices"]["Default"]["feature_names"])}
    # Build Creditcard matrix
    cc_X, cc_y, cc_idx = build_creditcard_features(cleansed_dfs["Creditcard"], cfg)
    # Store Creditcard
    static_matrices["Creditcard"] = {"X": cc_X, "y": cc_y, "original_index": cc_idx, "is_placeholder": False, "feature_names": tuple(cfg["feature_matrices"]["Creditcard"]["feature_names"])}
    # Set FraudEcom to None placeholder with metadata
    static_matrices["FraudEcom"] = None
    # Return
    return static_matrices

def _build_fraud_matrix_for_seed(
    cleansed_fraud: pd.DataFrame,
    cfg: Dict[str, Any],
    seed: int,
    feature_names: Sequence[str],
) -> Tuple[np.ndarray, np.ndarray]:
    """
    Build the FraudEcom matrix for one seed (training-only counts).

    The split is obtained on the fraud frame itself (a dummy row-count
    matrix plus the fraud labels; the splitter consumes only the row
    count and the labels), then the count features are built from the
    training partition only.

    Parameters
    ----------
    cleansed_fraud : pd.DataFrame
        Cleansed FraudEcom frame.
    cfg : Dict[str, Any]
        Normalised study configuration.
    seed : int
        Random seed of the split.
    feature_names : Sequence[str]
        FraudEcom feature names.

    Returns
    -------
    Tuple[np.ndarray, np.ndarray]
        Fraud matrix of shape (n, 6) and the fraud label vector.
    """
    # Extract the fraud label vector from the cleansed frame
    y_fraud_full: np.ndarray = cleansed_fraud["y"].to_numpy(dtype=np.int64)
    # Build a dummy matrix of the fraud row count (the splitter uses only
    # the row count and the labels for stratification)
    fraud_dummy_matrix: np.ndarray = np.zeros(
        (y_fraud_full.shape[0], len(feature_names))
    )
    # Obtain the stratified split on the fraud frame for the seed
    train_idx, _, _ = make_stratified_splits(
        fraud_dummy_matrix, y_fraud_full, seed, cfg
    )
    # Build the six-feature matrix from the training partition only
    X_fraud, y_fraud, _ = build_fraud_features(
        cleansed_fraud, train_idx, cfg
    )
    # Return the fraud matrix and its aligned labels
    return X_fraud, y_fraud


def _run_features_and_preprocessing(
    cleansed_dfs: Dict[str, pd.DataFrame],
    cfg: Dict[str, Any],
    output_dir: Path,
) -> Dict[str, Any]:
    """
    Run the Task 3 feature-matrix stage and the Task 4 preprocessing.

    The Task 3 builders produce the static matrices; the Task 4
    orchestrator (orchestrate_splitting_and_scaling) runs per dataset at
    seed 0, persisting the canonical splits/stats/scaled artifacts. The
    scaled seed-0 training splits are returned for the hyperparameter
    stage.

    Parameters
    ----------
    cleansed_dfs : Dict[str, pd.DataFrame]
        Cleansed frames keyed by dataset name.
    cfg : Dict[str, Any]
        Normalised study configuration.
    output_dir : Path
        Project output root.

    Returns
    -------
    Dict[str, Any]
        Static matrices, seed-0 preprocessing artifacts, and the scaled
        seed-0 training splits keyed by dataset.
    """
    # Build the static matrices for the non-FraudEcom datasets
    static_matrices: Dict[str, Dict[str, np.ndarray]] = _build_static_matrices(
        cleansed_dfs, cfg
    )
    # Build the FraudEcom seed-0 matrix for the preprocessing stage
    fraud_X0, fraud_y0 = _build_fraud_matrix_for_seed(
        cleansed_fraud=cleansed_dfs["FraudEcom"],
        cfg=cfg,
        seed=0,
        feature_names=cfg["feature_matrices"]["FraudEcom"]["feature_names"],
    )
    # Store the seed-0 fraud matrix as the static FraudEcom matrix
    static_matrices["FraudEcom"] = {"X": fraud_X0, "y": fraud_y0}
    # Initialise the preprocessing persistence and scaled training maps
    preprocessing_persistence: Dict[str, Any] = {}
    seed0_scaled_train: Dict[str, np.ndarray] = {}
    seed0_labels: Dict[str, np.ndarray] = {}
    # Iterate over the four datasets for the seed-0 preprocessing record
    for dataset_name in DATASETS:
        # Extract the feature names of the dataset
        feature_names: Sequence[str] = cfg["feature_matrices"][dataset_name][
            "feature_names"
        ]
        # Extract the matrix and labels of the dataset
        X_dataset: np.ndarray = static_matrices[dataset_name]["X"]
        y_dataset: np.ndarray = static_matrices[dataset_name]["y"]
        # Run the Task 4 orchestrator at seed 0 for the canonical record
        preprocessing_result: Dict[str, Any] = (
            orchestrate_splitting_and_scaling(
                X=X_dataset,
                y=y_dataset,
                cfg=cfg,
                dataset_name=dataset_name,
                feature_names=feature_names,
                seed=0,
                output_dir=output_dir,
            )
        )
        # Record the preprocessing persistence of the dataset
        preprocessing_persistence[dataset_name] = preprocessing_result[
            "persistence"
        ]
        # Extract the scaled seed-0 training matrix
        seed0_scaled_train[dataset_name] = preprocessing_result["scaled"][
            "train"
        ]
        # Extract the seed-0 training labels
        train_idx0: np.ndarray = preprocessing_result["splits"]["train"]
        # Slice the labels of the seed-0 training partition
        seed0_labels[dataset_name] = y_dataset[train_idx0]
    # Return the matrices, persistence, and seed-0 scaled training data
    return {
        "static_matrices": static_matrices,
        "preprocessing_persistence": preprocessing_persistence,
        "seed0_scaled_train": seed0_scaled_train,
        "seed0_labels": seed0_labels,
    }


# -----------------------------------------------------------------------------
# Task 16, Step 2: Quantitative and IHT stage helpers
# -----------------------------------------------------------------------------

def _run_hyperparameter_stage(
    cfg: Dict[str, Any],
    output_dir: Path,
    seed0_scaled_train: Dict[str, np.ndarray],
    seed0_labels: Dict[str, np.ndarray],
) -> Dict[str, Any]:
    """
    Run the Task 5 hyperparameter search stage per dataset.

    Parameters
    ----------
    cfg : Dict[str, Any]
        Normalised study configuration.
    output_dir : Path
        Project output root.
    seed0_scaled_train : Dict[str, np.ndarray]
        Scaled seed-0 training matrices keyed by dataset.
    seed0_labels : Dict[str, np.ndarray]
        Seed-0 training labels keyed by dataset.

    Returns
    -------
    Dict[str, Any]
        Hyperparameter records keyed by dataset.
    """
    # Initialise the hyperparameter record map
    hyperparameter_records: Dict[str, Any] = {}
    # Iterate over the four datasets
    for dataset_name in DATASETS:
        # Run the Task 5 orchestrator on the seed-0 training split
        search_result: Dict[str, Any] = orchestrate_hyperparameter_search(
            X_train=seed0_scaled_train[dataset_name],
            y_train=seed0_labels[dataset_name],
            cfg=cfg,
            dataset_name=dataset_name,
            output_dir=output_dir,
            seed=0,
        )
        # Record the configuration of the dataset
        hyperparameter_records[dataset_name] = search_result["configuration"]
    # Return the hyperparameter records
    return hyperparameter_records


def _reconstruct_scaled_splits(
    cfg: Dict[str, Any],
    cleansed_dfs: Dict[str, pd.DataFrame],
    static_matrices: Dict[str, Dict[str, np.ndarray]],
    dataset_name: str,
    seed: int,
) -> Dict[str, np.ndarray]:
    """
    Reconstruct the scaled splits for one dataset and seed.

    Reuses the Task 4 pure callables (make_stratified_splits,
    fit_preprocessing, apply_preprocessing) — implementation reuse, not
    duplication. For FraudEcom the per-seed count-feature matrix is
    built from the training partition first.

    Parameters
    ----------
    cfg : Dict[str, Any]
        Normalised study configuration.
    cleansed_dfs : Dict[str, pd.DataFrame]
        Cleansed frames keyed by dataset name.
    static_matrices : Dict[str, Dict[str, np.ndarray]]
        Static matrices keyed by dataset.
    dataset_name : str
        Dataset name.
    seed : int
        Random seed.

    Returns
    -------
    Dict[str, np.ndarray]
        Scaled train, val, and test matrices plus the label slices.
    """
    # Extract the feature names of the dataset
    feature_names: Sequence[str] = cfg["feature_matrices"][dataset_name][
        "feature_names"
    ]
    # Resolve the matrix and labels for the dataset
    if dataset_name == "FraudEcom":
        # Build the per-seed fraud matrix from the training partition
        X_dataset, y_dataset = _build_fraud_matrix_for_seed(
            cleansed_fraud=cleansed_dfs["FraudEcom"],
            cfg=cfg,
            seed=seed,
            feature_names=feature_names,
        )
    else:
        # Use the static matrix and labels of the dataset
        X_dataset = static_matrices[dataset_name]["X"]
        y_dataset = static_matrices[dataset_name]["y"]
    # Obtain the stratified split for the seed
    train_idx, val_idx, test_idx = make_stratified_splits(
        X_dataset, y_dataset, seed, cfg
    )
    # Fit the preprocessing on the training rows only
    fitted = fit_preprocessing(
        X_train=X_dataset[train_idx],
        cfg=cfg,
        dataset_name=dataset_name,
        feature_names=feature_names,
    )
    # Apply the fitted preprocessing to all three partitions
    scaled_train, scaled_val, scaled_test, _ = apply_preprocessing(
        X_train=X_dataset[train_idx],
        X_val=X_dataset[val_idx],
        X_test=X_dataset[test_idx],
        fitted=fitted,
    )
    # Slice the labels of the three partitions
    y_train: np.ndarray = y_dataset[train_idx]
    y_val: np.ndarray = y_dataset[val_idx]
    y_test: np.ndarray = y_dataset[test_idx]
    # Return the scaled splits and the label slices
    return {
        "train": scaled_train,
        "val": scaled_val,
        "test": scaled_test,
        "y_train": y_train,
        "y_val": y_val,
        "y_test": y_test,
    }


def _run_quantitative_and_iht_stages(
    cfg: Dict[str, Any],
    cleansed_dfs: Dict[str, pd.DataFrame],
    static_matrices: Dict[str, Dict[str, np.ndarray]],
    output_dir: Path,
    hyperparameter_records: Dict[str, Any],
) -> Dict[str, Any]:
    """
    Run the Task 11 quantitative engine and the Task 12 IHT stage.

    The engine (run_reproduction_study) is the single driver of the
    twelve-step per-dataset per-seed sequence and invokes the Tasks
    5/7/8/9/10 ORCHESTRATORS internally (the record resolver, pruning,
    directional rule, scorecard/tally, and calibration orchestrators),
    so those stages are not re-run at the top level — the documented
    no-duplication decision. The IHT stage then calls the Task 12
    orchestrator per dataset and seed with the scaled splits
    reconstructed via the Task 4 pure callables.

    Parameters
    ----------
    cfg : Dict[str, Any]
        Normalised study configuration.
    cleansed_dfs : Dict[str, pd.DataFrame]
        Cleansed frames keyed by dataset name.
    static_matrices : Dict[str, Dict[str, np.ndarray]]
        Static matrices keyed by dataset.
    output_dir : Path
        Project output root.
    hyperparameter_records : Dict[str, Any]
        Hyperparameter records keyed by dataset.

    Returns
    -------
    Dict[str, Any]
        Pipeline results and IHT comparison records.
    """
    # Run the Task 11 quantitative engine over all datasets and seeds
    pipeline_results: Dict[str, Any] = run_reproduction_study(
        cfg=cfg,
        cleaned_inputs=cleansed_dfs,
        output_dir=output_dir,
    )
    # Resolve the configured seed list
    seeds: Sequence[int] = tuple(cfg["data_splitting"]["seeds"])
    # Initialise the IHT comparison records
    iht_results: Dict[str, Any] = {}
    # Iterate over the datasets
    for dataset_name in DATASETS:
        # Initialise the per-dataset IHT records
        iht_results[dataset_name] = {}
        # Extract the feature names of the dataset
        feature_names: Sequence[str] = cfg["feature_matrices"][dataset_name][
            "feature_names"
        ]
        # Iterate over the seeds
        for seed in seeds:
            # Extract the per-seed record from the engine results
            seed_record: Dict[str, Any] = pipeline_results[dataset_name][
                str(seed)
            ]
            # Extract the fitted coefficient from the seed record
            alpha: float = float(seed_record["alpha"])
            # Extract the fitted exponent vector from the seed record
            beta: np.ndarray = np.asarray(seed_record["beta"], dtype=np.float64)
            # Extract the retained support from the seed record
            support: Sequence[int] = seed_record["representations"]["full"][
                "support"
            ]
            # Extract the pruned test PR-AUC from the seed record
            pr_auc_prune: float = float(
                seed_record["representations"]["pruned"]["pr_auc"]
            )
            # Reconstruct the scaled splits via the Task 4 callables
            scaled_splits: Dict[str, np.ndarray] = _reconstruct_scaled_splits(
                cfg=cfg,
                cleansed_dfs=cleansed_dfs,
                static_matrices=static_matrices,
                dataset_name=dataset_name,
                seed=int(seed),
            )
            # Load the hyperparameter record of the dataset
            hparams: Dict[str, Any] = hyperparameter_records[dataset_name]
            # Run the Task 12 IHT orchestrator for the dataset and seed
            iht_result: Dict[str, Any] = orchestrate_iht_comparison(
                alpha=alpha,
                beta=beta,
                support_prune=support,
                X_train=scaled_splits["train"],
                y_train=scaled_splits["y_train"],
                X_test=scaled_splits["test"],
                y_test=scaled_splits["y_test"],
                hparams=hparams,
                cfg=cfg,
                dataset_name=dataset_name,
                seed=int(seed),
                pr_auc_prune=pr_auc_prune,
                feature_names=feature_names,
                output_dir=output_dir,
            )
            # Record the IHT comparison of the dataset and seed
            iht_results[dataset_name][str(seed)] = iht_result["comparison"]
    # Return the pipeline results and the IHT records
    return {"pipeline_results": pipeline_results, "iht": iht_results}


# -----------------------------------------------------------------------------
# Task 16, Step 3: Display, survey, and aggregation helpers
# -----------------------------------------------------------------------------

def _run_seed45_display_models(
    cfg: Dict[str, Any],
    cleansed_dfs: Dict[str, pd.DataFrame],
    output_dir: Path,
) -> Dict[str, Any]:
    """
    Produce the seed-45 display models and PR-AUC values for the survey.

    The manuscript displays seed 45, which is not one of the five
    aggregation seeds; the engine is re-run restricted to Loan and
    FraudEcom at seed 45 with a configuration copy whose seed list
    includes 45 (documented replication of the display seed).

    Parameters
    ----------
    cfg : Dict[str, Any]
        Normalised study configuration.
    cleansed_dfs : Dict[str, pd.DataFrame]
        Cleansed frames keyed by dataset name.
    output_dir : Path
        Project output root.

    Returns
    -------
    Dict[str, Any]
        Seed-45 results, fitted-models contract, and PR-AUC values.
    """
    # Build the configuration copy whose seed list includes the display seed
    display_cfg: Dict[str, Any] = deepcopy(cfg)
    # Extend the seed list with the display seed
    display_cfg["data_splitting"]["seeds"] = tuple(
        list(cfg["data_splitting"]["seeds"]) + [DISPLAY_SEED]
    )
    # Run the engine restricted to the survey datasets at seed 45
    display_results: Dict[str, Any] = run_reproduction_study(
        cfg=display_cfg,
        cleaned_inputs=cleansed_dfs,
        output_dir=output_dir,
        datasets=SURVEY_DATASETS,
        seeds=(DISPLAY_SEED,),
    )
    # Initialise the fitted-models contract and PR-AUC values
    fitted_models: Dict[str, Dict[str, Any]] = {}
    pr_auc_by_representation: Dict[str, Dict[str, float]] = {}
    # Iterate over the survey datasets
    for dataset_name in SURVEY_DATASETS:
        # Extract the seed-45 record of the dataset
        seed_record: Dict[str, Any] = display_results[dataset_name][
            str(DISPLAY_SEED)
        ]
        # Extract the fitted coefficient
        alpha: float = float(seed_record["alpha"])
        # Extract the fitted exponent vector
        beta: np.ndarray = np.asarray(seed_record["beta"], dtype=np.float64)
        # Extract the retained support from the full representation
        support: Sequence[int] = seed_record["representations"]["full"][
            "support"
        ]
        # Extract the feature names of the dataset
        feature_names: Sequence[str] = cfg["feature_matrices"][dataset_name][
            "feature_names"
        ]
        # Load the seed-45 scorecard artifact for the instrument contract
        scorecard_path: Path = (
            output_dir
            / "artifacts"
            / "scorecard"
            / "v1"
            / f"scorecard_{dataset_name.lower()}_seed{DISPLAY_SEED}.json"
        )
        # Raise when the scorecard artifact is missing
        if not scorecard_path.exists():
            raise FileNotFoundError(
                f"Seed-45 scorecard artifact missing: {scorecard_path}."
            )
        # Parse the scorecard artifact from disk
        scorecard_artifact: Dict[str, Any] = json.loads(
            scorecard_path.read_text(encoding="utf-8")
        )
        # Assemble the fitted-models contract for the survey stage
        fitted_models[dataset_name] = {
            "alpha": alpha,
            "beta": beta.tolist(),
            "support": list(support),
            "feature_names": list(feature_names),
            "points": scorecard_artifact["points"],
            "cutpoints": scorecard_artifact["cutpoints"],
            "sides": scorecard_artifact["sides"],
            "scorecard_threshold": scorecard_artifact["scorecard_threshold"],
            "tally_threshold": scorecard_artifact["tally_threshold"],
        }
        # Assemble the seed-45 PR-AUC values per representation
        pr_auc_by_representation[dataset_name] = {
            representation: float(
                seed_record["representations"][representation]["pr_auc"]
            )
            for representation in REPRESENTATIONS
        }
    # Return the display results and the survey contracts
    return {
        "display_results": display_results,
        "fitted_models": fitted_models,
        "pr_auc_by_representation": pr_auc_by_representation,
    }


def _run_survey_stage(
    cfg: Dict[str, Any],
    output_dir: Path,
    survey_records: Optional[List[Dict[str, Any]]],
    fitted_models: Dict[str, Dict[str, Any]],
    pr_auc_by_representation: Dict[str, Dict[str, float]],
) -> Dict[str, Any]:
    """
    Run the Task 13 human-assessment stage.

    When the caller supplies no survey records, the stage applies an
    EXPLICIT skip policy recorded in the result (never silent, never
    fabricated).

    Parameters
    ----------
    cfg : Dict[str, Any]
        Normalised study configuration.
    output_dir : Path
        Project output root.
    survey_records : Optional[List[Dict[str, Any]]]
        Raw survey response records, or None to skip the stage.
    fitted_models : Dict[str, Dict[str, Any]]
        Seed-45 fitted-models contract keyed by dataset.
    pr_auc_by_representation : Dict[str, Dict[str, float]]
        Seed-45 PR-AUC values keyed dataset -> representation.

    Returns
    -------
    Dict[str, Any]
        Survey stage result, or the explicit skip-policy record.
    """
    # Apply the explicit skip policy when no survey records were supplied
    if survey_records is None:
        # Return the recorded skip policy (never a silent degradation)
        return {
            "skipped": True,
            "reason": "survey data absent",
            "policy": "explicit skip: supply survey_records to run the "
            "human assessment stage",
        }
    # Run the Task 13 human-assessment orchestrator
    survey_result: Dict[str, Any] = orchestrate_human_assessment(
        cfg=cfg,
        fitted_models=fitted_models,
        pr_auc_by_representation=pr_auc_by_representation,
        raw_records=survey_records,
        output_dir=output_dir,
        seed=DISPLAY_SEED,
    )
    # Return the survey stage result
    return survey_result


def _build_figure_data(
    survey_result: Dict[str, Any],
    display_models: Dict[str, Any],
) -> Dict[str, Any]:
    """
    Assemble the figure-data dictionaries for the Task 14 persistence.

    Figures 1-2 are text panels rendered from the seed-45 display
    models; Figures 3-4 are derived from the survey analysis.

    Parameters
    ----------
    survey_result : Dict[str, Any]
        Survey stage result (must contain the analysis).
    display_models : Dict[str, Any]
        Seed-45 display models and PR-AUC values.

    Returns
    -------
    Dict[str, Any]
        Figure-data dictionaries keyed "fig1".."fig4".
    """
    # Build the Figure 1 text panel from the Loan display model
    loan_model: Dict[str, Any] = display_models["fitted_models"]["Loan"]
    # Render the pruned monomial LaTeX for the Figure 1 panel

    # Compose the Figure 1 monomial text
    fig1_text: str = render_pruned_monomial_latex(
        alpha=float(loan_model["alpha"]),
        beta=np.asarray(loan_model["beta"], dtype=np.float64),
        support=loan_model["support"],
        feature_names=loan_model["feature_names"],
    )
    # Assemble the Figure 1 data dictionary
    fig1: Dict[str, Any] = {
        "kind": "text_panel",
        "text": f"Learned model (Loan): {fig1_text}",
        "seed": DISPLAY_SEED,
    }
    # Build the Figure 2 text panels from both display models
    fraud_model: Dict[str, Any] = display_models["fitted_models"]["FraudEcom"]
    # Render the FraudEcom pruned monomial for the Figure 2 panel
    fig2_fraud_text: str = render_pruned_monomial_latex(
        alpha=float(fraud_model["alpha"]),
        beta=np.asarray(fraud_model["beta"], dtype=np.float64),
        support=fraud_model["support"],
        feature_names=fraud_model["feature_names"],
    )
    # Assemble the Figure 2 data dictionary
    fig2: Dict[str, Any] = {
        "kind": "text_panel",
        "text": (
            f"Simplifications (Loan): {fig1_text}\n\n"
            f"Simplifications (FraudEcom): {fig2_fraud_text}"
        ),
        "seed": DISPLAY_SEED,
    }
    # Extract the survey analysis for the preference and trade-off figures
    analysis: Dict[str, Any] = survey_result["analysis"]
    # Assemble the Figure 3 preference-bar data dictionary
    fig3: Dict[str, Any] = {
        "kind": "preference_bars",
        "finance_risk": analysis["preferences"]["finance_risk"],
        "aiml": analysis["preferences"]["aiml"],
    }
    # Assemble the Figure 4 trade-off data dictionary
    trade_off_points: Dict[str, List[Dict[str, float]]] = {}
    # Iterate over the survey datasets
    for dataset_name in SURVEY_DATASETS:
        # Slice the trade-off rows of the dataset
        dataset_rows = analysis["trade_off"][
            analysis["trade_off"]["dataset"] == dataset_name
        ]
        # Convert the rows to point dictionaries
        trade_off_points[dataset_name] = [
            {
                "representation": str(row["representation"]),
                "ease_gain": float(row["ease_gain"]),
                "pr_auc_loss": float(row["pr_auc_loss"]),
            }
            for _, row in dataset_rows.iterrows()
        ]
    # Assemble the Figure 4 data dictionary
    fig4: Dict[str, Any] = {"kind": "ease_vs_loss", "points": trade_off_points}
    # Return the four figure-data dictionaries
    return {"fig1": fig1, "fig2": fig2, "fig3": fig3, "fig4": fig4}


def _run_aggregation_and_documentation(
    cfg: Dict[str, Any],
    output_dir: Path,
    pipeline_results: Dict[str, Any],
    iht_results: Dict[str, Any],
    survey_result: Dict[str, Any],
    display_models: Dict[str, Any],
    figures_data: Optional[Dict[str, Any]],
    all_artifacts: Dict[str, Any],
) -> Dict[str, Any]:
    """
    Run the Task 14 aggregation stage and the Task 15 documentation.

    Parameters
    ----------
    cfg : Dict[str, Any]
        Normalised study configuration.
    output_dir : Path
        Project output root.
    pipeline_results : Dict[str, Any]
        Engine results from run_reproduction_study.
    iht_results : Dict[str, Any]
        IHT comparison records.
    survey_result : Dict[str, Any]
        Survey stage result.
    display_models : Dict[str, Any]
        Seed-45 display models and PR-AUC values.
    figures_data : Optional[Dict[str, Any]]
        Caller-supplied figure data, or None to derive from artifacts.
    all_artifacts : Dict[str, Any]
        Collected artifacts for the methodological appendix.

    Returns
    -------
    Dict[str, Any]
        Aggregation, verification, documentation, and audit results.
    """
    # Resolve the figure data: caller-supplied or derived from artifacts
    if figures_data is None:
        # Raise when the survey was skipped (figures 3-4 need its data)
        if survey_result.get("skipped", False):
            raise ValueError(
                "Figures 3 and 4 require survey data; supply survey_records "
                "or figures_data to the top-level orchestrator."
            )
        # Derive the figure data from the survey analysis and models
        figures_data = _build_figure_data(survey_result, display_models)
    # Run the Task 14 aggregation and verification orchestrator
    aggregation_result: Dict[str, Any] = orchestrate_aggregation_and_verification(
        results=pipeline_results,
        cfg=cfg,
        output_dir=output_dir,
        figures=figures_data,
        iht_results=iht_results,
    )
    # Run the Task 15 documentation and audit orchestrator
    documentation_result: Dict[str, Any] = orchestrate_documentation_and_audit(
        all_artifacts=all_artifacts,
        cfg=cfg,
        output_dir=output_dir,
        results=pipeline_results,
    )
    # Return the aggregation, verification, documentation, and audit
    return {
        "aggregation": aggregation_result["aggregated"],
        "verification": aggregation_result["verification"],
        "figures_persistence": aggregation_result["persistence"],
        "documentation": documentation_result,
    }


# -----------------------------------------------------------------------------
# Task 16, Orchestrator Function
# -----------------------------------------------------------------------------

def run_end_to_end_pipeline(
    study_configuration: Dict[str, Any],
    raw_dfs: Dict[str, pd.DataFrame],
    output_dir: Path,
    survey_records: Optional[List[Dict[str, Any]]] = None,
    figures_data: Optional[Dict[str, Any]] = None,
) -> Dict[str, Any]:
    """
    Run the complete end-to-end reproduction pipeline (Tasks 1-15).

    This top-level integrative orchestrator runs each RELEVANT
    task-specific orchestrator exactly once, in dependency order,
    without implementation duplication:

    1. Task 1 (orchestrate_validation) and Task 2 (cleanse_all).
    2. Task 3 feature builders and the Task 4 orchestrator
       (orchestrate_splitting_and_scaling) at seed 0 per dataset.
    3. Task 5 (orchestrate_hyperparameter_search) per dataset.
    4. Task 11 (run_reproduction_study) as the quantitative engine —
       it invokes the Tasks 5/7/8/9/10 orchestrators internally (the
       documented no-duplication decision) — followed by Task 12
       (orchestrate_iht_comparison) per dataset and seed.
    5. Task 13 (orchestrate_human_assessment) with the seed-45 display
       models (explicit skip policy when survey records are absent).
    6. Task 14 (orchestrate_aggregation_and_verification) and Task 15
       (orchestrate_documentation_and_audit).

    Parameters
    ----------
    study_configuration : Dict[str, Any]
        The study configuration dictionary (single source of truth).
    raw_dfs : Dict[str, pd.DataFrame]
        The four raw DataFrames keyed by dataset name.
    output_dir : Path
        Project output root; every stage persists under
        output_dir/artifacts.
    survey_records : Optional[List[Dict[str, Any]]], optional
        Raw survey response records for the Task 13 stage; when None,
        the stage is skipped under the explicit recorded skip policy.
    figures_data : Optional[Dict[str, Any]], optional
        Figure-data dictionaries keyed "fig1".."fig4" for Task 14; when
        None, they are derived from the survey analysis and the seed-45
        display models.

    Returns
    -------
    Dict[str, Any]
        Artifacts dictionary keyed by stage: "validation", "cleansing",
        "feature_matrices", "preprocessing", "hyperparameters",
        "pipeline_results", "iht", "survey", "aggregation",
        "documentation", and "artifacts_root".

    Raises
    ------
    TypeError
        If `study_configuration` is not a dict or `output_dir` is not a
        pathlib.Path.
    ConfigError, SchemaValidationError, LabelRateError
        Propagated from the Task 1 validation stage.
    ValueError
        If the survey data or figure data are required but absent.
    """
    # Validate the configuration type
    if not isinstance(study_configuration, dict):
        raise TypeError(
            "study_configuration must be a dict; "
            f"received {type(study_configuration).__name__}."
        )
    # Validate the raw frames mapping type
    if not isinstance(raw_dfs, dict):
        raise TypeError(
            f"raw_dfs must be a dict; received {type(raw_dfs).__name__}."
        )
    # Validate the output directory type
    if not isinstance(output_dir, Path):
        raise TypeError(
            f"output_dir must be a pathlib.Path; "
            f"received {type(output_dir).__name__}."
        )
    # Create the output directory when it does not exist yet
    output_dir.mkdir(parents=True, exist_ok=True)
    # Stage 1: validation and cleansing (Tasks 1-2)
    validation_stage: Dict[str, Any] = _run_validation_and_cleansing(
        raw_dfs=raw_dfs, cfg=study_configuration, output_dir=output_dir
    )
    # Extract the normalised configuration from the validation report
    cfg: Dict[str, Any] = validation_stage["validation"]["config"]
    # Extract the cleansed frames from the cleansing stage
    cleansed_dfs: Dict[str, pd.DataFrame] = validation_stage["cleansed"]
    # Stage 2: feature matrices and the seed-0 preprocessing record
    features_stage: Dict[str, Any] = _run_features_and_preprocessing(
        cleansed_dfs=cleansed_dfs, cfg=cfg, output_dir=output_dir
    )
    # Extract the static matrices from the feature stage
    static_matrices: Dict[str, Dict[str, np.ndarray]] = features_stage[
        "static_matrices"
    ]
    # Stage 3: hyperparameter search per dataset (Task 5)
    hyperparameter_stage: Dict[str, Any] = _run_hyperparameter_stage(
        cfg=cfg,
        output_dir=output_dir,
        seed0_scaled_train=features_stage["seed0_scaled_train"],
        seed0_labels=features_stage["seed0_labels"],
    )
    # Stage 4: quantitative engine (Task 11) and IHT robustness (Task 12)
    quantitative_stage: Dict[str, Any] = _run_quantitative_and_iht_stages(
        cfg=cfg,
        cleansed_dfs=cleansed_dfs,
        static_matrices=static_matrices,
        output_dir=output_dir,
        hyperparameter_records=hyperparameter_stage,
    )
    # Extract the pipeline results from the quantitative stage
    pipeline_results: Dict[str, Any] = quantitative_stage["pipeline_results"]
    # Extract the IHT comparison records from the quantitative stage
    iht_results: Dict[str, Any] = quantitative_stage["iht"]
    # Produce the seed-45 display models for the survey stage
    display_models: Dict[str, Any] = _run_seed45_display_models(
        cfg=cfg, cleansed_dfs=cleansed_dfs, output_dir=output_dir
    )
    # Stage 5: human assessment (Task 13) with the display models
    survey_stage: Dict[str, Any] = _run_survey_stage(
        cfg=cfg,
        output_dir=output_dir,
        survey_records=survey_records,
        fitted_models=display_models["fitted_models"],
        pr_auc_by_representation=display_models["pr_auc_by_representation"],
    )
    # Assemble the artifact record for the methodological appendix
    all_artifacts: Dict[str, Any] = {
        dataset_name: {
            str(seed): {
                "feature_names": list(
                    cfg["feature_matrices"][dataset_name]["feature_names"]
                ),
                "hyperparameters": hyperparameter_stage[dataset_name],
            }
            for seed in tuple(cfg["data_splitting"]["seeds"])
        }
        for dataset_name in DATASETS
    }
    # Stage 6: aggregation, verification, documentation, and audit
    documentation_stage: Dict[str, Any] = _run_aggregation_and_documentation(
        cfg=cfg,
        output_dir=output_dir,
        pipeline_results=pipeline_results,
        iht_results=iht_results,
        survey_result=survey_stage,
        display_models=display_models,
        figures_data=figures_data,
        all_artifacts=all_artifacts,
    )
    # Assemble and return the complete artifacts dictionary
    return {
        "validation": validation_stage["validation"],
        "cleansing": validation_stage["cleansing_persistence"],
        "feature_matrices": features_stage["preprocessing_persistence"],
        "preprocessing": features_stage["preprocessing_persistence"],
        "hyperparameters": hyperparameter_stage,
        "pipeline_results": pipeline_results,
        "iht": iht_results,
        "survey": survey_stage,
        "aggregation": documentation_stage["aggregation"],
        "verification": documentation_stage["verification"],
        "figures_persistence": documentation_stage["figures_persistence"],
        "documentation": documentation_stage["documentation"],
        "artifacts_root": str(output_dir / "artifacts"),
    }
````
